```python
df_titanic = sns.load_dataset('titanic')
df_titanic
```

```python
df_titanic.isna().sum()[df_titanic.isna().sum() > 0]
```

```python
conditions = [
    df_titanic["age"].isna(),
    df_titanic["age"] < 13,
    (df_titanic["age"] >= 13) & (df_titanic["age"] < 18),
    (df_titanic["age"] >= 18) & (df_titanic["age"] < 36),
    (df_titanic["age"] >= 36) & (df_titanic["age"] < 60),
    df_titanic["age"] >= 60
]

choices = ["unknown", "child", "teen", "young_adult", "adult", "senior"]

df_titanic["age_group"] = np.select(conditions, choices, default="unknown")
df_titanic
```

```python
display(df_titanic[df_titanic['embarked'].isna()])

display(df_titanic[df_titanic['embark_town'].isna()])
```

```python
counts = df_titanic["embark_town"].value_counts()
print(counts)
```

```python
counts = df_titanic["embarked"].value_counts()
print(counts)
```

```python
counts = df_titanic["age_group"].value_counts()
print(counts)
```

```python
df_titanic.fillna({'age':df_titanic['age'].median(),
                   'embarked':df_titanic['embarked'].mode()[0],
                   'deck':df_titanic['deck'].mode()[0],
                   'embark_town': 'Unknown Town'}, inplace = True)
```
