```python
# Импорт библиотек
import seaborn as sns
import pandas as pd
import scipy.stats as stats
from statsmodels.stats.weightstats import ztest
```

##### <b><font color='teal'>Задание 1</font></b>
С помощью библиотеки scipy.stats выполните независимый t-тест на наборе данных tips из seaborn. Сравните средний общий счет между курящими и некурящими.
* Если в условии задачи не сказано про равенство дисперсий, то лучше предполагать, что дисперсии НЕ равны

```python
# Загрузите набор данных tips в датафрейм df_tips
df_tips = sns.load_dataset('tips')
display(df_tips.head())
```

```python
# В объект типа Series s_no_smoker запишите данные по общему счету для некурящих клиентов
s_no_smoker = df_tips[df_tips['smoker'] == 'No']['total_bill']
```

```python
# В объект типа Series s_smoker запишите данные по общему счету для курящих клиентов
s_smoker = df_tips[df_tips['smoker'] == 'Yes']['total_bill']
```

```python
# Рассчитайте двухвыборочный t-тест с НЕ равной дисперсией
# (Синтаксис в файле "Проверка гипотез" - п. 1.3.)
t_stat_1, p_value_1 = stats.ttest_ind(s_no_smoker, s_smoker, equal_var=False)
```

```python
# Выведите результат с интерпретацией
print(f't-statistic: {t_stat_1:.4f}')
print(f'p-value: {p_value_1:.6f}')
if p_value_1 < 0.05:
    print('H0 отклоняем. Принимаем H1: средний общий счет у курящих и некурящих различается.')
else:
    print('H0 принимаем: средний общий счет у курящих и некурящих одинаков.')
```

##### <b><font color='teal'>Задание 2</font></b>
С помощью библиотеки statsmodels выполните одновыборочный z-тест для набора данных по бриллиантам из seaborn. Проверьте, значительно ли отличается средняя цена бриллиантов от $4000.

```python
# Загрузите в датафрейм df_diamonds набор данных diamonds из библиотеки seaborn
df_diamonds = sns.load_dataset('diamonds')
display(df_diamonds.head())
```

```python
# В переменную mu запишите ожидаемое значение цены бриллиантов
mu = 4000
```

```python
# В объект типа Series diamonds_prices запишите данные о ценах бриллиантов
diamonds_prices = df_diamonds['price']
```

```python
# Рассчитайте одновыборочный z-тест 
# На уроке мы рассчитывали вручную, потому что у нас были уже рассчитанные значения
# Здесь нам дается выборка целиком, поэтому используем z_test с параметрами:
# x1 - объект Series с ценами бриллиантов
# value - ожидаемое значение цены
z_stat_2, p_value_2 = ztest(x1=diamonds_prices, value=mu)
```

```python
# Выведите результат с интерпретацией
print(f'z-statistic: {z_stat_2:.4f}')
print(f'p-value: {p_value_2:.6f}')
if p_value_2 < 0.05:
    print('H0 отклоняем. Принимаем H1: средняя цена бриллиантов отличается от 4000.')
else:
    print('H0 принимаем: средняя цена бриллиантов равна 4000.')
```

##### <b><font color='teal'>Задание 3</font></b>
Используйте библиотеку scipy.stats для расчета коэффициента корреляции Пирсона между total_bill и tip в наборе данных tips.

```python
# Рассчитайте коэффициент корреляции Пирсона для переменных total_bill и tip в датафрейме df_tips
# (Синтаксис в файле "Основы корреляции" - п. 1)
corr_coef_3, p_value_3 = stats.pearsonr(df_tips['total_bill'], df_tips['tip'])
```

```python
# Выведите и интерпретируйте результат:
# 1. На основании p-value определите, можно ли доверять высчитанному коэффициенту
# 2. Если ответ на предыдущий вопрос - да, то какова сила и направленность корреляции между этими двумя переменными
print(f'Pearson correlation: {corr_coef_3:.4f}')
print(f'p-value: {p_value_3}')

if p_value_3 < 0.05:
    print('Коэффициент статистически значим, ему можно доверять.')
    if abs(corr_coef_3) < 0.3:
        strength = 'слабая'
    elif abs(corr_coef_3) < 0.7:
        strength = 'умеренная'
    else:
        strength = 'сильная'
    direction = 'положительная' if corr_coef_3 > 0 else 'отрицательная'
    print(f'Сила корреляции: {strength}. Направленность: {direction}.')
else:
    print('Коэффициент статистически незначим, делать вывод о силе и направленности нельзя.')
```
