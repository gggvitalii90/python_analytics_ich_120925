```python
import seaborn as sns
import matplotlib.pyplot as plt
```

Основываясь на датасете iris из seaborn, постройте следующий график.



```python
df_iris = sns.load_dataset('iris')
sns.set_theme(style='ticks', font_scale=1.5)

fig, ax = plt.subplots(figsize=(9.6, 7), facecolor='white')
setosa = df_iris[df_iris['species'] == 'setosa']['petal_length']
versicolor = df_iris[df_iris['species'] == 'versicolor']['petal_length']
virginica = df_iris[df_iris['species'] == 'virginica']['petal_length']


sns.histplot(
    ax=ax, 
    data = setosa, 
    binwidth = 0.05, 
    multiple='stack', 
    edgecolor=None, 
    linewidth=0,
    alpha=0.5, 
    color='#1f77b4', 
    label='setosa')

sns.histplot(
    ax=ax, 
    data = versicolor, 
    binwidth = 0.11, 
    multiple='dodge', 
    edgecolor=None, 
    linewidth=0, 
    alpha=0.5, 
    color='#ff7f0e', 
    label='versicolor')

sns.histplot(
    ax=ax, 
    data = virginica, 
    binwidth = 0.11, 
    multiple='dodge', 
    edgecolor=None, 
    linewidth=0, 
    alpha=0.5, 
    color='#2ca02c', 
    label='virginica')

ax.tick_params(axis='both', which='major', direction='out', length=6, width=1.0, colors='black')
ax.set_title('Iris Species Distribution by Petal Length')
ax.set_xlabel('Petal Length (cm)')
ax.set_ylabel('Count')
ax.legend(title=None)
ax.grid(False)


plt.show()
```
