1. Работа с временными рядами в pandas:


Напишите программу, которая:

- Создаёт DataFrame с двумя столбцами: "Дата" (в формате дд-мм-гггг) и "Значение" (целые числа).

- Преобразует столбец "Дата" в тип datetime и устанавливает его в качестве индекса DataFrame.

- Генерирует временной ряд, заполняя столбец "Значение" случайными числами.

- Выводит сгенерированный временной ряд.

- Запрашивает у пользователя начальную и конечную дату периода для фильтрации.

- Выводит временной ряд за указанный период.


Пример вывода:


Сгенерированный временной ряд:

Дата

01-01-2023    120

02-01-2023    135

...

31-12-2023    110


Введите начальную дату (дд-мм-гггг): 01-01-2023

Введите конечную дату (дд-мм-гггг): 31-01-2023


Временной ряд за указанный период:

Дата

01-01-2023    120

02-01-2023    135

...

31-01-2023    140

```python
dates = pd.date_range('2023-01-01', '2023-12-31')
values = np.random.randint(0, 1000, len(dates))
# print(dates)
df = pd.DataFrame({'Дата': dates, 'Значение': values})
df['Дата'] = pd.to_datetime(df['Дата']) # Вроде как лишняя строка, так как при создании DataFrame 'Дата' уже в формате datetime, но по задаче следующий шаг преобразовать надо
df.set_index('Дата', inplace=True)
print('Сгенерированный временной ряд:')
print(df)
print("-"*50)
start_date = pd.to_datetime(input(f'Введите начальную дату (ГГГГ-ММ-ДД) в пределах {dates.min().date()} - {dates.max().date()}: '))
end_date = pd.to_datetime(input(f'Введите конечную дату (ГГГГ-ММ-ДД) в пределах {dates.min().date()} - {dates.max().date()}: '))
filtered_df = df.loc[start_date:end_date]
print('Временной ряд за указанный период:')
print(filtered_df)
```

2. Работа с несколькими DataFrame


Напишите программу, которая:

- Создаёт DataFrame с тремя столбцами: "Товар", “Откуда”, “Куда” (в последних двух столбцах указываются города).

- Создаёт второй DataFrame с двумя столбцами: "Город", “Страна”.

- На основе слияния двух предыдущих DataFrame создает третий, в котором для каждого товара вместо городов доставки будут указаны страны. Лишних столбцов в этом DataFrame быть не должно, только "Товар", “Откуда”, “Куда” (в последних двух столбцах указываются страны)

- Добавьте в последний DataFrame новый столбец “Внутри_страны”, который принимает логические значения (True, если доставка осуществляется внутри страны).

- Вычислите, сколько в наших данных доставок внутри страны.

```python
df_city_country = pd.DataFrame({
    'Город': [
        'Berlin', 'Munich', 'Hamburg', 'Frankfurt',
        'Paris', 'Lyon', 'Marseille', 'Toulouse',
        'Madrid', 'Barcelona', 'Valencia', 'Seville',
        'Rome', 'Milan', 'Naples', 'Turin',
        'Warsaw', 'Krakow', 'Gdansk', 'Wroclaw',
        'Prague', 'Brno', 'Vienna', 'Salzburg'
    ],
    'Страна': [
        'Germany', 'Germany', 'Germany', 'Germany',
        'France', 'France', 'France', 'France',
        'Spain', 'Spain', 'Spain', 'Spain',
        'Italy', 'Italy', 'Italy', 'Italy',
        'Poland', 'Poland', 'Poland', 'Poland',
        'Czechia', 'Czechia', 'Austria', 'Austria'
    ]
})
print('Города и страны:')
print(df_city_country)
print('-' * 50)

np.random.seed(42)  
n = 20
products = ['Phone', 'Laptop', 'Headphones', 'Camera', 'Monitor', 'Keyboard', 'Tablet', 'Smartwatch']
cities = df_city_country['Город'].tolist()

df_routes = pd.DataFrame({
    'Товар': np.random.choice(products, size=n),
    'Откуда': np.random.choice(cities, size=n),
    'Куда': np.random.choice(cities, size=n)
})
print('Маршруты доставки:')
print(df_routes)
print('-' * 50)

df_country_routes = pd.merge(df_routes, df_city_country, left_on='Откуда', right_on='Город', how='left') \
    .drop(['Город', 'Откуда'], axis=1) \
    .rename(columns={'Страна': 'Откуда'})

df_country_routes = pd.merge(df_country_routes, df_city_country, left_on='Куда', right_on='Город', how='left') \
    .drop(['Город', 'Куда'], axis=1) \
    .rename(columns={'Страна': 'Куда'})

print('Маршруты доставки с указанием стран:')
print(df_country_routes)
print('-' * 50)

df_country_routes['Внутри_страны'] = df_country_routes['Откуда'] == df_country_routes['Куда']
print('Маршруты доставки с указанием, внутри страны или между странами:')
print(df_country_routes)

print('Количество маршрутов внутри страны:', df_country_routes['Внутри_страны'].sum())
```

3. Напишите программу, которая:

Запрашивает у пользователя количество строк данных для ввода.

Для каждой строки запрашивает данные по трем столбцам: "Имя", "Возраст", "Регион", разделённые пробелом.

Создает DataFrame на основе введенных данных.

Выводит общую информацию о DataFrame, включая количество записей, средний возраст, и количество уникальных регионов.

Выбирает и выводит на экран данные о самом молодом участнике.

```python
n = int(input('Введите количество строк для вывода:'))
header = ["Имя", "Возраст", "Регион"]
data = []
for  i in range(n):
    row = input('Введите данные (Имя Возраст Регион):').split()
    data.append(dict(zip(header, row)))
    # print(data[-1])
df_data = pd.DataFrame(data)
df_data["Возраст"] = df_data["Возраст"].astype(int)
```

```python
# df_data.describe()
df_data[df_data['Возраст'] == df_data['Возраст'].min()]['Имя']
```

```python
print("Общая информация: "'\n')
print("Количество записей: ", len(df_data), '\n')
print("Средний возраст: ", round(df_data['Возраст'].mean(), 2), '\n')
print("Количество уникальных регионов: ", df_data['Регион'].nunique(), '\n')

print(f"""Данные о самом молодом участнике:\n
      Имя: {df_data[df_data['Возраст'] == df_data['Возраст'].min()]['Имя'].values[0]}\n
      Возраст: {df_data['Возраст'].min()}\n
      Регион: {df_data[df_data['Возраст'] == df_data['Возраст'].min()]['Регион'].values[0]}""")
```
