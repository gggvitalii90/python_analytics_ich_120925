```python
# Задача: Создайте простое приложение Dash, которое позволяет пользователям выбрать континент из выпадающего меню 
# и динамически отображает гистограмму ожидаемой продолжительности жизни для выбранного континента, используя набор данных Gapminder.

# Инструкции:

# 1.Инициализируйте приложение Dash.
# 2. Определите макет, включающий dcc.Dropdown для выбора континента и dcc.Graph для отображения гистограммы.
# 3. Наполните выпадающие опции континентами из набора данных Gapminder.
# 4. Напишите функцию обратного вызова, которая обновляет гистограмму в зависимости от выбранного континента.
# 5. Гистограмма должна отображать распределение lifeExp для выбранного континента.
# 6. Добавьте к гистограмме соответствующие заголовки и метки осей.
```

```python
import plotly.express as px
from dash import Dash, dcc, html, Input, Output, callback
import dash_bootstrap_components as dbc
```

```python
df_gapminder = px.data.gapminder()
df_gapminder
```

```python
# 1.Инициализируйте приложение Dash.
# 2. Определите макет, включающий dcc.Dropdown для выбора континента и dcc.Graph для отображения гистограммы.
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])
header=html.H1("Гистограмма ожидаемой продолжительности жизни по континентам", 
               style={"textAlign": "center"})

dropdown = dcc.Dropdown(
    id="continent-dropdown",
    options=[
        {"label": continent, "value": continent}
        for continent in df_gapminder["continent"].unique()
    ],
    value="Asia",
    style={"width": "300px", "margin": "20px auto", "backgroundColor": "white"}
)
histogramm = dcc.Graph(id="life-exp-histogram")

footer = html.P('Copyright (c) 2026 My Dashboard by coffee', 
                style={"textAlign": "center"})
app.layout = dbc.Container(
        [
            header,
            dropdown,
            histogramm,
            footer
        ], fluid=True, style={"backgroundColor": "white", "padding": "20px"}
    )       
@app.callback(
    Output("life-exp-histogram", "figure"),
    Input("continent-dropdown", "value"),
)
def update_histogram(selected_continent):
    filtered_df = df_gapminder[df_gapminder["continent"] == selected_continent]
    fig = px.histogram(
        filtered_df,
        x="lifeExp",
        title=f"Распределение ожидаемой продолжительности жизни в {selected_continent}",
        labels={"lifeExp": "Ожидаемая продолжительность жизни"},

    )
    fig.update_layout(title_x=0.5, title_xanchor="center", title_font=dict(size=14))
    fig.update_traces(marker=dict(line=dict(color='black', width=1)))
    return fig


app.run(port=1025, debug=True)
```
