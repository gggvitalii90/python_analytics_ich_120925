```python
# Импорт бибилиотек
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Совмещение двух графиков

Изобразите на одной системе координат графики функций y=sin(x) и y=log(x+1). Используйте две оси Y: одну для первого графика, и вторую для другого.

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.log(x + 1)
plt.figure(figsize=(9, 6))
ax1 = plt.subplot()
ax1.plot(x, y1, color='blue')
ax1.set_xlabel('Axis - x')
ax1.set_ylabel('Axis - np.sin(x)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

ax2 = ax1.twinx()
ax2.plot(x, y2, color='orange')
ax2.set_ylabel('Axis - np.log(x + 1)', color='orange')
ax2.tick_params(axis='y', labelcolor='orange')
plt.title('График функций sin(x) и log(x + 1)')
plt.show()
```

Создайте из этих данных датафрейм. Сгруппируйте данные по имени игроков. Методом plot() для датафреймов постройте график минимальных, максимальных и средних очков каждого игрока на одной системе координат.

```python
data = {

    'Name': ['Анна', 'Антон', 'Мария', 'Макс', 'Макс', 'Алекс', 'Антон', 'Мария', 'Макс', 'Алекс', 'Антон', 'Анна'],

     'Points': [145, 2524, 343, 4212, 5212, 6421, 745, 842, 524, 3224, 212, 1214]      

}

df = pd.DataFrame(data)
plt.figure(figsize=(9, 5))
df.groupby('Name').min()['Points'].plot(kind='line', marker='o', color='purple')
df.groupby('Name').mean()['Points'].plot(kind='line', marker='o', color='blue')
df.groupby('Name').max()['Points'].plot(kind='line', marker='o', color='green')
plt.xlabel('Name')
plt.ylabel('Points')
plt.title('График минимальных, средних и максимальных очков по именам')
plt.legend(['Min Points', 'Mean Points', 'Max Points'])
plt.show()
```

```python
data = {
    'Name': ['Анна', 'Антон', 'Мария', 'Макс', 'Макс', 'Алекс', 'Антон', 'Мария', 'Макс', 'Алекс', 'Антон', 'Анна'],
    'Points': [145, 2524, 343, 4212, 5212, 6421, 745, 842, 524, 3224, 212, 1214]
}

df = pd.DataFrame(data)

stats = df.groupby('Name')['Points'].agg(['min', 'mean', 'max'])
stats.columns = ['Min Points', 'Mean Points', 'Max Points']

ax = stats.plot(
    kind='bar',
    figsize=(9, 5),
    width=0.8,      # общая ширина группы
    stacked=False   # важно: не stack
)

plt.xlabel('Name')
plt.ylabel('Points')
plt.title('Минимальные, средние и максимальные очки по именам')
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```
