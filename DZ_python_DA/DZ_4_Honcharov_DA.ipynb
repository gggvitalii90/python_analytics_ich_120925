1. Использование пользовательских индексов в Series:

Напишите программу, которая:

- Запрашивает у пользователя ввод чисел, разделенных пробелами.

- Преобразует введенные данные в объект Series, используя в качестве индексов буквы алфавита начиная с 'a'.

- Выводит созданный объект Series на экран.


Пример вывода:

Введите числа, разделенные пробелами: 5 15 25 35

Созданный Series:

a     5

b    15

c    25

d    35

dtype: int64

```python
list_ask = list(map(int, input("Введите список числе через пробел: ").split()))
print("Список чисел:", list_ask)
series_ask = pd.Series(list_ask, index=[chr(i) for i in range(97, 97+len(list_ask))])
print(f"Серия чисел:\n{series_ask}")
```

2. Создание и индексация DataFrame:

Напишите программу, которая:

- Запрашивает у пользователя данные для создания DataFrame. Данные включают в себя имена столбцов и строки значений, вводимые через пробел.

- Создает DataFrame с использованием введенных данных.

- Выводит созданный DataFrame на экран.

- Запрашивает у пользователя имя столбца, по которому необходимо отфильтровать данные.

- Выводит значения указанного столбца.

Пример вывода:

Введите имена столбцов: Имя Возраст Город

Введите данные (для окончания введите 'end'): Анна 25 Брюгге

Введите данные (для окончания введите 'end'): Борис 30 Мюнхен

Введите данные (для окончания введите 'end'): end


   Имя  Возраст  Город

0  Анна       25  Брюгге

1  Борис      30  Мюнхен

Введите имя столбца для отображения: Возраст

Возраст

0    25

1    30

```python
header = input("Введите названия столбцов через пробел: ").split()
rows = []

while True:
    raw = input("Введите данные для строки через пробел (для окончания введите 'end' или Enter): ")

    if raw.lower() == "end" or raw == "":
        print("Ввод данных завершен.")
        break

    data = raw.split()

    if len(data) != len(header):
        print(f"Ошибка: количество данных должно быть равно количеству столбцов ({len(header)}). Попробуйте снова.")
        continue

    rows.append(data)
display(rows)
df_from_input = pd.DataFrame(rows, columns=header)
print("DataFrame из пользовательского ввода:")
display(df_from_input)

print("-" * 50)
column = input("Введите название столбца для отображения: ")

if column in df_from_input.columns:
    print(f"Столбец {column}:")
    display(df_from_input[[column]])

else:
    print(f"Такого столбца {column} нет в DataFrame.")
```
