```python
import plotly.express as px
import pandas as pd
```

```python
# Задание 1: Расширенная диаграмма рассеяния

# Создать диаграмму рассеивания с дополнительной сложностью для визуализации взаимосвязи между двумя переменными. 
# Используйте набор данных tips, доступный в Plotly Express. Постройте график total_bill на оси x и tip на оси y. 
# Раскрасьте точки по времени (обед или ужин). Добавьте на график линию тренда и настройте размер маркеров в зависимости от размера вечеринки.

df_tips = px.data.tips()
display(df_tips)
fig = px.scatter(df_tips, 
                x='total_bill',
                y='tip',
                color='time',
                trendline='ols',
                size='size',
                width=1000,
                height=600,
                title='Scatter Plot of Total Bill vs Tip Colored by Time')
fig.show()



```

```python
# Задача 2: Создать гистограмму с расширенными возможностями настройки. Используйте набор данных tips. 
# Создайте гистограмму для total_bill. Установите число бинов на 20 и цвет по времени.
fig = px.histogram(df_tips,
                x='total_bill', 
                nbins=20,
                color='time',
                text_auto=True,
                title='Histogram of Total Bill with 20 Bins Colored by Time',)
fig.update_traces(marker=dict(line=dict(color='black', width=1)))

fig.show()
```

```python
# Задача 3: Создайте линейный график с несколькими линиями для сравнения тенденций. Используйте набор данных gapminder, доступный в Plotly Express. 
# Отметьте год на оси x и gdpPercap на оси y. Отфильтруйте данные, чтобы включить в них только страны "Канада" и "Германия". 
# Используйте разные стили линий для каждой страны.
df_gapminder = px.data.gapminder()
display(df_gapminder)
fig = px.line(df_gapminder[df_gapminder['country'].isin(['Canada', 'Germany'])],
            x='year',
            y='gdpPercap',
            color='country',
            line_dash='country',
            markers=True,
            # text='gdpPercap',
             width=1000,
             height=600,
            title='Line Plot of GDP Per Capita Over Time for Canada and Germany with Different Line Styles')
fig.show()
```

```python
# Задача 4: Создать тепловую карту для визуализации корреляции между несколькими переменными с помощью интерактивных функций. 
# Используйте набор данных iris, доступный в Plotly Express. Рассчитайте корреляционную матрицу и визуализируйте ее с помощью тепловой карты. 
# Добавьте наведенную информацию, показывающую значения корреляции.
df_iris = px.data.iris()
display(df_iris)
correlation_matrix = df_iris.corr(numeric_only=True)

fig = px.imshow(correlation_matrix,
                text_auto=True,
                title='Correlation Heatmap of Iris Dataset',
                width=800,
                height=600,
                color_continuous_scale='RdBu_r',
                zmin=-1, zmax=1
                )
fig.show()
```
