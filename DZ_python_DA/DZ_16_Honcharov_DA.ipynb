##### <b><font color='teal'>Задание 1</font></b>
Компания оценивает своих сотрудников по трем показателям: продажи, удовлетворенность клиентов и пунктуальность. Среднее значение и стандартное отклонение для каждого показателя следующие:
- Продажи: Среднее = 500, Стандартное отклонение = 50
- Удовлетворенность клиентов: Среднее = 80, Стандартное отклонение = 10
- Пунктуальность: Среднее = 95, Стандартное отклонение = 5<br>

У сотрудника следующие показатели:
- Продажи: 550
- Удовлетворенность клиентов: 85
- Пунктуальность: 90<br>
Рассчитайте Z-баллы для каждой метрики и определите, в какой метрике сотрудник показал лучшие результаты по сравнению со средним показателем по компании.


```python
# Рассчитайте по формуле z-score для продаж
# Результат запишите в переменную z_sales
# (Синтаксис в файле "Теорема Байеса_практикум" - п. "Задачи на z-score")
from scipy import stats


mean_sales = 500
std_sales = 50
x = 550
z_sales = (x - mean_sales)/std_sales
print(z_sales)
par = stats.norm.cdf(z_sales)
print(f'Вероятность получить такой показатель продаж или меньше: {par*100:.2f}%')
```

```python
# Рассчитайте по формуле z-score для удовлетворенности клиентов
# Результат запишите в переменную z_customer
mean_customer = 80
std_customer = 10
x = 85
z_customer = (x - mean_customer)/std_customer
print(z_customer)
par = stats.norm.cdf(z_customer)
print(f'Вероятность получить такой показатель удовлетворенности клиентов или меньше: {par*100:.2f}%')
```

```python
# Рассчитайте по формуле z-score для пунктуальности
# Результат запишите в переменную z_punctuality
mean_punctuality = 95
std_punctuality = 5
x = 90
z_punctuality = (x - mean_punctuality)/std_punctuality
print(z_punctuality)
par = stats.norm.cdf(z_punctuality)
print(f'Вероятность получить такой показатель пунктуальности или меньше: {par*100:.2f}%')
```

```python
# Сравните z-scores и выведите ответ на вопрос задачи
# Сотрудник в показателе удовлетворенности клиентов показал лучшие результаты чем в остальных
```

ОТВЕТ: Сотрудник в показателе удовлетворенности клиентов показал лучшие результаты чем в остальных

##### <b><font color='teal'>Задание 2</font></b>
Среднее время выполнения задания на фабрике составляет 30 минут со стандартным отклонением 5 минут. Рассчитайте вероятность того, что выполнение случайно выбранной задачи займет от 25 до 35 минут. Используйте коэффициент Z и таблицу стандартного нормального распределения.

```python
# Рассчитайте z-score для 25 минут по формуле
# Значение запишите в переменную z_25
# (Синтаксис в файле "Теорема Байеса_практикум" - п. "Задачи на z-score")
mu = 30
std = 5 
x = 25
z_25 = (x - mu)/std
print(z_25)
par = stats.norm.cdf(z_25)
print(f'Вероятность получить такой показатель или меньше: {par*100:.2f}%')
```

```python
# Рассчитайте z-score для 35 минут по формуле
# Значение запишите в переменную z_35
x = 35
z_35 = (x - mu)/std
print(z_35)
par = stats.norm.cdf(z_35)
print(f'Вероятность получить такой показатель или меньше: {par*100:.2f}%')
```

```python
# Из таблицы стандартов нормального распределения найдите значения вероятностей для z_25 и z_35
# Значения запишите в переменные p_25 и p_35
# (Делали  в презентации "Summary Lesson" - сл. 3)
p_25 = stats.norm.cdf(z_25)
display(f"p_25 = {p_25*100:.2f}%")
p_35 = stats.norm.cdf(z_35)
display(f"p_35 = {p_35*100:.2f}%")
```

```python
# Вычислите вероятность того, что выполнение случайно выбранной задачи займет от 25 до 35 минут
# (Делали  в презентации "Summary Lesson" - сл. 3)
р = p_35 - p_25
print(f'Вероятность того, что выполнение случайно выбранной задачи займет от 25 до 35 минут: {р*100:.2f}%') 
```

##### <b><font color='teal'>Задание 3</font></b>
Определенное заболевание поражает 1% населения. Тест на это заболевание имеет 99 % чувствительности (истинно положительный результат) и 2 % ложноположительных результатов. Если у человека положительный тест, какова вероятность того, что у него действительно есть болезнь?

```python
# Эта задача на теорему Байеса
# (Пример в презентации "Python for DA_16-1_M - сл. 40 - 42")

P_A = 0.01
P_B_given_A = 0.99
P_B_given_not_A = 0.02
P_not_A = 1 - P_A
P_B = P_B_given_A * P_A + P_B_given_not_A * P_not_A
print(f'Вероятность того, что тест покажет положительный результат: {P_B*100:.2f}%')
P_A_given_B = (P_B_given_A * P_A) / P_B
print(f'Вероятность того, что сотрудник действительно болен, если тест показал положительный результат: {P_A_given_B*100:.2f}%')

```
