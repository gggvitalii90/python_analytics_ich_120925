### <font color='rebeccapurple'>2. Создание простого дашборда</font>

```python
# Подключение модулей
from dash import Dash, html, dcc
import plotly.express as px

# Создание экземпляра класса Dash
app = Dash()

# Создание макета приложения
df = px.data.iris()
fig = px.scatter(df, x = 'sepal_width', y = 'sepal_length', color = 'species')
app.layout = html.Div([
                html.H1("Визуализация набора данных Iris"),
                dcc.Graph(figure = fig)
])

# Запуск приложения
app.run()
```

### <font color='rebeccapurple'>3. HTML-компоненты макета</font>

```python
from dash import Dash, html, dcc
import plotly.express as px

df = px.data.iris()
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species')


app = Dash()

# Заголовок и навигация
header = html.Header(html.H1('My Dashboard'))
navbar = html.Div([
html.H2('Navigation'),
html.Ul([
        html.Li(html.A('Главная', href='/')),
        html.Li(html.A('Google', href='http://google.com')),
        html.Li(html.A('IT Career Hub', href='http://itcareerhub.de'))
        ])
])

# Основной блок
main_content = html.Section([
        html.Article([
        html.H3('Заголовок основного контента'),
        html.P('Некоторые важные материалы или визуализации данных находятся здесь.'),
        html.Div([
        html.H3("Визуализация набора данных Iris"),
        dcc.Graph(figure=fig)
])
])
])

# Нижний колонтитул
footer = html.Footer([
html.P('Copyright (c) 2026 My Dashboard'),
])

# Сборка макета
app.layout = html.Div([
header,
navbar,
main_content,
footer
])

app.run()
```

### <font color='rebeccapurple'>4. Интерактивные визуализации данных</font>

#### <font color='rebeccapurple'>4.1. Использование выпадающего списка</font>

```python
from dash import Dash, dcc, html, Input, Output, callback

# Создаем экземпляр приложения Dash
app = Dash()

# Компонент графика
graph_component = dcc.Graph(
    id='example-graph',
    figure={}
)

# Компонент выпадающего списка
dropdown_component = dcc.Dropdown(
    id='my-dropdown',
    options=[
        {'label': 'San Francisco', 'value': 'SF'},
        {'label': 'Montreal', 'value': 'MTL'}
    ],
    value='SF'
)

# Определяем layout приложения
app.layout = html.Div([
    dropdown_component,
    graph_component,
])

# Callback для обновления графика при выборе в выпадающем списке
@callback(
    Output('example-graph', 'figure'),
    Input('my-dropdown', 'value')
)
def update_graph(selected_value):
    SF = {'x': [1, 2, 3], 'y': [4, 5, 7], 'type': 'bar', 'name': 'San Francisco'}
    MTL = {'x': [1, 2, 3], 'y': [8, 3, 6], 'type': 'bar', 'name': "Montreal"}
    BERLIN = {'x': [1, 2, 3], 'y': [2, 4, 5], 'type': 'bar', 'name': "Berlin"}
    result = {'data': [eval(selected_value), BERLIN],
              'layout': {'title': 'Dash Data Visualization'}
              }
    return result

app.run()
```

#### <font color='rebeccapurple'>4.2. Использование слайдера</font>

```python
from dash import Dash, dcc, html, Input, Output, callback

# Создаем экземпляр приложения Dash
app = Dash(__name__)

# Компонент графика
graph_component = dcc.Graph(
    id='example-graph',
    figure={}
)

# Компонент слайдера
slider_component = dcc.Slider(
    id="my_slider",
    min=0,
    max=20,
    step=0.5,
    value=10,
    marks={i: str(i) for i in range(21)}
)

# Определяем layout приложения
app.layout = html.Div([
    graph_component,
    slider_component
])

# Callback для обновления графика при выборе в выпадающем списке
@callback(
    Output('example-graph', 'figure'),
    Input('my_slider', 'value')
)
def update_graph(selected_value):
       return {'data': [
            {'x': [1, 2, 3], 'y': [4, selected_value, 2], 'type': 'line', 'name': "SF"},
            {'x': [1, 2, 3], 'y': [2, 4, selected_value//2+4], 'type': 'line', 'name': "MTL"}
        ],
        'layout': {
            'title': 'Dash Data Visualization'
        }
              }

app.run(port = 8051)
```

#### <font color='rebeccapurple'>4.3. Использование радиокнопок (переключателей)</font>

```python
from dash import Dash, html, dcc, callback, Output, Input
import plotly.express as px

df = px.data.iris()

app = Dash()

app.layout = [
              html.Div(html.H2('My First App with Graph and Controls')),
              dcc.RadioItems(options=['sepal_width', 'petal_length', 'petal_width'], value='sepal_width', id='controls-and-radio-item'),
              dcc.Graph(figure={}, id='controls-and-graph')
             ]
@callback(
    Output(component_id='controls-and-graph', component_property='figure'),
    Input(component_id='controls-and-radio-item', component_property='value')
    )
def update_graph(col_chosen):
    fig = px.scatter(df, x=col_chosen, y='sepal_length', color='species')
    return fig

app.run(port = 8051)
```
