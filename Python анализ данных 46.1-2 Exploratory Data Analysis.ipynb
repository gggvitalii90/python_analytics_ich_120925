🔧 Преобразование данных

Преобразование — это подготовка данных для моделирования: удаление лишнего, создание новых признаков, исправление ошибок, работа с типами и единицами.

✅ Ключевые действия и зачем они нужны

| **Шаг**                           | **Код**                                                     | **Зачем это делается**                                                       |
| --------------------------------- | ----------------------------------------------------------- | ---------------------------------------------------------------------------- |
| Удаление неинформативных столбцов | `data.drop(['S.No.'], axis=1)`                              | Удаляем столбец, не влияющий на прогноз (идентификатор)                      |
| Создание возраста машины          | `data['Car_Age'] = date.today().year - data['Year']`        | Возраст лучше интерпретируется моделью, чем год                              |
| Выделение бренда и модели         | `data['Brand'] = data.Name.str.split().str[0]` и т.д.       | Название модели слишком детальное, а бренд — значимая категория              |
| Исправление ошибок в бренде       | `replace()`                                                 | Приведение к единому виду: «Mini» ≠ «Mini Cooper»                            |
| Удаление исходного Name           | `del data['Name']`                                          | После извлечения нужного — исходный текст не нужен                           |
| Разделение чисел и единиц         | `str.split()` + `astype(float)`                             | Нужно отделить значения от текста (например, "18.6 kmpl")                    |
| Преобразование km/kg в kmpl       | `data['Mileage'] = data['Mileage_num'] * (1 + 0.333 * ...)` | Унификация единиц — иначе модель будет обучаться на несопоставимых значениях |
| Удаление временных колонок        | `del data['Mileage_num'], del data['Mileage_units']`        | Очистка после обработки                                                      |

```python
# 🔹 1. Разделение переменных

cat_cols = data.select_dtypes(include=['object']).columns.tolist()
num_cols = data.select_dtypes(include=np.number).columns.tolist()

print("Categorical Variables:")
print(cat_cols)
print("Numerical Variables:")
print(num_cols)


# Категориальные переменные: тип object, например "Fuel", "Transmission", "Seller_Type".

# Числовые переменные: тип int или float, например "Price", "Kilometers_Driven".
```

```python
# 🔹 2. Анализ числовых переменных

for col in num_cols:
    print(col)
    plt.figure(figsize = (15, 4))
    plt.subplot(1, 2, 1)
    data[col].hist(grid=False)
    plt.ylabel('count')
    plt.subplot(1, 2, 2)
    sns.boxplot(x=data[col])
    plt.show()

# Строим гистограмму (hist) — показывает распределение: есть ли пики, нормальность, асимметрия.

# Строим boxplot — показывает медиану, квартиль, выбросы.

# Kilometers_Driven

# Гистограмма покажет: большинство машин проехали мало, но есть редкие случаи с 500 000 км.

# Boxplot покажет: основная масса данных сжата, есть длинные «усы» — выбросы.
```

📊 Многомерный анализ (Multivariate Analysis)

Многомерный анализ — это изучение сразу нескольких переменных (в отличие от одномерного анализа), чтобы понять взаимосвязи между ними, в том числе:

какие переменные влияют на целевую переменную (например, Price)

какие переменные дублируют друг друга

есть ли скрытые закономерности

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Убедимся, что числовые колонки в нужном виде
# Преобразуем 'Engine' и 'Power' из текстового формата в числа
data['Engine'] = data['Engine'].str.extract('(\d+\.?\d*)').astype(float)
data['Power'] = data['Power'].str.extract('(\d+\.?\d*)').astype(float)

# Создаём новые логарифмированные столбцы
if 'Kilometers_Driven_log' not in data.columns:
    data['Kilometers_Driven_log'] = np.log1p(data['Kilometers_Driven'])

if 'Price_log' not in data.columns:
    data['Price_log'] = np.log1p(data['Price'])

# Удалим строки с пропущенными значениями, чтобы pairplot не упал
pairplot_cols = ['Kilometers_Driven_log', 'Mileage', 'Engine', 'Power', 'Seats', 'Price_log', 'Car_Age']
data_clean = data[pairplot_cols].dropna()

# Строим pairplot
sns.pairplot(data_clean)
plt.show()
```

```python
import math
#Теперь построим столбчатые диаграммы для анализа влияния категориальных переменных на цену автомобиля.

# Категориальные переменные с не слишком большим числом уникальных значений
cat_cols_small = [col for col in cat_cols if data[col].nunique() <= 20]

# Расчёт размеров сетки графиков
n = len(cat_cols_small)
cols = 2
rows = math.ceil(n / cols)

fig, axes = plt.subplots(rows, cols, figsize=(10, 5 * rows))
axes = axes.flatten()

for i, col in enumerate(cat_cols_small):
    data.groupby(col)['Price_log'].mean().sort_values(ascending=False).plot.bar(ax=axes[i])
    axes[i].set_title(f'Средняя Price_log по {col}')
    axes[i].tick_params(axis='x', labelrotation=45)

# Удалить пустые графики
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()

```

```python
# Многомерный анализ (multivariate analysis) — это метод анализа, при котором одновременно рассматриваются
# три и более переменных для выявления взаимосвязей, закономерностей, влияний и потенциальной избыточности (например, дублирующие признаки).

# В контексте машинного обучения или EDA (разведочного анализа данных), многомерный анализ помогает:

# понять, какие переменные наиболее важны для прогнозирования;
# выявить мультиколлинеарность (две переменные дублируют друг друга);
# улучшить интерпретируемость и упростить модель.
```

📍 5. Заполнение пропущенных значений (Handling Missing Values)

```python
# 1. Обнаружение некорректных значений

data.loc[data["Mileage"] == 0.0, 'Mileage'] = np.nan

# Заменяем подозрительные 0.0 в расходе топлива на NaN — это логично, потому что машина с расходом 0 невозможна.
```

```python
data.Mileage.isnull().sum()
data['Mileage'] = data['Mileage'].fillna(value=np.mean(data['Mileage']))


median_values = data.groupby(['Model', 'Brand'])['Seats'].transform('median')
data.loc[data['Seats'].isnull(), 'Seats'] = data['Seats'].fillna(median_values)


# что тут происходит
# → Сколько % значений пропущено?
# → Это много или мало?
```

Модели одной марки часто имеют одинаковое число мест.

Мы группируем по Model + Brand, вычисляем медиану по группе, и подставляем её.

📌 Такой подход называется групповым заполнением по агрегату — он даёт гораздо более реалистичные результаты.

| Вопрос                                                              | Обсуждение                                                                       |
| ------------------------------------------------------------------- | -------------------------------------------------------------------------------- |
| 🔍 Где и почему могут быть пропуски?                                | Ошибки ввода, неуказанные параметры, редкие модели                               |
| 📈 Какой метод заполнения выбрать?                                  | Среднее, медиана, мода, группировка, ML                                          |

| ⚠️ Как влияет заполнение на последующий анализ/модель?              | Может искажать распределения, давать ложные выводы                               |
