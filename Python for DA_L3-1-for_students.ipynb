### <font color='teal'>2. Матрицы в практических задачах</font>

<b>Задача №1</b><br>
Кондитерская производит несколько видов десертов. Для каждого необходимы мука, яйца и сахар, но в разном количестве. Необходимое количество опишем матрицей, в которой каждая строка описывает одну партию какого-то десерта, при этом первый столбец показывает количество муки в килограммах, второй столбец – количество яиц в штуках, третий – количество сахара в килограммах.

```python
# Расход продуктов  
goods = np.array([[0.5, 6, 0.1],
                  [1, 10, 0.5],
                  [0.3, 10, 0.6],
                  [0.5, 7, 0.3]])

print(goods)
```

```python
goods.T
```

```python
#Стоимость единицы продукта
prices = np.array([0.75, 0.22, 0.88])
```

```python
#Посчитайте стоимость каждого вида десерта
prices @ goods.T 
```

```python
st = goods * prices
st.sum(axis=1)
print(st.sum(axis=1))
```

<b>Задача №2</b><br>
Фирма закупает ноутбуки и стационарные компьютеры для оборудования офисов на двух этажах здания. На первом этаже планируется поставить 20 стационарных компьютеров и 20 ноутбуков. На втором этаже – 12 стационарных компьютеров и 35 ноутбуков. Компьютер стоит 850 евро, ноутбук – 1125 евро. Запишите матрицу <i>equipment</i>, описывающую количество компьютеров и ноутбуков на каждом этаже. Запишите вектор-столбец <i>prices</i> с ценами компьютеров и ноутбуков. С помощью матричных операций вычислите стоимость оборудования на каждом этаже.

```python
equipment = np.array([  [20, 20],
                        [12, 35]])
prices_equipment = np.array([850, 1125])
st_equipment = equipment * prices_equipment
print("Стоимость", st_equipment)
tot_equipment = equipment @ prices_equipment.T
print("Общая стоимость", tot_equipment)
tot2_equipment = st_equipment.sum(axis=1)
print("Общая стоимость (сумма по строкам)", tot2_equipment)
```

### <font color='teal'>3. Системы уравнений и обратная матрица</font>

<b>Система уравнений</b> – это набор уравнений с несколькими переменными.<br>
<b>Решение системы уравнений</b> – это набор значений переменных, при подстановке которых в уравнения все они обращаются в верные равенства.

\begin{cases}
   20x + 20y = 40000 \\
   12x + 35y = 50000
\end{cases}

$$A = \begin{bmatrix}
   20 & 20 \\
   12 & 35
\end{bmatrix},
X = \begin{bmatrix}
   x \\
   y
\end{bmatrix},
B = \begin{bmatrix}
   40000 \\
   50000
\end{bmatrix}$$

$$AX = B$$

$$X = \dfrac{1}{A}* B = A^{-1}* B$$

Две матрицы называются <b>взаимнообратными</b>, если их произведение равно единичной матрице.

```python
equipment = np.array([[20, 20], [12, 35]])
expenses = np.array([[40000], [50000]])
#Получение обратной матрицы для equipment
inverse = np.linalg.inv(equipment)
print("Обратная матрица для equipment:\n", inverse)
print(inverse @ expenses)
```

```python
x = inverse @ expenses
print(equipment @ x)
```

### <font color='teal'>4. Определитель</font>
<i>Определитель</i> – специальная числовая характеристика матрицы. Система уравнений с данной матрицей имеет единственное решение, только если ее определитель не равен нулю.

```python
equipment = np.array([[20, 20], [30, 30]])
inverse = np.linalg.inv(equipment)
print(inverse)
```

```python
equipment = np.array([[20, 20], [30, 30]])
# inverse = np.linalg.inv(equipment)
# print(inverse)
det = np.linalg.det(equipment)

print(det)
```

```python
# Вычислите определитель предыдущей версии матрицы
equipment = np.array([[20, 20], [12, 35]])
det = np.linalg.det(equipment)

print(det)
```

### <font color='teal'>6. Решение задач</font>

Матрица goods, как в примере выше, описывает количество муки, яиц и сахара в трех видах десертов некоторой кондитерской. Известно, что затраты на выпечку первого десерта составляют 2 евро, второго – 4 евро, третьего – 3 евро. Найдите, сколько должны стоить мука, яйца и сахар для такой себестоимости десертов.

```python
# Расход продуктов  
goods = np.array([[0.5, 6, 0.1],
                  [1, 10, 0.5],
                  [0.3, 10, 0.6]])
price_keks = np.array([2, 4, 3])
det = np.linalg.det(goods)
print(det)
inverse_goods = np.linalg.inv(goods)
print("Обратная матрица для goods:\n", inverse_goods)
print("Результат умножения обратной матрицы на вектор цен:\n", inverse_goods @ price_keks.T)
```
