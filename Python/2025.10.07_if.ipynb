```python
# Напишите программу, которая принимает три числа и определяет, являются ли они упорядоченными по
# возрастанию.
# Пример вывода:
# Введите первое число: 3
# Введите второе число: 5
# Введите третье число: 7
# Числа упорядочены по возрастанию.

x = int(input("Введите первое число: "))
y = int(input("Введите второе число: "))
z = int(input("Введите третье число: "))
if x < y < z:
    print("Числа упорядочены по возрастанию.")
elif x > y > z:
    print("Числа упорядочены по убыванию.")
else:
    print("Числовой беспорядок")
```

```python
# Напишите программу, которая рассчитывает налог на доход в зависимости от вводимой суммы и ставок:
# ● До 10,000 включительно: налог 5%.
# ● От 10,001 до 50,000 включительно: налог 10%.
# ● Свыше 50,000: налог 20%.
# Обратите внимание, что на сумму выше 10000, налог оплачивается как 5% на 10000 из суммы, а следующие
# проценты на то что превышает границу.
# Программа должна вывести налог и итоговую сумму после уплаты налога.
# Пример вывода:
# Введите доход: 45000
# Налог: 4500.0
# Итоговая сумма после уплаты налога: 40500.0

kredit = int(input("Введите доход: "))
const = 10000
pro1 = 0.05
pro2 = 0.1
pro3 = 0.2
if kredit <= 10000:
    nal = kredit * pro1
elif kredit <= 50000:                   # 10000*0.05 + (x - 10000)*0.1
    nal = const * pro1  + (kredit - const) * pro2
else:
    nal = const * pro1 + (kredit - const) * pro3
itog = kredit - nal
print("Налог: ", nal)
print("Итоговая сумма после уплаты налога: ", itog)

```
