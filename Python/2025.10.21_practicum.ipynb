```python
# Задание 1
# Перевёрнутая лестница
# Напишите программу, которая рисует перевёрнутую лестницу из звёздочек заданной высоты.
# Пример вывода:
# Введите высоту лестницы: 5
# *****
# ****
# ***
# **
# *
n = 5 
for i in range(n, 0, -1):
    print("*"*i)

```

```python
# Задание 2
# Проверка текста на ASCII
# Напишите программу, которая принимает строку от пользователя и проверяет, являются ли все символы
# строки ASCIIсимволами (диапазон кодов от 0 до 127 включительно).
# Пример вывода:
# Введите текст: Python
# Все символы ASCII True
# Введите текст: Привет
# Все символы ASCII False

x = input("Введите текст: ")
for s in x:
    if ord(s) > 127:
        print("Все символы ASCII False")
        break
else:  
    print("Все символы ASCII True")
```

```python
# Задание 3
# Вывод строки с интервалами
# Напишите программу, которая принимает строку от пользователя и выводит её символы через подчеркивание
# “_ˮ на одной строке. Обратите внимание, что в конце и в начале не должно быть лишних подчеркиваний.
# Пример вывода:
# Введите строку: Python
# P_y_t_h_o_n

x = input("Введите текст: ")
l = ""
for s in x:
    l += s 
    l += "_"  
print(l[:-1])

x = input("Введите текст: ")
for s in range(len(x)):
    print(x[s], end="")
    if s !=  len(x)-1:
        print("_", end="")
        
    
    
```

```python
# Задание 4
# Проверка скобок
# Задание: Напишите программу, которая проверяет правильность расстановки круглых скобок в строке.
# Пример вывода:
# Введите строку: (a+b)*(c-d)
# Скобки расставлены правильно? True
# Введите строку: (a+b)(*((c-d)
# Скобки расставлены правильно? False

x = input("Введите текст: ")
l = ""
bal = 0
cor = True
for s in x:
    if s == "(" or s == ")":
       l += s
print(l, end = "\n")
for s in x:
    if s == "(":
        bal += 1
    elif s == ")":
        bal -= 1
        if bal != 0:
            cor = False 
print(cor and bal == 0)

    
        
```

```python
# Задание 5
# Поиск минимального символа
# Напишите программу, которая принимает строку от пользователя и выводит символ с минимальным кодом. Не
# используйте функцию min().
# Пример вывода:
# Введите строку: Hello!
# Минимальный символ: !
s = input("Введите текст: ")
m = s[0]
for l in s:
    if l < m:
        m = l
print(m)

```

```python
# Задание 7
# Строка из диапазона Unicode
# Напишите программу, которая принимает два числа (начало и конец диапазона Unicode) и выводит строку,
# состоящую из всех символов этого диапазона.
# Пример вывода:
# Введите начальный код: 65
# Введите конечный код: 70
# Символы: ABCDEF
x = int(input("Введите начало диапазона Unicode: "))
y = int(input("Введите конец диапазона Unicode: "))
for l in range(x, y +1):
    print(chr(l), end = " ")
    

```

```python
# Задание 8
# Одинаковая сумма кодов
# Напишите программу, которая проверяет, имеют ли две строки одинаковые суммы кодов символов. Например
# буквы в 'XYZ' имеют коды 88, 89, 90, в 'BACE' коды 66, 65, 67,69, но сумма у них одинаковая – 267.
# Пример вывода:
# Введите первую строку: acc
# Введите вторую строку: egg
# Одинаковый сдвиг? True
x = input("Введите строку 1: ")
y = input("Введите строку 2: ")
d_x = 0
d_y = 0
for i in x:
    d_x += ord(i)
for i in y:
    d_y += ord(i)   
print(d_x, d_y, sep = "\n")
print( True if d_x == d_y else False)

```

```python
# Задание 9
# Длина слов
# Напишите программу, которая принимает строку, разделяет её на слова по пробелам и выводит каждое слово
# с его длиной.
# Пример вывода:
# Введите строку: Программирование это весело
# Слово: Программирование, Длина: 16
# Слово: это, Длина: 3
# Слово: весело, Длина: 6

x = input("Введите строку: ")
sl = ""
for i in x:
    if i != " ":
        sl += i
    elif sl != "":
        print("Слово: " + sl + ", Длина:" + str(len(sl)))
        sl = ""
if sl != "":
    print("Слово: " + sl + ", Длина:" + str(len(sl)))        
   
```

```python
# Задание 10
# Текст палиндром
# Напишите программу, которая принимает строку и проверяет, является ли она палиндромом, игнорируя всё
# кроме букв русского и английского алфавита.
# Пример вывода:
# Введите строку: А роза упала на лапу Азора
# Палиндром? True

x = input("Введите строку: ")
sl = ""
for i in x:
    if ('а' <= i.lower() <= 'я') or ('a' <= i.lower() <= 'z'):
        sl += i.lower()
print(True if sl == sl[::-1] else False)
```

```python
# Задание 11
# Все буквы алфавита
# Напишите программу, которая проверяет, содержит ли введённая строка все буквы английского алфавита (без
# учёта регистра).
# Пример вывода:
# Введите строку: The quick brown fox jumps over the lazy dog
# Содержит все буквы алфавита? True
baza = 0
st1 = 0
st = ""
x = input("Введите строку: ")
for i in x:
    if ('a' <= i.lower() <= 'z'):
        if i.lower() not in st:
            st += i.lower()
            st1 += ord(i.lower())
    elif i == " ":
        continue
for l in range(ord('a'), ord('z') + 1):
    baza += l
if st1 == baza:
    print("Ваша строка содержит все буквы английского алфавита")
else: 
    print("Не все буквы английского алфавита есть ва вашей строке")
```

```python
# Задание 12
# Сравнение строк на уникальность
# Напишите программу, которая принимает две строки и сравнивает в какой из них больше уникальных
# символов.
# Пример вывода:
# Введите первую строку: Python
# Введите вторую строку: JavaJava
# Больше уникальных символов в строке: Python
st1 = ""
st2 = ""
x = input("Введите первую строку: ")
y = input("Введите вторую строку: ")
for i in x:
    if i.lower() not in st1:
        st1 += i.lower()
for i in y:
    if i.lower() not in st2:
        st2 += i.lower()
len(st1) > len(st2)
print("Больше уникальных символов в строке:", x if len(st1) > len(st2) else y)
        
    
```

```python
# Задание 13
# Счётчик символов
# Напишите программу, которая принимает строку и подсчитывает количество английских букв, цифр и
# остальных символов в строке.
# Пример вывода:
# Введите строку: Python 3.12!
# Букв: 6
# Цифр: 3
# Остальных символов: 3
b = 0
c = 0
s = 0
x = input("Введите строку: ")
for i in x:
    if ('a' <= i.lower() <= 'z'):
        b += 1
    elif ('0' <= i <= '9'):
        c += 1 
    else:
        s += 1
print("Букв:", b, "\n" + "Цифр:", c, "\n" + "Остальных символов:", s)
```
