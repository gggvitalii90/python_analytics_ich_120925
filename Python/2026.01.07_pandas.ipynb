```python
import sqlite3
# Подключение к базе данных (файл базы данных будет создан, если не существует)
conn = sqlite3.connect('my_database.db')
# Создание курсора для выполнения SQL-запросов
cursor = conn.cursor()
```

```python
cursor.execute('''
CREATE TABLE if not exists users (
 id INTEGER PRIMARY KEY,
 name TEXT NOT NULL,
 age INTEGER,
 email TEXT
)
''')
conn.commit() # Подтверждаем изменения в базе данных
```

```python
cursor.execute(''' INSERT INTO users (name, age, email) VALUES (?, ?, ?) ''', ("John Doe",
25, "johndoe@example.com")) 
conn.commit()
```

```python
cursor.execute('SELECT * FROM users')
rows = cursor.fetchall() # Получаем все записи
for row in rows:
 print(row)
```

```python
import pandas as pd
import sqlite3
# Создание DataFrame с данными
data = {
 "Name": ["Ivan", "Olga", "Petr"],
 "Age": ["20", "30", "40"],
 "City": ["Hamburg", "Dortmund", "Berlin"],
}
df = pd.DataFrame(data)
print(df)
```

```python
# Подключаемся к базе данных (файл базы данных создается при необходимости)
conn = sqlite3.connect('students.db')
# Запись данных DataFrame в базу данных SQLite
df.to_sql('student', conn, if_exists='replace', index=False)
```

```python
cursor.execute('SELECT * FROM student')
rows = cursor.fetchall() # Получаем все записи
for row in rows:
 print(row)
```

```python
# Подключаемся к базе данных (файл базы данных создается при необходимости)
conn1 = sqlite3.connect('students.db')

# Запись данных DataFrame в базу данных SQLite
df.to_sql('t_students', conn1, if_exists='replace', index=False)
# Создание курсора для выполнения SQL-запросов
cursor1 = conn1.cursor()

cursor1.execute('SELECT * FROM t_students')
rows1 = cursor1.fetchall()  # Получаем все записи
for row in rows1:
    print(row)
```

```python
# Выполняем SQL-запрос для получения данных
df_from_db = pd.read_sql_query('SELECT * FROM t_students', conn1) 
# Печатаем полученные данные 
print(df_from_db)
```
