```python
# Напишите программу, которая позволяет пользователю переводить слова между английским и русским языками. 
# Пользователь вводит слово, программа ищет его перевод в словаре. Если слово отсутствует, программа выводит сообщение об отсутствии перевода.
# Данные:
dictionary = {"Butterfly": "Бабочка",
"Training": "Обучение",
"Restaurant": "Ресторан",
"Programming": "Программирование"}
x = input("Переведи")
for key in dictionary:
    if x == key:
        print(f"Перевод: {dictionary[key]}")
        break
    elif x == dictionary[key]:
        print(f"Перевод: {key}")
        break
else:
    print("Перевод отсутствует")
        
```

```python
dictionary = {"Butterfly": "Бабочка","Training": "Обучение","Restaurant": "Ресторан","Programming": "Программирование"}

while True:

    x = input("Введите слово для перевода или 'exit' для выхода:")
    
    if x == 'exit':
        print("Программа завершена.")
        break
    
    if x in dictionary.keys():
        print(1)
        print(dictionary[x])
    elif x in dictionary.values():
        print(2)
        for key, value in dictionary.items():
            if value == x:
                print(key)
```
