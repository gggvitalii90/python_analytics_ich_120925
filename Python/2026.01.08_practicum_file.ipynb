```python
# Напишите программу, которая:
# Запрашивает у пользователя путь.
# Определяет, является ли путь файлом, папкой или он не существует.
# Пример ввода
# Введите путь: example.txt
# Пример вывода
# example.txt — это файл.


import sys 
import os
p = input("Укажите путь к файлы: ")
if os.path.isfile(p):
    print(f"'{p}' - это файл.")
elif os.path.isdir(p):
    print(f"'{p}' - это папка.")
else:
    print(f"'{p}' - не существует.")
```

```python
def fio_age(x):
    list_x = x.split()
    if len(list_x) != 4:
        raise ValueError("Кол-во не соотвевтствуе, ожидается фамилия, имя, год рождения, курс")
    if not list_x[0].isalpha():
        raise ValueError("Фамилия должно содержать только буквы.")
    if not list_x[1].isalpha():
        raise ValueError("Имя должно содержать только буквы.")    
    if not list_x[2].isdigit():
        raise ValueError("Год рождения должен содержать только цифры.")    
    if not list_x[3].isdigit():
        raise ValueError("Курс должен содержать только цифры.")
    if not (1965 <= int(list_x[2]) <= 2026): 
        raise ValueError("Год рождения должен должен быть в диапазон 1965-2026.")
    if not (1 <= int(list_x[3]) <= 5):
        raise ValueError("Курс должен должен быть в диапазон 1-4.")
        
    return {'last_name': list_x[0], 'first_name': list_x[1], 'birth_year': list_x[2], 'course': list_x[3]}
try:        
    print(fio_age("Smith John 2030 4"))
except ValueError as e: 
    print(f"Ошибка ввода: {e}")
```
