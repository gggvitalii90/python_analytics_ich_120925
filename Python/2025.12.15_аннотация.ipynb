```python
def count_words(text: str) -> dict[str, int]:
 """Принимает строку и возвращает словарь с подсчётом каждого слова."""
 words = text.split()
 # Словарь, где ключи — строки, значения — целые числа
 return {word: words.count(word) for word in words}
```

```python
def modify_value(n: int) -> None:
 """Изменяет значение переменной внутри функции (не влияет на оригинал)."""
 print(f"До изменения в функции: {n}, id: {id(n)}")
 n += 1 # Создаётся новый объект
 print(f"После изменения в функции: {n}, id: {id(n)}")
num: int = 10
modify_value(num)
print(f"Вне функции: {num}, id: {id(num)}")
```

```python
def modify_list(lst: list) -> None:
 """Добавляет элемент в список (изменяет оригинальный объект)."""
 print(f"До изменения в функции: {lst}, id: {id(lst)}")
 lst.append(99) # Изменение оригинального списка
 print(f"После изменения в функции: {lst}, id: {id(lst)}")
my_list = [1, 2, 3]
modify_list(my_list)
print(f"Вне функции: {my_list}, id: {id(my_list)}")
```

```python
def safe_modify_list(lst: list) -> list:
 """Работает с копией списка, оставляя оригинальный неизменным."""
 copy_lst = lst.copy()
 copy_lst.append(99)
 return copy_lst
original_list = [1, 2, 3]
new_list = safe_modify_list(original_list)
print(f"Оригинал: {original_list}")
print(f"Копия: {new_list}")
```

```python
# Обработка элементов списка функцией
# Напишите функцию, которая принимает другую функцию и список произвольных элементов. Примените переданную
# функцию ко всем элементам списка и верните новый список. 
# Добавьте документацию и аннотации типов для всех параметров и возвращаемого значения.
# Данные
from typing import Callable
numbers = [1, 2, 3, 4, 5]
# Пример вывода
# [2, 4, 6, 8, 10]
def mn_2(func: Callable, value: list[int]) -> list:
    """Принимаем любой типи выводим список со значениями в квадрате"""
    return [func(num) for num in value]
def num_1(num: int):
    """Принимаем лист и возвращает значеними в квадрате в виде списка """
    return num*2
```

```python
print(mn_2(num_1, numbers))
```

```python
from typing import Callable
numbers = [1, 2, 3, 4, 5]

def qwdr_korn(func: Callable, elem: list[int]) -> list[int]:
    """Функция возвращает список из чисел возведенных в квадрат"""
    return [func(num) for num in elem]
    
```

```python
print(qwdr_korn(lambda n: n*2, numbers))
```
