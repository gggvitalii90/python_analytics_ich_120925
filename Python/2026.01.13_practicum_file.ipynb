```python
import os

base_dir = "sales_by_date"
os.makedirs(base_dir, exist_ok=True)  # главная папка для результата
for b, y, m in os.walk(r"C:\Users\Vitaliy\OneDrive\It_career_hub\Python\drive-download-20250514T182028Z-1-001\sales_by_date"):
    print(b)
    for file in m:
        file_path = os.path.join(b, file) 
        os.remove(file_path)
        # print(file_path)
with open("sales_data.txt", "r", encoding="utf-8") as file:
    for line in file:     # идём по строкам файла
        line = line.strip()  # убираем пробелы
        if not line:
            continue  # если строка пустая, идем дальше, пропуская ее

        parts = line.split(",")  # убираем запятые

        name = parts[0]
        date = parts[1]
        amount = parts[2]
        category = parts[3]
        city = parts[4]

        year, month, day = date.split("-")  # разбиваем дату

        print(f"Продажа: {date} | {category} | {amount} | {city}")    # показываем, что сейчас обрабатываем

        # year_dir = os.path.join(base_dir, year)
        # month_dir = os.path.join(year_dir, month)
        
        month_dir = os.path.join(base_dir, year, month)   # создаём путь год/месяц
        
        os.makedirs(month_dir, exist_ok=True)    # создаём папки (если их ещё нет)

        print(f"Папка: {year}/{month}")

        all_file = os.path.join(month_dir, "sales.txt")   # путь к файлу со всеми продажами месяца
        
        with open(all_file, "a", encoding="utf-8") as f_all:  # создаем файл, путь указали ранее
            f_all.write(line + "\n")

        print("→ записано в sales.txt")

        category_file = os.path.join(month_dir, f"{category}.txt")  # путь к файлу, разбивка по категориям
        
        with open(category_file, "a", encoding="utf-8") as f_cat:  # создаем файл, путь указали ранее
            f_cat.write(line + "\n")

        print(f"→ записано в {category}.txt\n")
```
