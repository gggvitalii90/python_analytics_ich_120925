```python
# Задание 1.1
# Фильтрация четных с функцией
# Напишите функцию-предикат, которая проверяет четный ли переданный элемент. Также напишите функцию,
# которая принимает функцию-предикат и список чисел. Она должна возвращать новый список, содержащий
# только элементы, для которых предикат возвращает True.
# Данные:
from typing import Callable
nums = ["1", 2, 3, 4, 5, 6]
# Пример вывода:
# [2, 4, 6]

def chet(num: int) -> bool:
    """ Проверка на четность"""
    return num % 2 == 0
def one_list(func: Callable, nums: list[int]) -> list:
    """Возвращает список элементов для проверки на четность"""
    return [num for num in nums if func(num)]
```

```python
print(chet(4))
```

```python
print(one_list(chet, nums))
```

```python
# Задание 1.2
# Фильтрация четных с filter
# Выполните те же условия, что в задаче "Фильтрация списка с функцией", но решите с помощью filter и
# lambda.
# Данные:
# nums = [1, 2, 3, 4, 5, 6]
# Пример вывода:
# 2, 4, 6


from typing import Callable
nums = [1, 2, 3, 4, 5, 6]
# Пример вывода:
# [2, 4, 6]

# def chet(num: int) -> bool:
#     """ Проверка на четность"""
#     return num % 2 == 0
def one_list(nums: list[int]) -> list:
    """Возвращает список элементов для проверки на четность"""
    return list(filter(lambda num: num % 2 == 0, nums))
```

```python
print(one_list(nums))
```

```python
# Напишите рекурсивную функцию, которая подсчитывает количество вхождений определённого слова во вложенной структуре (список списков строк).
# Данные
nested_sentences = [
["Python is great", "I love Python"],
["Python is powerful", ["Python is everywhere", "Learn Python"]],
"Coding in Python is fun"
]
word = "in"
# Пример вывода
# Количество вхождений: 6
def count_word(nested_sentences: list, word: str) -> int:
    """Функция подсчитывает количество вхождений определенного слова в сложный список"""
    count = 0 
    if nested_sentences == []:
        return count
    if isinstance(nested_sentences, str) == True:
        return nested_sentences.split().count(word)
    elif isinstance(nested_sentences, list) == True:
        for list_ in nested_sentences:
            count += count_word(list_, word)
        return count

print(count_word(nested_sentences, word))
```

```python
nested_sentences = [["Python is great", "I love Python"],
["Python is powerful", ["Python is everywhere", "Learn Python"]],
"Coding in Python is fun"
]
text1 = "Python"
def count_words(list_str: list, text1: str) -> str:
    """подсчитывает количество вхождений определённого слова и выводит количество вхождений"""
    count = 0
    if isinstance(list_str, str) == True:
        return list_str.split().count(text1)
    elif isinstance(list_str, list) == True:
        for item in list_str:
            count += count_words(item, text1)
        return count 
    return 0
```

```python
print(count_words(nested_sentences, text1))
```
