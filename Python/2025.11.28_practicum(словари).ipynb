```python
# Напишите программу, которая заменяет числовые значения в
# словаре на их строковое представление (например, 1 →
# "один"). Используйте заранее подготовленный словарь чисел.
# Данные:
# # Словарь сопоставлений
# number_to_word = {1: "один", 2: "два", 3: "три"}
# # Исходные данные
# data = {"x": 1, "y": 2, "z": 3}
# Пример вывода:
# {'x': 'один', 'y': 'два', 'z': 'три'}

number_to_word = {1: "один", 2: "два", 3: "три"}
data = {"x": 1, "y": 2, "z": 3}
for i in  data:
    if data[i] in number_to_word:
        data[i] = number_to_word[data[i]]

data
```
