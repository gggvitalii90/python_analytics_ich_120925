```python
def kw_args(**kwargs):
    for x, y in kwargs:
        print(f"{x} - {y}")

kw_args(name="Ivan", age=12, city="Бремен")
```

```python
def fibanaci(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    if n > 1:
        return fibanaci(n-1) + fibanaci(n-2)
fibanaci(11)
```
