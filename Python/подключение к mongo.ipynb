```python
from pymongo import MongoClient

# Requires the PyMongo package.
# https://api.mongodb.com/python/current

client = MongoClient('mongodb://<username>:<password>@<host>/?authSource=<database>')
filter={
    'age': 74,
    'native_country': ' Mexico'
}
project={
    'age': 1, 
    'native_country': 1, 
    '_id': 0
}
sort=list({
    'age': -1
}.items())

result = client['ich']['US_Adult_Income'].find(
  filter=filter,
  projection=project,
  sort=sort
)
for a in result:
    print(a)
```
