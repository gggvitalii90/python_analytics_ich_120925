```python
# Создайте три списка: Month – с числами от 1 до 12, Income и Outcome – оба с 12 случайными числами от 2000 до
# 10000. Запишите эти три списка как столбцы в csv файл. Затем прочитайте этот csv-файл как список словарей.
# Запишите этот список словарей в json-файл.

import csv
import json
import random
Month = list(range(1,13))
Income = []
for inc in Month:
    Income.append(random.randint(2000,10000))
Outcome = [random.randint(2000,10000) for out in Month]

with open('list_1.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(["Month", "Income", "Outcome"])
    for i in range(0, len(Month)):
        writer.writerow([Month[i], Income[i], Outcome[i]])
with open('list_1.csv') as f1:
    reader = csv.DictReader(f1)
    slov = []
    for row in reader:
        row["Month"] = int(row["Month"])
        row["Income"] = int(row["Income"])
        row["Outcome"] = int(row["Outcome"])
        slov.append(row)
    print(slov)
```

```python
with open("Month.json", "w") as file2:
    json.dump(slov, file2, indent=4)
with open("Month.json", "r") as file2:
    x = file2.read()
    print(x)
```
