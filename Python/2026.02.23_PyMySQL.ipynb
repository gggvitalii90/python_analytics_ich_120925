Добавьте к предыдущей программе возможность выбора департамента. Пусть пользователь введёт название из
списка. Далее выведите всех сотрудников этого департамента, указав имя, фамилию и должность.
Пример вывода:
Enter department: Marketing
Michael Hartstein — Marketing Manager
Pat Fay — Marketing Representative
...

```python
import pymysql
with pymysql.connect(
    host='YOUR_MYSQL_HOST',
    user='ich1',
    password='password',
    database='hr'
) as connection:
    with connection.cursor() as cursor:
        cursor.execute(
            """SELECT t_d.department_name , t_e.first_name , t_e.last_name , t_e.job_id  FROM	departments AS t_d
            JOIN employees AS t_e
            ON t_d.department_id = t_e.department_id
            """
            
        )
        dep = input("Введите название департамента: ")
        for r in cursor:
            if r[0] == dep:
                print(f"{r[1]} {r[2]} - {r[3]}")
                
```
