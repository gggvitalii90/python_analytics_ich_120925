```python
from pymongo import MongoClient

# Requires the PyMongo package.
# https://api.mongodb.com/python/current

client = MongoClient('mongodb://ich1:password@YOUR_MONGO_HOST/?readPreference=primary&ssl=false&authMechanism=DEFAULT&authSource=ich')
result = client['ich_edit']['customers'].aggregate([
    {
        '$unwind': {
            'path': '$accounts'
        }
    }, {
        '$lookup': {
            'from': 'transactions', 
            'localField': 'accounts', 
            'foreignField': 'account_id', 
            'as': 'result'
        }
    }
])
# from pprint import pprint
# for doc in result:
#     pprint(doc)
# import json

# # cursor одноразовый, поэтому сначала соберём список
# docs = list(result)

# for doc in docs:
#     # ObjectId и другие “не‑json‑типа” превратятся в строку благодаря default=str
#     print(json.dumps(doc, indent=2, default=str))

# a = collections.defaultdict
a = {}
for i in result:
    # print(i["name"], i['result'])
    j = i['result']
    total_all = 0
    for k in j:
        # total_all += k['total']
        l = k['transactions']
        for m in l:
            total_all += float(m['total'])
    if i["name"] in a:
        a[i["name"]] += total_all
    else:
        a[i["name"]] = total_all
# print( a["name"], totll ) al_a
print(f"|{"Name":^20}: {"Total":^20}|  ")
print("-" * 44)
for key, value in a.items():
    a[key] = round(value, 2)
    print(f"|{key:<20}| {a[key]:^20}|  ")
```
