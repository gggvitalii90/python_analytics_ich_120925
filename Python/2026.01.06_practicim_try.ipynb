```python
# Напишите программу, которая запрашивает у пользователя число и обрабатывает возможные ошибки ввода, пока не получат корректное число.
# Пример вывода
# Введите число: qwe
# Ошибка: Введите корректное число.
# Введите число: 12.5
# Вы ввели число: 12.5
import logging
logging.basicConfig(filename="app1.log", format="%(asctime)s - %(filename)s - %(lineno)d - %(levelname)s -%(message)s", level=logging.error)
while True:
    try: 
        x = float(input("Введите число: "))
        break
    except ValueError as e:
        print(f"Ошибка {e}. Введите корректное число")
        logging.error("Ошибка error")
print(x)        
```

```python
# Напишите функцию, которая рекурсивно вычисляет количество четных цифр числа. 
# Сделайте второй вариант этой функции – через хвостовую рекурсию.
def count_num(num: int):

    if num < 10: 
        if num % 2 == 0:
            return 1
        else: 
            return 0
    else: 
        ost = num//10
        if ost % 2 == 0:
            return  1 + count_num(ost)
count_num(222)
```

```python
# Напишите функцию, которая рекурсивно вычисляет количество четных цифр числа.

def number_q(num):
    if num < 10:
        if num % 2 == 0:
            return 1
        else:
            return 0
    else:
        num_new = num%10
        if num_new % 2 == 0:
            return 1 + number_q(num//10)
        else:
            return number_q(num//10)
number_q(200)
```

```python
def number_q(num):
    if num < 10:
        return 1 if num % 2 == 0 else 0
    else:
        last_digit = num % 10
        return (1 if last_digit % 2 == 0 else 0) + number_q(num // 10)
```

```python
# Пользователь вводит строку с информацией о студенте: фамилия, имя, год рождения, курс – через пробел. Напишите функцию, 
# которая проверяет корректность ввода: первые два элемента должны быть строками (из букв, без лишних символов), 
# вторые два элемента – целые числа в подходящем диапазоне. В случае нарушения формата функция должна выбрасывать ошибку. 
# В случае корректного ввода функция возвращает словарь с информацией о студенте. При вызове функции обработайте исключение.

# Пример ввода 1: 
# Введите информацию о студенте (фамилия имя год_рождения курс):  Smith John 2000 3
# Пример вывода: 
# Введенная информация корректна: {'last_name': 'Smith', 'first_name': 'John', 'birth_year': 2000, 'course': 3}

# Пример ввода 2: 
# Введите информацию о студенте (фамилия имя год_рождения курс):  Smith Jo.hn 2000 3
# Пример вывода: 
# Ошибка ввода: Имя должно содержать только буквы.

def fio_age(x):
    list_x = x.split()
    if len(list_x) != 4:
        raise ValueError("Кол-во не соотвевтствуе, ожидается фамилия, имя, год рождения, курс")
    if not list_x[0].isalpha():
        raise ValueError("Фамилия должно содержать только буквы.")
    if not list_x[1].isalpha():
        raise ValueError("Имя должно содержать только буквы.")    
    if not list_x[2].isdigit():
        raise ValueError("Год рождения должен содержать только цифры.")    
    if not list_x[3].isdigit():
        raise ValueError("Курс должен содержать только цифры.")
    if not (1965 <= int(list_x[2]) <= 2026): 
        raise ValueError("Год рождения должен должен быть в диапазон 1965-2026.")
    if not (1 <= int(list_x[3]) <= 5):
        raise ValueError("Курс должен должен быть в диапазон 1-4.")
        
    return {'last_name': list_x[0], 'first_name': list_x[1], 'birth_year': list_x[2], 'course': list_x[3]}
try:        
    print(fio_age("Smith Jo.hn 2000 3 5"))
except ValueError as e: 
    print(f"Ошибка ввода: {e}")
```
