```python
import sqlite3
path = 'kunden.db'
conn2 = sqlite3.connect(path)
cursor = conn2.cursor()
cursor.execute('''
    CREATE Table if not exists Employees2 (
      ID INTEGER Primary Key, 
      Name varchar(50), 
      Age INTEGER, 
      Salary decimal(10,2)
      )
'''
)
conn2.commit()

```

```python
cursor.execute('''
Insert into Employees2(Name, Age, Salary)
Values
("Petr", 30, 2000.50),
("Ivan", 35, 1000)
''')
conn2.commit()
```

```python
cursor.execute('''
INSERT OR REPLACE INTO Employees2(Name, Age, Salary)
Values
("Oleg", 30, 70000.50),
("Vlad", 35, 10001),
("Goga", 35, 11200)
''')
conn2.commit()
```

```python
cursor.execute('select * from Employees2')
rows = cursor.fetchall()
for row in rows:
  print(row)
```

```python
cursor.execute('select * from Employees2  where Age > 30')
rows = cursor.fetchall()
for row in rows:
  print(row)
```

```python
cursor.execute('select * from Employees2 where Salary > 60000')
rows = cursor.fetchall()
for row in rows:
  print(row)
```
