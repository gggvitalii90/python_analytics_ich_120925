```python
# import os
# import sys
import pymysql
import pymongo
from collections import defaultdict
import datetime
# import logging
def mysqlconn():
    connection = pymysql.connect(
        host='YOUR_MYSQL_HOST',
        user='ich1',
        password='YOUR_PASSWORD',
        database='sakila'
    )
    return connection
    
def get_film(limit, offset):
    conn1 = mysqlconn()
    with conn1.cursor() as film:
        film.execute('SELECT film_id, title FROM film limit %s OFFSET %s', (limit, offset))

   

        dict_film = defaultdict(str)
        for row in film:
            id_film = row[0]
            name_film = row[1]
            # print(row)
            dict_film[id_film] = name_film

        return dict_film


def mongoconn():
    client = pymongo.MongoClient("mongodb://<username>:<password>@<host>/?authSource=<database>")
    db = client['ich_edit']
    collection = db['120925_Honcharov_film']
    return collection

def log_write(id_, name_film):
    log_line = {"id": id_, "name_film": name_film, "timestamp": datetime.datetime.now()}
    coll = mongoconn()
    coll.insert_one(log_line)
    print(f'Лог записан: {log_line}')
```

```python
offset = 0
limit = 20
films = get_film(limit, offset)
for id_, name_film in films.items():
    
    step = input(f"Сделайте свой выбор или нажмите '*' для продолжения: ")
    if step == '*':
        offset = offset + limit
        films = get_film(limit, offset)
        for id_, name_film in films.items():
            print(f'ID: {id_}, Name: {name_film}')
```

```python
offset = 0
limit = 20
while True:
    par = int(input("""
        Начнем выбор фильмов? 
        1 - Показать список фильмов 
        0 - Выход"""))
    if par == 0:
        break
    elif par == 1:
        while True:
            dict_film = get_film(limit, offset)
            if not dict_film:
                print("Фильмов больше нет.")
                break
            for id_, name_film in dict_film.items():
                print(f'{id_} - {name_film}')
            step = input("Введите ID фильма, '*' для следующей страницы или 'e' для выхода: ")
            if step == '*':
                offset += limit
                continue
            if step.lower() == 'e':
                break
            try:
                id_ = int(step)
            except ValueError:
                print("Пожалуйста, введите корректный ввод.")
                continue
            if id_ in dict_film:
                print(f'Вы выбрали фильм: {dict_film[id_]}')
                log_write(id_, dict_film[id_])
                break
            else:
                print(f'Фильм с таким ID {id_} не найден')
                continue
        # вернём offset к началу на случай повторного входа в меню
        offset = 0
        continue
    else:
        print("Выбери пункт меню 0 или 1")
        continue
```
