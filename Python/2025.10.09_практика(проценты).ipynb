```python
# Напишите программу, которая получает от
# пользователя цену на товар и количество товара.
# Программа должна рассчитать итоговую сумму
# заказа со скидкой, исходя из количества товара.
# Условия скидки:
# Если количество товара от 10 до 19 — скидка 5%.
# Если количество товара от 20 до 49 — скидка 10%.
# Если количество товара от 50 до 99 — скидка 15%.
# Если количество товара 100 и более — скидка 20%.

x = int(input("Введите цену товара: "))
y = int(input("Введите кол-во товара: "))
if 10 <= y <= 19:
    d = 0.05
elif  20 <= y <= 49:
    d = 0.1
elif  50 <= y <= 99:
    d = 0.1
elif  y > 100:
    d = 0.2
else:
    d = 0
res = x * y * (1-d)
print(round(res, 2))
```
