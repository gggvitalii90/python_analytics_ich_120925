```python
# У вас есть список сотрудников с их возрастами и зарплатами. Напишите программу, которая:
# ● Фильтрует сотрудников старше 30 лет.
# ● Увеличивает зарплату отфильтрованных сотрудников на 20%.
# ● Возвращает обновлённый список сотрудников.
# Пример вывода:
# [{'name': 'Bob', 'age': 35, 'salary': 72000.0},
# {'name': 'Charlie', 'age': 40, 'salary': 84000.0},
# {'name': 'Eve', 'age': 45, 'salary': 96000.0}]
# Данные о сотрудниках
# Данные:
employees = [
 {"name": "Alice", "age": 25, "salary": 50000},
 {"name": "Bob", "age": 35, "salary": 60000},
 {"name": "Charlie", "age": 40, "salary": 70000},
 {"name": "Diana", "age": 28, "salary": 55000},
 {"name": "Eve", "age": 45, "salary": 80000}
]
result = list(map(lambda x: {"name": x["name"], "age": x["age"], "salary": x["salary"] * 1.2}, filter(lambda x: x["age"] > 30, employees)))

print(result)
```

```python
# Популярные слова
# Реализуйте функцию, которая принимает любое количество строк с текстом.
# Функция должна возвращать подсчет самых популярных слов в количестве, переданном в функцию. 
# Программа должна игнорировать регистр слов. Выведите 5 самых популярных слов и их количество.
# Данные
text1 = "This is a sample text with some repeated words."
text2 = "Another sample text with different words."
text3 = "Text processing is fun when words repeat."

# Пример вывода
# 5 самых популярных слов:
#  words: 3
#  text: 3
#  with: 2
#  sample: 2
#  is: 2
from collections import Counter, defaultdict

supper_text = text1.replace(".", "") + " " + text2.replace(".","") + " " + text3.replace(".","")
counter = dict(Counter(supper_text.lower().split()))
sorted_counter = sorted(counter.items(), key= lambda x: (x[1], x[0]), reverse=True)
top5 = sorted_counter[:5]
for word, quant in top5:
    print(f"{word}: {quant}")
```

```python
# Группировка слов по длине
# Реализуйте функцию, которая группирует слова по их длине и возвращает словарь с ними.
# Данные
words = ["apple", "banana", "kiwi", "grape", "orange", "peach"]

# Пример вывода
# Слова по длине:
# 5: ['apple', 'grape', 'peach']
# 6: ['banana', 'orange']
# 4: ['kiwi']
def group_f2(words):
    new_text = defaultdict(list)
    for sl in words:
        new_text[len(sl)].append(sl)
    for key, value in new_text.items():
        print(f"{key}: {value}")

```

```python
group_f2(words)
```
