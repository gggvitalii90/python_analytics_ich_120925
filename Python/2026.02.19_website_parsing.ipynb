```python
import requests
from bs4 import BeautifulSoup
url = 'https://www.iban.com/country-codes'
response = requests.get(url)
if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    title = soup.title.text
    print('Заголовок страницы:', title)
else:
    print('Ошибка при запросе:', response.status_code)
```

```python
print(soup.find('h1')) # Первый <h1>
print(soup.find('h1').text)
```

```python
print(soup.find('a')) # Первый <a>
print(soup.find('a').text) # Текст первого <a>
print(soup.find('a').get('href')) # Ссылка из первого <a>
print(soup.find('a')['href'])
```

```python
list_p = soup.find_all('p') # Все <p>
print(list_p)
for tag in list_p:
    print(tag.text)
```

```python
import requests
from bs4 import BeautifulSoup
html = requests.get("https://realpython.com").text
soup = BeautifulSoup(html, "html.parser")
links = soup.find_all("a")
for i in links:
    # print(i.attrs.get("href"))
    
    href = i.attrs.get("href")  #  attrs - не обязателен для href, get("href") и так найдет все ссылки
    if href[:4]=="http":
        print(href)
```

```python
import requests
from bs4 import BeautifulSoup

html = requests.get("https://realpython.com").text
soup = BeautifulSoup(html, "html.parser")

links = soup.find_all("a")
#print(links)

for x in links:
    for key,value in x.attrs.items():
        if key == 'href':
            if value[:4] == 'http':
                print(value)
```

```python
table = soup.find('table')
rows = table.find_all('tr')
for row in rows:
    cols = row.find_all('td')
    data = [col.text.strip() for col in cols]
    print(data)
```

```python
url = "http://pckneww9.beget.tech/"
response = requests.get(url)
if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    # Находим первую таблицу
    table = soup.find("table", id="table1")
    # Извлекаем заголовки
    headers = [th.text.strip() for th in table.find_all("th")]
    print("Заголовки:", headers)
    # Извлекаем строки данных
    rows = table.find_all("tr")[1:] # Пропускаем заголовок
    for row in rows:
        cols = [td.text.strip() for td in row.find_all("td")]
        print(cols)
else:
    print("Ошибка запроса:", response.status_code)
```

```python
table = soup.find("table", id="table2")
headers = [th.text.strip() for th in table.find_all("th")]
print(headers)
print("-"*100)
# Парсинг в словари
data = []
rows = table.find_all("tr")[1:] # Пропускаем строку заголовков
for row in rows:
    cols = [td.text.strip() for td in row.find_all("td")]
    print(cols)
    row_dict = dict(zip(headers, cols))
    data.append(row_dict)
# Вывод результата
print("-"*100)
for entry in data:
    print(entry)
```

```python
# Получите список всех заголовков (от <h1> до <h6>) со страницы
url = "https://contextures.com/xlsampledata01.html"
response = requests.get(url)
if response.status_code == 200:
    site = BeautifulSoup(response.text, "html.parser")
    # print(site)
    h1 = site.find_all("h1")
    print(*h1)
    print(site.find("h1").text)
    print("-"*100)
    h2 = site.find_all("h2")
    print(*h2)
    print(site.find("h2").text)
    print("-"*100)
    h3 = site.find_all("h3")
    print(*h3)
    print(site.find("h3").text)
    print("-"*100)
    h4 = site.find_all("h4")
    print(*h4)
    # print(site.find("h4").text)
    print("-"*100)
    h5 = site.find_all("h5")
    print(*h5)
    # print(site.find("h5").text)
    print("-"*100)
    h6 = site.find_all("h6")
    print(*h6)
    # print(site.find("h6").text)
    print("-"*100)
else:
    print(response.status_code)
```

```python
url = "https://contextures.com/xlsampledata01.html"
response = requests.get(url)
if response.status_code == 200:
    site = BeautifulSoup(response.text, "html.parser")
    # print(site)
    for t in ["h1", "h2", "h3", "h4", "h5", "h6"]:
        tag_h = site.find_all(t)
        # print(tag_h)
        for tag in tag_h:
            print(tag.text)
        print("-"*100)    
else:
    print(response.status_code)
```
