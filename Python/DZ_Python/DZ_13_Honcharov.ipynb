```python
# Прогрессия увеличения
# Напишите программу, которая создаёт новый кортеж, состоящий из элементов изначального в том же порядке. 
# Добавить в него только элементы, которые больше всех предыдущих значений в исходном кортеже.
# Данные:
# numbers = (3, 7, 2, 8, 5, 10, 1)
# Пример вывода:
# Кортеж по возрастанию: (3, 7, 8, 10)
numbers = (3, 7, -2, -8, 5, 10, 1)
numbers_1 = ()
e_1 = numbers[0]
for e in numbers:
    if e >= e_1:
        numbers_1 += (e,)
        e_1 = e
print(f"Кортеж по возрастанию: {numbers_1}")
```

```python
# # Длины слов в предложении
# # Напишите программу, которая принимает строку от пользователя и разбивает ее на отдельные слова. 
# Затем программа должна создать новый кортеж, содержащий длину каждого слова в исходной строке. 
# Используйте методы строк и кортежей для обработки данных.
# # Пример вывода:
# # Введите предложение: Программирование это интересно и полезно
# # Длины слов в предложении: (16, 3, 9, 1, 7)
x = input("Введите предложение: ")
k = ()
for e in x.split():
    k += (len(e),)
print(f"Длины слов в предложении: {k}")
```
