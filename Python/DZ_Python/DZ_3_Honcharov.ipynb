```python
# Сумма цифр числа
# Напишите программу, которая получит четырехзначное число от пользователя и выведет на экран сумму цифр этого числа.
# Пример вывода:
# Введите четырехзначное число: 1634
# Сумма цифр числа: 14

num = int(input("Введите четырехзначное число:"))
d4 = num % 10 # 4
num1 = num // 10 # 163
d3 = num1 % 10 # 3
num2 = num1 // 10 # 16
d2 = num2 % 10 # 6 
num3 = num2 // 10 # 1 
d1 = num3 % 10 # 1
print(d1+d2+d3+d4)
```
