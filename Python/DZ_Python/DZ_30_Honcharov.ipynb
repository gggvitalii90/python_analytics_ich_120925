```python
# Комбинации одежды
# Напишите функцию, которая принимает списки типов одежды, цветов и размеров, 
# а затем генерирует все возможные комбинации в формате "Clothe - Color - Size".
clothes = ["T-shirt", "Jeans", "Jacket"]
colors = ["Red", "Blue", "Black"]
sizes = ["S", "M", "L"]
import itertools
def gen(clothes, colors, sizes):
    return itertools.product(clothes, colors, sizes)

var = gen(clothes, colors, sizes)
# print(list(var))
for clothes, colors, sizes in var:
    print(f"{clothes} - {colors} - {sizes}")
```

```python
# Генератор уникальных элементов
# Создайте генератор, который принимает список элементов и выдаёт только уникальные значения, сохраняя порядок их появления в исходном списке.
data = [3, 1, 2, 3, 4, 1, 5, 2, 6, 7, 8, 5]
def uniq_gen(list_):
    new_list = []
    for x in list_:
        if x not in new_list:
            new_list.append(x)
            yield x
gen = uniq_gen(data)
for i in gen:
    print(i)
```
