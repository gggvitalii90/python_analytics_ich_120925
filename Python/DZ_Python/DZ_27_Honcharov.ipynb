```python
# Фильтрация по ключевому слову
# Напишите программу, которая ищет в файле все строки, содержащие указанное пользователем слово, и сохраняет их в новый файл.
# Имя нового файла формируется как <keyword>_<original_filename>.
# Если файл не существует, программа должна вывести ошибку.
# Если совпадения не найдены, новый файл не создаётся.
# Используйте файл system_log.txt.
import os

while True:
    file = input("Введите 'имя файла' для поиска или 'exit' для завершения работы: ")
    if file == "exit":
        print("Вы вышли из программы")
        break
    if not os.path.exists(file):
        print(f"Файл '{file}' не найден.")
    else:
        while True:
            word = input("Введите ключевое 'слово' а для выхода 'exit2': ")
            if word == "exit2":
                print("Вы вышли из программы")
                break            
            new_file = ""
            with open(file, "r") as f:
                name_file = f.readlines()
                # print(name_file)
                for str_n in name_file:
                    str_o = str_n
                    for sym in "(){}[]:;/,.!?-":
                        if sym in str_o:
                            str_o = str_o.replace(sym," ")
                list_str_o = str_o.lower().split()
                # print(str_n)
                if word.lower() in list_str_o:
                     new_file += str_n
                # print(new_file)
            if new_file == "":
                print(f"Слово '{word}' в файле не найдено")
            else:
                name_new_file = word+"_"+file
                with open(name_new_file, "w") as file2:
                    file2.write(new_file)
                print(f"Строки, содержащие '{word}' сохранены в {name_new_file}")
                break
        break
```

```python
# Поиск и удаление дубликатов
# Напишите программу, которая удаляет дублирующиеся строки из файла и сохраняет результат в новый файл.
# Имя нового файла формируется как unique_<original_filename>.
# Если файл не существует, программа должна вывести ошибку.
# Исходный порядок строк должен сохраниться.
# Если в файле нет дубликатов, создаётся точная копия файла.
# Используйте файл movies_to_watch.txt.

import os

while True:
    file = input("Введите 'имя файла' для поиска или 'exit' для завершения работы: ")
    if file == "exit":
        print("Вы вышли из программы")
        break
    if not os.path.exists(file):
        print(f"Файл '{file}' не найден.")
        continue
    else:
        new_file = ""
        with open(file, "r") as f:
            name_file = f.readlines()
            for str_n in name_file:
                if str_n not in new_file:
                    new_file += str_n
                else:
                    continue
            name_new_file = "unique__"+file
            with open(name_new_file, "w") as file3:
                file3.write(new_file)
            print(f"Дубликаты удалены. Уникальные строки сохранены в {name_new_file}")
    break
```
