```python
# Правильные скобки
# Что нужно сделать:
# Напишите программу, которая принимает строку, содержащую любые виды скобок — круглые (), квадратные [] и фигурные {}.
# Необходимо проверить, правильно ли они расставлены. Скобки считаются правильно расставленными, если:

# Каждая открывающая скобка имеет соответствующую закрывающую.
# Скобки закрываются в правильном порядке.
# Пример данных:
# string = "({[}])"

# Пример вывода:
# Скобки: ({[}])
# Валидны: False
# Скобки: ([({}()){}])
# Валидны: True

string = input("Скобки: ")
test = []

for x in string:
    if x in '([{':
        test.append(x)
    elif x in ')]}':
        if len(test) == 0:
            print("Валидны: False")
            break
        if (x == ')' and test[-1] == '(') or (x == ']' and test[-1] == '[') or (x == '}' and test[-1] == '{'):
            test.pop()
        else:
            print("Валидны: False")
            break
else:
    if len(test) == 0:
        print("Валидны: True")
    else:
        print("Валидны: False")
```
