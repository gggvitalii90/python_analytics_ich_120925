```python
# Сумма цифр числа
# Напишите рекурсивную функцию, которая находит сумму всех цифр числа.
# Данные:
num = 43197
# Пример вывода:
# 24
def sumnum(num: int, total: int = 0) -> int:
    if num == 0: 
         return total
    return sumnum(num // 10, total + num % 10)

print(sumnum(num))
```

```python
# Сумма вложенных чисел
# Напишите рекурсивную функцию, которая суммирует все числа во вложенных списках.
# Данные:
nested_numbers = [1, [2, 3], [4, [5, 6]], 7]
# Пример вывода:
# 28
def sumlist(nested_numbers: list, total: int = 0) -> int:
    if nested_numbers == []: # Базовый случай
        return total
    if type(nested_numbers[0]) == int:
        total += nested_numbers[0]
    else:    
        total += sumlist(nested_numbers[0])
        
    return sumlist(nested_numbers[1:], total)
 
print(sumlist(nested_numbers))
```
