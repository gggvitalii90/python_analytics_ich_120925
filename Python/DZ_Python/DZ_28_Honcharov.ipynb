```python
# Анализ курсов студентов
# Реализовать программу, которая должна:
# Прочитать файл student_courses.json, содержащий:
# имя,
# дату рождения (birth_date) в формате дд.мм.гггг,
# дату поступления (enrollment_date) в том же формате,
# список курсов.
# Вычислить:
# Общее количество студентов.
# Средний возраст на момент поступления.
# Сохранить отчёт в JSON-файл student_courses_report.json.
```

```python
import os
import json
from datetime import datetime, timedelta

# os.chdir(r"C:\Users\Vitaliy\OneDrive\It_career_hub\Python\drive-download-20250514T182028Z-1-001")
# os.getcwd()
# os.listdir()
with open('student_courses.json', "r") as file:
    st_file = file.read()
    cont_info_st = len(json.loads(st_file))
    list_age = []
    # print(cont_info_st)
    for info_st in json.loads(st_file):
        # print(info_st)
        name_st = info_st.get('name')
        data0 = datetime.strptime(info_st.get('birth_date'), "%d.%m.%Y")
        data1 = datetime.strptime(info_st.get('enrollment_date'), "%d.%m.%Y")
        # age = data1 - data0
        # print(age/365)  - переводим в года
        # age = data1.year - data0.year - не точный вариант
        # print(age)
        age = (data1.year * 10000 + data1.month * 100 + data1.day - (data0.year * 10000 + data0.month * 100 + data0.day)) // 10000 
        # пользуемся методом excel и гугл таблиц когда они переводят даты в обычные цифры чтоб делать над ними математические вычисления
        list_age += [age]
        # print(list_age)
average_age = sum(list_age) / len(list_age)
info_file = {
    "total_students": cont_info_st,
    "average_enrollment_age": average_age
}
with open("student_courses_report.json", "w", encoding="utf-8") as f:
 json.dump(info_file, f)
    
with open("student_courses_report.json", "r", encoding="utf-8") as fread: 
    open_file = fread.read()
    print(open_file)    
```
