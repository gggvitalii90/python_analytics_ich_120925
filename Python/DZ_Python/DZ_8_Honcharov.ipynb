```python
# Проверка кодировки
# Напишите программу, которая принимает от пользователя один символ и выводит его код в таблице Unicode и его принадлежность к диапазону ASCII 
# (True/False).
# Пример вывода:
# Введите символ: A  
# Код Unicode: 65  
# ASCII символ: True

x = "  "
while len(x) > 1:
    print("Вы ввели больше одного символа. Попробуйте еще")
    x = input("Введите один символ: ")
    
print("Код Unicode:", ord(x))
print("ASCII символ: True" if ord(x) < 127 else "ASCII символ: False")
```

```python
# Подстрока с заполнением
# Напишите программу, которая принимает строку и индексы начала и конца. Программа должна вывести подстроку на указанных позициях. 
# Также если конечный индекс выходит за пределы строки, программа заполняет недостающие символы символами _.
# Пример вывода:
# Введите строку: Программирование  
# Введите начальный индекс: 3  
# Введите конечный индекс: 20
# Подстрока: граммирование_____
x = input("Введите строку: ")
a = int(input("Введите начальный индекс: "))
a_abs = a if a >= 0 else len(x) + a   # если а значение с минусом это тоже самое что вся длина строки плюс а 
while True:
    b = int(input("Введите конечный индекс: "))
    b_abs = b if b >= 0 else len(x) + b
    if a_abs >= b_abs and a_abs != b_abs:
        print("Конечная позиция не может быть достигнута, так как она меньше или равна начальной.")
        continue
    else:
        break
d = len(x) - len(x[a:b])  
if d < 0:
    d = b - a - len(x)

print("Подстрока:", x[a:b] + "_"*d)
```

```python
# Подсчёт символа
# Напишите программу, которая принимает строку и символ, а затем подсчитывает, сколько раз символ встречается в строке.
# Пример вывода:
# Введите строку: Hello, world!  
# Введите символ: o  
# Символ o встречается 2 раз(а).

x = input("Введите строку: ")
y = input("Введите символ: ")
p = 0 
a = 0
b = 1
while b <= len(x):
     if x[a:b] == y:
         p += 1
     a += 1
     b += 1
print("Символ o встречается", p, "раз(а).")
```

```python
# Инвертирование строки без цифр
# Напишите программу, которая принимает строку и выводит её в обратном порядке, пропуская все цифры.
# Пример вывода:
# Введите строку: n52uFs6Inoh67ty8P
# Результат: PythonIsFun

x = input("Введите строку: ")
inv_x = "" 
i = 0
while i < len(x):
    if ('0' <= x[i] <= '9'):
         i += 1 
    else:
        inv_x += x[i]
        i += 1
print("Результат:", inv_x[::-1])
```
