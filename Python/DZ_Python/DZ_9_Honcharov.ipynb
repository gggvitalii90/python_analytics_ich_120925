```python
# Таблица умножения
# Напишите программу, которая выводит таблицу умножения для чисел от 1 до n. Где n это число, которое введет пользователь. 
# Оформите вывод так, чтобы столбцы были ровные (число ровно под числом)
# Пример вывода:
# Введите число: 5
i = 1
n = int(input("Введите число: "))
for strok in range(i, n+1):
    for stolb in range(i, n+1): 
        result = strok * stolb
        print(result, end="\t")
    print("\n")
        
```

```python
# Лестница чисел
# Напишите программу, которая принимает число n и выводит "лестницу" из чисел. Каждая строка лестницы содержит числа от 1 до номера строки.
# Пример вывода:
# Введите число: 5

n = int(input("Введите число: "))
for p in range(1, n+1):
    st = ""
    et = 1
    while et <= p:
        st += str(et) + "\t"
        et += 1
    print(st)
    print("\n")
 

```

```python
# Удаление всех повторяющихся символов
# Напишите программу, которая принимает строку и удаляет из неё все повторяющиеся символы, сохраняя только первые их вхождения.
# Пример вывода:
# Введите строку: Python programming  
# Результат: Python prgami
st = ""
x = input("Введите строку: ")
for s in x:
    for s1 in st:
        if s1 == s:
            break
    else:
        st += s
print(st)
```
