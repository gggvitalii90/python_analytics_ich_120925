```python
# Список файлов и папок
# Напишите программу, которая принимает путь к директории через аргумент командной строки и выводит:
# Отдельно список папок
# Отдельно список файлов
# Пример запуска
# python script.py /home/user/documents
# Пример вывода
# Содержимое директории '/home/user/documents':

# Папки:
# - folder1
# - folder2

# Файлы:
# - file1.txt
# - file2.txt
# - notes.docx

import sys
import os
if len(sys.argv) > 1:
    path = sys.argv[1] 
else:
    sys.exit()

print(f"Работаем с папкой: {path}")
for root, dirs, files in os.walk(path):
    print("Папки: ")
    for dir_ in dirs:
        print(f"-{dir_}")
        
    print(f"\nФайлы: ")
    for file in files:
        print(f"- {file}")
    break
```
