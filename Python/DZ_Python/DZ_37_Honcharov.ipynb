Список всех стран

Используя базу данных world, выведите названия всех стран из таблицы country. Каждое название должно отображаться с новой строки и иметь номер.



```python
import pymysql
config = {
    'host': 'YOUR_MYSQL_HOST',
    'user': 'ich1',
    'password': 'password',
    'database': 'world'
}
with pymysql.connect(**config
) as connection:
    # if connection.open:
        # print("Connection successful!")
    with connection.cursor() as cursor:
        cursor.execute(
            """SELECT CONCAT(ROW_NUMBER() OVER (ORDER BY Name), '.') AS row_num_dot,
                Name
                FROM country"""
            
        )

        for r in cursor:
                print(*r)
```

Города выбранной страны

Добавьте к предыдущей программе возможность выбора страны. Пусть пользователь должен ввести название страны. Далее выведите все города этой страны и их численность населения.



```python
with pymysql.connect(**config
) as connection:
    with connection.cursor() as cursor:
            cursor.execute(
                    """
                    SELECT ci.Name, ci.Population, co.Name AS country 
                    FROM city AS ci 
                    LEFT JOIN country AS co
                    ON ci.CountryCode = co.Code
                    """
    )
    country = input("Введите страну: ")
    for r in cursor:
        if r[2] == country:
            print(f"{r[0]} - {r[1]}")
```
