```python
# На лекции при выполнении практического задания 2 был получен файл grades.csv. Напишите программу, которая:
# Читает этот файл.
# Создает на его основе три новых файла: grades-Science.csv, grades-Math.csv, grades-Physics.csv. 
# В каждом из этих файлов только два столбца - имя и количество баллов по тому предмету, который указан в названии файла.
# Создает четвертый файл grades-info.csv. В этом файле три строки (по названиям предметов) и 
# столбцы со статистическими характеристиками оценок: среднее арифметическое, минимум, максимум, медиана, стандартное отклонение.

import os
import statistics
# os.chdir(r"C:\Users\Vitaliy\OneDrive\It_career_hub\Python\dop_mat_download")
# os.getcwd()
# os.listdir()
import csv
g_Science = []
num_Science = []
g_Math = []
num_Math = []
g_Physics = []
num_Physics = []
with open("grades.csv", "r", encoding="utf-8-sig") as file_grades:
    read_f = csv.reader(file_grades)
    for row in read_f:
        # print(row)
        if row[1] == "Science": 
            num_Science.append(int(row[2]))
            del row[1]
            del row[2]
            g_Science.append(row)
        elif row[1] == "Math":
            num_Math.append(int(row[2]))
            del row[1]
            del row[2]
            g_Math.append(row)
        elif row[1] == "Physics":
            num_Physics.append(int(row[2]))
            del row[1]
            del row[2]
            g_Physics.append(row)
with open("grades-Science.csv", 'w', newline='') as file1:
    writer = csv.writer(file1)
    writer.writerow(['name', 'grade'])
    writer.writerows(g_Science)

with (
    open("grades-Science.csv", "w", newline="") as f_sci,
    open("grades-Math.csv", "w", newline="") as f_math,
    open("grades-Physics.csv", "w", newline="") as f_phys
):
    writer_sci = csv.writer(f_sci)
    writer_math = csv.writer(f_math)
    writer_phys = csv.writer(f_phys)

    header = ['name', 'grade']
    writer_sci.writerow(header)
    writer_math.writerow(header)
    writer_phys.writerow(header)

    writer_sci.writerows(g_Science)
    writer_math.writerows(g_Math)
    writer_phys.writerows(g_Physics)
    
    # print(g_Science)
    # print("---"*30)
    # print(g_Math)
    # print("---"*30)
    # print(g_Physics)
    # print("\n")
    # print(num_Science)
    # print("---"*30)
    # print(num_Math)
    # print("---"*30)
    # print(num_Physics)
    
info_rows = [
    ['Science', statistics.mean(num_Science), min(num_Science), max(num_Science), statistics.median(num_Science), statistics.stdev(num_Science)],
    ['Math', round(statistics.mean(num_Math),2), min(num_Math), max(num_Math), statistics.median(num_Math), statistics.stdev(num_Math)],
    ['Physics', statistics.mean(num_Physics), min(num_Physics), max(num_Physics), statistics.median(num_Physics), statistics.stdev(num_Physics)]
]

with open("grades-info.csv", "w", newline="", encoding="utf-8") as f_info:
    writer = csv.writer(f_info)
    writer.writerow(['subject', 'mean', 'min', 'max', 'median', 'stdev']) 
    writer.writerows(info_rows)   
```
