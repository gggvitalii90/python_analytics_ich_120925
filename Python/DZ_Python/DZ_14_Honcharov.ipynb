```python
# Буква в начале, цифра в конце
# Напишите программу, которая создает новый список. В него необходимо добавить только те строки из исходного списка, 
# которые начинаются с буквы, а заканчиваются цифрой.
# Данные:
# strings = ["apple23", "ban1ana45", "12cherry", "grape3", "blue23berry"]
# Пример вывода:
# Строки с буквами в начале, цифрами в конце: ['apple23', 'ban1ana45', 'grape3']

strings = ["apple23", "ban1ana45", "12cherry", "grape3", "blue23berry"]
new = []
for e in strings:
    if e[0].isalpha():
        if e[-1].isdigit():
            new.append(e)
print(f"Строки с буквами в начале, цифрами в конце: {new}")
            
        
```

```python
# Удаление кратных
# Напишите программу, которая удаляет из списка все значения, кратные числу, которое вводится пользователем.
# Данные:
# numbers = [1, 3, 6, 9, 10, 12, 15, 19, 20]
# Пример вывода:
# Введите число для удаления кратных ему элементов: 3
# Список без кратных значений: [1, 10, 19, 20]

x = int(input("Введите число для удаления кратных ему элементов: "))
numbers = [1, 3, 6, 9, 10, 12, 15, 19, 20]
for i in range(len(numbers)-1,-1,-1):
    if not numbers[i] % x:
        numbers.pop(i)
print(f"Список без кратных значений: {numbers}")

```
