```python
file_name = input("Введите имя файла:")
words_num = int(input("Введите количество популярных слов:"))
try:
    with open(file_name, "r", encoding = "utf-8") as file: 
        text = file.read()
except FileNotFoundError:
    print(f"Ошибка указанный файл {file_name} не существует")
else:
    text = text.lower().replace('.', "").replace('!', "").replace(',', "")
    words = text.split() 
    count = {}
    for word in words:
        if word in count:
            count[word] += 1
        else:
            count[word] = 1
    print("Популярные слова:")
    sorted_count = sorted(count.items(), key = lambda x: x[1], reverse = True)
    for wor_ds, num_new in sorted_count:
        if words_num > 0:
            print(f"{wor_ds}:{num_new}")
        else:
            break
        words_num -= 1
```
