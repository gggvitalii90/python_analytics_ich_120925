##### <b><font color='saddlebrown'>Задача на функцию распределения</font></b>
Аналитик исследует время загрузки страниц на сайте. Время (в секундах) распределено нормально: 𝜇 = 3, 𝜎 = 0.8. 
Какова вероятность того, что страница загрузится за не более чем 4 секунды?

```python
import numpy as np
from scipy import stats
```

```python
mu = 4 
sigma = 0.8
x = 4
v= stats.norm.cdf(x, loc=mu, scale=sigma)
print(f'Вероятность: {v*100:.4f}%')
```

##### <b><font color='saddlebrown'>Задача на функцию процентной точки</font></b>
Определите, какой порог загрузки (в секундах) укладывается 95 % пользователей.

```python
n = stats.norm.ppf(0.95, loc=mu, scale=sigma)
print(f'95-й перцентиль: {n:.4f}')
```

### <b><font color='saddlebrown'>Задачи на Z-score</font></b>
Время отклика API (в миллисекундах) распределено нормально cо средним значением 200 и стандартным отклонением 20.

```python
mu = 200
std = 20
```

##### <b><font color='saddlebrown'>Z-score для определения выбросов</font></b>
Запрос занял 310 мс. Считается ли это выбросом, если порог выбросов z > 3?

```python
z_score = (310 - mu) / std
display(f"z_score: {z_score}")
```

##### <b><font color='saddlebrown'>Z-score для оценки вероятности</font></b>
Какова вероятность того, что запрос займёт более 260 мс?

```python
z_score = (260 - mu) / std
display(f"z_score: {z_score}")
probability = 1 - stats.norm.cdf(z_score)
print(f"Вероятность: {probability * 100:.2f}%")
```

##### <b><font color='saddlebrown'>Z-score для сравнения разных наборов данных</font></b>
- API A: среднее = 200 мс, σ = 40 мс, запрос = 260 мс
- API B: среднее = 500 мс, σ = 60 мс, запрос = 580 мс <br>
Какой из запросов относительно своего API был «хуже» (имел больший z-score)

```python
mu_A = 200
std_A = 40
x_A = 260

z_score_A = (x_A - mu_A) / std_A
display(f"z_score_A: {z_score_A}")

mu_B = 500
std_B = 60
x_B = 580

z_score_B = (x_B - mu_B) / std_B
display(f"z_score_B: {z_score_B}")
```
