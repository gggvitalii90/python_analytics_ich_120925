### <font color='teal'>Задание 1</font>
Напишите программу, которая:
- Запрашивает у пользователя ввод двух списков – в первом 5 элементов, во втором 4. Списки вводятся в виде чисел, разделенных пробелами.
- Объединяет введенные списки в один новый список из 9 элементов.
- Преобразует полученный список в массив NumPy.
- Вычисляет минимальное, среднее и максимальное значение для полученного массива.

*** Дополнительно реализовать контроль выполнения условия (определить поведение программы при неправильной длине введенных списков)

```python
import numpy as np

#1. Запрашиваем ввод данных от пользователя с помощью команды input()
str_1 = input("Введите 5 чисел для первого списка через пробел: ")
print(str_1)
str_2 = input("Введите 4 числа для второго списка через пробел: ") 
print(str_2)
main_list = str_1 + " " + str_2

#2. Используем команду split(), чтобы разделить числа. По умолчанию в данной команде используется разделитель - пробел
list_ = main_list.split()
# print(list_)
#3. С помощью map() преобразуем все элементы списка в числа. Помним, что все, что вводит пользователь, python считывает как строка
list_ = list(map(int, list_))
# print(list_)
#4. "Обертываем" строку в список

#Все эти действия можно объединить в одну строку


#Объединяем два списка в 1

#Преобразуем список в массив NumPy
arr = np.array(list_)
# print(arr)
#Вычисляем минимальное, среднее, максимальное для полученного массива
print("Минимальное значение:", np.min(arr))
print("Среднее значение:", np.mean(arr))
print("Максимальное значение:", np.max(arr))
```

### <font color='teal'>Задание 3</font>

```python
#Создадим массив согласно условию с помощью функции arange
arr_3 = np.arange(1, 11, 1)
print(arr_3)
print("-"*80)
#Кумулятивное произведение вычисляется подобно кумулятивной сумме, которое мы рассматривали на занятии
print(np.cumprod(arr_3))
```

Продолжите предыдущую программу. Далее она:
- Создает матрицу такого же размера, но заполненную вычисленным средним значением (назовем ее матрицей B).
- Разделите матрицу A на матрицу B - это будет матрица С.
- Создайте новую матрицу, заполненную логическими значениями – True, если элемент матрицы С больше 1, False – если меньше 1.

```python
#Создаем матрицу B с помощью функции full (первый аргумент - размерность массива, второй аргумент - чем заполнить элементы массива)
arr_B = np.full_like(arr_A, np.mean(arr_A), dtype=float)
print(arr_B)
print("-"*80)
#Делим матрицу A на B
arr_C = arr_A / arr_B
print(arr_C)
print("-"*80)
#Создаем логическую матрицу по условию элемент > 1
arr_D = arr_C > 1
print(arr_D)
```
