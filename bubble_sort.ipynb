```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            print(f"Сравниваем {arr[j]} и {arr[j+1]}")
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                print(f"Меняем местами: {arr}")
        print(f"После {i+1}-й итерации: {arr}")
    return arr
```

```python
# task #2
# START
# READ number n
# numbers i = 0, j = 0, a = 0
# FOR i = n/2, i <= n, i + 1
#     FOR j = 2, j <= n, j * 2
#         a = a + n / 2
# END
```

```python
n = int(input("Введите число n: "))
a = 0
for i in range(n // 2, n + 1):
    j = 2
    while j <= n:
        a = a + n // 2
        j = j * 2
print("Результат a:", a)
```

```python
n = 2
a = 0
for i in range(n):
    for j in range(n, i, -1):
        a = a + i + j
print(a)
```
