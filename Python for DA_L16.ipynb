# <b><font color='saddlebrown'>Условная вероятность, теорема Байеса, Z-score</font></b>

```python
# Импорт библиотек
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd
```

### <b><font color='saddlebrown'>1. Функция распределения, функция процентной точки и функция выживания</font></b>

##### <b><font color='saddlebrown'>1.1. Функция распределения (CDF)</font></b>

Функция распределения (CDF) случайной величины 𝑋 - это функция, которая дает вероятность того, что 𝑋 примет значение, меньшее или равное 𝑥.

##### <b><font color='saddlebrown'>1.2. Функция процентной точки (PPF)</font></b>

Функция процентной точки (PPF), также известная как обратная CDF, - это функция, дающая значение, ниже которого окажется заданный процент наблюдений в распределении.

Банк анализирует время обслуживания клиентов в отделении.<br>
Известно, что это время подчиняется нормальному распределению с:
- средним временем обслуживания: 10 минут
- стандартным отклонением: 2 минуты

Банк хочет узнать:
- Какое время обслуживания не превышает 95% клиентов?
- Какое время обслуживания не превышает 5% клиентов?

```python
mu = 10
sigma = 2
```

```python
# Какое время обслуживания не превышает 95% клиентов
t_95 = stats.norm.ppf(0.95, mu, sigma)
t_95
```

```python
# Какое время обслуживания не превышает 5% клиентов
t_5 = stats.norm.ppf(0.05, mu, sigma)
t_5
```

```python
# Нарисуем график ppf
percentiles = np.linspace(0, 1, 1000)
ppf_values = stats.norm.ppf(percentiles, mu, sigma)

# Строим график PPF
df = pd.DataFrame({'percentiles':percentiles, 'ppf_values':ppf_values})
fig = px.bar(data_frame = df,
            x = 'percentiles', y = 'ppf_values')
fig.show()
```

##### <b><font color='saddlebrown'>1.3. Функция выживания (SF)</font></b>

Функция выживания, также известная как функция надежности, дает вероятность того, что случайная переменная 𝑋 больше определенного значения 𝑥. 

```python
stats.norm.sf(15, loc = 10, scale = 2)
```

### <b><font color='saddlebrown'>2. Z-score</font></b>

```python
# Для набора данных 55, 60, 65, 70, 75, 80, 85, 200 вычислите z-score
```

##### <b><font color='saddlebrown'>2.1. Стандартизация данных</font></b>
Стандартизация преобразует данные в общую шкалу со средним значением 0 и стандартным отклонением 1

```python
import numpy as np

exam1_scores = [85, 90, 78, 92, 88]
exam2_scores = [75, 80, 70, 85, 78]

mean_exam1 = np.mean(exam1_scores)
std_exam1 = np.std(exam1_scores)
mean_exam2 = np.mean(exam2_scores)
std_exam2 = np.std(exam2_scores)

z_scores_exam1 = [(x - mean_exam1) / std_exam1 for x in exam1_scores]
z_scores_exam2 = [(x - mean_exam2) / std_exam2 for x in exam2_scores]

print("Z-баллы для экзамена 1:", z_scores_exam1)
print("Z-баллы для экзамена 2:", z_scores_exam2)
```

##### <b><font color='saddlebrown'>2.2. Определение выбросов</font></b>
Как правило, точки данных с Z-score больше 3 или меньше -3 считаются выброшенными.

```python
heights = [160, 162, 165, 170, 175, 180, 185, 190, 195, 200, 320]

mean_height = np.mean(heights)
std_height = np.std(heights)

z_scores_heights = [(x - mean_height) / std_height for x in heights]

outliers = [x for x, z in zip(heights, z_scores_heights) if abs(z) > 3]

print("Выбросы:", outliers)
print(z_scores_heights)
```

##### <b><font color='saddlebrown'>2.3. Оценка вероятности</font></b>
Можно определить вероятность того, что значение будет выше или ниже определенного порога.

```python
# Оцените вероятность того, что студент наберет больше 90 баллов на экзамене со средним значением 80 и стандартным отклонением 10.
import scipy.stats as stats
score = 90
mean_score = 80
std_score = 10

z_score = (score - mean_score) / std_score

probability = 1 - stats.norm.cdf(z_score)
probability_1 = 1 - stats.norm.cdf(score, mean_score, std_score)
print("Z-score:", z_score)
print("Вероятность набрать больше 90 баллов:", probability)
print("Вероятность набрать больше 90 баллов:", probability_1)
```

```python
# Пример 1: Вероятность меньше определенного Z-score
z1 = 0.25
prob_less_than_z1 = stats.norm.cdf(z1)
print(f"Вероятность меньше {z1}: {prob_less_than_z1}")

# Пример 2: Вероятность больше определенного Z-score
z2 = -0.5
prob_greater_than_z2 = 1 - stats.norm.cdf(z2)
print(f"Вероятность больше {z2}: {prob_greater_than_z2}")

# Пример 3: Вероятность между двумя Z-score
z3 = 0.4
z4 = 1
prob_between_z3_z4 = stats.norm.cdf(z4) - stats.norm.cdf(z3)
print(f"Вероятность между {z3} и {z4}: {prob_between_z3_z4}")
```
