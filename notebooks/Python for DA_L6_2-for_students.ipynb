### <font color='chocolate'>2. Настройка стилей графиков</font>

<b>linestyle</b> - стиль линии <a href="https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html" title="LineStyles">Line Styles</a> <br> 
<b>color</b> - цвет линии <a href="https://matplotlib.org/stable/gallery/color/named_colors.html" title="Colors">Colors</a> <br>

```python
plt.plot(range(10), [x**2 for x in range(10)], linestyle = 'dotted', color = 'green', marker = 'o')
plt.title("Квадратичная функция")
plt.xlabel("X axis") # Подпись оси X
plt.ylabel("Y axis") # Подпись оси Y
plt.grid(False) # Показать линии сетки
plt.show()
```

```python
# Несколько графиков на одной системе координат
plt.plot(range(10), [x**2 for x in range(10)], linestyle = '--', color = 'green', marker = 'o', label = 'Парабола')
plt.plot(range(10), [x**3 for x in range(10)], linestyle = 'dotted', color = 'blue', marker = '*', label = 'Кубическая парабола')
plt.title("Степенные функции")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid(True)
plt.legend()
plt.show()
```

##### <font color='chocolate'>3.1. Гистограммы и распределения</font>

Гистограмма (plt.hist()) — это график, который показывает распределение числовых данных.
- <b>bins</b> – задает количество бинов гистограммы. По умолчанию используется значение 10. Если вместо целого числа в аргумент bins передать кортеж значений, то они будут использованы для задания границ бинов. Таким образом можно построить гистограмму с произвольным разбиением.
- <b>color</b> – цвет гистограммы
- <b>edgecolor</b> – цвет границы бинов
- <b>range: (float, float)</b> — диапазон значений, в котором строится гистограмма. Значения за пределами заданного диапазона игнорируются.
- <b>histtype</b> – может принимать значения {'bar', 'barstacked', 'step', 'stepfilled'}. Определяет тип отрисовки гистограммы.
- <b>cumulative: bool</b> – при указании этого параметра со значением True бины гистограммы будут строиться с эффектом накопления, т.е. все предыдущие значения будут входить в текущий бин.
- <b>orientation</b> – может принимать значения 'vertical' или 'horizontal'
- <b>alpha</b> – число от 0 до 1, определяющее прозрачность гистограммы

```python
import numpy as np
data = np.random.randint(0, 1000, 1000) #Стандартное нормальное распределение
plt.hist(data, bins = 30, color = 'steelblue', edgecolor = 'black', alpha = 0.5)
plt.title("Гистограмма распределения")
plt.xlabel("Значение")
plt.ylabel("Частота")
plt.show()
```

```python
print(data)
```

```python
# Несколько гистограмм
data1 = np.random.randn(10000)
data2 = np.random.randn(2000)
plt.hist([data1, data2], bins = 20)
plt.show()
plt.hist([data1, data2], bins = 40, histtype = 'step', color = ['red','green'], label = ['data1', 'data2'])
plt.show()
```

##### <font color='chocolate'>3.2. Круговые диаграммы</font>

Круговая диаграмма (plt.pie()) или пироговая диаграмма представляет собой круг, разделенный на сектора, показывающие процентное соотношение частей к целому.
- <b>x</b> – массив с размерами долей.
- <b>explode</b> – массив, если этот параметр передан, то некоторые сектора будут отодвинуты от центра диаграммы
- <b>labels</b> – список текстовых меток долей.
- <b>colors</b> – список цветов долей.
- <b>autopct</b> – формат текстовой метки внутри доли, текст - это численное значение доли.
- <b>shadow</b> – отображение тени для диаграммы.
- <b>startangle</b> – угол поворота диаграммы против часовой стрелки относительно оси х.
- <b>radius</b> – величина радиуса диаграммы.
- <b>frame</b> – если параметр равен True, то вокруг диаграммы будет отображена рамка.

```python
labels = 'A', 'B', 'C', 'D'
sizes = [15, 30, 45, 10]
explode = (0, 0.2, 0, 0.15)  # выделяем второй сектор

plt.pie(sizes, explode = explode, labels = labels, autopct = '%1.1f%%', shadow = True, startangle = 90, frame = True)
plt.title("Круговая диаграмма")
plt.show()
```

##### <font color='chocolate'>3.3. Столбчатые диаграммы</font>

Столбчатая диаграмма (plt.bar(), plt.barh()) показывает сравнения между отдельными элементами
- <b>x</b> – массив x-координат столбцов или список подписей столбцов.
- <b>height</b> – высоты столбцов.
- <b>widht</b> – ширина столбцов (число или список).
- <b>colors</b> – цвет столбцов диаграммы.
- <b>edgecolor</b> – цвет границы столбцов.
- <b>linewidth</b> – ширина границы.
- <b>tick_label</b> – метки для столбцов.

```python
objects = ('Python', 'C++', 'Java', 'Perl', 'Scala', 'Lisp')
performance2020 = [10,8,6,4,2,1]
performance2000 = [1,10,8,4,2,1]
index = np.arange(0,2*len(objects),2)

plt.bar(index, performance2020, color = 'blue', edgecolor = 'black', width = 0.8)
plt.bar(index + 0.8, performance2000, color = 'red', edgecolor = 'blue', width = 0.8)
plt.xticks(index + 0.4, objects)
plt.title("Популярность языков программирования")
plt.xlabel("Языки")
plt.ylabel("Популярность")
plt.legend(['2020', '2000'])
plt.show()
```

### <font color='chocolate'>8. Структура matplotlib</font>

pyplot - это один из модулей библиотеки matplotlib.
Другие модули:
- figure (вся область графика)
- axes (сам график)
- axis (оси графика)
- text (текст графика)
- и др.

<b> Через эти модули мы тоже можем управлять нашим графиком. В некоторых случаях они предоставляют больше возможностей, чем pyplot. </b>

```python
# Как обратиться к этим модулям
# Способ 1
plt.plot([1, 2, 3], [4, 5, 6])

ax = plt.gca()  # Get Current Axes (получить график)

ax.set_title("Мой график")
ax.set_xlabel("X")
ax.set_ylabel("Y")
```

```python
plt.plot([1, 2, 3], [4, 5, 6])

ax = plt.gca()  # Get Current Axes (получить текущие оси)

ax.set_title("Мой график")
ax.xaxis.set_visible(False)
ax.set_ylabel("Y")
```

```python
# Как обратиться к этим модулям
# Способ 2
fig, ax = plt.subplots()

ax.plot([1, 2, 3], [4, 5, 6])
ax.set_title("График")
fig.set_facecolor('lightgreen')
```
