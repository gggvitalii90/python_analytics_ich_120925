```python
fifa = pl.read_csv('../data/Fifa.csv')
fifa.null_count()
```

```python
fifa.describe()
```

<b>2. Фильтрация и селекция </b><br>
- Оставьте только ключевые метрики: имя, возраст, позицию, текущий рейтинг (Overall), потенциал (Potential), клуб и стоимость.
- Отфильтруйте «Молодых талантов»: возраст до 21 года и потенциал выше 85.
- Сравните средний рейтинг игроков в разных клубах.

```python
fifa_kl = fifa.select(pl.col('Name', 'Age', 'Position', 'Overall_Rating', 'Future Potential', 'Team', 'Value Per M$'))
fifa_kl.head(10)
```

```python
team_top_20 = fifa.group_by("Team").agg(
    pl.col("Overall_Rating").median().alias("Median Overall Rating")
).sort("Median Overall Rating", descending=True).head(20)


```

```python
fifa_19 = fifa.filter(pl.col('Age') < 19).filter(pl.col('Future Potential') > 75).filter(pl.col("Team").is_in(team_top_20["Team"]))
fifa_19
```

```python
fifa_rt = fifa.group_by('Team').agg(pl.col('Overall_Rating').mean().alias('Mean_Rating')).sort('Mean_Rating', descending=True)
fifa_rt
```

<b>3. Конструирование признаков</b><br>
- Создайте несколько новых вычисляемых столбцов.
    - <i>Growth (Рост)</i>: Разница между Potential и Overall. Кто может прогрессировать сильнее всех?
    - <i>Value per Potential</i>: Сколько стоит единица «потенциала» игрока?
    - <i>Age Category</i>: спользуйте конструкцию pl.when().then().otherwise() для сегментации игроков на «Young», «Prime», «Veteran».

```python

fifa_01 = fifa.with_columns([
    (pl.col('Future Potential') - pl.col('Overall_Rating')).alias('Growth'),
    (pl.col('Value Per M$') / pl.col('Future Potential')).alias('Value per Potential'),
    pl.when(pl.col('Age') < 25)
    .then(pl.lit('Young'))
    .when((pl.col('Age') >= 25) & (pl.col('Age') <= 30))
    .then(pl.lit('Prime'))
    .otherwise(pl.lit('Veteran')).alias('Age Category')
])
fifa_01
```

```python
fifa_01.filter(pl.col('Growth') < 0)
```

```python
fifa_01.describe()
```

```python
name_top = fifa_01.with_columns((
    pl.col('Growth') / pl.col('Value Per M$'))
    .alias('Growth per Value')).filter(pl.col('Growth per Value') > 0).sort('Growth per Value', descending=True)
name_top
```

<b>4. Группировка и агрегация</b><br>
<br>
Посчитайте статистику по позициям (ST, GK, CM и т.д.):
- На какой позиции игроки в среднем самые дорогие?
- В каком возрасте игроки достигают пика своей рыночной стоимости?

```python
fifa_pos = fifa.group_by('Position').agg(pl.col('Value Per M$').mean().alias('Price')).sort('Price', descending=True)
fifa_pos
```

<b>5. Визуализация</b><br>
деи для графиков:
- Scatter Plot: Зависимость стоимости (Value) от рейтинга (Overall) с цветовой разбивкой по возрастной категории. Поищите «выбросы» — очень дорогих игроков с низким рейтингом (маркетинговые звезды) и наоборот.
- Histogram: Распределение разницы между потенциалом и текущим рейтингом.
- Bar Chart: Топ-10 стран с самым высоким средним потенциалом игроков.

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 6))
sns.barplot(x='Position', y='Price', data=fifa_pos.to_pandas())
plt.xticks(rotation=45)
plt.title('Average Value Per M$ by Position')
plt.xlabel('Position')
plt.ylabel('Average Value Per M$')
plt.tight_layout()
plt.show()

```
