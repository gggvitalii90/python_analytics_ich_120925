РєР°РєРёРµ РїРµСЂРµРјРµРЅРЅС‹Рµ РІР»РёСЏСЋС‚ РЅР° С†РµР»РµРІСѓСЋ РїРµСЂРµРјРµРЅРЅСѓСЋ (РЅР°РїСЂРёРјРµСЂ, Price)

РєР°РєРёРµ РїРµСЂРµРјРµРЅРЅС‹Рµ РґСѓР±Р»РёСЂСѓСЋС‚ РґСЂСѓРі РґСЂСѓРіР°

РµСЃС‚СЊ Р»Рё СЃРєСЂС‹С‚С‹Рµ Р·Р°РєРѕРЅРѕРјРµСЂРЅРѕСЃС‚Рё

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# РЈР±РµРґРёРјСЃСЏ, С‡С‚Рѕ С‡РёСЃР»РѕРІС‹Рµ РєРѕР»РѕРЅРєРё РІ РЅСѓР¶РЅРѕРј РІРёРґРµ
# РџСЂРµРѕР±СЂР°Р·СѓРµРј 'Engine' Рё 'Power' РёР· С‚РµРєСЃС‚РѕРІРѕРіРѕ С„РѕСЂРјР°С‚Р° РІ С‡РёСЃР»Р°
data['Engine'] = data['Engine'].str.extract('(\d+\.?\d*)').astype(float)
data['Power'] = data['Power'].str.extract('(\d+\.?\d*)').astype(float)

# РЎРѕР·РґР°С‘Рј РЅРѕРІС‹Рµ Р»РѕРіР°СЂРёС„РјРёСЂРѕРІР°РЅРЅС‹Рµ СЃС‚РѕР»Р±С†С‹
if 'Kilometers_Driven_log' not in data.columns:
    data['Kilometers_Driven_log'] = np.log1p(data['Kilometers_Driven'])

if 'Price_log' not in data.columns:
    data['Price_log'] = np.log1p(data['Price'])

# РЈРґР°Р»РёРј СЃС‚СЂРѕРєРё СЃ РїСЂРѕРїСѓС‰РµРЅРЅС‹РјРё Р·РЅР°С‡РµРЅРёСЏРјРё, С‡С‚РѕР±С‹ pairplot РЅРµ СѓРїР°Р»
pairplot_cols = ['Kilometers_Driven_log', 'Mileage', 'Engine', 'Power', 'Seats', 'Price_log', 'Car_Age']
data_clean = data[pairplot_cols].dropna()

# РЎС‚СЂРѕРёРј pairplot
sns.pairplot(data_clean)
plt.show()
```

```python
import math
#РўРµРїРµСЂСЊ РїРѕСЃС‚СЂРѕРёРј СЃС‚РѕР»Р±С‡Р°С‚С‹Рµ РґРёР°РіСЂР°РјРјС‹ РґР»СЏ Р°РЅР°Р»РёР·Р° РІР»РёСЏРЅРёСЏ РєР°С‚РµРіРѕСЂРёР°Р»СЊРЅС‹С… РїРµСЂРµРјРµРЅРЅС‹С… РЅР° С†РµРЅСѓ Р°РІС‚РѕРјРѕР±РёР»СЏ.

# РљР°С‚РµРіРѕСЂРёР°Р»СЊРЅС‹Рµ РїРµСЂРµРјРµРЅРЅС‹Рµ СЃ РЅРµ СЃР»РёС€РєРѕРј Р±РѕР»СЊС€РёРј С‡РёСЃР»РѕРј СѓРЅРёРєР°Р»СЊРЅС‹С… Р·РЅР°С‡РµРЅРёР№
cat_cols_small = [col for col in cat_cols if data[col].nunique() <= 20]

# Р Р°СЃС‡С‘С‚ СЂР°Р·РјРµСЂРѕРІ СЃРµС‚РєРё РіСЂР°С„РёРєРѕРІ
n = len(cat_cols_small)
cols = 2
rows = math.ceil(n / cols)

fig, axes = plt.subplots(rows, cols, figsize=(10, 5 * rows))
axes = axes.flatten()

for i, col in enumerate(cat_cols_small):
    data.groupby(col)['Price_log'].mean().sort_values(ascending=False).plot.bar(ax=axes[i])
    axes[i].set_title(f'РЎСЂРµРґРЅСЏСЏ Price_log РїРѕ {col}')
    axes[i].tick_params(axis='x', labelrotation=45)

# РЈРґР°Р»РёС‚СЊ РїСѓСЃС‚С‹Рµ РіСЂР°С„РёРєРё
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()

```

```python
# РњРЅРѕРіРѕРјРµСЂРЅС‹Р№ Р°РЅР°Р»РёР· (multivariate analysis) вЂ” СЌС‚Рѕ РјРµС‚РѕРґ Р°РЅР°Р»РёР·Р°, РїСЂРё РєРѕС‚РѕСЂРѕРј РѕРґРЅРѕРІСЂРµРјРµРЅРЅРѕ СЂР°СЃСЃРјР°С‚СЂРёРІР°СЋС‚СЃСЏ
# С‚СЂРё Рё Р±РѕР»РµРµ РїРµСЂРµРјРµРЅРЅС‹С… РґР»СЏ РІС‹СЏРІР»РµРЅРёСЏ РІР·Р°РёРјРѕСЃРІСЏР·РµР№, Р·Р°РєРѕРЅРѕРјРµСЂРЅРѕСЃС‚РµР№, РІР»РёСЏРЅРёР№ Рё РїРѕС‚РµРЅС†РёР°Р»СЊРЅРѕР№ РёР·Р±С‹С‚РѕС‡РЅРѕСЃС‚Рё (РЅР°РїСЂРёРјРµСЂ, РґСѓР±Р»РёСЂСѓСЋС‰РёРµ РїСЂРёР·РЅР°РєРё).

# Р’ РєРѕРЅС‚РµРєСЃС‚Рµ РјР°С€РёРЅРЅРѕРіРѕ РѕР±СѓС‡РµРЅРёСЏ РёР»Рё EDA (СЂР°Р·РІРµРґРѕС‡РЅРѕРіРѕ Р°РЅР°Р»РёР·Р° РґР°РЅРЅС‹С…), РјРЅРѕРіРѕРјРµСЂРЅС‹Р№ Р°РЅР°Р»РёР· РїРѕРјРѕРіР°РµС‚:

# РїРѕРЅСЏС‚СЊ, РєР°РєРёРµ РїРµСЂРµРјРµРЅРЅС‹Рµ РЅР°РёР±РѕР»РµРµ РІР°Р¶РЅС‹ РґР»СЏ РїСЂРѕРіРЅРѕР·РёСЂРѕРІР°РЅРёСЏ;
# РІС‹СЏРІРёС‚СЊ РјСѓР»СЊС‚РёРєРѕР»Р»РёРЅРµР°СЂРЅРѕСЃС‚СЊ (РґРІРµ РїРµСЂРµРјРµРЅРЅС‹Рµ РґСѓР±Р»РёСЂСѓСЋС‚ РґСЂСѓРі РґСЂСѓРіР°);
# СѓР»СѓС‡С€РёС‚СЊ РёРЅС‚РµСЂРїСЂРµС‚РёСЂСѓРµРјРѕСЃС‚СЊ Рё СѓРїСЂРѕСЃС‚РёС‚СЊ РјРѕРґРµР»СЊ.
```

```python
plt.figure()
sns.heatmap(data[['Kilometers_Driven_log', 'Kilometers_Driven', 'Mileage', 'Engine', 'Power', 'Seats', 'Price_log',  'Price', 'Car_Age']].corr(), annot=True)
plt.show()

# .corr() вЂ” СЃС‡РёС‚Р°РµС‚ РјР°С‚СЂРёС†Сѓ РєРѕСЂСЂРµР»СЏС†РёР№ РџРёСЂСЃРѕРЅР° РјРµР¶РґСѓ С‡РёСЃР»РѕРІС‹РјРё РїРµСЂРµРјРµРЅРЅС‹РјРё:

# Р—РЅР°С‡РµРЅРёСЏ РѕС‚ -1 РґРѕ 1

# +1 в†’ РёРґРµР°Р»СЊРЅРѕ РїРѕР»РѕР¶РёС‚РµР»СЊРЅР°СЏ Р»РёРЅРµР№РЅР°СЏ Р·Р°РІРёСЃРёРјРѕСЃС‚СЊ

# -1 в†’ РёРґРµР°Р»СЊРЅРѕ РѕС‚СЂРёС†Р°С‚РµР»СЊРЅР°СЏ Р»РёРЅРµР№РЅР°СЏ

# 0 в†’ Р»РёРЅРµР№РЅРѕР№ СЃРІСЏР·Рё РЅРµС‚

# sns.heatmap(..., annot=True) вЂ” РІРёР·СѓР°Р»СЊРЅРѕ РїРѕРєР°Р·С‹РІР°РµС‚:

# С†РІРµС‚: РѕС‚ СЃРёРЅРµРіРѕ (РѕС‚СЂРёС†Р°С‚РµР»СЊРЅРѕ) РґРѕ РєСЂР°СЃРЅРѕРіРѕ (РїРѕР»РѕР¶РёС‚РµР»СЊРЅРѕ)

# С†РёС„СЂР° РІРЅСѓС‚СЂРё: С‚РѕС‡РЅРѕРµ Р·РЅР°С‡РµРЅРёРµ РєРѕСЂСЂРµР»СЏС†РёРё
```

рџ“Ќ 5. Р—Р°РїРѕР»РЅРµРЅРёРµ РїСЂРѕРїСѓС‰РµРЅРЅС‹С… Р·РЅР°С‡РµРЅРёР№ (Handling Missing Values)

```python
# 1. РћР±РЅР°СЂСѓР¶РµРЅРёРµ РЅРµРєРѕСЂСЂРµРєС‚РЅС‹С… Р·РЅР°С‡РµРЅРёР№

data.loc[data["Mileage"] == 0.0, 'Mileage'] = np.nan

# Р—Р°РјРµРЅСЏРµРј РїРѕРґРѕР·СЂРёС‚РµР»СЊРЅС‹Рµ 0.0 РІ СЂР°СЃС…РѕРґРµ С‚РѕРїР»РёРІР° РЅР° NaN вЂ” СЌС‚Рѕ Р»РѕРіРёС‡РЅРѕ, РїРѕС‚РѕРјСѓ С‡С‚Рѕ РјР°С€РёРЅР° СЃ СЂР°СЃС…РѕРґРѕРј 0 РЅРµРІРѕР·РјРѕР¶РЅР°.
```
