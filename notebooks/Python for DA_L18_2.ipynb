```python
# Импорт библиотек
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import seaborn as sns
from sklearn.datasets import load_wine
```

### <b><font color='saddlebrown'>1. Коэффициент корреляции Пирсона</font></b>
Функция pearsonr из библиотеки scipy.stats возвращает:
- r - Коэффициент корреляции Пирсона (от –1 до +1)
- p_value - p-значение для проверки гипотезы H0 (p = 0, значит, линейная связь есть)

Рассчитайте коэффициент корреляции Пирсона между длиной и шириной лепестка в наборе данных Iris.

```python
# Запишем в DataFrame df_iris данные из датасета Iris
df_iris = sns.load_dataset('iris')
df_iris
```

```python
# В переменные corr, p_value запишем результат работы функции pearsonr
corr, p_value = pearsonr(df_iris['petal_length'], df_iris['petal_width'])
```

```python
# Выведем результат и интерпретируем его
print(corr)
print(p_value)
```

### <b><font color='saddlebrown'>2. Коэффициент корреляции Спирмена</font></b>
Функция spearmanr из библиотеки scipy.stats возвращает:
- rho - коэффициент ранговой корреляции Спирмена (от –1 до +1)
- p_value - p-значение для проверки гипотезы H0 (p = 0, значит, монотонная связь есть)

Рассчитайте коэффициент корреляции Спирмена между содержанием color_intensity и hue в наборе данных Wine Quality.

```python
# Запишем в DataFrame df_wine данные из датасета wine
wine_data = load_wine()
df_wine = pd.DataFrame(data = wine_data.data, columns = wine_data.feature_names)
df_wine['target'] = wine_data.target
df_wine
```

```python
# В переменные corr, p_value запишем результат работы функции spearmanr
corr, p_value = spearmanr(df_wine['color_intensity'], df_wine['hue'])
```

```python
# Выведем результат и интерпретируем его
print(corr)
print(p_value)
```
