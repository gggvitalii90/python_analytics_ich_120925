# рџЋµ Spotify Songs вЂ” Exploratory Data Analysis Dashboard
### Practicum 14 | Python for Data Analytics

> **Dataset:** ~32 800 tracks from Spotify API (6 genres, 2020)  
> **Goal:** Uncover audio patterns, genre DNA, popularity drivers, and musical evolution trends  
> **Tools:** `pandas` В· `numpy` В· `plotly`

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

PALETTE = {
    'edm':   '#1DB954',
    'latin': '#FF6B6B',
    'pop':   '#4ECDC4',
    'r&b':   '#9B59B6',
    'rap':   '#F39C12',
    'rock':  '#E74C3C',
}
GENRE_ORDER = ['edm', 'latin', 'pop', 'r&b', 'rap', 'rock']
COLOR_SEQ   = [PALETTE[g] for g in GENRE_ORDER]

LAYOUT_DEFAULTS = dict(
    template='plotly_dark',
    font=dict(family='Arial', size=13),
    paper_bgcolor='#191414',
    plot_bgcolor='#191414',
    margin=dict(t=60, b=40, l=60, r=40),
)
```

## 1. Dataset Overview

```python
df = pd.read_csv('../data/spotify_songs.csv')

df['year']         = pd.to_datetime(df['track_album_release_date'], errors='coerce').dt.year
df['duration_min'] = df['duration_ms'] / 60_000
df['party_index']  = (df['danceability'] + df['energy'] + df['valence']) / 3
df['mode_label']   = df['mode'].map({1: 'Major', 0: 'Minor'})

AUDIO_FEATURES = ['danceability', 'energy', 'loudness', 'speechiness',
                   'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo']

kpis = pd.DataFrame({
    'Metric': ['Total Tracks', 'Unique Artists', 'Unique Albums',
               'Avg Popularity', 'Avg Duration (min)', 'Year Range'],
    'Value': [
        f"{df.shape[0]:,}",
        f"{df['track_artist'].nunique():,}",
        f"{df['track_album_name'].nunique():,}",
        f"{df['track_popularity'].mean():.1f} / 100",
        f"{df['duration_min'].mean():.2f}",
        f"{int(df['year'].min())} - {int(df['year'].max())}",
    ]
})
print(kpis.to_string(index=False))
```

## 2. Genre Landscape

```python
genre_counts = df['playlist_genre'].value_counts().reindex(GENRE_ORDER).reset_index()
genre_counts.columns = ['genre', 'count']
genre_counts['pct'] = (genre_counts['count'] / genre_counts['count'].sum() * 100).round(1)

fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type': 'domain'}, {'type': 'xy'}]],
    subplot_titles=['Share of Tracks by Genre', 'Track Count per Genre']
)
fig.add_trace(go.Pie(
    labels=genre_counts['genre'], values=genre_counts['count'],
    hole=0.55, marker_colors=COLOR_SEQ, textinfo='label+percent', textfont_size=13, name='',
), row=1, col=1)
fig.add_trace(go.Bar(
    x=genre_counts['genre'], y=genre_counts['count'], marker_color=COLOR_SEQ,
    text=genre_counts['count'].apply(lambda x: f'{x:,}'), textposition='outside', name='',
), row=1, col=2)
fig.update_layout(title_text='Genre Distribution', showlegend=False, height=420, **LAYOUT_DEFAULTS)
fig.update_yaxes(showgrid=False, row=1, col=2)
fig.show()
```

**Insight:** Р”Р°С‚Р°СЃРµС‚ С…РѕСЂРѕС€Рѕ СЃР±Р°Р»Р°РЅСЃРёСЂРѕРІР°РЅ РїРѕ 6 Р¶Р°РЅСЂР°Рј. EDM РЅРµРјРЅРѕРіРѕ РїРµСЂРµРїСЂРµРґСЃС‚Р°РІР»РµРЅ (~18%), Rock вЂ” РЅРµРґРѕРїСЂРµРґСЃС‚Р°РІР»РµРЅ (~15%).

## 3. Audio DNA вЂ” Genre Radar Chart

```python
df_radar = df.copy()
df_radar['loudness_norm'] = (
    (df_radar['loudness'] - df_radar['loudness'].min()) /
    (df_radar['loudness'].max() - df_radar['loudness'].min())
)

RADAR_FEATURES = ['danceability', 'energy', 'loudness_norm',
                  'speechiness', 'acousticness', 'valence', 'liveness']
RADAR_LABELS   = ['Danceability', 'Energy', 'Loudness',
                  'Speechiness', 'Acousticness', 'Valence', 'Liveness']

genre_means = df_radar.groupby('playlist_genre')[RADAR_FEATURES].mean()

fig = go.Figure()
for genre in GENRE_ORDER:
    vals = genre_means.loc[genre].tolist() + genre_means.loc[genre].tolist()[:1]
    fig.add_trace(go.Scatterpolar(
        r=vals, theta=RADAR_LABELS + RADAR_LABELS[:1],
        fill='toself', fillcolor=PALETTE[genre],
        line=dict(color=PALETTE[genre], width=2),
        opacity=0.45, name=genre.upper(),
    ))
fig.update_layout(
    title_text='Audio DNA -- Average Feature Profile by Genre',
    polar=dict(
        bgcolor='#242424',
        radialaxis=dict(visible=True, range=[0, 1], gridcolor='#444', tickfont_size=10),
        angularaxis=dict(tickfont_size=13, gridcolor='#444'),
    ),
    legend=dict(orientation='h', y=-0.12, x=0.15),
    height=560, **LAYOUT_DEFAULTS,
)
fig.show()
```

**Key Insights:**

| Genre | Superpower | Weakness |
|-------|-----------|----------|
| **EDM** | Highest Energy (0.80) + Loudness | Lowest Valence -- energetic but not happy |
| **Latin** | Highest Valence (0.61) + Danceability | Moderate energy |
| **Rap** | Highest Danceability (0.72) + Speechiness | Lower energy vs EDM/Rock |
| **Rock** | High Energy (0.73) + Liveness feel | Lowest Danceability (0.52) |
| **R&B** | Most Acoustic (0.26) | Lower loudness / energy |
| **Pop** | Most balanced profile | No dominant superpower |

## 4. Popularity Deep Dive

```python
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=['Popularity Distribution by Genre', 'Top 15 Artists (min 5 tracks)']
)
for genre in GENRE_ORDER:
    fig.add_trace(go.Violin(
        y=df[df['playlist_genre'] == genre]['track_popularity'],
        name=genre.upper(), box_visible=True, meanline_visible=True,
        fillcolor=PALETTE[genre], line_color=PALETTE[genre], opacity=0.8,
    ), row=1, col=1)

top_artists = (
    df.groupby('track_artist')
    .agg(avg_pop=('track_popularity', 'mean'), count=('track_id', 'count'))
    .query('count >= 5')
    .sort_values('avg_pop', ascending=True).tail(15).reset_index()
)
fig.add_trace(go.Bar(
    x=top_artists['avg_pop'], y=top_artists['track_artist'], orientation='h',
    marker=dict(color=top_artists['avg_pop'], colorscale='Viridis', showscale=False),
    text=top_artists['avg_pop'].round(1), textposition='outside', name='',
), row=1, col=2)
fig.update_layout(title_text='Popularity Analysis', showlegend=False, height=520, **LAYOUT_DEFAULTS)
fig.update_xaxes(title_text='Avg Popularity', row=1, col=2, range=[75, 100])
fig.update_yaxes(title_text='Popularity (0-100)', row=1, col=1)
fig.show()
```

## 5. Top 10 Most Popular Tracks

```python
top_tracks = (
    df.nlargest(20, 'track_popularity')
    .drop_duplicates(subset='track_name').head(10)
    [['track_name', 'track_artist', 'track_popularity', 'playlist_genre',
      'danceability', 'energy', 'valence', 'tempo']]
    .reset_index(drop=True)
)
top_tracks.index += 1
top_tracks.columns = ['Track', 'Artist', 'Popularity', 'Genre', 'Danceability', 'Energy', 'Valence', 'BPM']
top_tracks['BPM'] = top_tracks['BPM'].round(0).astype(int)

print('=' * 90)
print(f"{'#':>2}  {'Track':<35} {'Artist':<20} {'Pop':>3}  {'Genre':<7} {'Dance':>5}  {'BPM':>3}")
print('=' * 90)
for idx, row in top_tracks.iterrows():
    print(f"{idx:>2}  {str(row['Track'])[:35]:<35} {str(row['Artist'])[:20]:<20} "
          f"{row['Popularity']:>3}  {row['Genre']:<7} {row['Danceability']:>5.2f}  {row['BPM']:>3}")
print('=' * 90)
```

## 6. The Party Index

> *Custom metric = (Danceability + Energy + Valence) / 3*  
> Measures how suitable a track is for a party.

```python
party_by_genre = (
    df.groupby('playlist_genre')['party_index']
    .agg(['mean', 'std'])
    .reindex(GENRE_ORDER).reset_index()
)
party_by_genre.columns = ['genre', 'mean', 'std']
party_by_genre = party_by_genre.sort_values('mean', ascending=True)

fig = go.Figure()
fig.add_trace(go.Bar(
    x=party_by_genre['mean'],
    y=party_by_genre['genre'].str.upper(),
    orientation='h',
    marker=dict(
        color=party_by_genre['mean'],
        colorscale=[[0, '#2C3E50'], [0.4, '#F39C12'], [1.0, '#1DB954']],
        showscale=True,
        colorbar=dict(title='Party Index', tickformat='.2f'),
    ),
    error_x=dict(type='data', array=party_by_genre['std'], visible=True, color='#aaa'),
    text=party_by_genre['mean'].apply(lambda x: f'{x:.3f}'),
    textposition='outside',
))
fig.add_vline(x=party_by_genre['mean'].mean(), line_dash='dash', line_color='white',
              opacity=0.5, annotation_text='Overall avg', annotation_position='top')
fig.update_layout(
    title_text='Party Index by Genre',
    xaxis=dict(title='Party Index (0-1)', range=[0.55, 0.73]),
    height=380, **LAYOUT_DEFAULTS,
)
fig.show()
```

## 7. Mood Map вЂ” Energy vs Valence

```python
sample = df.sample(3000, random_state=42)

fig = px.scatter(
    sample, x='valence', y='energy',
    color='playlist_genre', color_discrete_map=PALETTE,
    size='track_popularity', size_max=12, opacity=0.65,
    hover_data={'track_name': True, 'track_artist': True,
                'valence': ':.2f', 'energy': ':.2f', 'track_popularity': True},
    labels={'valence': 'Valence (Sadness to Happiness)', 'energy': 'Energy (Calm to Intense)'},
    title='Mood Map: Where Do Genres Live in Emotional Space?',
    category_orders={'playlist_genre': GENRE_ORDER},
)
for x, y, text, align in [
    (0.05, 0.97, 'Turbulent / Angry', 'left'),
    (0.72, 0.97, 'Happy / Euphoric', 'right'),
    (0.05, 0.05, 'Sad / Depressing', 'left'),
    (0.72, 0.05, 'Peaceful / Chill', 'right'),
]:
    fig.add_annotation(x=x, y=y, text=text, showarrow=False,
                       font=dict(size=11, color='#888'), xanchor=align)
fig.add_hline(y=0.5, line_dash='dot', line_color='#555')
fig.add_vline(x=0.5, line_dash='dot', line_color='#555')
fig.update_layout(height=560, legend_title_text='Genre', **LAYOUT_DEFAULTS)
fig.show()
```

## 8. Music Evolution (2000-2020)

```python
year_df = (
    df[df['year'].between(2000, 2020)]
    .groupby('year')
    .agg(
        avg_danceability=('danceability', 'mean'),
        avg_energy=('energy', 'mean'),
        avg_valence=('valence', 'mean'),
        avg_acousticness=('acousticness', 'mean'),
        track_count=('track_id', 'count'),
    )
    .reset_index()
)

fig = make_subplots(
    rows=2, cols=1,
    subplot_titles=['Audio Features Trend (2000-2020)', 'Track Count per Year'],
    vertical_spacing=0.14, shared_xaxes=True,
)
for col, label, color in [
    ('avg_danceability', 'Danceability', '#1DB954'),
    ('avg_energy',       'Energy',       '#FF6B6B'),
    ('avg_valence',      'Valence',      '#4ECDC4'),
    ('avg_acousticness', 'Acousticness', '#9B59B6'),
]:
    fig.add_trace(go.Scatter(
        x=year_df['year'], y=year_df[col], mode='lines+markers', name=label,
        line=dict(color=color, width=2), marker=dict(size=5),
    ), row=1, col=1)
fig.add_trace(go.Bar(
    x=year_df['year'], y=year_df['track_count'],
    name='Track Count', marker_color='#F39C12', opacity=0.75,
), row=2, col=1)
fig.update_layout(
    title_text='How Has Music Changed Over Two Decades?',
    height=620, legend=dict(orientation='h', y=1.05, x=0), **LAYOUT_DEFAULTS,
)
fig.update_yaxes(title_text='Feature Value (0-1)', row=1, col=1)
fig.update_yaxes(title_text='Track Count', row=2, col=1)
fig.show()
```

**РљР»СЋС‡РµРІРѕР№ С‚СЂРµРЅРґ:** Valence СѓРїР°Р» РЅР° **23%** (2000в†’2020), Acousticness РЅР° **49%**. РњСѓР·С‹РєР° СЃС‚Р°Р»Р° С‚РµРјРЅРµРµ Рё СЌР»РµРєС‚СЂРѕРЅРЅРµРµ, РїСЂРё СЌС‚РѕРј С‚Р°РЅС†РµРІР°Р»СЊРЅРѕСЃС‚СЊ Рё СЌРЅРµСЂРіРёСЏ РѕСЃС‚Р°Р»РёСЃСЊ СЃС‚Р°Р±РёР»СЊРЅС‹РјРё.

## 9. Audio Feature Correlations

```python
corr_matrix = df[AUDIO_FEATURES].corr().round(2)

fig = px.imshow(
    corr_matrix, text_auto=True,
    color_continuous_scale='RdBu_r', zmin=-1, zmax=1,
    aspect='auto', title='Pearson Correlation Matrix -- Audio Features',
)
fig.update_layout(height=560, coloraxis_colorbar=dict(title='Correlation'), **LAYOUT_DEFAULTS)
fig.update_traces(textfont_size=11)
fig.show()
```

## 10. Subgenre Popularity Heatmap

```python
pivot = (
    df.groupby(['playlist_genre', 'playlist_subgenre'])
    .agg(avg_pop=('track_popularity', 'mean'))
    .reset_index()
    .pivot(index='playlist_genre', columns='playlist_subgenre', values='avg_pop')
)
fig = px.imshow(
    pivot, text_auto='.1f', color_continuous_scale='Viridis',
    aspect='auto', title='Average Popularity by Genre x Subgenre',
    labels=dict(color='Avg Popularity'),
)
fig.update_layout(height=400, xaxis_tickangle=-35, **LAYOUT_DEFAULTS)
fig.show()
```
