- multiple - РєР°Рє РёРјРµРЅРЅРѕ РЅР°РєР»Р°РґС‹РІР°С‚СЊ СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ СЂР°Р·РЅС‹С… РєР°С‚РµРіРѕСЂРёР№ РґСЂСѓРі РЅР° РґСЂСѓРіР° (РІ СЃР»СѓС‡Р°Рµ РїРѕСЃС‚СЂРѕРµРЅРёСЏ РЅРµСЃРєРѕР»СЊРєРёС… РіРёСЃС‚РѕРіСЂР°РјРј)
    - "layer" (РїРѕ СѓРјРѕР»С‡Р°РЅРёСЋ) вЂ” РїСЂРѕР·СЂР°С‡РЅС‹Рµ СЃС‚РѕР»Р±РёРєРё, РЅР°Р»РѕР¶РµРЅРЅС‹Рµ РґСЂСѓРі РЅР° РґСЂСѓРіР°.

    - "dodge" вЂ” РєР°С‚РµРіРѕСЂРёРё СЃС‚РѕСЏС‚ СЂСЏРґРѕРј (СЂР°Р·РЅРµСЃРµРЅС‹).
    - "fill" вЂ” РЅРѕСЂРјРёСЂРѕРІРєР° С‚Р°Рє, С‡С‚Рѕ РІС‹СЃРѕС‚Р° РєР°Р¶РґРѕРіРѕ СЃС‚РѕР»Р±РёРєР° = 1 (РїРѕРєР°Р·С‹РІР°РµС‚ РґРѕР»Рё).

- kde - РЅР°СЂРёСЃРѕРІР°С‚СЊ РёР»Рё РЅРµС‚ РіСЂР°С„РёРє СЏРґРµСЂРЅРѕР№ РѕС†РµРЅРєРё РїР»РѕС‚РЅРѕСЃС‚Рё
- palette - С†РІРµС‚РѕРІР°СЏ СЃС…РµРјР°. РСЃРїРѕР»СЊР·СѓРµС‚СЃСЏ С‚РѕР»СЊРєРѕ c hue ([Palette](https://seaborn.pydata.org/generated/seaborn.color_palette.html#seaborn.color_palette))

- legend - РїРѕРєР°Р·С‹РІР°С‚СЊ/РЅРµ РїРѕРєР°Р·С‹РІР°С‚СЊ Р»РµРіРµРЅРґСѓ

<b>sns.jointplot()</b><br>

- x - РєР°РєРёРµ Р·РЅР°С‡РµРЅРёСЏ СѓРєР°Р·Р°С‚СЊ РїРѕ РѕСЃРё X
- y - РєР°РєРёРµ Р·РЅР°С‡РµРЅРёСЏ СѓРєР°Р·Р°С‚СЊ РїРѕ РѕСЃРё Y
- hue - СЃС‚РѕР»Р±РµС†, РµСЃР»Рё Р·РЅР°С‡РµРЅРёСЏ РЅРµРѕР±С…РѕРґРёРјРѕ СЂР°СЃРїСЂРµРґРµР»РёС‚СЊ РїРѕ РЅРµСЃРєРѕР»СЊРєРёРј РєР°С‚РµРіРѕСЂРёСЏРј
- kind - С‚РёРї РіСЂР°С„РёРєР° РґР»СЏ РІРёР·СѓР°Р»РёР·Р°С†РёРё:
    - scatter - РїРѕ СѓРјРѕР»С‡Р°РЅРёСЋ
    - kde
    - hist
    - hex
    - reg
    - resid
- height - СЂР°Р·РјРµСЂ С„РёРіСѓСЂС‹ (РІ РґСЋР№РјР°С…). 1 РґСЋР№Рј = 2.54 СЃРј
- space - РїСЂРѕСЃС‚СЂР°РЅСЃС‚РІРѕ РјРµР¶РґСѓ РѕСЃРЅРѕРІРЅС‹Рј РіСЂР°С„РёРєРѕРј Рё РіСЂР°С„РёРєР°РјРё РїРѕ Р±РѕРєР°Рј. Р’ РґСЋР№РјР°С….

- color - С†РІРµС‚ РіСЂР°С„РёРєРѕРІ
- palette - С†РІРµС‚РѕРІР°СЏ СЃС…РµРјР°. РСЃРїРѕР»СЊР·СѓРµС‚СЃСЏ С‚РѕР»СЊРєРѕ c hue
- hue_order - РїРѕСЂСЏРґРѕРє РєР°С‚РµРіРѕСЂРёР№

```python
# Р·Р°РіСЂСѓР·РёРј РґР°РЅРЅС‹Рµ brain networks
# РјР°С‚СЂРёС†Р° РєРѕСЂСЂРµР»СЏС†РёР№ РјРµР¶РґСѓ СЂР°Р·Р»РёС‡РЅС‹РјРё СѓС‡Р°СЃС‚РєР°РјРё РјРѕР·РіР° (brain networks), 
# РїРѕР»СѓС‡РµРЅРЅС‹РјРё СЃ РїРѕРјРѕС‰СЊСЋ С„СѓРЅРєС†РёРѕРЅР°Р»СЊРЅРѕР№ РјР°РіРЅРёС‚РЅРѕ-СЂРµР·РѕРЅР°РЅСЃРЅРѕР№ С‚РѕРјРѕРіСЂР°С„РёРё (fMRI)
data = sns.load_dataset("brain_networks", header=[0, 1, 2], index_col=0)
data
```

```python
# РІС‹Р±РµСЂРµРј РїРѕРґРјРЅРѕР¶РµСЃС‚РІРѕ РѕС‚РґРµР»РѕРІ РјРѕР·РіР°
used_networks = [1, 5, 6, 7, 8, 12, 13, 17]
used_columns = (data.columns.get_level_values("network")
                            .astype(int)
                            .isin(used_networks))
data = data.loc[:, used_columns]
data
```

```python
# СЃРѕР·РґР°РґРёРј РґР»СЏ РЅРёС… РєР°С‚РµРіРѕСЂРёР°Р»СЊРЅСѓСЋ РїР°Р»РёС‚СЂСѓ
# husl_palette СЃРѕР·РґР°С‘С‚ РЅР°Р±РѕСЂ РєР°С‚РµРіРѕСЂРёР°Р»СЊРЅС‹С… С†РІРµС‚РѕРІ, 
# СЂР°РІРЅРѕРјРµСЂРЅРѕ СЂР°СЃРїСЂРµРґРµР»С‘РЅРЅС‹С… РїРѕ С†РІРµС‚РѕРІРѕРјСѓ РєСЂСѓРіСѓ РІ РїСЂРѕСЃС‚СЂР°РЅСЃС‚РІРµ HUSL (Hue, Saturation, Lightness).
# H (Hue) вЂ” РѕС‚С‚РµРЅРѕРє (С†РІРµС‚РѕРІРѕР№ С‚РѕРЅ, СѓРіРѕР» РЅР° С†РІРµС‚РѕРІРѕРј РєСЂСѓРіРµ).
# S (Saturation) вЂ” РЅР°СЃС‹С‰РµРЅРЅРѕСЃС‚СЊ С†РІРµС‚Р°.
# L (Lightness) вЂ” СЏСЂРєРѕСЃС‚СЊ (СЃРІРµС‚Р»РѕС‚Р°)
# sns.husl_palette(n_colors=6, s=.75, l=.65)
# n_colors в†’ СЃРєРѕР»СЊРєРѕ С†РІРµС‚РѕРІ РЅСѓР¶РЅРѕ (РїРѕ СѓРјРѕР»С‡Р°РЅРёСЋ 6).
# s в†’ РЅР°СЃС‹С‰РµРЅРЅРѕСЃС‚СЊ (РѕС‚ 0 РґРѕ 1). РњРµРЅСЊС€Рµ = Р±РѕР»РµРµ В«СЃРµСЂС‹Р№В» С†РІРµС‚.
# l в†’ СЃРІРµС‚Р»РѕС‚Р° (РѕС‚ 0 РґРѕ 1). РњРµРЅСЊС€Рµ = С‚РµРјРЅРµРµ, Р±РѕР»СЊС€Рµ = СЃРІРµС‚Р»РµРµ.
# h в†’ СѓРіРѕР» РЅР° С†РІРµС‚РѕРІРѕРј РєСЂСѓРіРµ, РјРѕР¶РЅРѕ Р·Р°РґР°С‚СЊ РЅР°С‡Р°Р»СЊРЅС‹Р№ hue.
network_pal = sns.husl_palette(len(used_networks), s=.45)
network_lut = dict(zip(map(str, used_networks), network_pal))
# РѕРєСЂР°СЃРёРј СЃС‚СЂРѕРєРё Рё СЃС‚РѕР»Р±С†С‹ РІ СЃРѕРѕС‚РІРµС‚СЃС‚РІРёРё СЃ РѕС‚РґРµР»РѕРј
networks = data.columns.get_level_values("network")
network_colors = pd.Series(networks, index=data.columns).map(network_lut)
# РЅР°СЂРёСЃСѓРµРј РіСЂР°С„РёРє РєРѕСЂСЂРµР»СЏС†РёРё РїР°С‚С‚РµСЂРЅРѕРІ Р°РєС‚РёРІРЅРѕСЃС‚Рё СѓС‡Р°СЃС‚РєРѕРІ РјРѕР·РіР°
# Рё СѓР±РµРґРёРјСЃСЏ, С‡С‚Рѕ СѓС‡Р°СЃС‚РєРё РёР· РѕРґРЅРѕРіРѕ РѕС‚РґРµР»Р° РїРѕРїР°РґР°СЋС‚ РІ РѕРґРЅСѓ РіСЂСѓРїРїСѓ
# РїРѕСЃР»Рµ РёРµСЂР°СЂС…РёС‡РµСЃРєРѕР№ РєР»Р°СЃС‚РµСЂРёР·Р°С†РёРё
sns.clustermap(data.corr(), center=0, cmap="vlag",
               row_colors=network_colors, col_colors=network_colors,
               linewidths=.75, figsize=(13, 13));
```

```python
def sinplot(flip=1):
    """РџСЂРѕСЃС‚РµР№С€РёР№ СЃРёРЅСѓСЃРѕРёРґР°Р»СЊРЅС‹Р№ РїСЂРёРјРµСЂ РґР»СЏ РґРµРјРѕРЅСЃС‚СЂР°С†РёРё seaborn"""
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i * .5) * (7 - i) * flip)
```
