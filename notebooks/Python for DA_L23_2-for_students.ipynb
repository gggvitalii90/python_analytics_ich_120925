### <b><font color='skyblue'>4. Слияние и объединение в Pandas</font></b>

#### <b><font color='skyblue'>4.1. merge</font></b>

```python
df1 = pd.DataFrame({'a': ['a1', 'a2', 'a3'], 'b': ['b1', 'b2', 'b3']})
df2 = pd.DataFrame({'a': ['a1', 'a2', 'a4'], 'c': ['c1', 'c2', 'c4']})
print("DataFrame 1:\n", df1)
print("-"*20)
print("DataFrame 2:\n", df2)
print("-"*20)
merged_df = pd.merge(df1, df2, on = 'a')
print(merged_df)
```

##### <b><font color='skyblue'>Типы слияний</font></b>

```python
# Внутреннее соединение: Возвращает только строки с совпадающими ключами в обоих фреймах данных.
pd.merge(df1, df2, how = 'inner', on = 'a')
```

```python
# Левое объединение: Возвращает все строки из левого DataFrame и совпадающие строки из правого DataFrame.
pd.merge(df1, df2, how = 'left', on = 'a')
```

```python
# Правое объединение: Возвращает все строки из правого DataFrame и совпадающие строки из левого DataFrame.
pd.merge(df1, df2, how = 'right', on = 'a')
```

```python
# Внешнее объединение: Возвращает все строки, если есть совпадение в левом или правом DataFrame.
pd.merge(df1, df2, how = 'outer', on = 'a')
```

```python
# Указание столбцов для слияния:
df2 = df2.rename(columns = {'a':'a_'})
merged_df = pd.merge(df1, df2, left_on = 'a', right_on = 'a_')
# df1.merge(df2, left_on = 'a', right_on = 'a_')
print(merged_df)
```

### <u><b><font color='skyblue'>5. Группировка данных и сводные таблицы в Pandas</font></b></u>

#### <b><font color='skyblue'>5.1. groupby()</font></b>

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value': [10, 20, 30, 40, 50, 60]
}
df = pd.DataFrame(data)

# Группируем по "Category" и вычисляем сумму "Value"
grouped_df = df.groupby('Category').sum()
print(grouped_df)
```

```python
grouped_df = df.groupby('Category').agg({'Value': ['sum', 'mean', 'max']})
print(grouped_df)
```

#### <b><font color='skyblue'>5.2. pivot_table()</font></b>
Ключевые параметры в pivot_table()
- data: DataFrame для разворота.
- value: Столбец(и) для агрегирования.
- index: Столбец(и) для группировки по строкам.
- columns: Столбец(и) для группировки по столбцам.
- aggfunc: Функция(и) агрегирования (например, 'mean', 'sum').
- fill_value (заполнить значение): Значение для замены отсутствующих значений.
- margins: Добавить итоговые значения всех строк/столбцов (по умолчанию False).

```python
data = {
    'A': ['foo', 'foo', 'foo', 'bar', 'bar', 'bar'],
    'B': ['one', 'one', 'two', 'two', 'one', 'one'],
    'C': ['маленький', 'большой', 'маленький', 'маленький', 'большой', 'маленький'],
    'D': [1, 2, 2, 3, 3, 4]
}
df = pd.DataFrame(data)
df
```

```python
# Создание сводной таблицы
pivot_table = pd.pivot_table(df, 
                             values = 'D', 
                             index = ['A', 'B'], 
                             columns = ['C'], 
                             aggfunc = 'sum', 
                             fill_value = 0,
                             margins = True,
                             margins_name = 'Total')
print(pivot_table)
```

```python
# Создайте сводную таблицу с несколькими агрегатами
pivot_table = pd.pivot_table(df, 
                             values = ['D'], 
                             index = ['A', 'C'], 
                             aggfunc = {'D': ['mean', 'sum']})
print(pivot_table)
```

```python
# Построение графиков по таблице pivot
pivot_table = pd.pivot_table(df, 
                             values = 'D', 
                             index = 'A', 
                             columns = 'B', 
                             aggfunc = 'mean')
pivot_table.plot(kind = 'bar', title = 'Средние значения по A и B')
pivot_table.plot(kind = 'hist', title = 'Средние значения по A и B')
```
