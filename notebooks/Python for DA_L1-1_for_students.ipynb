### <font color='teal'>2. Создание массивов</font>
- <i>вектор - одномерный массив</i>
- <i>матрица - двумерный массив</i>
- <i>вектор-столбец - двумерный массив с одним столбцом </i>

```python
#Способ №1 - из списка
#Вектор
one_d_array = np.array([1, 2, 3])
print('Вектор: ',one_d_array)
#Матрица
two_d_array = np.array([[1, 2, 3], [4, 5, 6]])
print('Матрица: \n',two_d_array)
#Вектор-столбец
col = np.array([[10], [20], [30]])
print('Вектор-столбец: \n',col)
```

```python
#Способ №2 - с помощью функции arange(начало диапазона, конец диапазона, шаг)
a = np.arange(0, 9, 2)
print(a)
```

```python
print("Целочисленный массив: ", np.arange(0, 9, 2))
print("Массив десятичных чисел: ", np.arange(0.0, 8.0, 0.5))
```

```python
#Способ №3 - с помощью функции linspace(начало диапазона, конец диапазона, количество чисел в массиве)
a = np.linspace(0, 8, 7)
print(a)
```

#### <font color='teal'>2.1. Специальные функции для создания массивов</font>
- <i>empty_like(), empty()</i>
- <i>ones_like(), ones()</i>
- <i>zeros_like(), zeros()</i>
- <i>full_like, full()</i>

```python
#empty_like()
#Возвращает новый массив той же формы и типа, что и заданный массив
a = ([1,2,3], [4,5,6])
np.empty_like(a)
```

```python
#empty()
#Возвращает новый массив заданной формы и типа
np.empty([2, 2], float)
```

```python
#ones_like()
#Возвращает массив единиц той же формы и типа, что и заданный массив
a = ([1,2,3], [4,5,6])
np.ones_like(a)
```

```python
#ones()
#Возвращает новый массив заданной формы и типа, заполненный единицами
np.ones((5,5), dtype=int)
```

```python
#zeros_like()
#Возвращает массив нулей той же формы и типа, что и заданный массив
a = ([1,2,3], [4,5,6])
np.zeros_like(a)
```

```python
#zeros()
#Возвращает новый массив заданной формы и типа, заполненный нулями
np.zeros((10,10))
```

```python
#full_like()
#Возвращает заполненный массив той же формы и типа, что и заданный массив
a = ([1,2,3], [4,5,6])
np.full_like(a, 255)
```

```python
#full()
#Возвращает новый массив заданной формы и типа, заполненный указанными значениями
np.full((3,3), 255)
```
