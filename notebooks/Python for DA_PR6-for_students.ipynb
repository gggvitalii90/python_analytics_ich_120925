### <font color='chocolate'>Анализ победителей Формулы-1 (1950–2025)</font>
<b>Описание данных</b><br>
<i>date</i> — дата проведения Гран-при (например, 2023-07-23)<br>
<i>continent</i> — континент, на котором проходила гонка (Europe, Asia, Americas, Africa, Oceania)<br>
<i>grand_prix</i> — название конкретного Гран-при (например, British Grand Prix)<br>
<i>circuit</i> — название гоночной трассы (например, Silverstone)<br>
<i>winner_name</i> — имя победителя гонки (например, Lewis Hamilton)<br>
<i>team</i> — команда (конструктор), за которую выступал победитель (например, Mercedes)<br>
<i>time</i> — время гонки победителя в формате HH:MM:SS<br>
<i>laps</i> — количество кругов, пройденных в гонке<br>
<i>year</i> — год проведения гонки<br>

```python
# мпорт библиотек
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np  
```

```python
# Загрузка данных. Назовите DataFrame df_formula_1
df_formula_1 = pd.read_csv('../data/winners_f1_1950_2025_v2.csv')
df_formula_1
```

```python
# 1. Посчитайте количество побед каждого пилота за весь период.
df_formula_1.groupby('winner_name')\
            .count()[['continent']]\
            .sort_values(by='continent', ascending=False)\
            .rename(columns= {'continent' : 'Count winners'})
```

```python
df_formula_1= pd.read_csv('../data/winners_f1_1950_2025_v2.csv')
df_formula_1
count_winners = df_formula_1.groupby('winner_name').count()[['continent']]
count_winners.sort_values(by='continent', ascending=False).rename(columns= {'continent' : 'Count winners'})
```

```python
# 2. Определите, сколько побед у каждой команды.
df_formula_1.groupby('team', as_index=False)['team']\
            .count()\
            .sort_values(by='team', ascending=False)\
            .rename(columns={'team': 'team winners'})
```

```python
# 3. Создайте сводную таблицу, где по строкам будут континенты, по столбцам - года (последние 10 лет), значения - количество гонок.
pd.pivot_table(df_formula_1[df_formula_1['year'] >= 2015], 
               index='continent', 
               columns='year', 
               aggfunc='count',
               values='laps', 
               fill_value=0)
```

```python
# 4. Постройте гистограмму (hist) распределения количества кругов в гонках.
plt.hist(df_formula_1['laps'], bins=20, edgecolor='black', color='green', alpha=0.7)
plt.title("Гистограмма распределения количества кругов в гонках")
plt.xlabel("Количество кругов")
plt.ylabel("Частота")
plt.show()
```

```python
df_formula_1[df_formula_1['laps'] > 120].sort_values(by='laps', ascending=False)
```

```python
sizes = df_formula_1.groupby('continent').count()['winner_name']
sizes
```

```python
# Создайте круговую диаграмму (pie) для распределения гонок по континентам.
# Шаг 1. Сгруппируйте данные по континентам и сохраните получившиеся данные в новый DataFrame
labels = df_formula_1.groupby('continent').count().index
sizes = df_formula_1.groupby('continent').count()['winner_name']
explode = [0.1 if s == max(sizes) else 0 for s in sizes]
# Шаг 2. Постройте круговую диаграмму, где labels - названия континентов, 
# размеры частей круга - соотношение количества гонок по континентам
patches, texts, autotexts = plt.pie(sizes,
        labels=labels,
        autopct='%1.1f%%',
        startangle=0,
        shadow=True,
        explode=explode
        )
plt.title("Распределение гонок по континентам")
# for text in autotexts:
#     text.set_rotation(0)
plt.legend(labels, loc='center left', bbox_to_anchor=(1.3, 0.5))
plt.show()
```
