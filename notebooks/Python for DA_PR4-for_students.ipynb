1. Напишите программу, которая генерирует Series c данными о ежедневных доходах от трейдинга за прошлый год (индексы – это даты, суммы дохода – случайные, но учтите, что они могут быть и отрицательными). Программа должны вывести: 
- Годовую доходность.
- Среднюю дневную доходность.
- Медианную дневную доходность.
- Наивысшую дневную доходность.
- Наибольший дневной убыток.

```python
import numpy as np
import pandas as pd

# Фиксируем seed для воспроизводимости результата
np.random.seed(42)

# Формируем даты за прошлый календарный год
current_year = pd.Timestamp.today().year
last_year = current_year - 1
dates = pd.date_range(start=f"{last_year}-01-01", end=f"{last_year}-12-31", freq="D")

# Генерируем случайные дневные доходы (могут быть и отрицательные)
# mean=50, std=300 дает смесь прибыли и убытков
daily_income = np.random.normal(loc=50, scale=300, size=len(dates))
income_series = pd.Series(daily_income, index=dates, name="daily_income")

# Считаем метрики
annual_return = income_series.sum()               # Годовая доходность (сумма за год)
avg_daily_return = income_series.mean()           # Средняя дневная доходность
median_daily_return = income_series.median()      # Медианная дневная доходность
highest_daily_return = income_series.max()        # Наивысшая дневная доходность
largest_daily_loss = income_series.min()          # Наибольший дневной убыток

# Вывод
print(f"Период: {dates.min().date()} - {dates.max().date()}")
print(f"Годовая доходность: {annual_return:.2f}")
print(f"Средняя дневная доходность: {avg_daily_return:.2f}")
print(f"Медианная дневная доходность: {median_daily_return:.2f}")
print(f"Наивысшая дневная доходность: {highest_daily_return:.2f}")
print(f"Наибольший дневной убыток: {largest_daily_loss:.2f}")

# При необходимости посмотреть первые значения:
# income_series.head()
```
