1. РќР°РїРёС€РёС‚Рµ РїСЂРѕРіСЂР°РјРјСѓ, РєРѕС‚РѕСЂР°СЏ РіРµРЅРµСЂРёСЂСѓРµС‚ Series c РґР°РЅРЅС‹РјРё Рѕ РµР¶РµРґРЅРµРІРЅС‹С… РґРѕС…РѕРґР°С… РѕС‚ С‚СЂРµР№РґРёРЅРіР° Р·Р° РїСЂРѕС€Р»С‹Р№ РіРѕРґ (РёРЅРґРµРєСЃС‹ вЂ“ СЌС‚Рѕ РґР°С‚С‹, СЃСѓРјРјС‹ РґРѕС…РѕРґР° вЂ“ СЃР»СѓС‡Р°Р№РЅС‹Рµ, РЅРѕ СѓС‡С‚РёС‚Рµ, С‡С‚Рѕ РѕРЅРё РјРѕРіСѓС‚ Р±С‹С‚СЊ Рё РѕС‚СЂРёС†Р°С‚РµР»СЊРЅС‹РјРё). РџСЂРѕРіСЂР°РјРјР° РґРѕР»Р¶РЅС‹ РІС‹РІРµСЃС‚Рё: 
- Р“РѕРґРѕРІСѓСЋ РґРѕС…РѕРґРЅРѕСЃС‚СЊ.
- РЎСЂРµРґРЅСЋСЋ РґРЅРµРІРЅСѓСЋ РґРѕС…РѕРґРЅРѕСЃС‚СЊ.
- РњРµРґРёР°РЅРЅСѓСЋ РґРЅРµРІРЅСѓСЋ РґРѕС…РѕРґРЅРѕСЃС‚СЊ.
- РќР°РёРІС‹СЃС€СѓСЋ РґРЅРµРІРЅСѓСЋ РґРѕС…РѕРґРЅРѕСЃС‚СЊ.
- РќР°РёР±РѕР»СЊС€РёР№ РґРЅРµРІРЅРѕР№ СѓР±С‹С‚РѕРє.

```python
import numpy as np
import pandas as pd

# Р¤РёРєСЃРёСЂСѓРµРј seed РґР»СЏ РІРѕСЃРїСЂРѕРёР·РІРѕРґРёРјРѕСЃС‚Рё СЂРµР·СѓР»СЊС‚Р°С‚Р°
np.random.seed(42)

# Р¤РѕСЂРјРёСЂСѓРµРј РґР°С‚С‹ Р·Р° РїСЂРѕС€Р»С‹Р№ РєР°Р»РµРЅРґР°СЂРЅС‹Р№ РіРѕРґ
current_year = pd.Timestamp.today().year
last_year = current_year - 1
dates = pd.date_range(start=f"{last_year}-01-01", end=f"{last_year}-12-31", freq="D")

# Р“РµРЅРµСЂРёСЂСѓРµРј СЃР»СѓС‡Р°Р№РЅС‹Рµ РґРЅРµРІРЅС‹Рµ РґРѕС…РѕРґС‹ (РјРѕРіСѓС‚ Р±С‹С‚СЊ Рё РѕС‚СЂРёС†Р°С‚РµР»СЊРЅС‹Рµ)
# mean=50, std=300 РґР°РµС‚ СЃРјРµСЃСЊ РїСЂРёР±С‹Р»Рё Рё СѓР±С‹С‚РєРѕРІ
daily_income = np.random.normal(loc=50, scale=300, size=len(dates))
income_series = pd.Series(daily_income, index=dates, name="daily_income")

# РЎС‡РёС‚Р°РµРј РјРµС‚СЂРёРєРё
annual_return = income_series.sum()               # Р“РѕРґРѕРІР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ (СЃСѓРјРјР° Р·Р° РіРѕРґ)
avg_daily_return = income_series.mean()           # РЎСЂРµРґРЅСЏСЏ РґРЅРµРІРЅР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ
median_daily_return = income_series.median()      # РњРµРґРёР°РЅРЅР°СЏ РґРЅРµРІРЅР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ
highest_daily_return = income_series.max()        # РќР°РёРІС‹СЃС€Р°СЏ РґРЅРµРІРЅР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ
largest_daily_loss = income_series.min()          # РќР°РёР±РѕР»СЊС€РёР№ РґРЅРµРІРЅРѕР№ СѓР±С‹С‚РѕРє

# Р’С‹РІРѕРґ
print(f"РџРµСЂРёРѕРґ: {dates.min().date()} - {dates.max().date()}")
print(f"Р“РѕРґРѕРІР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ: {annual_return:.2f}")
print(f"РЎСЂРµРґРЅСЏСЏ РґРЅРµРІРЅР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ: {avg_daily_return:.2f}")
print(f"РњРµРґРёР°РЅРЅР°СЏ РґРЅРµРІРЅР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ: {median_daily_return:.2f}")
print(f"РќР°РёРІС‹СЃС€Р°СЏ РґРЅРµРІРЅР°СЏ РґРѕС…РѕРґРЅРѕСЃС‚СЊ: {highest_daily_return:.2f}")
print(f"РќР°РёР±РѕР»СЊС€РёР№ РґРЅРµРІРЅРѕР№ СѓР±С‹С‚РѕРє: {largest_daily_loss:.2f}")

# РџСЂРё РЅРµРѕР±С…РѕРґРёРјРѕСЃС‚Рё РїРѕСЃРјРѕС‚СЂРµС‚СЊ РїРµСЂРІС‹Рµ Р·РЅР°С‡РµРЅРёСЏ:
# income_series.head()
```
