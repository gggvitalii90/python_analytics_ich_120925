### <b><font color='chocolate'>1. GeoPandas</font></b>

<b>Описание набора данных</b><br>
Расходы полиции, уровень преступности и социально-экономические характеристики округов штата Миссисипи, 1982 год.

<b>Основные столбцы (Features):</b><br>
- <i>AREA</i> - Площадь полигона округа (рассчитана программой ArcView)
- <i>PERIMETER</i> - Периметр полигона округа (рассчитанный с помощью ArcView)
- <i>CNTY_</i> - Внутренний идентификатор округа
- <i>CNTY_ID</i> -	дентификатор округа
- <i>NAME</i> - Название округа
- <i>STATE_NAME</i> - Название штата
- <i>STATE_FIPS</i> - Код FIPS для штата (символьный)
- <i>CNTY_FIPS</i> - Код FIPS для округа (символьный)
- <i>FIPS</i> - Полный код FIPS, штат в сочетании с округом (символьно)
- <i>FIPSNO</i> - Код FIPS (числовой)
- <i>POLICE</i> - Расходы полиции на душу населения, 1982 год (доллары на человека)
- <i>POP</i> - Население округа, 1982 год
- <i>TAX</i> - Общая сумма налогов на душу населения, 1982 год (доллары на человека)
- <i>TRANSFER</i> - Доходы штата и федерального бюджета на душу населения, 1982 год (1000 долларов на человека)
- <i>INC</i> - Личный доход на одного жителя округа, 1982 год (доллары на человека)
- <i>CRIME</i> - Количество тяжких преступлений на 100 000 жителей, 1981 год
- <i>UNEMP</i> - Процент безработных в 1980 году
- <i>OWN</i> - Процент жилых единиц, находящихся в собственности, 1980 год
- <i>COLLEGE</i> - Процент жителей округа, получивших высшее образование, 1980 год 
- <i>WHITE</i> - Процент белых жителей округа, 1980 год
- <i>COMMUTE</i> - Процент работников округа, которые ездили на работу за пределы округа, 1980 год

```python
gdf_police = gpd.read_file(gds.get_path("geoda.police"))
# gds.data.query_name("geoda.police")
# get_path = gpd.datasets.get_path
gdf_police.info()
```

```python
gdf_police.crs
gdf_police.to_crs(epsg=3857, inplace=True)
gdf_police.crs
```

```python
gdf_police['cenroid'] = gdf_police.geometry.centroid
gdf_police
```

```python
from shapely.geometry import Point
jackson = Point(-90.1848, 32.2988)
jackson = gpd.GeoDataFrame(
    {   'name': 'Jackson',
        'geometry': [jackson]}, crs='EPSG:4326').to_crs(epsg=3857).iloc[0]
```

```python
gdf_police['distance_to_jackson'] = gdf_police.geometry.distance(jackson.geometry)/1000
fix, ax = plt.subplots(figsize=(15, 10))
gdf_police.plot(column='distance_to_jackson', ax=ax, legend=True, cmap='gist_heat', edgecolor='black')
plt.title("Distance to Jackson")
gdf_police['cenroid'].plot(ax=ax, color='blue', markersize=10)
gpd.GeoSeries(jackson.geometry).plot(ax=ax, color='blue', marker='*', markersize=100)
plt.legend(['Centroid', 'Jackson'])
ax.axis('off')

plt.show()
```

```python
import seaborn as sns
# Создание макета дашборда
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize = (15, 15))
# График 1: Распределение Police Expenditure per Capita
sns.histplot(gdf_police['POLICE'], bins=20, ax=ax1)
ax1.set_xlabel('Distribution of Police Expenditure per Capita')
# График 2: Скатерплот Crime Rate & Police Expenditure per Capita
sns.scatterplot(x='CRIME', y='POLICE', data=gdf_police, ax=ax2, color='green')
ax2.set_xlabel('Crime Rate')
ax2.set_xlim(0, 500)
ax2.set_ylabel('Police Expenditure per Capita')
# График 3: Свечи POP_GROUP & POLICE
gdf_police['POP_GROUP'] = pd.qcut(gdf_police['POP'], q=5, labels=['Q1', 'Q2', 'Q3', 'Q4', 'Q5'])
sns.boxplot(x='POP_GROUP', y='POLICE', data=gdf_police, ax=ax3, color='orange')
ax3.set_xlabel('POP_GROUP')
ax3.set_ylabel('POLICE')
ax3.set_ylim(0, 4000)
# График 4: БАР NAME & POLICE
sns.barplot(x='NAME', y='POLICE', data=gdf_police, ax=ax4, color='firebrick', order=gdf_police.sort_values('POLICE', ascending=False)['NAME'].head(10))
ax4.set_xlabel('NAME')
ax4.set_ylabel('POLICE')
ax4.set_xticklabels(ax4.get_xticklabels(), rotation=90, ha='right')


plt.show()
```
