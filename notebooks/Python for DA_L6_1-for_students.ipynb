### <font color='lightcoral'>1. РўР°Р±Р»РёС†С‹ СЃРѕРїСЂСЏР¶РµРЅРЅРѕСЃС‚Рё (crosstab)</font>
<b>РЎРёРЅС‚Р°РєСЃРёСЃ:</b>
-------------------------------
##### pd.crosstab(index, columns, values=None, rownames=None, colnames=None, aggfunc=None, margins=False, margins_name='All', dropna=True)
-------------------------------
- <i>index</i> вЂ” Р·РЅР°С‡РµРЅРёСЏ РґР»СЏ РіСЂСѓРїРїРёСЂРѕРІРєРё РїРѕ СЃС‚СЂРѕРєР°Рј;
- <i>columns</i> вЂ” Р·РЅР°С‡РµРЅРёСЏ РґР»СЏ РіСЂСѓРїРїРёСЂРѕРІРєРё РїРѕ СЃС‚РѕР»Р±С†Р°Рј;
- <i>values</i> вЂ” Р°РіСЂРµРіРёСЂСѓРµРјС‹Р№ СЃС‚РѕР»Р±РµС† (РёР»Рё СЃС‚РѕР»Р±С†С‹), РµРіРѕ Р·РЅР°С‡РµРЅРёСЏ РЅРµРїРѕСЃСЂРµРґСЃС‚РІРµРЅРЅРѕ РѕРїСЂРµРґРµР»СЏСЋС‚ Р·РЅР°С‡РµРЅРёСЏ С‚Р°Р±Р»РёС†С‹ СЃРѕРїСЂСЏР¶РµРЅРЅРѕСЃС‚Рё;
- <i>aggfunc</i> вЂ” С„СѓРЅРєС†РёСЏ, РєРѕС‚РѕСЂР°СЏ Р±СѓРґРµС‚ РїСЂРёРјРµРЅРµРЅР° Рє РєР°Р¶РґРѕР№ РіСЂСѓРїРїРµ Р·РЅР°С‡РµРЅРёР№ values, СЃРіСЂСѓРїРїРёСЂРѕРІР°РЅРЅС‹Рј РїРѕ Р·РЅР°С‡РµРЅРёСЏРј index Рё columns. Р—РЅР°С‡РµРЅРёСЏ СЌС‚РѕР№ С„СѓРЅРєС†РёРё Рё РµСЃС‚СЊ Р·РЅР°С‡РµРЅРёСЏ СЃРІРѕРґРЅРѕР№ С‚Р°Р±Р»РёС†С‹;
- <i>rownames Рё colnames</i> вЂ” РёРјРµРЅР° СЃС‚СЂРѕРє Рё СЃС‚РѕР»Р±С†РѕРІ С‚Р°Р±Р»РёС†С‹ СЃРѕРїСЂСЏР¶РµРЅРЅРѕСЃС‚Рё;
- <i>margins</i> вЂ” РґРѕР±Р°РІР»СЏРµС‚ СЂРµР·СѓР»СЊС‚РёСЂСѓСЋС‰РёР№ СЃС‚РѕР»Р±РµС†/СЃС‚СЂРѕРєСѓ;
- <i>margins_name</i> вЂ” РёРјСЏ СЂРµР·СѓР»СЊС‚РёСЂСѓСЋС‰РµРіРѕ СЃС‚РѕР»Р±РµС†/СЃС‚СЂРѕРєСѓ;
- <i>dropna</i> вЂ” РЅРµ РІРєР»СЋС‡Р°С‚СЊ СЃС‚РѕР»Р±С†С‹, РєРѕС‚РѕСЂС‹Рµ СЃРѕСЃС‚РѕСЏС‚ С‚РѕР»СЊРєРѕ РёР· NaN;

```python
pd.crosstab(appointments['Р’СЂР°С‡'], appointments['РџРѕСЃС‚Р°РІР»РµРЅРЅС‹Р№ РґРёР°РіРЅРѕР·'], margins = True, margins_name = 'РС‚РѕРі')
```

### <font color='lightcoral'>2. РЎРІРѕРґРЅС‹Рµ С‚Р°Р±Р»РёС†С‹ (pivot-table)</font>
<b>РЎРёРЅС‚Р°РєСЃРёСЃ:</b>
-------------------------------
##### pd.pivot_table(data, values=None, index=None, columns=None, aggfunc='mean', fill_value=None, margins=False, dropna=True, margins_name='All')
-------------------------------
- <i>data</i> вЂ” РёСЃС…РѕРґРЅР°СЏ С‚Р°Р±Р»РёС†Р°;
- <i>values</i> вЂ” Р°РіСЂРµРіРёСЂСѓРµРјС‹Р№ СЃС‚РѕР»Р±РµС†, РµРіРѕ Р·РЅР°С‡РµРЅРёСЏ РЅРµРїРѕСЃСЂРµРґСЃС‚РІРµРЅРЅРѕ РѕРїСЂРµРґРµР»СЏСЋС‚ Р·РЅР°С‡РµРЅРёСЏ СЃРІРѕРґРЅРѕР№ С‚Р°Р±Р»РёС†С‹;
- <i>index</i> вЂ” РєР»СЋС‡Рё РґР»СЏ РіСЂСѓРїРїРёСЂРѕРІРєРё, РѕС‚РЅРѕСЃСЏС‚СЃСЏ Рє РёРЅРґРµРєСЃР°Рј СЃРІРѕРґРЅРѕР№ С‚Р°Р±Р»РёС†С‹;
- <i>columns</i> вЂ” РєР»СЋС‡Рё РґР»СЏ РіСЂСѓРїРїРёСЂРѕРІРєРё, РѕС‚РЅРѕСЃСЏС‚СЃСЏ Рє СЃС‚РѕР»Р±С†Р°Рј СЃРІРѕРґРЅРѕР№ С‚Р°Р±Р»РёС†С‹;
- <i>aggfunc</i> вЂ” С„СѓРЅРєС†РёСЏ, РєРѕС‚РѕСЂР°СЏ Р±СѓРґРµС‚ РїСЂРёРјРµРЅРµРЅР° Рє РєР°Р¶РґРѕР№ РіСЂСѓРїРїРµ Р·РЅР°С‡РµРЅРёР№ values, СЃРіСЂСѓРїРїРёСЂРѕРІР°РЅРЅС‹Рј РїРѕ Р·РЅР°С‡РµРЅРёСЏРј index Рё columns. Р—РЅР°С‡РµРЅРёСЏ СЌС‚РѕР№ С„СѓРЅРєС†РёРё Рё РµСЃС‚СЊ Р·РЅР°С‡РµРЅРёСЏ СЃРІРѕРґРЅРѕР№ С‚Р°Р±Р»РёС†С‹. Р•СЃР»Рё РїРµСЂРµРґР°РµС‚СЃСЏ СЃРїРёСЃРѕРє С„СѓРЅРєС†РёР№, С‚Рѕ СЃРІРѕРґРЅР°СЏ С‚Р°Р±Р»РёС†Р° РёРјРµРµС‚ РёРµСЂР°СЂС…РёС‡РµСЃРєРёРµ РёРјРµРЅР° РєРѕР»РѕРЅРѕРє, РІРµСЂС…РЅРёРµ Р·РЅР°С‡РµРЅРёСЏ РєРѕС‚РѕСЂС‹С… вЂ” РёРјРµРЅР° С„СѓРЅРєС†РёР№;
- <i>fill_value</i> вЂ” Р·РЅР°С‡РµРЅРёСЏ РґР»СЏ Р·Р°РјРµРЅС‹ РїСЂРѕРїСѓСЃРєРѕРІ;
- <i>dropna</i> вЂ” РЅРµ РІРєР»СЋС‡Р°С‚СЊ СЃС‚РѕР»Р±С†С‹, РєРѕС‚РѕСЂС‹Рµ СЃРѕСЃС‚РѕСЏС‚ С‚РѕР»СЊРєРѕ РёР· NaN;
- <i>margins</i> вЂ” РґРѕР±Р°РІР»СЏРµС‚ СЂРµР·СѓР»СЊС‚РёСЂСѓСЋС‰РёР№ СЃС‚РѕР»Р±РµС†/СЃС‚СЂРѕРєСѓ;
- <i>margins_name</i> вЂ” РёРјСЏ СЂРµР·СѓР»СЊС‚РёСЂСѓСЋС‰РµРіРѕ СЃС‚РѕР»Р±РµС†/СЃС‚СЂРѕРєСѓ.

### <font color='lightcoral'>3. Р—Р°РґР°С‡Р°</font>
РўР°Р±Р»РёС†Р° sales_dataset.csv СЃРѕРґРµСЂР¶РёС‚ РїРѕР»СЏ:
- date вЂ” РґР°С‚Р° РїСЂРѕРґР°Р¶Рё
- region вЂ” СЂРµРіРёРѕРЅ
- product вЂ” С‚РѕРІР°СЂ
- units вЂ” РєРѕР»РёС‡РµСЃС‚РІРѕ РїСЂРѕРґР°РЅРЅС‹С… РµРґРёРЅРёС†
- price вЂ” С†РµРЅР° Р·Р° РµРґРёРЅРёС†Сѓ
1. РџРѕСЃС‡РёС‚Р°Р№С‚Рµ, СЃРєРѕР»СЊРєРѕ СЂР°Р· РєР°Р¶РґС‹Р№ С‚РѕРІР°СЂ РїСЂРѕРґР°РІР°Р»СЃСЏ РІ РєР°Р¶РґРѕРј СЂРµРіРёРѕРЅРµ (С‡Р°СЃС‚РѕС‚Р° С‚СЂР°РЅР·Р°РєС†РёР№). РСЃРїРѕР»СЊР·СѓР№СЊС‚Рµ crosstab.
2. РџРѕСЃС‚СЂРѕР№С‚Рµ С‚Р°Р±Р»РёС†Сѓ, РіРґРµ РїРѕ СЃС‚СЂРѕРєР°Рј - РјРµСЃСЏС† РїСЂРѕРґР°Р¶Рё, РїРѕ СЃС‚РѕР»Р±С†Р°Рј - СЂРµРіРёРѕРЅС‹ Рё РїСЂРѕРґСѓРєС‚С‹. РџСЂРѕРґСѓРєС‚С‹ РІРѕР·СЊРјРёС‚Рµ С‚РѕР»СЊРєРѕ Laptop Рё Tablet. Р—РЅР°С‡РµРЅРёСЏ - СЃСѓРјРјР° РІС‹СЂСѓС‡РєРё. РСЃРїРѕР»СЊР·СѓР№С‚Рµ pivot_table.

```python
# Р—Р°РіСЂСѓР·РєР° РґР°РЅРЅС‹С…
df_sales = pd.read_csv("../data/sales_dataset.csv")
df_sales
```

```python
display(pd.crosstab(df_sales['product'], df_sales['region']))


df_sales['date'] = pd.to_datetime(df_sales['date'])
df_sales['revenue'] = df_sales['units'] * df_sales['price']
df_sales.info()
```

```python
# РџРѕСЃС‡РёС‚Р°Р№С‚Рµ, СЃРєРѕР»СЊРєРѕ СЂР°Р· РєР°Р¶РґС‹Р№ С‚РѕРІР°СЂ РїСЂРѕРґР°РІР°Р»СЃСЏ РІ РєР°Р¶РґРѕРј СЂРµРіРёРѕРЅРµ (С‡Р°СЃС‚РѕС‚Р° С‚СЂР°РЅР·Р°РєС†РёР№).
pd.crosstab(df_sales['product'], df_sales['region'], aggfunc = 'sum', values = df_sales['units'],margins = True, margins_name = 'РС‚РѕРі')
```

```python
# РџРѕСЃС‚СЂРѕР№С‚Рµ С‚Р°Р±Р»РёС†Сѓ, РіРґРµ РїРѕ СЃС‚СЂРѕРєР°Рј - РјРµСЃСЏС† РїСЂРѕРґР°Р¶Рё, РїРѕ СЃС‚РѕР»Р±С†Р°Рј - СЂРµРіРёРѕРЅС‹ Рё РїСЂРѕРґСѓРєС‚С‹
# РџСЂРѕРґСѓРєС‚С‹ РІРѕР·СЊРјРёС‚Рµ С‚РѕР»СЊРєРѕ Laptop Рё Tablet
# Р—РЅР°С‡РµРЅРёСЏ - СЃСѓРјРјР° РІС‹СЂСѓС‡РєРё

pd.pivot_table(df_sales[df_sales['product'].isin(['Laptop', 'Tablet'])], index = df_sales['date'].dt.month_name(), 
               columns = ['region', 'product'], aggfunc = 'sum', values = ['revenue'])
```

```python
# Р¤СѓРЅРєС†РёСЏ, РіРµРЅРµСЂРёСЂСѓСЋС‰Р°СЏ СЃР»СѓС‡Р°Р№РЅРѕРµ С‡РёСЃР»Рѕ Рё Р·Р°СЃС‹РїР°СЋС‰Р°СЏ РЅР° 1 СЃРµРєСѓРЅРґСѓ, РµСЃР»Рё С‡РёСЃР»Рѕ РЅРµС‡РµС‚РЅРѕРµ
import time
import numpy as np

def sleep_if_odd():
    rvs = np.random.randint(low=0, high=int(1e9))
    randbit = rvs % 2
    if randbit % 2 == 1:
        time.sleep(1)
```

```python
%time sleep_if_odd()
```
