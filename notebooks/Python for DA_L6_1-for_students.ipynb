<i> Задача: </i><br>
Посчитать, сколько раз за предыдущий месяц каждый врач ставил какой-либо диагноз. <br>
Результаты представить в виде таблицы, в которой посчитать суммы по строкам и столбцам, т.е. сколько врач сделал приемов за месяц и сколько раз конкретный диагноз поставлен всеми врачами.


### <font color='lightcoral'>1. Таблицы сопряженности (crosstab)</font>
<b>Синтаксис:</b>
-------------------------------
##### pd.crosstab(index, columns, values=None, rownames=None, colnames=None, aggfunc=None, margins=False, margins_name='All', dropna=True)
-------------------------------
- <i>index</i> — значения для группировки по строкам;
- <i>columns</i> — значения для группировки по столбцам;
- <i>values</i> — агрегируемый столбец (или столбцы), его значения непосредственно определяют значения таблицы сопряженности;
- <i>aggfunc</i> — функция, которая будет применена к каждой группе значений values, сгруппированным по значениям index и columns. Значения этой функции и есть значения сводной таблицы;
- <i>rownames и colnames</i> — имена строк и столбцов таблицы сопряженности;
- <i>margins</i> — добавляет результирующий столбец/строку;
- <i>margins_name</i> — имя результирующего столбец/строку;
- <i>dropna</i> — не включать столбцы, которые состоят только из NaN;

```python
pd.crosstab(appointments['Врач'], appointments['Поставленный диагноз'], margins = True, margins_name = 'тог')
```

### <font color='lightcoral'>2. Сводные таблицы (pivot-table)</font>
<b>Синтаксис:</b>
-------------------------------
##### pd.pivot_table(data, values=None, index=None, columns=None, aggfunc='mean', fill_value=None, margins=False, dropna=True, margins_name='All')
-------------------------------
- <i>data</i> — исходная таблица;
- <i>values</i> — агрегируемый столбец, его значения непосредственно определяют значения сводной таблицы;
- <i>index</i> — ключи для группировки, относятся к индексам сводной таблицы;
- <i>columns</i> — ключи для группировки, относятся к столбцам сводной таблицы;
- <i>aggfunc</i> — функция, которая будет применена к каждой группе значений values, сгруппированным по значениям index и columns. Значения этой функции и есть значения сводной таблицы. Если передается список функций, то сводная таблица имеет иерархические имена колонок, верхние значения которых — имена функций;
- <i>fill_value</i> — значения для замены пропусков;
- <i>dropna</i> — не включать столбцы, которые состоят только из NaN;
- <i>margins</i> — добавляет результирующий столбец/строку;
- <i>margins_name</i> — имя результирующего столбец/строку.

```python
pd.pivot_table(appointments, index='Врач', columns='Поставленный диагноз', 
               aggfunc = "count", values = 'Время приема', fill_value = 0, margins=True)
```

```python
pd.pivot_table(appointments, index=['Специальность', 'Врач'], columns='Поставленный диагноз', 
               aggfunc = "count", values = 'Время приема', margins=True)
```

### <font color='lightcoral'>3. Задача</font>
Таблица sales_dataset.csv содержит поля:
- date — дата продажи
- region — регион
- product — товар
- units — количество проданных единиц
- price — цена за единицу
1. Посчитайте, сколько раз каждый товар продавался в каждом регионе (частота транзакций). спользуйьте crosstab.
2. Постройте таблицу, где по строкам - месяц продажи, по столбцам - регионы и продукты. Продукты возьмите только Laptop и Tablet. Значения - сумма выручки. спользуйте pivot_table.

```python
# Загрузка данных
df_sales = pd.read_csv("../data/sales_dataset.csv")
df_sales
```

```python
display(pd.crosstab(df_sales['product'], df_sales['region']))


df_sales['date'] = pd.to_datetime(df_sales['date'])
df_sales['revenue'] = df_sales['units'] * df_sales['price']
df_sales.info()
```

```python
# Посчитайте, сколько раз каждый товар продавался в каждом регионе (частота транзакций).
pd.crosstab(df_sales['product'], df_sales['region'], aggfunc = 'sum', values = df_sales['units'],margins = True, margins_name = 'тог')
```

```python
# Постройте таблицу, где по строкам - месяц продажи, по столбцам - регионы и продукты
# Продукты возьмите только Laptop и Tablet
# Значения - сумма выручки

pd.pivot_table(df_sales[df_sales['product'].isin(['Laptop', 'Tablet'])], index = df_sales['date'].dt.month_name(), 
               columns = ['region', 'product'], aggfunc = 'sum', values = ['revenue'])
```

##### <font color='lightcoral'>5.1.  %time, %timeit, %%time</font>

```python
# Функция, генерирующая случайное число и засыпающая на 1 секунду, если число нечетное
import time
import numpy as np

def sleep_if_odd():
    rvs = np.random.randint(low=0, high=int(1e9))
    randbit = rvs % 2
    if randbit % 2 == 1:
        time.sleep(1)
```

```python
%time sleep_if_odd()
```

<b> CPU times </b>
Это время, которое процессор реально потратил на выполнение кода.
- user — время, затраченное пользовательским кодом (Python-кодом).
- sys — время, затраченное системными вызовами (например, работа с файлами, памятью, ввод/вывод).
- total — сумма user + sys.

<b> Wall time </b>
Это реальное прошедшее время по часам (как будто засекли секундомером).

Отличие: wall time может быть чуть больше CPU total, если процесс ждал чего-то (например, доступа к памяти или ОС переключала задачи).

<b>Синтаксис %timeit [-n<N> -r<R>]</b>
- n - Выполнить заданный код N раз в цикле. Если значение N не указано, оно определяется автоматически, обеспечивая достаточную точность.
- r - Количество повторений R, каждое из которых состоит из N циклов, и вычисляется средний результат. По умолчанию: 7.

```python
%timeit -n 10 -r 10 sleep_if_odd()
```
