### <font color='coral'>4. РћР±СЂР°Р±РѕС‚РєР° РїСЂРѕРїСѓС‰РµРЅРЅС‹С… Р·РЅР°С‡РµРЅРёР№</font>

<b> fillna() </b> - СЌС‚Рѕ РјРµС‚РѕРґ, РєРѕС‚РѕСЂС‹Р№ РёСЃРїРѕР»СЊР·СѓРµС‚СЃСЏ РґР»СЏ Р·Р°РїРѕР»РЅРµРЅРёСЏ РѕС‚СЃСѓС‚СЃС‚РІСѓСЋС‰РёС… Р·РЅР°С‡РµРЅРёР№ РІ DataFrame СѓРєР°Р·Р°РЅРЅС‹РјРё Р·РЅР°С‡РµРЅРёСЏРјРё. РџРѕР·РІРѕР»СЏРµС‚ Р·Р°РјРµРЅРёС‚СЊ NaN (Not a Number) РёР»Рё РѕС‚СЃСѓС‚СЃС‚РІСѓСЋС‰РёРµ Р·РЅР°С‡РµРЅРёСЏ РЅР° СѓРєР°Р·Р°РЅРЅРѕРµ СЃРєР°Р»СЏСЂРЅРѕРµ Р·РЅР°С‡РµРЅРёРµ, СЃР»РѕРІР°СЂСЊ РёР»Рё РёСЃРїРѕР»СЊР·РѕРІР°С‚СЊ СЂР°Р·Р»РёС‡РЅС‹Рµ РјРµС‚РѕРґС‹, С‚Р°РєРёРµ РєР°Рє 'ffill' (РїСЂСЏРјРѕРµ Р·Р°РїРѕР»РЅРµРЅРёРµ) РёР»Рё 'bfill' (РѕР±СЂР°С‚РЅРѕРµ Р·Р°РїРѕР»РЅРµРЅРёРµ).

<b> РЎРёРЅС‚Р°РєСЃРёСЃ: </b> <br>
DataFrame.fillna(
                value=None, 
                method=None,
                axis=None, 
                inplace=False,
                limit=None, 
                downcast=None
                )<br>
<i>РџР°СЂР°РјРµС‚СЂС‹:</i>
- value - РєРѕРЅСЃС‚Р°РЅС‚Р°, СЃР»РѕРІР°СЂСЊ (dict), Series РёР»Рё DataFrame. Р—РЅР°С‡РµРЅРёРµ, РёСЃРїРѕР»СЊР·СѓРµРјРѕРµ РґР»СЏ Р·Р°РїРѕР»РЅРµРЅРёСЏ РЅРµРґРѕСЃС‚Р°СЋС‰РёС… Р·РЅР°С‡РµРЅРёР№.
- method - {'backfill', 'bfill', 'pad', 'ffill', None}, РїРѕ СѓРјРѕР»С‡Р°РЅРёСЋ None. РњРµС‚РѕРґ, РёСЃРїРѕР»СЊР·СѓРµРјС‹Р№ РґР»СЏ Р·Р°РїРѕР»РЅРµРЅРёСЏ РїСЂРѕРїСѓС‰РµРЅРЅС‹С… Р·РЅР°С‡РµРЅРёР№.
    - ffill/pad - Р·Р°РїРѕР»РЅРµРЅРёРµ РїСЂРѕРїСѓСЃРєРѕРІ РїСЂРµРґС‹РґСѓС‰РёРј РёР·РІРµСЃС‚РЅС‹Рј Р·РЅР°С‡РµРЅРёРµРј;
    - bfill/backfill - Р·Р°РїРѕР»РЅРµРЅРёРµ РїСЂРѕРїСѓСЃРєРѕРІ СЃР»РµРґСѓСЋС‰РёРј РёР·РІРµСЃС‚РЅС‹Рј Р·РЅР°С‡РµРЅРёРµРј;
    - РќРµ СЂР°Р±РѕС‚Р°РµС‚ РѕРґРЅРѕРІСЂРµРјРµРЅРЅРѕ СЃ РєРѕРЅСЃС‚Р°РЅС‚Р°РјРё Рё СЃР»РѕРІР°СЂСЏРјРё;
- axis - {0 РёР»Рё 'index', 1 РёР»Рё 'columns'}. РћСЃСЊ, РІРґРѕР»СЊ РєРѕС‚РѕСЂРѕР№ Р·Р°РїРѕР»РЅСЏСЋС‚СЃСЏ РїСЂРѕРїСѓС‰РµРЅРЅС‹Рµ Р·РЅР°С‡РµРЅРёСЏ.
- inplace - bool, РїРѕ СѓРјРѕР»С‡Р°РЅРёСЋ False. РњРѕРґРёС„РёС†РёСЂРѕРІР°С‚СЊ DataFrame РЅР° РјРµСЃС‚Рµ (РЅРµ СЃРѕР·РґР°РІР°С‚СЊ РЅРѕРІС‹Р№ РѕР±СЉРµРєС‚).
- limit - int, РїРѕ СѓРјРѕР»С‡Р°РЅРёСЋ None. РњР°РєСЃРёРјР°Р»СЊРЅРѕРµ РєРѕР»РёС‡РµСЃС‚РІРѕ РїРѕСЃР»РµРґРѕРІР°С‚РµР»СЊРЅС‹С… NaN РґР»СЏ Р·Р°РїРѕР»РЅРµРЅРёСЏ. Р Р°Р±РѕС‚Р°РµС‚ СЃ ffill Рё bfill;
- downcast - dict item -> dtype. РџРѕР·РІРѕР»СЏРµС‚ СЃСЂР°Р·Сѓ РёР·РјРµРЅРёС‚СЊ С‚РёРї РґР°РЅРЅС‹С… РїРѕСЃР»Рµ Р·Р°РїРѕР»РЅРµРЅРёСЏ NaN. Р Р°Р±РѕС‚Р°РµС‚ С‚РѕР»СЊРєРѕ РґР»СЏ С‡РёСЃР»РѕРІС‹С… СЃС‚РѕР»Р±С†РѕРІ.

```python
# Р—Р°РїРѕР»РЅРµРЅРёРµ РєРѕРЅСЃС‚Р°РЅС‚РѕР№
df.fillna(0)
```

```python
# Р—Р°РїРѕР»РЅРµРЅРёРµ СЃР»РѕРІР°СЂРµРј
df.fillna({'Age': 100, 'Gender': 'Female'})
```

```python
# Р—Р°РїРѕР»РЅРµРЅРёРµ Series
df.fillna(pd.Series({'Age': 100, 'Gender': 'Female'}))
```

```python
# Р—Р°РїРѕР»РЅРµРЅРёРµ DataFrame
# СЂР°Р±РѕС‚Р°РµС‚ С‚РѕР»СЊРєРѕ РµСЃР»Рё РёРЅРґРµРєСЃС‹ СЃРѕРІРїР°РґР°СЋС‚.
df2 = pd.DataFrame({
    "ID": [None, 3, 11],
    "Name": [None, 'Alice', 'Piter'],
    "Age": [None, 33, 45]
})
display(df2)
df.fillna(df2)
```

```python
# ffill + limit
df_temp = pd.DataFrame({"A":[1, np.nan, np.nan, np.nan, 5, np.nan, np.nan]})
df_temp
```

```python
df_temp.fillna(method = 'ffill', limit = 2) #df_temp.ffill(limit = 1)
```

РњРµС‚РѕРґ <b> dropna() </b> РёСЃРїРѕР»СЊР·СѓРµС‚СЃСЏ РґР»СЏ СѓРґР°Р»РµРЅРёСЏ РѕС‚СЃСѓС‚СЃС‚РІСѓСЋС‰РёС… Р·РЅР°С‡РµРЅРёР№ РёР· DataFrame. Р­С‚РѕС‚ РјРµС‚РѕРґ РїРѕР·РІРѕР»СЏРµС‚ СѓРґР°Р»СЏС‚СЊ СЃС‚СЂРѕРєРё РёР»Рё СЃС‚РѕР»Р±С†С‹, РІ РєРѕС‚РѕСЂС‹С… Р»СЋР±РѕР№ РёР»Рё РІСЃРµ СЌР»РµРјРµРЅС‚С‹ СЏРІР»СЏСЋС‚СЃСЏ NaN (Not a Number).

<b> РЎРёРЅС‚Р°РєСЃРёСЃ: </b> <br>
DataFrame.dropna(
                axis = 0, 
                how = 'any',
                thresh = None, 
                subset = None,
                inplace = False, 
                )<br>
<i>РџР°СЂР°РјРµС‚СЂС‹:</i>

```python
df_temp = pd.DataFrame({
    'TextColumn': ['Р­С‚Рѕ РѕР±СЂР°Р·РµС†     РўР•РљРЎРўРђ СЃРѕ РЎРџР•Р¦РРђР›Р¬РќР«РњР СЃРёРјРІРѕР»Р°РјРё!',
                   '^^^^Р•С‰Рµ РѕРґРёРЅ РѕР±СЂР°Р·РµС†, СЃ РїСѓРЅРєС‚СѓР°С†РёРµР№!!!',
                   'РќРѕСЂРјР°Р»РёР·Р°С†РёСЏ :::С‚РµРєСЃС‚РѕРІС‹С… РґР°РЅРЅС‹С… РІР°Р¶РЅР° РґР»СЏ Р·Р°РґР°С‡ РќР›Рџ']
})
df_temp
```

```python
# r - Р±СѓРєРІР°Р»СЊРЅРѕ "СЃС‹СЂР°СЏ СЃС‚СЂРѕРєР°". Рў.Рµ. РѕР±СЂР°С‚РЅС‹Рµ СЃР»СЌС€Рё РѕСЃС‚Р°СЋС‚СЃСЏ РєР°Рє РµСЃС‚СЊ, РЅРµ РёРЅС‚РµСЂРїСЂРµС‚РёСЂСѓСЋС‚СЃСЏ Python
# [] - Р»СЋР±РѕР№ РёР· СЃРёРјРІРѕР»РѕРІ, РЅР°С…РѕРґСЏС‰РёС…СЃСЏ РІРЅСѓС‚СЂРё СЃРєРѕР±РѕРє
# ^ - РѕС‚СЂРёС†Р°РЅРёРµ
# \w - Р»СЋР±РѕР№ Р±СѓРєРІРµРЅРЅРѕ-С†РёС„СЂРѕРІРѕР№ СЃРёРјРІРѕР» Рё _
# \s - Р»СЋР±РѕР№ РїСЂРѕР±РµР»СЊРЅС‹Р№ СЃРёРјРІРѕР»: РїСЂРѕР±РµР», С‚Р°Р±СѓР»СЏС†РёСЏ, РїРµСЂРµРЅРѕСЃ СЃС‚СЂРѕРєРё
# РёС‚РѕРіРѕ - Р»СЋР±РѕР№ СЃРёРјРІРѕР», РєРѕС‚РѕСЂС‹Р№ РЅРµ СЏРІР»СЏРµС‚СЃСЏ Р±СѓРєРІРѕР№, С†РёС„СЂРѕР№, РЅРёР¶РЅРёРј РїРѕРґС‡С‘СЂРєРёРІР°РЅРёРµРј Рё РїСЂРѕР±РµР»РѕРј
df_temp['TextColumn'] = df_temp['TextColumn'].str.replace(pat = r'[^\w\s]', repl = '', regex=True)
df_temp
```

### <font color='coral'>10. Р РµС€РµРЅРёРµ Р·Р°РґР°С‡</font>
РЎРґРµР»Р°Р№С‚Рµ РїСЂРµРґРІР°СЂРёС‚РµР»СЊРЅСѓСЋ РѕР±СЂР°Р±РѕС‚РєСѓ Рё РѕС‡РёСЃС‚РєСѓ РґР°РЅРЅС‹С… РґР°С‚Р°С„СЂРµР№РјР°

```python
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emily'],
                  'Age': [25, 30, np.nan, 35, 40],
                  'Gender': ['Female', 'Male', 'Male', 'Male', np.nan],
                  'Height (cm)': [160, 175, 180, np.nan, 165],
                  'Weight (kg)': [55, np.nan, 75, 80, 60]})
df
```

```python
# 1. РџРµСЂРµРёРјРµРЅРѕРІР°РЅРёРµ РєРѕР»РѕРЅРѕРє Height (cm), Weight (kg)
df.rename(columns={'Height (cm)': 'Height', 'Weight (kg)': 'Weight'}, inplace=True)
df
```

```python
# 2. Р—Р°РїРѕР»РЅРёРј NaN РІ СЃС‚РѕР»Р±С†Рµ Age РјРµРґРёР°РЅРЅС‹Р№ Р·РЅР°С‡РµРЅРёРµРј, Height Рё Weight - СЃСЂРµРґРЅРёРјРё, Gender - РјРѕРґРѕР№
df.fillna({'Age':df['Age'].median(),
          'Height':df['Height'].mean(),
          'Weight':df['Weight'].mean(),
          'Gender':df['Gender'].mode()[0]}, inplace = True)
df
```

```python
# 3. РџРµСЂРµРІРµРґРµРј С‚РёРї РґР°РЅРЅС‹С… РІ СЃС‚РѕР»Р±С†e Age РІ int
df['Age'] = df['Age'].astype(int, copy = True, errors = 'raise')
df
```
