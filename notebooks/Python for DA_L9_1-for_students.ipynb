### <font color='coral'>4. Обработка пропущенных значений</font>

<b> fillna() </b> - это метод, который используется для заполнения отсутствующих значений в DataFrame указанными значениями. Позволяет заменить NaN (Not a Number) или отсутствующие значения на указанное скалярное значение, словарь или использовать различные методы, такие как 'ffill' (прямое заполнение) или 'bfill' (обратное заполнение).

<b> Синтаксис: </b> <br>
DataFrame.fillna(
                value=None, 
                method=None,
                axis=None, 
                inplace=False,
                limit=None, 
                downcast=None
                )<br>
<i>Параметры:</i>
- value - константа, словарь (dict), Series или DataFrame. Значение, используемое для заполнения недостающих значений.
- method - {'backfill', 'bfill', 'pad', 'ffill', None}, по умолчанию None. Метод, используемый для заполнения пропущенных значений.
    - ffill/pad - заполнение пропусков предыдущим известным значением;
    - bfill/backfill - заполнение пропусков следующим известным значением;
    - Не работает одновременно с константами и словарями;
- axis - {0 или 'index', 1 или 'columns'}. Ось, вдоль которой заполняются пропущенные значения.
- inplace - bool, по умолчанию False. Модифицировать DataFrame на месте (не создавать новый объект).
- limit - int, по умолчанию None. Максимальное количество последовательных NaN для заполнения. Работает с ffill и bfill;
- downcast - dict item -> dtype. Позволяет сразу изменить тип данных после заполнения NaN. Работает только для числовых столбцов.

```python
# Заполнение константой
df.fillna(0)
```

```python
# Заполнение словарем
df.fillna({'Age': 100, 'Gender': 'Female'})
```

```python
# Заполнение Series
df.fillna(pd.Series({'Age': 100, 'Gender': 'Female'}))
```

```python
# Заполнение DataFrame
# работает только если индексы совпадают.
df2 = pd.DataFrame({
    "ID": [None, 3, 11],
    "Name": [None, 'Alice', 'Piter'],
    "Age": [None, 33, 45]
})
display(df2)
df.fillna(df2)
```

```python
# method = 'ffill'
# df.fillna(method = 'ffill') # df.ffill()
df.ffill()
```

```python
# method = 'bfill'
df.fillna(method = 'bfill') # df.bfill()
```

```python
df.fillna(method = 'pad', axis = 1)
```

```python
# ffill + limit
df_temp = pd.DataFrame({"A":[1, np.nan, np.nan, np.nan, 5, np.nan, np.nan]})
df_temp
```

```python
df_temp.fillna(method = 'ffill', limit = 2) #df_temp.ffill(limit = 1)
```

```python
# downcast
# df.fillna(method = 'ffill', downcast = 'int')
df.ffill(downcast = 'int')
df1 = df.ffill()
pd.to_numeric(df1['Age'], downcast="integer")
```

Метод <b> dropna() </b> используется для удаления отсутствующих значений из DataFrame. Этот метод позволяет удалять строки или столбцы, в которых любой или все элементы являются NaN (Not a Number).

<b> Синтаксис: </b> <br>
DataFrame.dropna(
                axis = 0, 
                how = 'any',
                thresh = None, 
                subset = None,
                inplace = False, 
                )<br>
<i>Параметры:</i>
- axis: {0 или 'index', 1 или 'columns'}. Ось, по которой нужно отбросить пропущенные значения.
- how: {'any', 'all'}, по умолчанию 'any'. Определяет, удалять ли строку или столбец из DataFrame, где любой или все элементы являются NaN.
- thresh: int, по умолчанию None. Минимальное количество непустых значений, которое должно быть в строке или столбце, чтобы она не была удалена.
- subset: имена столбцов, которые необходимо включить в работу функции
- inplace: bool, по умолчанию False. Модифицировать DataFrame на месте (не создавать новый объект).

```python
df.dropna(axis = 0)
```

```python
df.dropna(axis = 0, how = 'all')
```

```python
df.dropna(axis = 0, thresh = 5)
```

```python
df.dropna(axis = 0, subset = ['Age', 'Salary'])
```

### <font color='coral'>6. Удаление дубликатов</font>

Дублирующиеся строки в наборе данных могут исказить результаты анализа и привести к неверным выводам. Очень важно выявлять и удалять дубликаты строк, чтобы обеспечить целостность и качество данных.

<b>Синтаксис:</b>
DataFrame.drop_duplicates(subset = None, keep = 'first', inplace = False, ignore_index = False) <br>
Параметры:
- subset - Подмножество столбцов для выявления дубликатов. По умолчанию учитываются все столбцы.
- keep: {'first', 'last', False}, по умолчанию 'first'. Определяет, какой дубликат сохранять. 'first' сохраняет первое вхождение, 'last' сохраняет последнее вхождение, а False удаляет все дубликаты.
- inplace: bool, по умолчанию False. Модифицировать DataFrame на месте (не создавать новый объект).
- ignore_index: bool, по умолчанию False. Если значение True, то результирующий DataFrame будет иметь новый RangeIndex.

```python
# получить все дублирующиеся строки
df[df.duplicated(keep = False)]
```

```python
df.drop_duplicates(keep = 'first', inplace = True, ignore_index = False)
df
```

### <font color='coral'>8. зменение типов данных</font>

Метод <b> astype() </b> используется для изменения типов данных столбцов в DataFrame на указанный тип данных.<br>
<b>Синтаксис:</b> DataFrame.astype(dtype, copy = True, errors = 'raise')<br>
Параметры:

- dtype - тип данных, или dict из имени столбца и типа данных. Тип данных, в который нужно преобразовать DataFrame. Это может быть один тип данных для всех столбцов или словарь, определяющий тип данных для каждого столбца.
- copy - bool, по умолчанию True. Возвращать ли новый DataFrame с указанными типами данных или модифицировать существующий DataFrame на месте.
- errors - {'raise', 'ignore'}, по умолчанию 'raise'. Определяет, как обрабатывать ошибки при преобразовании типов данных. Если 'raise', то при любой ошибке преобразования типа данных будет выдана ошибка. Если 'ignore', то ошибки будут проигнорированы, и будут возвращены исходные данные.

```python
import pandas as pd
df = pd.read_csv('../data/data_for_cleaning.csv')
df.head()
```

```python
df_copy = df.astype({'Department': 'category'}, copy = False, errors = 'raise')
df_copy.dtypes
```

### <font color='coral'>9. Нормализация текстовых данных</font>

<b> Нормализация текстовых данных </b> - это этап предварительной обработки, который часто выполняется перед задачами обработки естественного языка (NLP), такими как классификация текстов, анализ настроения или моделирование тем. Она включает в себя преобразование текстовых данных к единообразному формату путем применения таких операций, как удаление строчных букв и специальных символов.

##### <font color='coral'>9.1. Перевод текстовых данных в нижний регистр</font>

```python
df['Department'].str.lower()
```

##### <font color='coral'>9.2. Удаление специальных символов</font>
<b>Синтаксис:</b> DataFrame['текстовое_поле'].str.replace(pat, repl = '', regex = False)<br>
Параметры:
- pat - какой текст заменить или шаблон регулярного выражения;
- repl - на что заменить;
- regex - True, если переданную строку считывать как регулярное выражение, иначе False.

```python
df_temp = pd.DataFrame({
    'TextColumn': ['Это образец     ТЕКСТА со СПЕЦАЛЬНЫМ символами!',
                   '^^^^Еще один образец, с пунктуацией!!!',
                   'Нормализация :::текстовых данных важна для задач НЛП']
})
df_temp
```

```python
# r - буквально "сырая строка". Т.е. обратные слэши остаются как есть, не интерпретируются Python
# [] - любой из символов, находящихся внутри скобок
# ^ - отрицание
# \w - любой буквенно-цифровой символ и _
# \s - любой пробельный символ: пробел, табуляция, перенос строки
# итого - любой символ, который не является буквой, цифрой, нижним подчёркиванием и пробелом
df_temp['TextColumn'] = df_temp['TextColumn'].str.replace(pat = r'[^\w\s]', repl = '', regex=True)
df_temp
```

### <font color='coral'>10. Решение задач</font>
Сделайте предварительную обработку и очистку данных датафрейма

```python
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emily'],
                  'Age': [25, 30, np.nan, 35, 40],
                  'Gender': ['Female', 'Male', 'Male', 'Male', np.nan],
                  'Height (cm)': [160, 175, 180, np.nan, 165],
                  'Weight (kg)': [55, np.nan, 75, 80, 60]})
df
```

```python
# 1. Переименование колонок Height (cm), Weight (kg)
df.rename(columns={'Height (cm)': 'Height', 'Weight (kg)': 'Weight'}, inplace=True)
df
```

```python
# 2. Заполним NaN в столбце Age медианный значением, Height и Weight - средними, Gender - модой
df.fillna({'Age':df['Age'].median(),
          'Height':df['Height'].mean(),
          'Weight':df['Weight'].mean(),
          'Gender':df['Gender'].mode()[0]}, inplace = True)
df
```

```python
# 3. Переведем тип данных в столбцe Age в int
df['Age'] = df['Age'].astype(int, copy = True, errors = 'raise')
df
```
