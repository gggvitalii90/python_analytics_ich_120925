### <b><font color='saddlebrown'>Анализ временных рядов</font></b>

##### <b><font color='saddlebrown'>1. Декомпозиция ряда на компоненты</font></b>

```python
# мпорт библиотек
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose # Модуль декмопозиции
from statsmodels.datasets import get_rdataset # Здесь датасеты
import statsmodels.api as sm
import matplotlib.dates as mdates # Для красивой отрисовки графика
```

```python
# Загрузка данных
df_air = get_rdataset('AirPassengers').data
df_air.index = pd.date_range(start = '1949-01',
                            periods = len(df_air.index),
                            freq = 'ME').to_period()
df_air.index = df_air.index.to_timestamp()
df_air.head(20)
```

```python
get_rdataset('AirPassengers').title
```

```python
# Декомопозиция временного ряда
decomposition = seasonal_decompose(df_air['value'], model = 'multiplicative', period = 12)
```

```python
# Выделение компонент в результате декомпозиции
trend = decomposition.trend
seasonal = decomposition.seasonal
residual = decomposition.resid
```

```python
trend.head(20)
```

```python
# Построение графика компонентов

locator = mdates.MonthLocator(bymonth = (1, 4, 7, 10))  # метки на Янв, Апр, юль, Окт
formatter = mdates.DateFormatter('%b %Y')          # формат: Месяц Год

plt.figure(figsize = (12, 8))

# График с данными
plt.subplot(4, 1, 1)
plt.plot(df_air['value'], label = 'Original')
plt.legend(loc = 'best')
plt.gca().xaxis.set_major_locator(locator)
plt.gca().xaxis.set_major_formatter(formatter)
plt.xticks(rotation = 90)

# График тренда
plt.subplot(4, 1, 2)
plt.plot(trend, label = 'Trend')
plt.legend(loc = 'best')

# График сезонности
plt.subplot(4, 1, 3)
plt.plot(seasonal, label = 'Seasonality')
plt.legend(loc = 'best')
plt.gca().xaxis.set_major_locator(locator)
plt.gca().xaxis.set_major_formatter(formatter)
plt.xticks(rotation = 90)

# График шума
plt.subplot(4, 1, 4)
plt.plot(residual, label = 'Residuals')
plt.legend(loc = 'best')

plt.tight_layout()
plt.show()
```

<b>Выводы:</b>
1. сходные данные (Original)
- График показывает количество авиапассажиров по месяцам с 1949 по 1960 год.
- Видно устойчивый рост с течением времени
- Также заметны регулярные колебания — сезонность, летом пик, зимой спад.
- Амплитуда колебаний растёт вместе с уровнем ряда (что подтверждает использование мультипликативной модели).
Вывод: временной ряд нестационарен, имеет как тренд, так и сезонность, и сезонность масштабируется с уровнем.

2. Тренд (Trend)
- Тренд показывает долгосрочный рост пассажиропотока.

3. Сезонность (Seasonality)
- Сезонность повторяется каждые 12 месяцев (по period = 12), что видно по одинаковым паттернам.

4. Шумы (Residuals)
- Шумы показывают нерегулярные колебания, небольшие выбросы и шум ряда
- Если остатки имеют структуру, значит, модель может быть не совсем точной или выбранный период не идеально подходит.

##### <b><font color='saddlebrown'>3. Автокореляционные функции</font></b>

###### <b><font color='saddlebrown'>3.1. ACF</font></b>

```python
# мпорт необходимого модуля
from statsmodels.graphics.tsaplots import plot_acf
```

```python
# Построение графика автокорреляционной функции
plot_acf(df_air['value'], lags = 50)
plt.title('Автокорреляционная функция (АКФ)')
plt.show()
```

"Смотрим" график:
- ось X: лаги (от 0 до 50)
- ось Y: значения автокорреляции (от -1 до 1)
- синие вертикальные линии: автокорреляция для каждого лага
- светло-серые горизонтальные линии: доверительный интервал (95%)

"Читаем" график:
- Лаг 0: всегда = 1 (ряд полностью коррелирован с самим собой).
- Лаг 1: насколько текущее значение связано с предыдущим.
- Лаг k > 1: насколько ряд зависит от значения k шагов назад.
- Если линия выходит за границы доверительного интервала, то автокорреляция значима на этом лаге.
- Если линия внутри серого коридора, то автокорреляция незначима (может быть случайной).
- Постепенное "затухание" признак нестационарного ряда (тренд).

###### <b><font color='saddlebrown'>3.2. PACF</font></b>

```python
# мпорт необходимого модуля
from statsmodels.graphics.tsaplots import plot_pacf
```

```python
# Построение графика частичной автокорреляционной функции
plot_pacf(df_air['value'], lags = 50)
plt.title('Частичная автокорреляционная функция (PACF)')
plt.show()
```

"Смотрим" график:
- ось X: лаги (от 0 до 50)
- ось Y: значения частичной автокорреляции (от -1 до 1)
- синие вертикальные линии: автокорреляция для каждого лага
- светло-серые горизонтальные линии: доверительный интервал (95%)

"Читаем" график:
- Лаг 0: всегда = 1 (ряд полностью коррелирован с самим собой).
- Лаг 1: насколько текущее значение связано с предыдущим.
- Лаг k > 1: прямая зависимость на лаге k.
- Если линия выходит за границы доверительного интервала, то частичная автокорреляция значима на этом лаге.
- Если линия внутри серого коридора, то автокорреляция незначима (может быть случайной).
- Чередование положительных и отрицательных значений может указывать на сезонность.

##### <b><font color='saddlebrown'>4. Повторная выборка</font></b>

Воспользуемся набором данных Superstore.xlsx, чтобы продемонстрировать повторную выборку. Этот набор данных содержит данные о продажах с временными метками, которые мы можем агрегировать до различных временных частот.

```python
# Загрузка данных
df_superstore = pd.read_excel('../data/Superstore.xlsx', sheet_name = 'Orders')
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
df_superstore.set_index('Order Date', inplace = True)
df_superstore.head(20)
```

```python
df_superstore['Category'].value_counts()
```

```python
# Аггрегирование продажи по месяцам
monthly_sales = df_superstore['Sales'].resample('ME').sum()
monthly_sales.head(3)
```

```python
# Построение графика продаж за месяц
plt.figure(figsize = (10, 6))
plt.plot(monthly_sales, label = 'Monthly Sales')
plt.title('Ежемесячные продажи магазина')
plt.xlabel('Дата')
plt.ylabel('Продажи')
plt.legend()
plt.show()
```

```python
# Агрегирование продаж по неделям
weekly_sales = df_superstore['Sales'].resample('W').sum()
weekly_sales.head(3)
```

```python
# Построение графика недельных продаж
plt.figure(figsize = (10, 6))
plt.plot(weekly_sales, label = 'Weekly Sales')
plt.title('Еженедельные продажи в магазина')
plt.xlabel('Дата')
plt.ylabel('Продажи')
plt.legend()
plt.show()
```

```python
# Агрегирование продаж по дням
daily_sales = df_superstore['Sales'].resample('D').sum()
daily_sales.head(3)
```

```python
# Построение графика ежедневных продаж
plt.figure(figsize = (10, 6))
plt.plot(daily_sales, label = 'Ежедневные продажи')
plt.title('Ежедневные продажи супермаркета')
plt.xlabel('Дата')
plt.ylabel('Продажи')
plt.legend()
plt.show()
```
