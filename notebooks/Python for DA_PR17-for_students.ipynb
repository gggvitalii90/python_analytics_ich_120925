##### <b><font color='saddlebrown'>Задача 1. Средний чек в интернет-магазине</font></b>
<b>Описание:</b>
Маркетологи запустили европейскую версию сайта. Им нужно оценить истинный средний чек пользователей в евро, чтобы спрогнозировать выручку. Из генеральной совокупности взяли выборку из 150 заказов. Средний чек в выборке — 45 €, выборочное стандартное отклонение — 12 €. Истинная дисперсия ГС неизвестна. <br>

<b>Вопросы:</b>
- Постройте 95% ДИ для среднего чека.
- Интерпретируйте результат для бизнеса (Что Вы предложите при расчете финансовой модели на следующий месяц?)

```python
# <b><font color='saddlebrown'>Задача 1. Средний чек в интернет-магазине</font></b>
# <b>Описание:</b>
# Маркетологи запустили европейскую версию сайта. Им нужно оценить истинный средний чек пользователей в евро, чтобы спрогнозировать выручку. Из генеральной совокупности взяли выборку из 150 заказов. Средний чек в выборке — 45 €, выборочное стандартное отклонение — 12 €. Истинная дисперсия ГС неизвестна. <br>

# <b>Вопросы:</b>
# - Постройте 95% ДИ для среднего чека.
# - Интерпретируйте результат для бизнеса (Что Вы предложите при расчете финансовой модели на следующий месяц?)

n = 150
sigma = 12
mean = 45
confidence = 0.95

sem = sigma / np.sqrt(n)
ci = stats.t.interval(confidence = confidence, df = n-1, loc = mean, scale = sem)
print(f"Доверительный интервал (t-распределение): ({ci[0]:.2f}, {ci[1]:.2f})")

# Интерпретация для бизнеса:
# Мы можем с 95% уверенностью утверждать, что истинный средний чек пользователей находится в этом интервале.
# Для финансовой модели на следующий месяц можно использовать среднее значение интервала или его границы для оценки возможного диапазона выручки.
# А какую границу лучше использовать для финансовой модели? Нижнюю, верхнюю или среднее значение интервала?
# Ответ зависит от целей бизнеса и уровня риска, который компания готова принять.
# Если компания хочет быть консервативной и учитывать возможные риски, можно использовать нижнюю границу интервала. 
# Это позволит избежать переоценки выручки и обеспечить более устойчивое финансовое планирование.
# Если же компания готова принимать более высокий риск и рассчитывать на оптимистичный сценарий, можно использовать верхнюю границу интервала. 
# Это может помочь в планировании роста и инвестиций.
# Среднее значение интервала может быть использовано как наиболее вероятная оценка среднего чека, 
# но важно помнить, что оно не учитывает возможные колебания и неопределенности, связанные с выборкой.


```

##### <b><font color='saddlebrown'>Задача 3. Сравнение затрат на привлечение (CAC)</font></b>
<b> Описание: </b> Рекламная кампания была переведена на европейский рынок. По старым кампаниям исторически известно, что истинное стандартное отклонение стоимости привлечения (CAC) стабильно и равно $\sigma = 5$ €. По новой крупной выборке из 250 кампаний средний CAC составил 18 €.

<b>Вопросы:</b>

- Постройте 95% ДИ для нового среднего CAC
- Пересекает ли он старое плановое значение в 19 €?
- Можно ли масштабировать кампанию?

```python
mean = 18
mean_st = 19
sigma = 5
n = 250
confidence = 0.95
sem = sigma / np.sqrt(n)
ci = stats.norm.interval(confidence = confidence, loc = mean, scale = sem)
print(f"Доверительный интервал (нормальное распределение): ({ci[0]:.2f}, {ci[1]:.2f})")
ci_st = stats.norm.interval(confidence = confidence, loc = mean_st, scale = sem)
print(f"Доверительный интервал (t-распределение): ({ci_st[0]:.2f}, {ci_st[1]:.2f})")
```

##### <b><font color='saddlebrown'>Задача 4. Срок службы курьерских сумок</font></b>
<b> Описание: </b> Служба доставки в Берлине тестирует новый материал для сумок курьеров. Протестировали всего 12 сумок. Они прослужили в среднем 180 дней со стандартным отклонением 25 дней. Предполагается, что время службы распределено нормально.

<b>Вопросы:</b>

- Постройте 95% ДИ.
- Как бы Вы рекомендовали закладывать бюджет на замену сумок?

```python
# Задача 4. Срок службы курьерских сумок
n = 12
mean_life = 180
std_life = 25  # выборочное стандартное отклонение
confidence = 0.95

# Маленькая выборка и неизвестная sigma -> используем t-распределение
sem = std_life / np.sqrt(n)
ci = stats.t.interval(confidence=confidence, df=n - 1, loc=mean_life, scale=sem)

print(f"95% ДИ для среднего срока службы: ({ci[0]:.2f}, {ci[1]:.2f}) дней")

# Рекомендация для бюджета: ориентироваться на консервативную (нижнюю) границу ДИ
print(f"Для бюджетирования замены разумно закладывать срок около {ci[0]:.0f} дней (нижняя граница ДИ).")
print("Для базового сценария можно использовать среднюю оценку 180 дней.")
```

##### <b><font color='saddlebrown'>Задача 6. Конверсия сайта (CR) после редизайна</font></b>
<b> Описание: </b> После изменения интерфейса европейского сайта из 1000 случайных посетителей покупку совершили 50 человек. Точечная доля (конверсия) равна $5\%$.

<b>Вопросы:</b>

- Постройте 95% ДИ для истинной конверсии генеральной совокупности
- Если для окупаемости платного трафика бизнесу критически необходима конверсия не менее 4%, можем ли мы гарантировать успешность редизайна?

```python
# Задача 6. Конверсия сайта (CR) после редизайна
n = 1000
x = 50

# Из условия: конверсия 5%
p = 0.05

# Для 95% ДИ используем z = 1.96
z = 1.96

# Стандартная ошибка доли
sem = np.sqrt(p * (1 - p) / n)

# Границы доверительного интервала (ручной расчет)
ci_low = p - z * sem
ci_high = p + z * sem

print(f"95% ДИ для CR (ручной расчет): ({ci_low * 100:.2f}%, {ci_high * 100:.2f}%)")

# Тот же расчет через функцию scipy
ci_norm = stats.norm.interval(confidence=0.95, loc=p, scale=sem)
print(f"95% ДИ для CR (stats.norm.interval): ({ci_norm[0] * 100:.2f}%, {ci_norm[1] * 100:.2f}%)")

# Проверка бизнес-условия: нужна конверсия не ниже 4%
if ci_low >= 0.04:
    print("Можно гарантировать успешность редизайна (нижняя граница >= 4%).")
else:
    print("Гарантировать успешность нельзя (нижняя граница < 4%).")
```

##### <b><font color='saddlebrown'>Задача 7. Доля оттока пользователей (Churn Rate)</font></b>
<b> Описание: </b> Необанк в еврозоне зафиксировал, что из 400 случайных пользователей, зарегистрировавшихся в прошлом месяце, 84 прекратили пользоваться сервисом (отток = 21%).

<b>Вопросы:</b>

- Построить 95% ДИ для Churn Rate
- Что вы укажете в отчете для инвесторов?

```python
# Задача 7. Доля оттока пользователей (Churn Rate)
n = 400
x = 84

# Из условия: отток 21%
p = 0.21

# Стандартная ошибка доли
sem = np.sqrt(p * (1 - p) / n)

# 95% ДИ через stats.norm.interval
ci = stats.norm.interval(confidence=0.95, loc=p, scale=sem)

print(f"Точечная оценка churn: {p * 100:.2f}%")
print(f"95% ДИ для churn: ({ci[0] * 100:.2f}%, {ci[1] * 100:.2f}%)")
print("Для отчета инвесторам: ожидаемый churn находится примерно в этом диапазоне при текущих данных.")
```

##### <b><font color='saddlebrown'>Задача 8. Кликбейтность баннеров</font></b>
<b> Описание: </b> Отдел маркетинга тестирует баннер под европейский рынок. Из 5000 показов пользователи кликнули 120 раз. Руководитель утверждает: "Наш истинный CTR (Click-Through Rate) точно выше 2%!".

<b>Вопросы:</b>

- Проверить утверждение руководителя. Постройте 95% ДИ через Z-оценку для долей и посмотрите, где находится отметка в 2%.

```python
# Задача 8. Кликбейтность баннеров
n = 5000
x = 120
p = x / n
confidence = 0.95

# Стандартная ошибка доли
sem = np.sqrt(p * (1 - p) / n)

# 95% ДИ через stats.norm.interval
ci = stats.norm.interval(confidence=confidence, loc=p, scale=sem)

print(f"Точечный CTR: {p * 100:.2f}%")
print(f"95% ДИ для CTR: ({ci[0] * 100:.2f}%, {ci[1] * 100:.2f}%)")

target = 0.02
if ci[0] > target:
    print("Утверждение руководителя подтверждается: истинный CTR выше 2%.")
else:
    print("Утверждение руководителя не подтверждается: 2% попадает в ДИ или выше нижней границы.")
```

##### <b><font color='saddlebrown'>Задача 9. Выявление брака при приемке товара</font></b>
<b> Описание: </b> Немецкий ритейлер закупает крупную партию премиальных девайсов. По договору доля брака не должна превышать 1%. Из партии случайным образом выбрали и протестировали 500 штук. Брак обнаружился в 3 устройствах (точечная доля — 0.6%).

<b>Вопросы:</b>

- Постройте 90% доверительный интервал для доли брака. Можно ли без опасений принимать всю партию?

```python
# Задача 9. Выявление брака при приемке товара
n = 500
x = 3
p = x / n
confidence = 0.90

# Стандартная ошибка доли
sem = np.sqrt(p * (1 - p) / n)

# 90% ДИ через stats.norm.interval
ci = stats.norm.interval(confidence=confidence, loc=p, scale=sem)

print(f"Точечная доля брака: {p * 100:.2f}%")
print(f"90% ДИ для доли брака: ({ci[0] * 100:.2f}%, {ci[1] * 100:.2f}%)")

limit = 0.01
if ci[1] <= limit:
    print("Партию можно принимать без опасений: верхняя граница ДИ не превышает 1%.")
else:
    print("Безопасно принять партию нельзя: верхняя граница ДИ превышает 1%.")
```

##### <b><font color='saddlebrown'>Задача 10. Уровень удовлетворенности клиентов (CSAT) после релиза</font></b>
<b> Описание: </b> После обновления интерфейса службы поддержки европейского филиала опросили 300 случайных клиентов. 255 из них поставили оценку "Удовлетворен". До обновления исторический показатель удовлетворенности (CSAT) составлял ровно 88%.

<b>Вопросы:</b>

- Постройте 95% ДИ для новой доли удовлетворенных клиентов с помощью Z-критерия. Изменился ли уровень удовлетворенности клиентов по сравнению со старым значением?

```python
# Задача 10. Уровень удовлетворенности клиентов (CSAT) после релиза
n = 300
x = 255
p = x / n
confidence = 0.95

# Стандартная ошибка доли
sem = np.sqrt(p * (1 - p) / n)

# 95% ДИ через stats.norm.interval
ci = stats.norm.interval(confidence=confidence, loc=p, scale=sem)

print(f"Новая точечная оценка CSAT: {p * 100:.2f}%")
print(f"95% ДИ для нового CSAT: ({ci[0] * 100:.2f}%, {ci[1] * 100:.2f}%)")

old_csat = 0.88
if ci[0] <= old_csat <= ci[1]:
    print("Статистически значимого изменения относительно 88% не видно (старое значение внутри ДИ).")
else:
    print("Есть основания считать, что уровень удовлетворенности изменился (88% вне ДИ).")
```
