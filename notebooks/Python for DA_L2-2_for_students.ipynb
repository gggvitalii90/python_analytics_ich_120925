### <font color='teal'>1. Р•РґРёРЅРёС‡РЅР°СЏ РјР°С‚СЂРёС†Р°, РѕРїРµСЂР°С†РёРё РЅР°Рґ РјР°С‚СЂРёС†Р°РјРё</font>

<b> Р•РґРёРЅРёС‡РЅР°СЏ РјР°С‚СЂРёС†Р° </b>вЂ“ СЌС‚Рѕ РєРІР°РґСЂР°С‚РЅР°СЏ РјР°С‚СЂРёС†Р°, Сѓ РєРѕС‚РѕСЂРѕР№ РЅР° РґРёР°РіРѕРЅР°Р»Рё СЃС‚РѕСЏС‚ РµРґРёРЅРёС†С‹, Р° РІРЅРµ вЂ“ РЅСѓР»Рё. РЎРІРѕР№СЃС‚РІРѕ: РїСЂРё СѓРјРЅРѕР¶РµРЅРёРё РЅР° РµРґРёРЅРёС‡РЅСѓСЋ РјР°С‚СЂРёС†Сѓ РёСЃС…РѕРґРЅР°СЏ РјР°С‚СЂРёС†Р° РЅРµ РёР·РјРµРЅРёС‚СЃСЏ.

```python
import numpy as np
#РЎРѕР·РґР°РЅРёРµ РµРґРёРЅРёС‡РЅРѕР№ РјР°С‚СЂРёС†С‹
#Р¤СѓРЅРєС†РёСЏ eye (РѕР±СЏР·Р°С‚РµР»СЊРЅС‹Р№ РїР°СЂР°РјРµС‚СЂ - РєРѕР»-РІРѕ СЃС‚СЂРѕРє Рё СЃС‚РѕР»Р±С†РѕРІ)
I = np.eye(4)
print(I)
```

```python
#РЈРјРЅРѕР¶РµРЅРёРµ РјР°С‚СЂРёС†С‹ РЅР° РµРґРёРЅРёС‡РЅСѓСЋ
a = np.arange(16).reshape(4,4)
print(a)
print(a @ I)
```

```python
#РЎРѕР·РґР°РЅРёРµ РјР°С‚СЂРёС†С‹ СЃ РїРѕРјРѕС‰СЊСЋ С„СѓРЅРєС†РёРё fromfunction()
# f - С„СѓРЅРєС†РёСЏ, СЃ РїРѕРјРѕС‰СЊСЋ РєРѕС‚РѕСЂРѕР№ Р±СѓРґСѓС‚ Р·Р°РїРѕР»РЅСЏС‚СЊСЃСЏ СЌР»РµРјРµРЅС‚С‹ РјР°СЃСЃРёРІР°
# i - РёРЅРґРµРєСЃС‹ СЃС‚СЂРѕРє
# j - РёРЅРґРµРєСЃС‹ СЃС‚РѕР»Р±С†РѕРІ

def f(i, j):
    print(i)
    print(j)
    return 10 * i + j

print(np.fromfunction(f, (3, 4), dtype=np.int64))
```

```python
def f(i, j):
    return (i+1) * (j + 1)

print(np.fromfunction(f, (9, 9)))
```

```python
def f(i, j):
    return (i+j)**2

print(np.fromfunction(f, (5, 5)))
```

```python
#РћР±СЉРµРґРёРЅРµРЅРёРµ РјР°СЃСЃРёРІРѕРІ
a = np.array([1, 3, 2])
b = np.array([4, 5, 6])
horizontal = np.hstack((a, b))
vertical = np.vstack((a, b))
print(horizontal)
print(vertical)
```

```python
#Р Р°СЃС‰РµРїР»РµРЅРёРµ РјР°СЃСЃРёРІРѕРІ
#Р¤СѓРЅРєС†РёСЏ hsplit (РїРµСЂРІС‹Р№ Р°СЂРіСѓРјРµРЅС‚ - РєР°РєСѓСЋ РјР°С‚СЂРёС†Сѓ, РІС‚РѕСЂРѕР№ - РїРѕСЃР»Рµ РєР°РєРѕРіРѕ СЌР»РµРјРµРЅС‚Р° "СЂР°Р·СЂРµР·Р°С‚СЊ", СЃС‡РёС‚Р°СЏ РѕС‚ 1)
matrix_a = np.arange(1,10).reshape(3,3)
print(matrix_a)
print("--------------")
print(np.hsplit(matrix_a, [2]))
a = np.hsplit(matrix_a, [2])
print(a[0])
print("--------------")
print(a[1])
```

```python
#Р¤СѓРЅРєС†РёСЏ vsplit (РїРµСЂРІС‹Р№ Р°СЂРіСѓРјРµРЅС‚ - РєР°РєСѓСЋ РјР°С‚СЂРёС†Сѓ, РІС‚РѕСЂРѕР№ - РїРѕСЃР»Рµ РєР°РєРѕРіРѕ СЌР»РµРјРµРЅС‚Р° "СЂР°Р·СЂРµР·Р°С‚СЊ", СЃС‡РёС‚Р°СЏ РѕС‚ 1)
print(matrix_a)
print("--------------")
np.vsplit(matrix_a, [1])
```

```python
#РЎСѓРјРјР° РІСЃРµС… СЌР»РµРјРµРЅС‚РѕРІ, СЌР»РµРјРµРЅС‚РѕРІ СЃС‚РѕР»Р±С†РѕРІ, СЌР»РµРјРµРЅС‚РѕРІ СЃС‚СЂРѕРє
b = np.array([[4, 5, 6], [7, 8, 9]])
print(b)
print('РЎСѓРјРјР° РІСЃРµС… СЌР»РµРјРµРЅС‚РѕРІ РјР°СЃСЃРёРІР°', b.sum()) 
print('РЎСѓРјРјР° СЌР»РµРјРµРЅС‚РѕРІ РїРѕ СЃС‚РѕР»Р±С†Р°Рј', b.sum(axis=0))
print('РЎСѓРјРјР° СЌР»РµРјРµРЅС‚РѕРІ РїРѕ СЃС‚СЂРѕРєР°Рј',b.sum(axis=1))
```

<b> РЎР»РµРґ РјР°С‚СЂРёС†С‹ </b>вЂ“ СЃСѓРјРјР° РґРёР°РіРѕРЅР°Р»СЊРЅС‹С… СЌР»РµРјРµРЅС‚РѕРІ.

```python
b = np.array([[4, 5, 6], [7, 8, 9]])
print(np.trace(b))
```

### <font color='teal'>2. РўРµРЅР·РѕСЂС‹ (РјРЅРѕРіРѕРјРµСЂРЅС‹Рµ РјР°СЃСЃРёРІС‹)</font>

```python
print(X.sum(axis=1)) #axis = 1 РѕР±РѕР·РЅР°С‡Р°РµС‚, С‡С‚Рѕ РјС‹ РІС‹РєР»СЋС‡Р°РµРј СЃС‚СЂРѕРєРё, С‚Рѕ РµСЃС‚СЊ РІ РєР°Р¶РґРѕРј РІРЅРµС€РЅРµРј РёР·РјРµСЂРµРЅРёРё СЃСѓРјРјРёСЂСѓРµРј РїРѕ СЃС‚РѕР»Р±С†Р°Рј
```

### <font color='teal'>4. Р РµС€РµРЅРёРµ Р·Р°РґР°С‡</font>

1) Р§С‚Рѕ Р±СѓРґРµС‚ РІС‹РІРµРґРµРЅРѕ РІ СЃР»РµРґСѓСЋС‰РёС… СЃР»СѓС‡Р°СЏС… Рё РїРѕС‡РµРјСѓ?

```python
a = np.array([[0, 0, 0],
              [10, 10, 10],
              [20, 20, 20],
              [30, 30, 30]])
b = np.array([0, 1, 2])
print(a)
print("--------------")
print(b)
result = a / (b + 1)
print(result)
```

РЎРѕР·РґР°Р№С‚Рµ РґРІР° РјР°СЃСЃРёРІР°: РїРµСЂРІС‹Р№ - РґРІСѓРјРµСЂРЅС‹Р№ РјР°СЃСЃРёРІ (РјР°С‚СЂРёС†Р°) СЂР°Р·РјРµСЂРѕРј 6x4, Р·Р°РїРѕР»РЅРµРЅРЅС‹Р№ С‡РёСЃР»Р°РјРё РѕС‚ 10 РґРѕ 33; РІС‚РѕСЂРѕР№ - РѕРґРЅРѕРјРµСЂРЅС‹Р№ РјР°СЃСЃРёРІ РґР»РёРЅС‹ 4 СЃ С‡РёСЃР»Р°РјРё РѕС‚ 15 РґРѕ 18. Р’С‹РїРѕР»РЅРёС‚Рµ РїРѕСЌР»РµРјРµРЅС‚РЅРѕРµ РґРµР»РµРЅРёРµ РјР°С‚СЂРёС†С‹ РЅР° РѕРґРЅРѕРјРµСЂРЅС‹Р№ РјР°СЃСЃРёРІ, РёСЃРїРѕР»СЊР·СѓСЏ РїСЂР°РІРёР»Р° Broadcasting, Рё РІС‹С‡РёСЃР»РёС‚Рµ СЃР»РµРґ СЂРµР·СѓР»СЊС‚РёСЂСѓСЋС‰РµР№ РјР°С‚СЂРёС†С‹.

```python
a = np.arange(10, 34).reshape(6, 4)
print(a)
print("--------------")
b = np.arange(15,19)
СЃ = a / b
print(СЃ)
print("--------------")
res_c = np.trace(СЃ) 
print(res_c)
```

РЎРѕР·РґР°Р№С‚Рµ РґРІСѓРјРµСЂРЅС‹Р№ РјР°СЃСЃРёРІ СЂР°Р·РјРµСЂРѕРј 3x8, Р·Р°РїРѕР»РЅРµРЅРЅС‹Р№ С‡РёСЃР»Р°РјРё РѕС‚ 1 РґРѕ 24. 
Р’С‹РїРѕР»РЅРёС‚Рµ СЃСѓРјРјРёСЂРѕРІР°РЅРёРµ РїРѕ РїРµСЂРІРѕР№ РѕСЃРё (axis=0), РїРѕСЃС‡РёС‚Р°Р№С‚Рµ СЃСЂРµРґРЅРµРµ Р·РЅР°С‡РµРЅРёРµ РїРѕ РІС‚РѕСЂРѕР№ РѕСЃРё. 
Р’С‹РІРµРґРёС‚Рµ РїРѕР»СѓС‡РµРЅРЅС‹Р№ СЂРµР·СѓР»СЊС‚Р°С‚.

```python
x = np.arange(1, 25).reshape(3,8)
print(x)
print("--------------")
y = np.sum(x, axis=0)
print(y)
print("--------------")
z = np.sum(x, axis=1)
print(z)  
print("--------------") 
СЃ = np.sum(x, axis=(0, 1))
print(СЃ)
print("--------------") 
avg = np.mean(x, axis=1)
print(avg) 
```
