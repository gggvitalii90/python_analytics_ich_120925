```python
import pandas as pd
import numpy as np
d = {'one': pd.Series(range(6), index=list('abdefg')),
     'two': pd.Series(range(7), index=list('abcdefg')),
     'three': pd.Series(np.random.normal(size=7), index=list('abcdefg'))}
df = pd.DataFrame(d)
df
```

```python
# Р’С‹С‡РёСЃР»РµРЅРёРµ СѓСЃР»РѕРІРёР№
df['flag'] = df['two'] > 2
df
```

```python
# Р—Р°РїРѕР»РЅРµРЅРёРµ РєРѕРЅСЃС‚Р°РЅС‚Р°РјРё
df['ones'] = 1
df
```

### <font color='lightcoral'>6. Р РµС€РµРЅРёРµ Р·Р°РґР°С‡</font>

Р’СЂР°С‡ РЅР° РїСЂРѕС‚СЏР¶РµРЅРёРё РґРЅСЏ РёР·РјРµСЂСЏРµС‚ РїР°С†РёРµРЅС‚Сѓ С‚РµРјРїРµСЂР°С‚СѓСЂСѓ РєР°Р¶РґС‹Рµ 3 С‡Р°СЃР° РІ С‚РµС‡РµРЅРёРµ 2 РЅРµРґРµР»СЊ. РўР°РєР¶Рµ РїР°С†РёРµРЅС‚Сѓ РЅРµРѕР±С…РѕРґРёРјРѕ СЃРїР°С‚СЊ СЃ 11 РІРµС‡РµСЂР° РґРѕ 7 СѓС‚СЂР°. РљР°Р¶РґС‹Р№ РґРµРЅСЊ РёР·РјРµСЂРµРЅРёСЏ С‚РµРјРїРµСЂР°С‚СѓСЂС‹ РЅР°С‡РёРЅР°СЋС‚СЃСЏ РІ 8 С‡Р°СЃРѕРІ. РџРµСЂРІРѕРµ РёР·РјРµСЂРµРЅРёРµ 13 Р°РїСЂРµР»СЏ 2026 РіРѕРґР°. РћРїСЂРµРґРµР»РёС‚Рµ РјРѕРјРµРЅС‚С‹ РІСЂРµРјРµРЅРё, РєРѕРіРґР° РЅСѓР¶РЅРѕ РёР·РјРµСЂРёС‚СЊ РїР°С†РёРµРЅС‚Сѓ С‚РµРјРїРµСЂР°С‚СѓСЂСѓ.

```python
# РЁР°Рі 1
# РћРїСЂРµРґРµР»РёРј РїРµСЂРёРѕРґС‹ РёР·РјРµСЂРµРЅРёСЏ С‚РµРјРїРµСЂР°С‚СѓСЂС‹ РґР»СЏ РѕРґРЅРѕРіРѕ РґРЅСЏ
periods = pd.timedelta_range(start='8h', freq='3h', end='23h')
periods
```

```python
# РЁР°Рі 2
# РџРѕСЃС‚СЂРѕРёРј РІСЂРµРјРµРЅРЅРѕР№ СЂСЏРґ, РІ С‚РµС‡РµРЅРёРµ РєРѕС‚РѕСЂРѕРіРѕ РЅРµРѕР±С…РѕРґРёРјРѕ РїР°С†РёРµРЅС‚Сѓ РёР·РјРµСЂСЏС‚СЊ С‚РµРјРїРµСЂР°С‚СѓСЂСѓ
dates = pd.to_datetime('2026.04.13') + pd.timedelta_range(start=0, freq='1D', end='2W')
dates
```

```python
# РЁР°Рі 3
# РЎРѕСЃС‚Р°РІРёРј СЃРїРёСЃРѕРє РґР°С‚С‹ Рё РІСЂРµРјРµРЅРё, РІ РєРѕС‚РѕСЂРѕРµ РЅРµРѕР±С…РѕРґРёРјРѕ РёР·РјРµСЂСЏС‚СЊ С‚РµРјРїРµСЂР°С‚СѓСЂСѓ
# 3.1. РўР°Рє РєР°Рє РєР°Р¶РґС‹Р№ РґРµРЅСЊ РЅРµРѕР±С…РѕРґРёРјРѕ РёР·РјРµСЂСЏС‚СЊ С‚РµРјРїРµСЂР°С‚СѓСЂСѓ 6 СЂР°Р·, С‚Рѕ, РєР°Р¶РґР°СЏ РґР°С‚Р° Р±СѓРґРµС‚ РїРѕРІС‚РѕСЂСЏС‚СЊСЃСЏ 6 СЂР°Р·
dates_new = dates.repeat(len(periods))
dates_new
```

```python
# 3.2. РЎРµР№С‡Р°СЃ РЅРµРѕР±С…РѕРґРёРјРѕ РІСЂРµРјСЏ РёР·РјРµСЂРµРЅРёСЏ С‚РµРјРїРµСЂР°С‚СѓСЂС‹ РїРѕРІС‚РѕСЂРёС‚СЊ СЃС‚РѕР»СЊРєРѕ СЂР°Р·, РєР°РєРѕРµ РєРѕР»РёС‡РµСЃС‚РІРѕ РґРЅРµР№ 
# periods.repeat(len(dates))
# np.tile(A, reps) 
# A вЂ” РёСЃС…РѕРґРЅС‹Р№ РјР°СЃСЃРёРІ
# reps вЂ” СЃРєРѕР»СЊРєРѕ СЂР°Р· РїРѕРІС‚РѕСЂСЏС‚СЊ (РјРѕР¶РµС‚ Р±С‹С‚СЊ С‡РёСЃР»РѕРј РёР»Рё РєРѕСЂС‚РµР¶РµРј)
periods_new = pd.to_timedelta(np.tile(periods, len(dates)))
periods_new
```

```python
# 3.3. Рљ РєР°Р¶РґРѕР№ РґР°С‚Рµ РґРѕР±Р°РІР»СЏРµРј timedelta Рё РїРѕР»СѓС‡РёРј СЃРїРёСЃРѕРє РґР°С‚С‹/РІСЂРµРјРµРЅРё 
time = dates_new + periods_new
time
```
