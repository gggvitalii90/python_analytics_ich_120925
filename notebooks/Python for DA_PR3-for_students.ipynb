### <font color='teal'>1. Анализ веб-логов</font>
Условия:<br>
Есть данные о посещениях сайта: каждая строка — пользователь, столбцы — дни недели (Пн–Вс).

Задачи:
1. Сгенерировать случайные данные о 10 пользователях сайта.
2. Найти пользователя, который посещал сайт чаще всего.
3. Найти день недели с максимальной посещаемостью.
4. Создать массив, где указано, сколько раз каждый пользователь посещал сайт больше 3 раз в день.

```python
import numpy as np
np.random.seed(42)
days = ["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Вс"]
print(days)
arr = np.random.randint(0, 6, size=(10, 7))
print(arr)
print("-------------------------------")
pt_user = np.sum(arr, axis=1)
print(f"User: {np.argmax(pt_user)} - {np.max(pt_user)} ")
print("-------------------------------")
pt_day = np.sum(arr, axis=0)
print(f"Day: {days[np.argmax(pt_day)]} - {np.max(pt_day)} ")
print("-------------------------------")
print(arr > 3) 
print("-------------------------------") 
pt_total = np.sum(arr > 3, axis=1)
print(f"Total: \n {pt_total[:, np.newaxis]}")
print("-------------------------------") 
new_gt_3_1 = pt_total.reshape(-1, 1)
print(new_gt_3_1)
```

### <font color='teal'>2. Продажа товаров со склада</font>
Условия:<br>
Компания управляет тремя складами, на которых хранятся три вида товаров: A, B, C.<br>
Количество товаров на складе (строки — склады, столбцы — товары):<br>
| Склад       | A   | B   | C  |
| ----------- | ----|-----|----|
| 1           | 50  | 30  | 20 |
| 2           | 40  | 50  | 30 |
| 3           | 60  | 20  | 40 |

Цена за единицу товара:<br>
- A — 10 евро
- B — 20 евро
- C — 15 евро<br>
Задачи:
1. Создать необходимую матрицу
2. Рассчитать, на какую сумму хранятся товары на каждом складе, и общую сумму всех товаров на складе
3. В день Х было куплено с каждого склада по 10 единиц каждого вида товара.
4. На следующий день остатки товара были проданы оптовому покупателю по заниженной стоимости. Известно, что с первого склада продали на сумму 560 евро, со второго - на 770 евро и с третьего - на 690 евро. За какую стоимость были проданы товары данному покупателю?

```python
matrix = np.array([[50,30,20], [40,50,30], [60,20,40]])
price = np.array([10, 20, 15])
sum_matrix = matrix*price
print("SK: \n", sum_matrix)

print("-------------------------------")
total_sk = np.sum(sum_matrix, axis=1)
print(f"Total SK: {total_sk}")
total = np.sum(sum_matrix)
print(f"Total: {total}")
matrix_10 = matrix - 10
print("SK - 10: \n", matrix_10)
print("-------------------------------")
price = np.array([560, 770, 690])
print("Price: \n", price)
print("-------------------------------")
det = np.linalg.det(matrix_10)
print(f"Determinant: {det}")   
print("-------------------------------")
inv_matirx_10 = np.linalg.inv(matrix_10)
print("Inverse Matrix: \n", inv_matirx_10)
print("-------------------------------")
result = inv_matirx_10 @ price
print("Result: \n", result)

```

### <font color='teal'>3. Стоимость акций</font>
Пусть у вас есть данные о ежедневных изменениях стоимости акций компании в течение недели. Найдите изменение цены за каждый день, а также дни с минимальным и максимальным изменением. 
(Подсказка: Не забудьте, что когда вы ищите изменение в ценах, то кол-во значений уменьшается на 1, т.к. первое значение не с чем сравнивать.)

Пример вывода:<br>
Изменения цен за день: [ 5  5 -8 -4 17  5]<br>
Лучший день: суббота<br>
Худший день: четверг<br>

```python
prices = np.array([100, 105, 110, 102, 98, 115, 120])
days = ["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Вс"]

changes = []
for i in range(1, len(prices)):
    changes.append(int(prices[i] - prices[i-1]))
print("Изменения цен за день:", changes)

best_day_idx = np.argmax(changes) + 1
worst_day_idx = np.argmin(changes) + 1

print("Лучший день:", days[best_day_idx])
print("Худший день:", days[worst_day_idx])
```

```python
np.random.seed(42)
prices = np.random.randint(100, 110, size=7)
days = ["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Вс"]
print("Цены на акции:", prices)
changes = np.diff(prices)
print("Изменения цен за день:", changes)
best_day_idx = np.argmax(changes) + 1
worst_day_idx = np.argmin(changes) + 1
abs_best_day_idx = np.argmax(abs(changes)) + 1
abs_worst_day_idx = np.argmin(abs(changes)) + 1
print("Лучший день:", days[best_day_idx])
print("Худший день:", days[worst_day_idx])
print("День с наибольшим абсолютным изменением:", days[abs_best_day_idx])
print("День с наименьшим абсолютным изменением:", days[abs_worst_day_idx])
```
