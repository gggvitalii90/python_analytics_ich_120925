```python
from dash import Dash, html, dcc, callback, Output, Input
import plotly.express as px

df = pd.read_csv('../data/spotify_songs.csv')

df_track = df[df['track_popularity'] > 80]

df_genres = df['playlist_genre'].unique().tolist()


app = Dash()

app.layout = [
html.Div(html.H2('My First App with Graph and Controls')),
dcc.RadioItems(options=df_genres, 
               value=df_genres[0], id='controls-and-radio-item'),
dcc.Graph(figure={}, id='controls-and-graph')
]
@callback(
Output(component_id='controls-and-graph', component_property='figure'),
Input(component_id='controls-and-radio-item', component_property='value')
)
def update_graph(col_chosen):
    fig = px.scatter(df_track[df_track['playlist_genre'] == col_chosen]
                 , x = 'track_popularity'
                 , y = 'duration_ms'
           
) 
    return fig

app.run(port= 8052)
```
