### <u><b><font color='skyblue'>3. Подграфики</font></b></u>

```python
# Создаем примерные данные
x = np.linspace(0, 2 * np.pi, 400)
y = np.sin(x ** 2)

plt.figure(figsize = (12, 6))
# Создайте рисунок с двумя подплощадками
plt.subplot(1, 2, 1)

# Нанесите данные на первый подграфик
plt.plot(x, y)
plt.title('Line Plot')

# Постройте данные на втором подграфике
plt.subplot(1, 2, 2)
plt.scatter(x, y)
plt.title('Scatter Plot')

# Отобразите рисунок
plt.show()
```

##### <b><font color='skyblue'>GridSpec</font></b>

```python
fig = plt.figure(figsize = (10, 8))
gs = GridSpec(3, 3, figure = fig)

ax1 = fig.add_subplot(gs[0, :])
ax2 = fig.add_subplot(gs[1, 0:2]) # gs[1, :-1]
ax3 = fig.add_subplot(gs[1:, 2]) # gs[1:, -1]
ax4 = fig.add_subplot(gs[2, 0]) # gs[-1, 0]
ax5 = fig.add_subplot(gs[2, 1]) # gs[-1, -2]

# Построение данных на подграфиках
ax1.plot(x, y)
ax2.plot(x, y)
ax3.plot(x, y)
ax4.plot(x, y)
ax5.plot(x, y)

plt.show()
```

##### <b><font color='skyblue'>subplot_mosaic</font></b>
Способ создать несколько подграфиков с произвольным расположением, задавая макет с помощью текстовой схемы

```python
# Задаем схему расположения графиков
# тройные кавычки - для многострочности
layout = """
AB
CC
"""

fig, axes = plt.subplot_mosaic(layout, figsize = (8, 6))

axes['A'].plot(x, y)
axes['B'].scatter(x, y)
axes['C'].bar(np.arange(10), np.random.rand(10))

plt.show()
```

### <b><font color='skyblue'>4. Seaborn</font></b>
<u>Seaborn - это библиотека Python, построенная поверх Matplotlib и предназначенная для создания привлекательных и информативных статистических графиков. </u>

##### <b><font color='skyblue'>4.1. Графики распределения</font></b>
Эти графики помогают понять распределение и плотность точек данных.

```python
# Гистограмма с графиком плотности вероятности
# Загружаем данные выборки
data = sns.load_dataset("iris")

# Гистограмма с помощью KDE
sns.histplot(data['sepal_length'], kde = True)
plt.title('Гистограмма с KDE')
plt.show()
```

```python
# rug график - показывает каждую отдельную точку данных как маленькую черточку на оси.
sns.kdeplot(data['sepal_length'], fill = True)
sns.rugplot(data['sepal_length'])
plt.title('KDE Plot with Rug Plot')
plt.show()
```

```python
# rug + kde + гистограмма
sns.histplot(data['sepal_length'], kde = True, color = 'skyblue')
sns.rugplot(data['sepal_length'], color ='black')
```

##### <b><font color='skyblue'>4.2. Столбчатая диаграмма</font></b>

```python
sns.barplot(x = 'species', y = 'sepal_length', data = data)
plt.title('Bar Plot')
plt.show()
```

##### <b><font color='skyblue'>4.3. Boxplot</font></b>

```python
sns.boxplot(x = 'species', y = 'sepal_length', data = data)
plt.title('Box Plot')
plt.show()
```

##### <b><font color='skyblue'>4.4. Violinplot</font></b>
Внутри - boxplot<br>

##### <b><font color='skyblue'>4.6. Lineplot</font></b>
Синня зона показывает диапазон, в котором с 95%-ной вероятностью находится истинное среднее значение y при данном x.<br>
Отображением этой зоны можно управлять с помощью параметра errorbar:
- errorbar = None - не отображать
- errorbar ('ci', 95) - по умолчанию
- errorbar ('ci', 68) - построить 68% ДИ

##### <u><b><font color='skyblue'>4.8. Тепловые карты</font></b></u>
Использует цвет для представления различных уровней величины данных. Полезны для визуализации корреляционной матрицы между несколькими переменными, позволяя с первого взгляда определить высококоррелированные или обратнокоррелированные переменные.

```python
import pandas as pd

# Загрузка данных выборки
data = sns.load_dataset("iris")
data = data[["sepal_length", "sepal_width", "petal_length", "petal_width"]]

# Рассчитайте корреляционную матрицу
correlation_matrix = data.corr()
print(correlation_matrix)

# Создайте тепловую карту
sns.heatmap(correlation_matrix, cmap = 'Blues', annot = True)
plt.show()
```

##### <b><font color='skyblue'>1. Покажите количество продаж машин во времени (в группировке по месяцам)</font></b>

```python
sales['month'] = pd.to_datetime(sales['Sale_Date']).dt.to_period('M')

sales_month = sales.groupby('month')['Sale_ID'].count().reset_index()
sales_month.head()
```

```python
# plt.plot(sales_month['month'].astype(str), sales_month['Sale_ID'], marker = 'o', color = 'blue')
figure = plt.figure(figsize = (15, 6))
sns.lineplot(x = sales_month['month'].astype(str), y = sales_month['Sale_ID'], marker = 'o', color = 'blue', )
plt.title('Количество продаж по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Количество продаж')
plt.xticks(rotation = 90)
# надо сделать еще подписи данных на точках же чуть выше точки
for i, value in enumerate(sales_month['Sale_ID']):
    plt.text(i, value + 0.5, str(value), ha = 'center', va = 'bottom')
plt.show()
```

##### <b><font color='skyblue'>2. Какие бренды продаются чаще всего?</font></b>

```python
brend_sales = pd.merge(sales, cars, on = 'Car_ID').groupby('Brand')['Quantity'].count().sort_values(ascending=False).reset_index()
brend_sales
```

```python
figure = plt.figure(figsize = (10, 6))
sns.barplot(x = 'Brand', y = 'Quantity', data = brend_sales, palette = 'Set2')
plt.title('Количество продаж по брендам')
plt.xlabel('Бренд')
plt.ylabel('Количество продаж')
# тоже надо подписи данных на барах
for i, value in enumerate(brend_sales['Quantity']):
    plt.text(i, value + 0.5, str(value), ha = 'center', va = 'bottom')
plt.show()
```

##### <b><font color='skyblue'>3. Предыдущий график дополните разделением по полу.</font></b>
