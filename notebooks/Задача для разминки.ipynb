У интернет-магазина есть данные о продажах за 7 дней:

- количество заказов в день
- средний чек заказа (в рублях)
- количество уникальных покупателей

```python
import numpy as np
```

```python
# Дни недели
days = np.array(["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Вс"])

# Кол-во заказов в день
orders = np.array([120, 135, 150, 160, 200, 250, 300])

# Средний чек
avg_check = np.array([1200, 1150, 1300, 1250, 1400, 1500, 1600])

# Кол-во уникальных покупателей
customers = np.array([100, 110, 130, 140, 180, 220, 260])
```

<b>Задания:</b>

- Посчитайте общую выручку по каждому дню.
- Определите день с максимальной выручкой.
- Найдите среднюю дневную выручку за неделю.
- Рассчитайте среднее количество заказов на одного покупателя по каждому дню.
- Постройте массив с накопительным итогом по дням.

```python
revenue = orders * avg_check
print("Выручка:", revenue)
max_revenue_day = days[np.argmax(revenue)]
print("День с максимальной выручкой:", max_revenue_day)
avg_revenue = round(np.mean(revenue), 2)
print("Средняя выручка:", avg_revenue)
customer_orders = np.round(orders / customers, 2)
print("Заказы на одного покупателя:", customer_orders)
sum_revenue = np.cumsum(revenue)
print("Накопленная выручка:", sum_revenue)

```
