### <font color='teal'>3. Сравнения, различные встроенные функции</font>

```python
#Синус
a = np.array([0, 2, 1])
print(np.sin, type(np.sin))
print(np.sin(a))
```

```python
#Сравнения
a = np.array([0, 2, 1])
b = np.array([3, 2, 5])
```

```python
print(a > b)
```

```python
print(a == b)
```

```python
c = a >= 2
print(c)
```

```python
#Any возвращает True, если хотя бы один элемент в массиве True
print(np.any(c))
```

```python
#All возвращает True, если все элементы в массиве True
print(np.all(c))
```

```python
#Модификация на месте
b *= 2
print(b)
```

```python
#Различные статистические функции
a = np.array([2, 1, 4, 5, 6])
#Сумма всех элементов массива
# char.chararray.sum(axis=None, dtype=None, out=None, *, keepdims=<no value>, initial=<no value>, where=<no value>)
np.sum(a)
#Произведение всех элементов массива
# char.chararray.prod(axis=None, dtype=None, out=None, *, keepdims=<no value>, initial=<no value>, where=<no value>)
np.prod(a)
#Максимальный и минимальный элементы массива
# char.chararray.max(axis=None, out=None, *, keepdims=<no value>, initial=<no value>, where=<no value>)
np.max(a)
# char.chararray.min(axis=None, out=None, *, keepdims=<no value>, initial=<no value>, where=<no value>)
np.min(a)
#Среднее арифметическое элементов массива
# char.chararray.mean(axis=None, dtype=None, out=None, *, keepdims=<no value>, where=<no value>)
np.mean(a)
np.average(a)
#Среднеквадартичное отклонение элементов массива
# char.chararray.std(axis=None, dtype=None, out=None, ddof=0, *, keepdims=<no value>, where=<no value>, mean=<no value>)
np.std(a)
```

```python
print(np.sum(a))
print(np.prod(a))
print(np.max(a))
print(np.min(a))
print(np.mean(a))
print(np.average(a))
print(np.std(a))
```

```python
#Немного полезных функций и констант
print(np.sqrt(b))
print(np.exp(b))
print(np.log(b))
print(np.sin(b))
print(np.e, np.pi)
```

```python
#Кумулятивные суммы
print(b.cumsum())
```
