```python
# Подключение модулей
from dash import Dash, html, dcc
import plotly.express as px

# Создание экземпляра класса Dash
app = Dash()

df = px.data.iris()
fig = px.scatter(
    df,
    x='sepal_width',
    y='sepal_length',
    color='species',
    template='plotly_white'
)

app.layout = html.Div(
    [
        html.H1('Визуализация набора данных Iris', style={'textAlign': 'center'}),
        dcc.Graph(figure=fig),
    ],
    style={'backgroundColor': 'white', 'color': 'black'}
)
# Запуск приложения
app.run(port=8050)
```

```python
from dash import Dash, dcc, html, Input, Output, callback

# Создаем экземпляр приложения Dash
app = Dash()

# Компонент графика
graph_component = dcc.Graph(
        id='example-graph',
        figure={}
)

# Компонент выпадающего списка
dropdown_component = dcc.Dropdown(
        id='my-dropdown',
        options=[
        {'label': 'San Francisco', 'value': 'SF'},
        {'label': 'Montreal', 'value': 'MTL'}
        ],
        value='SF',
        style={'width': '300px', 'paddingLeft': '100px', 'paddingTop': '24px'}
)

# Определяем layout приложения
app.layout = html.Div([
                dropdown_component,
                graph_component,
], style={'backgroundColor': 'white', 'color': 'black', 'paddingLeft': '24px'}
)

# Callback для обновления графика при выборе в выпадающем списке
@callback(
Output('example-graph', 'figure'),
Input('my-dropdown', 'value')
)
def update_graph(selected_value):
    SF = {'x': [1, 2, 3], 'y': [4, 5, 7], 'type': 'bar', 'name': 'San Francisco'}
    MTL = {'x': [1, 2, 3], 'y': [8, 3, 6], 'type': 'bar', 'name': "Montreal"}
    BERLIN = {'x': [1, 2, 3], 'y': [2, 4, 5], 'type': 'bar', 'name': "Berlin"}
    result = {'data': [eval(selected_value), BERLIN],
    'layout': {'title': 'Dash Data Visualization'}
    }
    return result

app.run(port=8053)
```

```python
from dash import Dash, html, dcc, callback, Output, Input
import pandas as pd
import plotly.express as px

# Данные
df_coffee = pd.read_csv("../data/Coffe_sales.csv")
weekday_order = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
coffee_values = sorted(df_coffee["coffee_name"].dropna().unique())

app = Dash(__name__)

app.layout = html.Div(
    [
        html.Div(
            html.H2("Coffee Sales by Weekday"),
            style={"paddingLeft": "100px", "paddingTop": "24px"}
        ),

        dcc.RadioItems(
            options=[
                {"label": "Все типы кофе", "value": "all_coffee"},
                {"label": "Один тип кофе", "value": "one_coffee"},
            ],
            value="all_coffee",
            id="controls-and-radio-item",
            style={"paddingLeft": "120px", "paddingTop": "8px"}
        ),

        html.Div(
            dcc.RadioItems(
                options=[
                    {"label": "С разделением по типам кофе", "value": "split"},
                    {"label": "Без разделения (сплошные бары)", "value": "solid"},
                ],
                value="split",
                id="view-mode-radio"
            ),
            id="view-mode-container",
            style={"paddingLeft": "120px", "paddingTop": "8px", "display": "block"}
        ),

        html.Div(
            dcc.Dropdown(
                id="my-dropdown",
                options=[{"label": name, "value": name} for name in coffee_values],
                value=coffee_values[0],
                style={"width": "300px"}
            ),
            id="dropdown-container",
            style={"paddingLeft": "120px", "paddingTop": "8px", "display": "none"}
        ),

        dcc.Graph(id="coffee-sales-graph")
    ],
    style={"backgroundColor": "white", "color": "black"}
)

@callback(
    Output("view-mode-container", "style"),
    Output("dropdown-container", "style"),
    Input("controls-and-radio-item", "value")
)
def toggle_controls(mode):
    view_style = {"paddingLeft": "120px", "paddingTop": "8px"}
    drop_style = {"paddingLeft": "120px", "paddingTop": "8px"}

    if mode == "all_coffee":
        return {**view_style, "display": "block"}, {**drop_style, "display": "none"}
    return {**view_style, "display": "none"}, {**drop_style, "display": "block"}

@callback(
    Output("coffee-sales-graph", "figure"),
    Input("controls-and-radio-item", "value"),
    Input("view-mode-radio", "value"),
    Input("my-dropdown", "value")
)
def update_graph(mode, view_mode, selected_coffee):
    if mode == "one_coffee":
        data = df_coffee[df_coffee["coffee_name"] == selected_coffee]
        fig = px.histogram(
            data,
            y="Weekday",
            category_orders={"Weekday": weekday_order},
            title=f"Coffee Sales by Weekday: {selected_coffee}"
        )
        return fig

    if view_mode == "split":
        fig = px.histogram(
            df_coffee,
            y="Weekday",
            color="coffee_name",
            category_orders={"Weekday": weekday_order},
            title="Coffee Sales by Weekday (Split by Coffee Type)"
        )
    else:
        fig = px.histogram(
            df_coffee,
            y="Weekday",
            category_orders={"Weekday": weekday_order},
            title="Coffee Sales by Weekday (Solid Bars)"
        )

    return fig

app.run(port=8057)
```
