Р”Р»СЏ РїСЂРѕСЃС‚РѕС‚С‹ РґР°РЅРЅС‹Рµ РїСЂРµРґСЃС‚Р°РІР»РµРЅС‹ РІ РІРёРґРµ РјР°С‚СЂРёС† 10С…5, РіРґРµ СЃС‚СЂРѕРєРё вЂ” СЌС‚Рѕ РєР°С‚РµРіРѕСЂРёРё С‚РѕРІР°СЂРѕРІ (РЅР°РїСЂРёРјРµСЂ: РѕРґРµР¶РґР°, СЌР»РµРєС‚СЂРѕРЅРёРєР°, РёРіСЂСѓС€РєРё), Р° СЃС‚РѕР»Р±С†С‹ вЂ” СЌС‚Рѕ СЃС‚СЂР°РЅС‹ РїСЂРѕРґР°Р¶ (РСЃРїР°РЅРёСЏ, Р“РµСЂРјР°РЅРёСЏ, Рё С‚.Рґ).

- РњР°С‚СЂРёС†Р° <i>july_orders</i> вЂ” РєРѕР»РёС‡РµСЃС‚РІРѕ Р·Р°РєР°Р·Р°РЅРЅС‹С… С‚РѕРІР°СЂРѕРІ РІ РёСЋР»Рµ.<br>
- РњР°С‚СЂРёС†Р° <i>aug_orders</i> вЂ” РєРѕР»РёС‡РµСЃС‚РІРѕ Р·Р°РєР°Р·Р°РЅРЅС‹С… С‚РѕРІР°СЂРѕРІ РІ Р°РІРіСѓСЃС‚Рµ.<br>
<br>
<b>Р—Р°РґР°РЅРёСЏ:</b> <br>
1. Р Р°СЃСЃС‡РёС‚Р°Р№С‚Рµ, РєР°Рє РёР·РјРµРЅРёР»РёСЃСЊ РїСЂРѕРґР°Р¶Рё РІ Р°РІРіСѓСЃС‚Рµ РѕС‚РЅРѕСЃРёС‚РµР»СЊРЅРѕ РёСЋР»СЏ (РІ РїСЂРѕС†РµРЅС‚Р°С…). РќР°Р·РѕРІРёС‚Рµ С‚Р°Р±Р»РёС†Сѓ <i>growth.</i><br>
2. РџРѕСЃС‚СЂРѕР№С‚Рµ РјР°С‚СЂРёС†Сѓ <i>is_growth</i>, РіРґРµ <i>True</i>, РµСЃР»Рё РїСЂРѕРґР°Р¶Рё РІС‹СЂРѕСЃР»Рё, Рё <i>False</i>, РµСЃР»Рё СѓРїР°Р»Рё РёР»Рё РЅРµ РёР·РјРµРЅРёР»РёСЃСЊ.<br>
3. РџРѕСЃС‚СЂРѕР№С‚Рµ РјР°С‚СЂРёС†Сѓ <i>diff_sales</i>, РїРѕРєР°Р·С‹РІР°СЋС‰СѓСЋ, РЅР°СЃРєРѕР»СЊРєРѕ Р±РѕР»СЊС€Рµ/РјРµРЅСЊС€Рµ С‚РѕРІР°СЂРѕРІ Р±С‹Р»Рѕ РїСЂРѕРґР°РЅРѕ РІ Р°РІРіСѓСЃС‚Рµ РїРѕ СЃСЂР°РІРЅРµРЅРёСЋ СЃ РёСЋР»РµРј.

```python
#Р“РµРЅРµСЂР°С†РёСЏ РґР°РЅРЅС‹С…
import numpy as np
np.random.seed(42)
categories = [
    "РћРґРµР¶РґР°", "РћР±СѓРІСЊ", "Р­Р»РµРєС‚СЂРѕРЅРёРєР°", "РРіСЂСѓС€РєРё", "РљРЅРёРіРё",
    "РљРѕСЃРјРµС‚РёРєР°", "РЎРїРѕСЂС‚С‚РѕРІР°СЂС‹", "РњРµР±РµР»СЊ", "РџСЂРѕРґСѓРєС‚С‹", "РўРµС…РЅРёРєР° РґР»СЏ РґРѕРјР°"
]

countries = ["РСЃРїР°РЅРёСЏ", "РС‚Р°Р»РёСЏ", "Р“РµСЂРјР°РЅРёСЏ", "Р¤СЂР°РЅС†РёСЏ", "РџРѕСЂС‚СѓРіР°Р»РёСЏ"]

# Р“РµРЅРµСЂРёСЂСѓРµРј РґР°РЅРЅС‹Рµ: РїСЂРѕРґР°Р¶Рё Р·Р° РёСЋР»СЊ Рё Р°РІРіСѓСЃС‚
july_sales = np.random.randint(50, 500, size=(len(categories), len(countries)))
aug_sales = np.random.randint(50, 500, size=(len(categories), len(countries)))
print("РџСЂРѕРґР°Р¶Рё РІ РёСЋР»Рµ:\n", july_sales)
print("РџСЂРѕРґР°Р¶Рё РІ Р°РІРіСѓСЃС‚Рµ:\n", aug_sales)
```

```python
#1. Р Р°СЃСЃС‡РёС‚Р°Р№С‚Рµ, РєР°Рє РёР·РјРµРЅРёР»РёСЃСЊ РїСЂРѕРґР°Р¶Рё РІ Р°РІРіСѓСЃС‚Рµ РѕС‚РЅРѕСЃРёС‚РµР»СЊРЅРѕ РёСЋР»СЏ (РІ РїСЂРѕС†РµРЅС‚Р°С…). РќР°Р·РѕРІРёС‚Рµ С‚Р°Р±Р»РёС†Сѓ growth.
growth = np.round(((aug_sales / july_sales) - 1) * 100, 1)
print("Р РѕСЃС‚ РїСЂРѕРґР°Р¶ РІ РїСЂРѕС†РµРЅС‚Р°С…:\n", growth)
```

```python
#2. РџРѕСЃС‚СЂРѕР№С‚Рµ РјР°С‚СЂРёС†Сѓ is_growth, РіРґРµ True, РµСЃР»Рё РїСЂРѕРґР°Р¶Рё РІС‹СЂРѕСЃР»Рё, Рё False, РµСЃР»Рё СѓРїР°Р»Рё РёР»Рё РЅРµ РёР·РјРµРЅРёР»РёСЃСЊ.
is_growth = growth > 0
print("РњР°С‚СЂРёС†Р° СЂРѕСЃС‚Р° РїСЂРѕРґР°Р¶ (True - СЂРѕСЃС‚, False - РїР°РґРµРЅРёРµ РёР»Рё Р±РµР· РёР·РјРµРЅРµРЅРёР№):\n", is_growth)
```

```python
#3. РџРѕСЃС‚СЂРѕР№С‚Рµ РјР°С‚СЂРёС†Сѓ diff_sales, РїРѕРєР°Р·С‹РІР°СЋС‰СѓСЋ, РЅР°СЃРєРѕР»СЊРєРѕ Р±РѕР»СЊС€Рµ/РјРµРЅСЊС€Рµ С‚РѕРІР°СЂРѕРІ Р±С‹Р»Рѕ РїСЂРѕРґР°РЅРѕ РІ Р°РІРіСѓСЃС‚Рµ РїРѕ СЃСЂР°РІРЅРµРЅРёСЋ СЃ РёСЋР»РµРј.
diff_sales = aug_sales - july_sales
print("Р Р°Р·РЅРёС†Р° РІ РїСЂРѕРґР°Р¶Р°С… РјРµР¶РґСѓ Р°РІРіСѓСЃС‚РѕРј Рё РёСЋР»РµРј:\n", diff_sales)
```
