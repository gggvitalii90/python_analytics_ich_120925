##### <font color='teal'>Задача №1</font>
Вы работаете аналитиком данных в интернет-магазине. У вас есть данные о заказах за два месяца — июль и август. Для простоты данные представлены в виде матриц 10х5, где строки — это категории товаров (например: одежда, электроника, игрушки), а столбцы — это страны продаж (спания, Германия, и т.д).
<p></p>
<b>Данные:</b> <br>
- Матрица <i>july_orders</i> — количество заказанных товаров в июле.<br>
- Матрица <i>aug_orders</i> — количество заказанных товаров в августе.<br>
<br>
<b>Задания:</b> <br>
1. Рассчитайте, как изменились продажи в августе относительно июля (в процентах). Назовите таблицу <i>growth.</i><br>
2. Постройте матрицу <i>is_growth</i>, где <i>True</i>, если продажи выросли, и <i>False</i>, если упали или не изменились.<br>
3. Постройте матрицу <i>diff_sales</i>, показывающую, насколько больше/меньше товаров было продано в августе по сравнению с июлем.

<p align="center">
  <img src="../Images/Example_of_table.png" alt="Пример данных" width="400"/>
</p>

```python
#Генерация данных
import numpy as np
np.random.seed(42)
categories = [
    "Одежда", "Обувь", "Электроника", "грушки", "Книги",
    "Косметика", "Спорттовары", "Мебель", "Продукты", "Техника для дома"
]

countries = ["спания", "талия", "Германия", "Франция", "Португалия"]

# Генерируем данные: продажи за июль и август
july_sales = np.random.randint(50, 500, size=(len(categories), len(countries)))
aug_sales = np.random.randint(50, 500, size=(len(categories), len(countries)))
print("Продажи в июле:\n", july_sales)
print("Продажи в августе:\n", aug_sales)
```

```python
#1. Рассчитайте, как изменились продажи в августе относительно июля (в процентах). Назовите таблицу growth.
growth = np.round(((aug_sales / july_sales) - 1) * 100, 1)
print("Рост продаж в процентах:\n", growth)
```

```python
#2. Постройте матрицу is_growth, где True, если продажи выросли, и False, если упали или не изменились.
is_growth = growth > 0
print("Матрица роста продаж (True - рост, False - падение или без изменений):\n", is_growth)
```

```python
#3. Постройте матрицу diff_sales, показывающую, насколько больше/меньше товаров было продано в августе по сравнению с июлем.
diff_sales = aug_sales - july_sales
print("Разница в продажах между августом и июлем:\n", diff_sales)
```
