### <b><font color='skyblue'>1. Смоделировать тепловую карту продаж в магазине электроники</font></b>
<i>Условие:</i><br>
У вас есть магазин электроники, в котором 12 категорий товара (колонки), и 30 дней месяца (строки). <br>
Нужно сгенерировать и проанализировать матрицу “продаж” (штучно, целые числа).<br>
Все операции — только через <b>numpy</b>.

```python
# 1. Создайте пустую матрицу sales 30×12 с нулями
zero = np.zeros((30, 12))
display(zero)
```

```python
# 2. Сгенерируйте вектор base - “средняя базовая продажа по категориям” длиной 12, равномерно от 5 до 50
base = np.random.uniform(5, 50, 12)
print(base)
```

```python
# 3. Сгенерируйте матрицу случайных отклонений noise (случайные числа) 30х12, которые будут прибавляться к базовым продажам (значения от -10 до +10)
noise = np.random.uniform(-10, 10, (30, 12))
print(noise)
```

```python
# 4. Заполните матрицу sales суммой base и noise 
sales = pd.DataFrame(np.round(base + noise, 2))
display(sales)
```

```python
# 5. Продажи не могут быть отрицательными. Замените все отрицательные значения на ноль.
sales = np.where(sales < 0, 0, sales)
df_sales = pd.DataFrame(sales)
display(df_sales)
display(sales)
```

```python
# 6. Добавьте небольшой тренд спроса для каждого товара. Для этого:
# 6.1. Создайте тренд вектор длиной 30 в переменную trend
trend = np.random.randint(0, 11, 30)
print(trend)
```

```python
# 6.2. Прибавьте значения тренда к таблице sales, но прежде вектор trend необходимо "превратить" в столбец
trend = pd.Series(trend)
trend

sales['trend'] = trend
display(sales)
```

```python
# 7. Для sales посчитайте:
# - сумму всех элементов
# - среднее
# - стандартное отклонение
print('Сумма всех элементов:', sales.sum().sum())
print('Среднее:', sales.mean().mean())
print('Стандартное отклонение:', sales.std().std())
```

```python
# 8. Найдите категорию (индекс столбца), которая дала максимальные суммарные продажи за месяц.
# Запишите результат в переменную best_category
best_category = sales.sum()
print('Категория с максимальными суммарными продажами за месяц:', best_category)
```

```python
# 9. Постройте heatmap на основе итоговой матрицы sales
import matplotlib.pyplot as plt
import seaborn as sns
sns.heatmap(sales, annot=True, cmap='YlGnBu')
plt.title('Heatmap продаж')
plt.xlabel('Категории')
plt.ylabel('Дни')
plt.show()
```

### <b><font color='skyblue'>2. Video Game Sales with Ratings</font></b>

<i>Условие:</i><br>
Вы - аналитик в геймдев.
Вам нужно сделать “быструю разведку” данных рынка видеоигр.

```python
# 1. Загрузите датасет Video_Games_Sales.csv в переменную df_videogames, посмотрите info(), head()
df_videogames = pd.read_csv('..//data//Video_Games_Sales.csv')
display(df_videogames)
df_videogames.info()
```

```python
display(df_videogames.sort_index())
```

```python
df_videogames['Year_of_Release'].isnull().sum()
```

```python
df_videogames[df_videogames['Year_of_Release'].isnull()]
```

```python
# 2. Приведите столбец Year_of_Release к типу int, пропуски заполните None 
df_videogames['Year_of_Release'].replace({None: pd.NA}, inplace=True)
df_videogames['Year_of_Release'] = df_videogames['Year_of_Release'].astype('Int64')
df_videogames.info()
```

```python
# 3. Приведите столбец User_Score к float, странные значения вроде “tbd” замените на NaN
df_videogames['User_Score'] = pd.to_numeric(df_videogames['User_Score'], errors='coerce')
df_videogames.info()
```

```python
# 4. Оставьте в датафрейме видеоигры, у которых релиз старше 2000 года
df_videogames_2000 = df_videogames[df_videogames['Year_of_Release'] > 2000]
df_videogames_2000
```

```python
# 5. Добавьте столбец total_sales как сумма всех региональных продаж (NA_Sales, EU_Sales, JP_Sales, Other_Sales)
df_videogames_2000['total_sales'] = df_videogames_2000[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum(axis=1)
df_videogames_2000
```

```python
# 6. Уберите игры, у которых total_sales = 0
df_videogames_2000 = df_videogames_2000[df_videogames_2000['total_sales'] > 0]
df_videogames_2000
```

```python
# 7. Посмотрите top-10 платформ по total_sales
df_videogames_2000.groupby('Platform')['total_sales'].sum().sort_values(ascending=False).head(10)
```

```python
# 8. В переменную genre_gb для каждого жанра посчитайте:
# - количество игр
# - среднюю user_score (игнорируя NaN)
# - суммарный total_sales
# Выведите топ-5 жанров по сумме продаж
genre_gb = df_videogames_2000.groupby('Genre').agg({
    'Name': 'count',
    'User_Score': 'mean',
    'total_sales': 'sum'
})
display(genre_gb.sort_values(by='total_sales', ascending=False).head(5))
```

```python
# 9. Постройте pivot_table:
# - строки: Genre
# - колонки: Rating
# - значение: средний User_Score
df_pivot = df_videogames_2000.pivot_table(index='Genre', columns='Rating', values='User_Score', aggfunc='mean')
display(df_pivot)
```

```python
# 10. Постройте barplot 10 лучших платформ по total_sales
platform_sales = df_videogames_2000.groupby('Platform')['total_sales'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x=platform_sales.index, y=platform_sales.values, palette='viridis')
plt.title('Топ-10 платформ по total_sales')
plt.xlabel('Платформа')
plt.ylabel('Суммарные продажи')
plt.xticks(rotation=45)
plt.show()
```

```python
# 11. Постройте lineplot, отражающий по годам средний User_Score
yearly_user_score = df_videogames_2000.groupby('Year_of_Release')['User_Score'].mean()
display(yearly_user_score)
plt.figure(figsize=(12, 6))
sns.lineplot(x=yearly_user_score.index, y=yearly_user_score.values, marker='o')
plt.title('Средний User_Score по годам')
plt.xlabel('Год релиза')
plt.ylabel('Средний User_Score')
plt.grid()
plt.show()
```
