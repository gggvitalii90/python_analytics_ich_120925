### <font color='coral'>4. РџСЂРёРјРµСЂ</font>

```python
# РРјРїРѕСЂС‚РёСЂСѓРµРј РѕР±СЉРµРєС‚ Point РёР· Р±РёР±Р»РёРѕС‚РµРєРё shapely.geometry
from shapely.geometry import Point
import pandas as pd
# from shapely.geometry import Point, LineString, Polygon # РРјРїРѕСЂС‚ РЅРµСЃРєРѕР»СЊРєРёС… РѕР±СЉРµРєС‚РѕРІ
# from shapely.geometry as geom # РРјРїРѕСЂС‚ РІСЃРµРіРѕ РјРѕРґСѓР»СЏ Рё РґР°Р»РµРµ РѕР±СЂР°С‰РµРЅРёРµ Рє РѕР±СЉРµРєС‚Р°Рј С‡РµСЂРµР· geom.Point

# РЎРѕР·РґР°РЅРёРµ DataFrame СЃ РґРІСѓРјСЏ СЃС‚РѕР»Р±С†Р°РјРё: City, Population
data = {
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston'],
    'Population': [8419600, 3990456, 2705994, 2325502]
}
df = pd.DataFrame(data)

# СЃРїРёСЃРѕРє РєРѕРѕСЂРґРёРЅР°С‚
geometry = [Point(-74.006, 40.7128),  # New York (longitude, latitude)
            Point(-118.2437, 34.0522),  # Los Angeles
            Point(-87.6298, 41.8781),  # Chicago
            Point(-95.3698, 29.7604)]  # Houston

# СЃРѕР·РґР°РЅРёРµ РѕР±СЉРµРєС‚Р° GeoDataFrame
# РІ РєР°С‡РµСЃС‚РІРµ РґР°РЅРЅС‹С… - DataFrame df
# РІ РєР°С‡РµСЃС‚РІРµ СЃС‚РѕР»Р±С†Р° СЃ РіРµРѕРјРµС‚СЂРёРµР№ - СЃРїРёСЃРѕРє geometry
# РЅР°Р·РІР°РЅРёРµ CRS 'EPSG:4326', С‡С‚Рѕ СЃРѕРѕС‚РІРµС‚СЃРІСѓРµС‚ WGS 84
gdf = gpd.GeoDataFrame(df, geometry = geometry, crs='EPSG:4326')
gdf
```

### <font color='coral'>8. Р РµС€РµРЅРёРµ Р·Р°РґР°С‡Рё</font>
1. Р—Р°РіСЂСѓР·РёС‚Рµ РґР°РЅРЅС‹Рµ Рѕ СЂР°Р№РѕРЅР°С… Р‘Р°СЂСЃРµР»РѕРЅС‹ РїРѕ СЃСЃС‹Р»РєРµ (https://raw.githubusercontent.com/jcanalesluna/bcn-geodata/master/districtes/districtes.geojson).
2. РР·РѕР±СЂР°Р·РёС‚Рµ РЅР° РєР°СЂС‚Рµ РѕР±Р»Р°СЃС‚Рё РІРјРµСЃС‚Рµ СЃ С†РµРЅС‚СЂРѕРёРґР°РјРё. РћР±Р»Р°СЃС‚Рё РґРѕР»Р¶РЅС‹ Р±С‹С‚СЊ РѕРєСЂР°С€РµРЅС‹ С†РІРµС‚РѕРј РІ Р·Р°РІРёСЃРёРјРѕСЃС‚Рё РѕС‚ РёС… РѕС‚РґР°Р»С‘РЅРЅРѕСЃС‚Рё РѕС‚ РЎР°РіСЂР°РґС‹ Р¤Р°РјРёР»РёРё:

```python
# Р—Р°РіСЂСѓР·РєР° С„Р°Р№Р»Р°:
url = 'https://raw.githubusercontent.com/jcanalesluna/bcn-geodata/master/districtes/districtes.geojson'
bars = gpd.read_file(url)

bars.head()
```

Р•СЃР»Рё CRS РЅРµРёР·РІРµСЃС‚РµРЅ, С‚Рѕ РґРµР№СЃС‚РІСѓРµРј РїРѕ СЃР»РµРґСѓСЋС‰РёРј РїСЂР°РІРёР»Р°Рј:
- Р•СЃР»Рё Р·РЅР°С‡РµРЅРёСЏ РѕС‚ в€’180 РґРѕ 180 (X) Рё РѕС‚ в€’90 РґРѕ 90 (Y), С‚Рѕ СЃРєРѕСЂРµРµ РІСЃРµРіРѕ, EPSG:4326.
- Р•СЃР»Рё РєРѕРѕСЂРґРёРЅР°С‚С‹ Р±РѕР»СЊС€РёРµ (РѕРєРѕР»Рѕ РјРёР»Р»РёРѕРЅР°: 2 000 000, 5 000 000) - СЌС‚Рѕ, РѕР±С‹С‡РЅРѕ, РјРµС‚СЂС‹ РІ РїСЂРѕРµРєС†РёРё (РЅР°РїСЂРёРјРµСЂ, EPSG:3857).
- Р•СЃР»Рё С‡РёСЃР»Р° РїРѕСЂСЏРґРєР° 300 000 вЂ“ 500 000 Рё 4 500 000 вЂ“ 4 800 000 - СЌС‚Рѕ UTM-Р·РѕРЅР°

```python
# Р”Р»СЏ РІС‹С‡РёСЃР»РµРЅРёСЏ СЂР°СЃСЃС‚РѕСЏРЅРёР№ РЅР°Рј РЅРµРѕР±С…РѕРґРёРјРѕ РїРµСЂРµРІРµСЃС‚Рё РєРѕРѕСЂРґРёРЅР°С‚С‹ РІ РјРµС‚СЂРёС‡РµСЃРєСѓСЋ СЃРёСЃС‚РµРјСѓ (РЅР°РїСЂРёРјРµСЂ, РІ Web Mercator)
display(bars.crs)
bars = bars.to_crs(epsg=3857)
```

```python
# Р’С‹С‡РёСЃР»РёРј С†РµРЅС‚СЂРѕРёРґС‹
bars['centroid'] = bars.centroid
bars.head()
```

```python
# Р—Р°РїРёС€РµРј РІ РїРµСЂРµРјРµРЅРЅСѓСЋ РєРѕРѕСЂРґРёРЅР°С‚С‹ Sagrada Familia Рё СЃСЂР°Р·Сѓ РїРµСЂРµРІРµРґРµРј РІ Web Mercator
sagrada_fam = Point(2.1743680500855005, 41.403656946781304)
sagrada_fam = gpd.GeoDataFrame(
    {   'name': 'Sagrada Familia',
        'geometry': [sagrada_fam]}, crs='EPSG:4326').to_crs(epsg=3857).iloc[0]
sagrada_fam
```

```python
# Р’С‹С‡РёСЃР»РёРј СЂР°СЃСЃС‚РѕСЏРЅРёРµ РґР»СЏ РєР°Р¶РґРѕРіРѕ СЂР°Р№РѕРЅР° Р‘Р°СЂСЃРµР»РѕРЅС‹ РѕС‚ Sagrada Familia
bars['distance'] = bars['centroid'].distance(sagrada_fam['geometry'])
bars.head()
```

```python
# РЎРѕР·РґР°РґРёРј РѕР±СЉРµРєС‚ subplots - РѕРЅ РїРѕРЅР°РґРѕР±РёС‚СЃСЏ РґР»СЏ СЂРµРіСѓР»РёСЂРѕРІРєРё РѕСЃРµР№ ax
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(10, 10))
bars.plot(column='distance', ax=ax, legend=True)
# Р”Р»СЏ С‚РѕРіРѕ, С‡С‚РѕР±С‹ РЅР°РЅРµСЃС‚Рё С‚РѕС‡РєРё РЅР° РєР°СЂС‚Сѓ (С†РµРЅС‚СЂРѕРёРґС‹), РІРѕР·СЊРјРµРј СЃС‚РѕР»Р±РµС† centroid РёР· РЅР°С€РµРіРѕ РґР°С‚Р°С„СЂРµР№РјР° Рё РІС‹РґРµР»РёРј РІ РѕС‚РґРµР»СЊРЅС‹Р№ Series
# bars_centroids = bars['centroid']
# РќР°РЅРµСЃРµРј С†РµРЅС‚СЂРѕРёРґС‹ СЂР°Р№РѕРЅРѕРІ РЅР° РєР°СЂС‚Сѓ
bars['centroid'].plot(ax=ax, color='blue', markersize=50, label='Centroids')

# РќР°РЅРµСЃРµРј С‚РѕС‡РєСѓ Sagrada Familia
gpd.GeoSeries([sagrada_fam['geometry']], crs=bars.crs).plot(
    ax=ax, color='red', markersize=100, label='Sagrada Familia'
)

plt.legend(loc='upper left')
plt.show()
```
