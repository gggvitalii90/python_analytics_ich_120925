```python
# Р”Р»СЏ РґР°С‚Р°СЃРµС‚Р° Coffee_sales.csv РІС‹С‡РёСЃР»РёС‚Рµ РјРµРґРёР°РЅСѓ С†РµРЅС‹ 3 СЃРїРѕСЃРѕР±Р°РјРё: numpy, pandas, polars
median_numpy = np.median(df_coffee['money'])
median_pandas = df_coffee['money'].median()
median_polars = df_coffee_polars['money'].median()
print('NumPy: ', median_numpy, '\n',
      'Pandas: ', median_pandas, '\n',
      'Polars: ', median_polars)
```

```python
# Р”Р»СЏ РґР°С‚Р°СЃРµС‚Р° Coffee_sales.csv РІС‹С‡РёСЃР»РёС‚Рµ РґРёР°РїР°Р·РѕРЅ С†РµРЅС‹ 4 СЃРїРѕСЃРѕР±Р°РјРё: СЃС‚Р°РЅРґР°СЂС‚РЅС‹РјРё С„СѓРЅРєС†РёСЏРјРё python, numpy, pandas, polars
range_value_simple = max(df_coffee['money']) - min(df_coffee['money'])
range_value_numpy = np.ptp(df_coffee['money'])
range_value_pandas = df_coffee['money'].max() - df_coffee['money'].min()
range_value_polars = df_coffee_polars['money'].max() - df_coffee_polars['money'].min()
print('Simple: ', range_value_simple, '\n' 
      'NumPy: ', range_value_numpy, '\n'
      'Pandas: ', range_value_pandas, '\n'
      'Polars: ', range_value_polars)
```

```python
# Р”Р»СЏ РґР°С‚Р°СЃРµС‚Р° Coffee_sales.csv РІС‹С‡РёСЃР»РёС‚Рµ РґРёСЃРїРµСЂСЃРёСЋ С†РµРЅС‹ (С„СѓРЅРєС†РёСЏ var()) 3 СЃРїРѕСЃРѕР±Р°РјРё: numpy, pandas, polars
variance_numpy = np.var(df_coffee['money'], ddof=1)
variance_pandas = df_coffee['money'].var()
variance_polars = df_coffee_polars['money'].var()
print('NumPy: ', variance_numpy, '\n',
      'Pandas: ', variance_pandas, '\n',
      'Polars: ', variance_polars)
```

```python
# Р”Р»СЏ РґР°С‚Р°СЃРµС‚Р° Coffee_sales.csv РІС‹С‡РёСЃР»РёС‚Рµ РґРёСЃРїРµСЂСЃРёСЋ С†РµРЅС‹ (С„СѓРЅРєС†РёСЏ var()) 3 СЃРїРѕСЃРѕР±Р°РјРё: numpy, pandas, polars
variance_numpy = np.var(df_coffee['money'], ddof=1)
variance_pandas = df_coffee['money'].var()
variance_polars = df_coffee_polars['money'].var()
print('NumPy: ', variance_numpy, '\n',
      'Pandas: ', variance_pandas, '\n',
      'Polars: ', variance_polars)
```

##### <b><font color='saddlebrown'>РЎС‚Р°РЅРґР°СЂС‚РЅРѕРµ РѕС‚РєР»РѕРЅРµРЅРёРµ</font></b>

```python
# Р”Р»СЏ РґР°С‚Р°СЃРµС‚Р° Coffee_sales.csv РІС‹С‡РёСЃР»РёС‚Рµ СЃС‚Р°РЅРґР°СЂС‚РЅРѕРµ РѕС‚РєР»РѕРЅРµРЅРёРµ С†РµРЅС‹ (С„СѓРЅРєС†РёСЏ std()) 3 СЃРїРѕСЃРѕР±Р°РјРё: numpy, pandas, polars
std_numpy = np.std(df_coffee['money'], ddof=1)
std_pandas = df_coffee['money'].std()
std_polars = df_coffee_polars['money'].std()
print('NumPy: ', std_numpy, '\n',
      'Pandas: ', std_pandas, '\n',
      'Polars: ', std_polars)
```

```python
# Р”Р»СЏ РґР°С‚Р°СЃРµС‚Р° Coffee_sales.csv РІС‹С‡РёСЃР»РёС‚Рµ IQR С†РµРЅС‹ 3 СЃРїРѕСЃРѕР±Р°РјРё: numpy, pandas, polars
IQR_numpy = np.percentile(df_coffee['money'], 75) - np.percentile(df_coffee['money'], 25)
IQR_pandas = df_coffee['money'].quantile(0.75) - df_coffee['money'].quantile(0.25)
IQR_polars = df_coffee_polars['money'].quantile(0.75) - df_coffee_polars['money'].quantile(0.25)
print('NumPy: ', IQR_numpy, '\n',
      'Pandas: ', IQR_pandas, '\n',
      'Polars: ', IQR_polars)
```

### <b><font color='saddlebrown'>Р‘РёРЅРѕРјРёР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ</font></b>

```python
from scipy.stats import binom
# РўРµРѕСЂРµС‚РёС‡РµСЃРєР°СЏ РІРµСЂРѕСЏС‚РЅРѕСЃС‚СЊ
# Р¤СѓРЅРєС†РёСЏ pmf (Probability mass function) РґР»СЏ Р±РёРЅРѕРјРёРЅР°Р»СЊРЅРѕРіРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ РїСЂРёРЅРёРјР°РµС‚ СЃР»РµРґСѓСЋС‰РёРµ Р°СЂРіСѓРјРµРЅС‚С‹:
# k - РєРѕР»РёС‡РµСЃС‚РІРѕ СѓСЃРїРµС…РѕРІ
# n - РєРѕР»РёС‡РµСЃС‚РІРѕ РёСЃРїС‹С‚Р°РЅРёР№
# p - РІРµСЂРѕСЏС‚РЅРѕСЃС‚СЊ СѓСЃРїРµС…Р° РІ РєР°Р¶РґРѕРј РёСЃРїС‹С‚Р°РЅРёРё
binom.pmf(k = 6, n = 10, p = 0.5)
```

```python
from scipy.stats import binom
import seaborn as sns
import matplotlib.pyplot as plt
# РџР°СЂР°РјРµС‚СЂС‹
n = 10 # РєРѕР»РёС‡РµСЃС‚РІРѕ РёСЃРїС‹С‚Р°РЅРёР№
p = 0.5 # РІРµСЂРѕСЏС‚РЅРѕСЃС‚СЊ СѓСЃРїРµС…Р°

# Р“РµРЅРµСЂРёСЂСѓРµРј СЃР»СѓС‡Р°Р№РЅС‹Рµ РІРµР»РёС‡РёРЅС‹ СЃ Р±РёРЅРѕРјРёР°Р»СЊРЅС‹Рј СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµРј
# Р¤СѓРЅРєС†РёСЏ rvs - Random variates РїСЂРёРЅРёРјР°РµС‚ СЃР»РµРґСѓСЋС‰РёРµ Р°СЂРіСѓРјРµРЅС‚С‹:
# n - РєРѕР»РёС‡РµСЃС‚РІРѕ РёСЃРїС‹С‚Р°РЅРёР№
# p - РІРµСЂРѕСЏС‚РЅРѕСЃС‚СЊ СѓСЃРїРµС…Р°
# size - СЃРєРѕР»СЊРєРѕ СЂР°Р· РїСЂРѕРІРµСЃС‚Рё СЃРµСЂРёСЋ РёСЃРїС‹С‚Р°РЅРёР№
data_binom = binom.rvs(n = n, p = p, size = 10000)

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ
ax = sns.histplot(data_binom, kde = False, color = 'skyblue', bins = 30)
ax.set(xlabel='Р‘РёРЅРѕРјРёР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ', ylabel = 'Р§Р°СЃС‚РѕС‚Р°')
plt.title('Р‘РёРЅРѕРјРёР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ (n=10, p=0.5)')
plt.show()
```

### <b><font color='saddlebrown'>РџСѓР°СЃСЃРѕРЅРѕРІСЃРєРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ</font></b>

```python
from scipy.stats import poisson
# РўРµРѕСЂРµС‚РёС‡РµСЃРєРѕРµ Р·РЅР°С‡РµРЅРёРµ
# Р¤СѓРЅРєС†РёСЏ pmf (Probability mass function) РґР»СЏ РїСѓР°СЃСЃРѕРЅРѕРІСЃРєРѕРіРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ РїСЂРёРЅРёРјР°РµС‚ СЃР»РµРґСѓСЋС‰РёРµ Р°СЂРіСѓРјРµРЅС‚С‹:
# k - РѕР¶РёРґР°РµРјРѕРµ РєРѕР»РёС‡РµСЃС‚РІРѕ СЃРѕР±С‹С‚РёР№ 
# mu - СЃСЂРµРґРЅРµРµ РєРѕР»РёС‡РµСЃС‚РІРѕ СЃРѕР±С‹С‚РёР№
poisson.pmf(k = 5, mu = 3)
```

```python
from scipy.stats import poisson
# РџР°СЂР°РјРµС‚СЂ
mu = 3 # СЃСЂРµРґРЅРµРµ РєРѕР»РёС‡РµСЃС‚РІРѕ СЃРѕР±С‹С‚РёР№

# Р“РµРЅРµСЂРёСЂСѓРµРј СЃР»СѓС‡Р°Р№РЅС‹Рµ РїРµСЂРµРјРµРЅРЅС‹Рµ СЃ РїСѓР°СЃСЃРѕРЅРѕРІСЃРєРёРј СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµРј
# Р¤СѓРЅРєС†РёСЏ rvs - Random variates РїСЂРёРЅРёРјР°РµС‚ СЃР»РµРґСѓСЋС‰РёРµ Р°СЂРіСѓРјРµРЅС‚С‹:
# mu - СЃСЂРµРґРЅРµРµ РєРѕР»РёС‡РµСЃС‚РІРѕ СЃРѕР±С‹С‚РёР№
# size - СЃРєРѕР»СЊРєРѕ СЂР°Р· РїСЂРѕРІРµСЃС‚Рё СЃРµСЂРёСЋ РёСЃРїС‹С‚Р°РЅРёР№
data_poisson = poisson.rvs(mu = mu, size = 10000)

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ
ax = sns.histplot(data_poisson, kde = False, color = 'skyblue', bins = 30)
ax.set(xlabel = 'Р Р°СЃРїСЂРµРґРµР»РµРЅРёРµ РџСѓР°СЃСЃРѕРЅР°', ylabel = 'Р§Р°СЃС‚РѕС‚Р°')
plt.title('Р Р°СЃРїСЂРµРґРµР»РµРЅРёРµ РџСѓР°СЃСЃРѕРЅР° (Р»СЏРјР±РґР°=3)')
plt.show()
```

### <b><font color='saddlebrown'>РќРѕСЂРјР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ</font></b>

```python
from scipy.stats import norm
# РўРµРѕСЂРµС‚РёС‡РµСЃРєРѕРµ Р·РЅР°С‡РµРЅРёРµ
# Р¤СѓРЅРєС†РёСЏ СЃdf (Cumulative distribution function) РґР»СЏ РЅРѕСЂРјР°Р»СЊРЅРѕРіРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ РїСЂРёРЅРёРјР°РµС‚ СЃР»РµРґСѓСЋС‰РёРµ Р°СЂРіСѓРјРµРЅС‚С‹:
# x - РѕР¶РёРґР°РµРјРѕРµ СЃРѕР±С‹С‚РёРµ 
# loc - СЃСЂРµРґРЅРµРµ Р·РЅР°С‡РµРЅРёРµ
# scale - СЃС‚Р°РЅРґР°СЂС‚РЅРѕРµ РѕС‚РєР»РѕРЅРµРЅРёРµ
norm.cdf(73, 70, 3) - norm.cdf(67, 70, 3)
```

```python
# РџР°СЂР°РјРµС‚СЂС‹
mu = 70 # СЃСЂРµРґРЅРµРµ Р·РЅР°С‡РµРЅРёРµ
sigma = 3 # СЃС‚Р°РЅРґР°СЂС‚РЅРѕРµ РѕС‚РєР»РѕРЅРµРЅРёРµ

# Р“РµРЅРµСЂРёСЂСѓРµРј РЅРѕСЂРјР°Р»СЊРЅРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРЅС‹Рµ СЃР»СѓС‡Р°Р№РЅС‹Рµ РІРµР»РёС‡РёРЅС‹
data_normal = np.random.normal(mu, sigma, 1000)

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ
plt.hist(data_normal, bins = 30, density = True, alpha = 0.6, color = 'g')

plt.title("РќРѕСЂРјР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ (mu=70, sigma=3)")
plt.show()
```

```python
# РџР°СЂР°РјРµС‚СЂС‹
mu = 70 # СЃСЂРµРґРЅРµРµ Р·РЅР°С‡РµРЅРёРµ
sigma = 3 # СЃС‚Р°РЅРґР°СЂС‚РЅРѕРµ РѕС‚РєР»РѕРЅРµРЅРёРµ
a, b = 67, 73
# Р“РµРЅРµСЂРёСЂСѓРµРј РЅРѕСЂРјР°Р»СЊРЅРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРЅС‹Рµ СЃР»СѓС‡Р°Р№РЅС‹Рµ РІРµР»РёС‡РёРЅС‹
data_normal = np.random.normal(mu, sigma, 1000)

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ
plt.hist(data_normal, bins = 30, density = True, alpha = 0.6, color = 'g')

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° PDF
prob = norm.cdf(b, mu, sigma) - norm.cdf(a, mu, sigma)
xmin, xmax = plt.xlim() # Р‘РµСЂС‘Рј С‚РµРєСѓС‰РёРµ РіСЂР°РЅРёС†С‹ РѕСЃРё X РїРѕСЃР»Рµ РїРѕСЃС‚СЂРѕРµРЅРёСЏ РіРёСЃС‚РѕРіСЂР°РјРјС‹, С‡С‚РѕР±С‹ Р·РЅР°С‚СЊ РґРёР°РїР°Р·РѕРЅ Р·РЅР°С‡РµРЅРёР№ (РЅР°РїСЂРёРјРµСЂ, РѕС‚ 60 РґРѕ 80).
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mu, sigma)
plt.plot(x, p, 'k', linewidth=2)
plt.title(f"РќРѕСЂРјР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ (Ој={mu}, Пѓ={sigma})\nP({a} в‰¤ X в‰¤ {b}) = {prob:.3f}")
x_fill = np.linspace(a, b, 200)
plt.fill_between(x_fill, norm.pdf(x_fill, mu, sigma), color='skyblue', alpha=0.5, label='P(68 в‰¤ X в‰¤ 72)')
plt.show()
```

### <b><font color='saddlebrown'>Р Р°РІРЅРѕРјРµСЂРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ</font></b>

```python
from scipy.stats import uniform

# РўРµРѕСЂРµС‚РёС‡РµСЃРєРѕРµ Р·РЅР°С‡РµРЅРёРµ
# Р¤СѓРЅРєС†РёСЏ СЃdf (Cumulative distribution function) РґР»СЏ СЂР°РІРЅРѕРјРµСЂРЅРѕРіРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ РїСЂРёРЅРёРјР°РµС‚ СЃР»РµРґСѓСЋС‰РёРµ Р°СЂРіСѓРјРµРЅС‚С‹:
# x - РѕР¶РёРґР°РµРјРѕРµ СЃРѕР±С‹С‚РёРµ 
# loc - Р»РµРІР°СЏ РіСЂР°РЅРёС†Р°
# scale - СЂР°Р·РЅРѕСЃС‚СЊ РјРµР¶РґСѓ РїСЂР°РІРѕР№ Рё Р»РµРІРѕР№ РіСЂР°РЅРёС†РµР№
uniform.cdf(2, 0, 10) - uniform.cdf(0, 0, 10)
```

```python
# РџР°СЂР°РјРµС‚СЂС‹
a = 0 # РЅРёР¶РЅСЏСЏ РіСЂР°РЅРёС†Р°
b = 10 # РІРµСЂС…РЅСЏСЏ РіСЂР°РЅРёС†Р°

# Р“РµРЅРµСЂРёСЂСѓРµРј СЂР°РІРЅРѕРјРµСЂРЅРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРЅС‹Рµ СЃР»СѓС‡Р°Р№РЅС‹Рµ РІРµР»РёС‡РёРЅС‹
data_uniform = uniform.rvs(loc = a, scale = b - a, size = 1000)

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ
plt.hist(data_uniform, bins = 30, density = True, alpha = 0.6, color = 'b')

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° PDF
x = np.linspace(a, b, 100)
p = uniform.pdf(x, loc = a, scale = b-a)
plt.plot(x, p, 'k', linewidth = 2)
plt.title("Р Р°РІРЅРѕРјРµСЂРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ (a=0, b=10)")
plt.show()
```

### <b><font color='saddlebrown'>Р­РєСЃРїРѕРЅРµРЅС†РёР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ</font></b>

```python
from scipy.stats import expon
# РўРµРѕСЂРµС‚РёС‡РµСЃРєРѕРµ Р·РЅР°С‡РµРЅРёРµ
# Р¤СѓРЅРєС†РёСЏ СЃdf (Cumulative distribution function) РґР»СЏ СЌРєСЃРїРѕРЅРµРЅС†РёР°Р»СЊРЅРѕРіРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ РїСЂРёРЅРёРјР°РµС‚ СЃР»РµРґСѓСЋС‰РёРµ Р°СЂРіСѓРјРµРЅС‚С‹:
# x - РѕР¶РёРґР°РµРјРѕРµ СЃРѕР±С‹С‚РёРµ 
# scale - 1 / СЃРєРѕСЂРѕСЃС‚СЊ СЃРѕР±С‹С‚РёСЏ
expon.cdf(x = 10, scale = 10) - expon.cdf(x = 0, scale = 10)
```

```python
from scipy.stats import expon

# РїР°СЂР°РјРµС‚СЂ
lambda_ = 1/10 # РџР°СЂР°РјРµС‚СЂ СЃРєРѕСЂРѕСЃС‚Рё

# Р“РµРЅРµСЂРёСЂСѓРµРј СЌРєСЃРїРѕРЅРµРЅС†РёР°Р»СЊРЅРѕ СЂР°СЃРїСЂРµРґРµР»РµРЅРЅС‹Рµ СЃР»СѓС‡Р°Р№РЅС‹Рµ РІРµР»РёС‡РёРЅС‹
data_exponential = expon.rvs(scale = 1/lambda_, size = 1000)

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° СЂР°СЃРїСЂРµРґРµР»РµРЅРёСЏ
plt.hist(data_exponential, bins=30, density=True, alpha=0.6, color='r')

# РџРѕСЃС‚СЂРѕРµРЅРёРµ РіСЂР°С„РёРєР° PDF
x = np.linspace(0, np.max(data_exponential), 100)
p = expon.pdf(x, scale=1/lambda_)
plt.plot(x, p, 'k', linewidth=2)
plt.title("Р­РєСЃРїРѕРЅРµРЅС†РёР°Р»СЊРЅРѕРµ СЂР°СЃРїСЂРµРґРµР»РµРЅРёРµ (Р»СЏРјР±РґР°=1/5)")
plt.show()
```
