### <font color='lightcoral'>2. Тип данных Series</font>

- Series поддерживает автоматическое индексирование и возможность задавать пользовательские индексы. 
- Пользовательские индексы могут быть числами, строками или даже датами.
- Каждому элементу в Series можно присвоить метку (через индекс).
- Множество встроенных методов для анализа данных: статистические функции, операции сортировки и ранжирования, методы для обработки пропущенных значений.
- Series - строительный блок для DataFrame в pandas. 
- DataFrame - контейнер для Series.

### <font color='lightcoral'>4. Индексация</font>

```python
s = pd.Series([1, 3, 5, np.nan, 6, 8])
s
```

```python
#Элемент с индексом 2
print(s[2])
```

```python
#Присвавиваем новое значение для элемента с индексом 2
s[2] = 7
s
```

```python
#Получаем срез элементов с индексами от 2 до 4 включительно
s[2:5]
```

```python
#Получаем срез элементов с индексами от 1 и до последнего
s1 = s[1:]
s1
```

```python
#Получаем срез элементов с индексами от 0 и до предпоследнего
s2 = s[:-1]
s2
```

```python
#Попробуем сложить срезы s1 и s2
s1 + s2
```

```python
#Использование функций NumPy к Series
print(s1)
np.exp(s1)
```

```python
#Если индекс - строка
series_str = pd.Series([10, 20, 30, 40, 50], index = ['Store_1', 'Store_2', 'Store_3', 'Store_4', 'Store_5'])
series_str
```

```python
series_str.Store_3
```

```python
print(series_str['Store_3'])
```

```python
#Даты вместо индексов
dr = pd.date_range('1.1.2018', periods = 6, freq = 'W')
dr
```

```python
series_from_dr = pd.Series([1, 2, 3, 5, 7, 9], 
                           index = pd.date_range('1.1.2018', periods = 6, freq = 'W'))
series_from_dr
```

```python
series_from_dr['2018-02-11']
```

### <font color='lightcoral'>5. Продвинутые операции с Series</font>

```python
#diff - считает разность между соседними элементами
print(series_from_dr)
series_from_dr.diff()
```

```python
#Можно указать шаг между элементами
series_from_dr.diff(periods = 2)
```

```python
#dropna - избавиться от значений NaN
series_diff = series_from_dr.diff(periods = 2)
series_diff.fillna(series_diff.median())
# series_diff.dropna()
```

```python
#Кумулятивные операции
series_from_dr.cummin()
```

```python
series_from_dr.cummax()
```

```python
series_from_dr.cumsum()
```

### <font color='lightcoral'>6. Решение задач</font>

Вы ведёте учёт ежедневных расходов на продукты в течение месяца. Данные собирались разными способами: часть данных записана вручную, часть пришла из банковского приложения в виде словаря, часть – из массива, выгруженного из сторонней системы.

Вам нужно объединить эти данные, провести базовый анализ и подготовить выводы.

```python
#-----------------
#Исходные данные
#-----------------
#1 часть - список расходов за первую неделю (c 1-го по 7-е августа)
list_first_week = [500, 450, 520, 610, 480, 0, 700]
#2 часть - словарь расходов за вторую неделю (c 8-го по 14-е августа)
dict_second_week = {"Mon": 100, "Tue": 450, "Wed": 450, "Thu": 450, "Fri": 450, "Sat": 450, "Sun": 450}
#3 часть - записана в блокноте: 15.08 - 1000, 16.08 - 600
#4 часть - из массива NumPy (c 17-го по 31-е августа)
array_last_part = np.random.randint(100, 1000, 15)
```

```python
#-----------------
#Объедините данные в одну общую Series (можно использовать concat).
#-----------------
week_1 = pd.Series(list_first_week)
# print(week_1)
# print("-----------------------------")
week_2 = pd.Series(dict_second_week)
# print(week_2)
# print("-----------------------------")
week_3 = pd.Series([1000, 600], index = ['15.08', '16.08'])
# print(week_3)
# print("-----------------------------")
week_4 = pd.Series(array_last_part, index = pd.date_range('2018-08-17', periods = 15, freq = 'D'))
# print(week_4)
# print("-----------------------------")
total_series = pd.concat([week_1, week_2, week_3, week_4])
print(total_series)
```

```python
#-----------------
#Измените индексы на даты с 01.03.2026 по 31.03.2026 (можно использовать функцию rename)
#-----------------
dr = pd.date_range('08.01.2025', periods = 31)
new_array = pd.Series(total_series.values, index = dr)
print(new_array)
```

```python
#-----------------
#Выведите статистическую информацию по расходам
#-----------------
new_array.describe()
```

```python
#-----------------
#Сделайте срезы по каждой неделе
#-----------------
week_1 = new_array[0 : 7]
week_2 = new_array[7 : 14]
week_3 = new_array[14 : 21]
week_4 = new_array[21 : 28]
week_5 = new_array[28 :]
print(week_1)
print("-----------------------------")
print(week_2)
print("-----------------------------")
print(week_3)
print("-----------------------------")
print(week_4)
print("-----------------------------")
print(week_5)
#-----------------
```

```python
#-----------------
#Посчитайте разницу между одинаковыми днями недели и сохраните результат в новый Series
#-----------------
diff_series = new_array.diff(periods = 7)
print(diff_series)
```

```python
#-----------------
#Посчитайте накопительный итог расходов
#-----------------
total_sum = new_array.cumsum()
print(total_sum)
```

```python
#-----------------
#Найдите день с максимальными расходами
#-----------------
max_day = new_array.idxmax()
max_total = new_array.max()
print(f"День с максимальными расходами: {max_day}, Сумма расходов: {max_total}")
```
