Создайте диаграмму рассеяния для визуализации взаимосвязи между двумя переменными. Используйте набор данных tips, доступный в Plotly Express. Постройте график total_bill на оси x и tip на оси y. Раскрасьте точки по времени (обед или ужин).

```python
import plotly.express as px
import pandas as pd
```

### <font color='rebeccapurple'>Задача 4</font>

Создайте линейный график для наблюдения тенденций за определенный период. Используйте набор данных gapminder, доступный в Plotly Express. Отметьте год на оси x и lifeExp на оси y. Отфильтруйте данные, чтобы включить в них только Германию.

```python
df_gapminder = px.data.gapminder()
df_gapminder
```

```python
from plotly.subplots import make_subplots
df_Germany = df_gapminder.query("country == 'Germany'")
fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(
    px.line(data_frame=df_Germany,
            x='year',
            y='lifeExp',
            markers=True,
            color_discrete_sequence=['green']).data[0],
    secondary_y=False
)

fig.add_trace(
    px.line(data_frame=df_Germany,
            x='year',
            y='gdpPercap',
            markers=True,
            color_discrete_sequence=['red']).data[0],
    secondary_y=True
)

# Имена для легенды
fig.data[0].name = 'Life Expectancy'
fig.data[1].name = 'GDP per Capita'
fig.data[0].showlegend = True
fig.data[1].showlegend = True

# Подписи осей
fig.update_yaxes(title_text='Life Expectancy (years)', secondary_y=False)
fig.update_yaxes(title_text='GDP per Capita (USD)', secondary_y=True)
fig.update_xaxes(title_text='Year')

fig.update_layout(
    title='Life Expectancy and GDP per Capita in Germany Over Time',
    legend=dict(x=0.01, y=0.99)
)

fig.show()
```

### <font color='rebeccapurple'>Задача 5</font>

Создайте круговую диаграмму для анализа соотношения категорий. Используйте набор данных tips. Создайте круговую диаграмму, показывающую долю общих счетов по дням.

```python
fig = px.pie(data_frame = df_tips,
             names='day',
             values='total_bill',
             hole = 0.4,
             category_orders={'day': ['Thur', 'Fri', 'Sat', 'Sun']},
             title='Total Bill Distribution by Day of Week',
             width=700,
             height=500,)
fig.update_traces( # обновить данные объектов
    textinfo = 'value+label', # Как подписать данные на диаграмме
    pull = [0, 0, 0, 0.2] # Выделение отдельного кусочка пирога
)             
fig.show()
```

### <font color='rebeccapurple'>Задача 6</font>

Создайте тепловую карту для визуализации корреляции между несколькими переменными. Используйте набор данных iris, доступный в Plotly Express. Рассчитайте корреляционную матрицу и визуализируйте ее с помощью тепловой карты.

```python
df_iris = px.data.iris()
df_iris.head()
```

```python
iris_correlation = df_iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species_id']].corr()
iris_correlation
```

```python
fig = px.imshow(iris_correlation,
                text_auto=True,
                color_continuous_scale='tempo',
                title='Correlation Matrix of Iris Dataset',
                width=700,
                height=500
                )
fig.show()
```
