### <b><font color='saddlebrown'>1. Тест Левена</font></b>
Статистический метод для проверки того, имеют ли две или более групп данных одинаковые (однородные) дисперсии

Станция контроля качества на заводе проверяет металлические втулки. По техусловиям средняя длина втулки должна быть 51.2 мм. На линии A и линии B используют разные инструменты, и инженер хочет проверить, отличаются ли средние длины втулок между линиями.

Выбрали случайные независимые образцы:

line_A = [50.8, 51.0, 50.6, 51.2, 50.7, 50.9, 50.5, 50.8, 51.1, 50.6, 50.7, 50.8]<br>
line_B = [51.5, 51.7, 51.8, 51.6, 51.4, 51.9, 51.7, 51.3, 51.5, 51.8, 51.6, 51.7]

Нужно проверить на уровне значимости α=0.05, есть ли статистически значимая разница в средних между линиями

```python
line_A = [50.8, 51.0, 50.6, 51.2, 50.7, 50.9,  # Линия A
          50.5, 50.8, 51.1, 50.6, 50.7, 50.8]

line_B = [51.5, 51.7, 51.8, 51.6, 51.4, 51.9,  # Линия B
          51.7, 51.3, 51.5, 51.8, 51.6, 51.7]
```

```python
from scipy.stats import levene  # Тест Левена
import scipy.stats as stats
stat, p_value = levene(line_A, line_B)  # Расчет

print("Levene statistic:", stat)  # Статистика
print("p-value:", p_value)  # Значимость
```

```python
# Гипотеза для еста левена:
# H_0: дисперсии по возрасту y/n одинаковы.
# H_1: дисперсии по возрасту y/n различаются.
```

```python
t_statistic, p_value = stats.ttest_ind(a = age_y, b = age_n, equal_var = False)
print(f"T-статистика: {t_statistic}")
print(f"P-value: {p_value}")
```

2.3. Влияет ли длительность звонка на вероятность успеха?

```python
bank_duration = bank.groupby('y')['duration'].mean()  # Средняя длительность
print(bank_duration)  # Вывод
```

```python
# H_0: средняя длительность звонка одинакова для y=yes и y=no  # Нулевая
# H_1: средняя длительность звонка различается для y=yes и y=no  # Альтернатива
from scipy.stats import ttest_ind  # t-тест

duration_yes = bank[bank['y'] == 'yes']['duration']  # Успешные звонки
duration_no = bank[bank['y'] == 'no']['duration']  # Неуспешные звонки

print('Средняя длительность (yes):', round(duration_yes.mean(), 2))  # Среднее yes
print('Средняя длительность (no):', round(duration_no.mean(), 2))  # Среднее no

t_stat, p_value = ttest_ind(duration_yes, duration_no, equal_var=False)  # Welch
print(f't-stat={t_stat:.4f}, p-value={p_value}')  # Научная нотация
print('p-value (полное):', p_value)  # Без округления
print('Вывод: длительность статистически связана с успехом' if p_value < 0.05 else 'Вывод: статистически значимой связи не найдено')  # Вывод
```

2.1. Проблема «забывчивости» и СМС-напоминаний 

Руководство клиники тратит бюджет на отправку СМС-напоминаний о приёме. Окупаются ли эти затраты? Помогают ли сообщения снизить долю неявок, или они никак не влияют на поведение пациентов? Оцените эффективность этой практики.

```python
# H_0: доля no-show одинакова у пациентов с SMS и без SMS  # Нулевая
# H_1: доля no-show различается между группами SMS_received=0 и SMS_received=1  # Альтернатива
```

```python
from scipy.stats import chi2_contingency  # Хи-квадрат

sms_table = pd.crosstab(clinic['SMS_received'], clinic['No-show'])  # Таблица
display(sms_table)  # Просмотр

sms_no_show_rate = clinic.groupby('SMS_received')['no_show'].mean()  # Доли
print('Доля no-show без SMS:', round(sms_no_show_rate.loc[0], 4))  # Без SMS
print('Доля no-show с SMS:', round(sms_no_show_rate.loc[1], 4))  # С SMS

chi2, p_value, dof, expected = chi2_contingency(sms_table)  # Тест
print(f'chi2={chi2:.4f}, p-value={p_value:.6f}')  # Результат
print('Вывод: статистически значимая связь есть' if p_value < 0.05 else 'Вывод: статистически значимой связи не выявлено')  # Вывод
```

```python
print('Решение по H0 (2.1): H0 отклоняем, связь SMS и no-show статистически значима (p < 0.05).')  # тог
```

2.2. Демографический портрет «нарушителя дисциплины»

Есть ли разница в возрасте между дисциплинированными пациентами и теми, кто пропускает визиты? Помогите составить возрастной профиль группы риска, чтобы администраторы могли уделять им больше внимания при подтверждении записи.

```python
# H_0: средний возраст одинаков у пришедших и у no-show  # Нулевая
# H_1: средний возраст различается между группами No-show=No и No-show=Yes  # Альтернатива
```

```python
from scipy.stats import ttest_ind  # t-тест

show_age = clinic.loc[clinic['No-show'] == 'No', 'Age']  # Пришли
no_show_age = clinic.loc[clinic['No-show'] == 'Yes', 'Age']  # Не пришли

print('Средний возраст пришедших:', round(show_age.mean(), 2))  # Среднее
print('Средний возраст пропустивших:', round(no_show_age.mean(), 2))  # Среднее
print('Медиана возраста пришедших:', round(show_age.median(), 2))  # Медиана
print('Медиана возраста пропустивших:', round(no_show_age.median(), 2))  # Медиана

t_stat, p_value = ttest_ind(show_age, no_show_age, equal_var=False, nan_policy='omit')  # Welch
print(f't-stat={t_stat:.4f}, p-value={p_value:.6f}')  # Результат
print('Вывод: возрастные группы различаются статистически значимо' if p_value < 0.05 else 'Вывод: статистически значимого различия по возрасту не найдено')  # Вывод
```

```python
print('Решение по H0 (2.2): H0 отклоняем, возрастные группы отличаются статистически значимо (p < 0.05).')  # тог
```

2.4. Оценка надежности текущих показателей

На основе имеющихся исторических данных мы видим определенный процент неявок. Насколько точно этот показатель отражает реальную долгосрочную ситуацию в сети клиник? В каких пределах может колебаться истинная доля пропусков в будущем?

```python
# H_0: истинная доля no-show равна наблюдаемой p_hat  # Нулевая (точечная)
# H_1: истинная доля no-show отличается от p_hat  # Альтернатива
# Далее оцениваем неопределенность через 95% доверительный интервал  # Подход
```

```python
from scipy.stats import norm  # Квантиль

n = clinic['no_show'].shape[0]  # Размер
p_hat = clinic['no_show'].mean()  # Оценка
alpha = 0.05  # Уровень
z = norm.ppf(1 - alpha / 2)  # z-критерий
se = np.sqrt(p_hat * (1 - p_hat) / n)  # Ошибка
ci_low = p_hat - z * se  # Нижняя граница
ci_high = p_hat + z * se  # Верхняя граница

print(f'Оценка доли no-show: {p_hat:.4f}')  # Точечная
print(f'95% доверительный интервал: ({ci_low:.4f}, {ci_high:.4f})')  # нтервал
```

```python
print('Решение по 2.4: 95% Д для доли no-show = (0.1996; 0.2043), текущая оценка статистически надежна.')  # тог
```

Сеть клиник работает в разных районах города (Neighbourhood). Как руководству понять, в каких районах уровень неявок критически высок и требует оперативного вмешательства (например, изменения графика работы или логистики), а где отклонения носят чисто случайный характер?

```python
# H_0: доля no-show в районе не превышает среднюю по сети  # Нулевая
# H_1: доля no-show в районе выше средней по сети  # Альтернатива
# Проверка выполняется для каждого района с поправкой Бонферрони  # Множественные проверки
```

```python
from scipy.stats import binomtest  # Биномиальный тест

overall_no_show_rate = clinic['no_show'].mean()  # База

neighbourhood_stats = clinic.groupby('Neighbourhood')['no_show'].agg(['sum', 'count']).reset_index()  # По районам
neighbourhood_stats = neighbourhood_stats.rename(columns={'sum': 'no_show_count', 'count': 'total'})  # Переименовать
neighbourhood_stats['no_show_rate'] = neighbourhood_stats['no_show_count'] / neighbourhood_stats['total']  # Доля

neighbourhood_stats['p_value'] = neighbourhood_stats.apply(
    lambda row: binomtest(
        int(row['no_show_count']),  # Успехи
        int(row['total']),  # Попытки
        overall_no_show_rate,  # Базовая доля
        alternative='greater'  # Только выше
    ).pvalue,
    axis=1
)

bonf_alpha = 0.05 / len(neighbourhood_stats)  # Бонферрони
neighbourhood_stats['critical_zone'] = neighbourhood_stats['p_value'] < bonf_alpha  # Критичность

problem_areas = neighbourhood_stats[neighbourhood_stats['critical_zone']].sort_values('no_show_rate', ascending=False)  # Топ районов

print('Средняя доля no-show по сети:', round(overall_no_show_rate, 4))  # Средняя
print('Порог Bonferroni alpha:', round(bonf_alpha, 6))  # Порог
display(problem_areas.head(10))  # Топ-10
```

```python
print('Решение по H0 (2.5): выявлены районы, где доля no-show статистически выше средней по сети.')  # тог
```
