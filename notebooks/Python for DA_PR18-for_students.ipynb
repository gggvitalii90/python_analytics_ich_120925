### <b><font color='saddlebrown'>1. РўРµСЃС‚ Р›РµРІРµРЅР°</font></b>
РЎС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРёР№ РјРµС‚РѕРґ РґР»СЏ РїСЂРѕРІРµСЂРєРё С‚РѕРіРѕ, РёРјРµСЋС‚ Р»Рё РґРІРµ РёР»Рё Р±РѕР»РµРµ РіСЂСѓРїРї РґР°РЅРЅС‹С… РѕРґРёРЅР°РєРѕРІС‹Рµ (РѕРґРЅРѕСЂРѕРґРЅС‹Рµ) РґРёСЃРїРµСЂСЃРёРё

РЎС‚Р°РЅС†РёСЏ РєРѕРЅС‚СЂРѕР»СЏ РєР°С‡РµСЃС‚РІР° РЅР° Р·Р°РІРѕРґРµ РїСЂРѕРІРµСЂСЏРµС‚ РјРµС‚Р°Р»Р»РёС‡РµСЃРєРёРµ РІС‚СѓР»РєРё. РџРѕ С‚РµС…СѓСЃР»РѕРІРёСЏРј СЃСЂРµРґРЅСЏСЏ РґР»РёРЅР° РІС‚СѓР»РєРё РґРѕР»Р¶РЅР° Р±С‹С‚СЊ 51.2 РјРј. РќР° Р»РёРЅРёРё A Рё Р»РёРЅРёРё B РёСЃРїРѕР»СЊР·СѓСЋС‚ СЂР°Р·РЅС‹Рµ РёРЅСЃС‚СЂСѓРјРµРЅС‚С‹, Рё РёРЅР¶РµРЅРµСЂ С…РѕС‡РµС‚ РїСЂРѕРІРµСЂРёС‚СЊ, РѕС‚Р»РёС‡Р°СЋС‚СЃСЏ Р»Рё СЃСЂРµРґРЅРёРµ РґР»РёРЅС‹ РІС‚СѓР»РѕРє РјРµР¶РґСѓ Р»РёРЅРёСЏРјРё.

Р’С‹Р±СЂР°Р»Рё СЃР»СѓС‡Р°Р№РЅС‹Рµ РЅРµР·Р°РІРёСЃРёРјС‹Рµ РѕР±СЂР°Р·С†С‹:

line_A = [50.8, 51.0, 50.6, 51.2, 50.7, 50.9, 50.5, 50.8, 51.1, 50.6, 50.7, 50.8]<br>
line_B = [51.5, 51.7, 51.8, 51.6, 51.4, 51.9, 51.7, 51.3, 51.5, 51.8, 51.6, 51.7]

РќСѓР¶РЅРѕ РїСЂРѕРІРµСЂРёС‚СЊ РЅР° СѓСЂРѕРІРЅРµ Р·РЅР°С‡РёРјРѕСЃС‚Рё О±=0.05, РµСЃС‚СЊ Р»Рё СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё Р·РЅР°С‡РёРјР°СЏ СЂР°Р·РЅРёС†Р° РІ СЃСЂРµРґРЅРёС… РјРµР¶РґСѓ Р»РёРЅРёСЏРјРё

```python
line_A = [50.8, 51.0, 50.6, 51.2, 50.7, 50.9,  # Р›РёРЅРёСЏ A
          50.5, 50.8, 51.1, 50.6, 50.7, 50.8]

line_B = [51.5, 51.7, 51.8, 51.6, 51.4, 51.9,  # Р›РёРЅРёСЏ B
          51.7, 51.3, 51.5, 51.8, 51.6, 51.7]
```

```python
from scipy.stats import levene  # РўРµСЃС‚ Р›РµРІРµРЅР°
import scipy.stats as stats
stat, p_value = levene(line_A, line_B)  # Р Р°СЃС‡РµС‚

print("Levene statistic:", stat)  # РЎС‚Р°С‚РёСЃС‚РёРєР°
print("p-value:", p_value)  # Р—РЅР°С‡РёРјРѕСЃС‚СЊ
```

```python
bank_count_age = bank[(bank['age'] >= 30) & (bank['age'] <= 40)]['age'].count()
part_age = bank_count_age / bank['age'].count()  # РљРѕР»-РІРѕ РІ РІРѕР·СЂР°СЃС‚Рµ 30-40
print(bank_count_age)  # Р’С‹РІРѕРґ
print(f'Р”РѕР»СЏ РєР»РёРµРЅС‚РѕРІ РІ РІРѕР·СЂР°СЃС‚Рµ 30-40 Р»РµС‚: {part_age:.4%}')  # Р”РѕР»СЏ РІ РїСЂРѕС†РµРЅС‚Р°С…

# Р”Р РЅР°РґРѕ РїРѕСЃС‚РѕСЂС‚РёС‚СЊ РѕС‚ РіРµРЅРµСЂР°Р»СЊРЅРѕР№ СЃРѕРІРѕРєСѓРїРЅРѕСЃС‚Рё Р·РЅР°СЏ С‡С‚Рѕ Сѓ РЅР°СЃ РІС‹Р±РѕСЂР°РєР° РІ РґР°С‚Р°С„СЂРµР№Рј РІСЃРµРіРѕ 10 РїСЂРѕС†РµРЅС‚РѕРІ РѕС‚ РіРµРЅРµСЂР°Р»СЊРЅРѕР№ СЃРѕРІРѕРєСѓРїРЅРѕСЃС‚Рё
general_population_size = bank['age'].count() * 10  # Р“РµРЅРµСЂР°Р»СЊРЅР°СЏ СЃРѕРІРѕРєСѓРїРЅРѕСЃС‚СЊ

```

```python
from statsmodels.stats.proportion import proportion_confint

# Р”Р°РЅРЅС‹Рµ
n = bank['age'].count()  # Р Р°Р·РјРµСЂ РіРµРЅРµСЂР°Р»СЊРЅРѕР№ СЃРѕРІРѕРєСѓРїРЅРѕСЃС‚Рё
x = bank_count_age  # РљРѕР»РёС‡РµСЃС‚РІРѕ РІ РІРѕР·СЂР°СЃС‚Рµ 30-40
confidence = 0.95
alpha = 1 - confidence

# Р Р°СЃС‡С‘С‚ РґРѕРІРµСЂРёС‚РµР»СЊРЅРѕРіРѕ РёРЅС‚РµСЂРІР°Р»Р°
lower, upper = proportion_confint(count = x, nobs = n, alpha = alpha, method = 'normal')

print(f"95%-РЅС‹Р№ РґРѕРІРµСЂРёС‚РµР»СЊРЅС‹Р№ РёРЅС‚РµСЂРІР°Р» РґР»СЏ РґРѕР»Рё: ({lower*100:.3f}%, {upper*100:.3f}%)")
```

```python
# Р“РёРїРѕС‚РµР·Р° РґР»СЏ РµСЃС‚Р° Р»РµРІРµРЅР°:
# H_0: РґРёСЃРїРµСЂСЃРёРё РїРѕ РІРѕР·СЂР°СЃС‚Сѓ y/n РѕРґРёРЅР°РєРѕРІС‹.
# H_1: РґРёСЃРїРµСЂСЃРёРё РїРѕ РІРѕР·СЂР°СЃС‚Сѓ y/n СЂР°Р·Р»РёС‡Р°СЋС‚СЃСЏ.
```

```python
# РўРµСЃС‚ Р›РµРІРµРЅР° РїСЂРѕРІРµСЂСЏРµС‚, СЂР°РІРЅС‹ Р»Рё РґРёСЃРїРµСЂСЃРёРё РІ РґРІСѓС… РіСЂСѓРїРїР°С….
age_y = bank[bank['y'] == 'yes']['age'].tolist()  # Р’РѕР·СЂР°СЃС‚ РґР»СЏ y=yes
age_n = bank[bank['y'] == 'no']['age'].tolist()  # Р’РѕР·СЂР°СЃС‚ РґР»СЏ y=no
stat, p_value = levene(age_y, age_n)  # Р Р°СЃС‡РµС‚

print("Levene statistic:", stat)  # РЎС‚Р°С‚РёСЃС‚РёРєР°
print("p-value:", p_value)  # Р—РЅР°С‡РёРјРѕСЃС‚СЊ

```

```python
# H_0: СЃСЂРµРґРЅРёР№ РІРѕР·СЂР°СЃС‚ РєР»РёРµРЅС‚РѕРІ, РєРѕС‚РѕСЂС‹Рµ РѕС„РѕСЂРјРёР»Рё РґРµРїРѕР·РёС‚ (y=yes), СЂР°РІРµРЅ СЃСЂРµРґРЅРµРјСѓ РІРѕР·СЂР°СЃС‚Сѓ РєР»РёРµРЅС‚РѕРІ, РєРѕС‚РѕСЂС‹Рµ РѕС„РѕСЂРјРёР»Рё РґРµРїРѕР·РёС‚ (y=no).
# H_1: СЃСЂРµРґРЅРёР№ РІРѕР·СЂР°СЃС‚ РєР»РёРµРЅС‚РѕРІ, РєРѕС‚РѕСЂС‹Рµ РѕС„РѕСЂРјРёР»Рё РґРµРїРѕР·РёС‚ (y=yes), РЅРµ СЂР°РІРµРЅ СЃСЂРµРґРЅРµРјСѓ РІРѕР·СЂР°СЃС‚Сѓ РєР»РёРµРЅС‚РѕРІ, РєРѕС‚РѕСЂС‹Рµ РѕС„РѕСЂРјРёР»Рё РґРµРїРѕР·РёС‚ (y=no).
```

```python
t_statistic, p_value = stats.ttest_ind(a = age_y, b = age_n, equal_var = False)
print(f"T-СЃС‚Р°С‚РёСЃС‚РёРєР°: {t_statistic}")
print(f"P-value: {p_value}")
```

### 2.2. РЎРІСЏР·Р°РЅРѕ Р»Рё РЅР°Р»РёС‡РёРµ РёРїРѕС‚РµС‡РЅРѕРіРѕ РєСЂРµРґРёС‚Р° СЃ СЃРѕРіР»Р°СЃРёРµРј РѕС„РѕСЂРјРёС‚СЊ РґРµРїРѕР·РёС‚?

```python
bank_housing = bank.groupby('y')['housing'].value_counts().unstack()  # РРїРѕС‚РµРєР°
print(bank_housing)  # Р’С‹РІРѕРґ
```

```python
# H_0: РЅР°Р»РёС‡РёРµ РёРїРѕС‚РµРєРё РЅРµ СЃРІСЏР·Р°РЅРѕ СЃ РѕС„РѕСЂРјР»РµРЅРёРµРј РґРµРїРѕР·РёС‚Р°.
# H_1: РЅР°Р»РёС‡РёРµ РёРїРѕС‚РµРєРё СЃРІСЏР·Р°РЅРѕ СЃ РѕС„РѕСЂРјР»РµРЅРёРµРј РґРµРїРѕР·РёС‚Р°.
from scipy.stats import chi2_contingency
contingency_table = pd.crosstab(bank['housing'], bank['y'])  # РљРѕРЅС‚РёРЅРіРµРЅС‚РЅР°СЏ С‚Р°Р±Р»РёС†Р°
chi2, p, dof, expected = chi2_contingency(contingency_table)  # РўРµСЃС‚
print(f"Chi2 statistic: {chi2}")
print(f"P-value: {p}")

print("Р“РёРїРѕС‚РµР·Р° 1 РїСЂРёРЅРёРјР°РµС‚СЃСЏ, С‚Р°Рє РєР°Рє p-value < 0.05, Р·РЅР°С‡РёС‚ РЅР°Р»РёС‡РёРµ РёРїРѕС‚РµРєРё СЃРІСЏР·Р°РЅРѕ СЃ РѕС„РѕСЂРјР»РµРЅРёРµРј РґРµРїРѕР·РёС‚Р°.")
```

```python
A = ((bank['y'] == 'yes')).sum()  # A: РѕС„РѕСЂРјРёР» РґРµРїРѕР·РёС‚
B = ((bank['housing'] == 'yes')).sum()  # B: РµСЃС‚СЊ РёРїРѕС‚РµРєР°
AB = ((bank['y'] == 'yes') & (bank['housing'] == 'yes')).sum()  # A Рё B РІРјРµСЃС‚Рµ
N = len(bank)  # РћР±С‰РµРµ С‡РёСЃР»Рѕ РЅР°Р±Р»СЋРґРµРЅРёР№

P_A = A / N  # P(A)
P_B = B / N  # P(B)
P_AB = AB / N  # P(A в€© B)
P_A_given_B = AB / B  # P(A | B)

print(f"A = {A}, B = {B}, AB = {AB}, N = {N}")
print(f"P(A) = {P_A:.4f}")
print(f"P(B) = {P_B:.4f}")
print(f"P(A в€© B) = {P_AB:.4f}")
print(f"P(A | B) = {P_A_given_B:.4f}")
```

2.3. Р’Р»РёСЏРµС‚ Р»Рё РґР»РёС‚РµР»СЊРЅРѕСЃС‚СЊ Р·РІРѕРЅРєР° РЅР° РІРµСЂРѕСЏС‚РЅРѕСЃС‚СЊ СѓСЃРїРµС…Р°?

```python
bank_duration = bank.groupby('y')['duration'].mean()  # РЎСЂРµРґРЅСЏСЏ РґР»РёС‚РµР»СЊРЅРѕСЃС‚СЊ
print(bank_duration)  # Р’С‹РІРѕРґ
```

```python
# H_0: СЃСЂРµРґРЅСЏСЏ РґР»РёС‚РµР»СЊРЅРѕСЃС‚СЊ Р·РІРѕРЅРєР° РѕРґРёРЅР°РєРѕРІР° РґР»СЏ y=yes Рё y=no  # РќСѓР»РµРІР°СЏ
# H_1: СЃСЂРµРґРЅСЏСЏ РґР»РёС‚РµР»СЊРЅРѕСЃС‚СЊ Р·РІРѕРЅРєР° СЂР°Р·Р»РёС‡Р°РµС‚СЃСЏ РґР»СЏ y=yes Рё y=no  # РђР»СЊС‚РµСЂРЅР°С‚РёРІР°
from scipy.stats import ttest_ind  # t-С‚РµСЃС‚

duration_yes = bank[bank['y'] == 'yes']['duration']  # РЈСЃРїРµС€РЅС‹Рµ Р·РІРѕРЅРєРё
duration_no = bank[bank['y'] == 'no']['duration']  # РќРµСѓСЃРїРµС€РЅС‹Рµ Р·РІРѕРЅРєРё

print('РЎСЂРµРґРЅСЏСЏ РґР»РёС‚РµР»СЊРЅРѕСЃС‚СЊ (yes):', round(duration_yes.mean(), 2))  # РЎСЂРµРґРЅРµРµ yes
print('РЎСЂРµРґРЅСЏСЏ РґР»РёС‚РµР»СЊРЅРѕСЃС‚СЊ (no):', round(duration_no.mean(), 2))  # РЎСЂРµРґРЅРµРµ no

t_stat, p_value = ttest_ind(duration_yes, duration_no, equal_var=False)  # Welch
print(f't-stat={t_stat:.4f}, p-value={p_value}')  # РќР°СѓС‡РЅР°СЏ РЅРѕС‚Р°С†РёСЏ
print('p-value (РїРѕР»РЅРѕРµ):', p_value)  # Р‘РµР· РѕРєСЂСѓРіР»РµРЅРёСЏ
print('Р’С‹РІРѕРґ: РґР»РёС‚РµР»СЊРЅРѕСЃС‚СЊ СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё СЃРІСЏР·Р°РЅР° СЃ СѓСЃРїРµС…РѕРј' if p_value < 0.05 else 'Р’С‹РІРѕРґ: СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё Р·РЅР°С‡РёРјРѕР№ СЃРІСЏР·Рё РЅРµ РЅР°Р№РґРµРЅРѕ')  # Р’С‹РІРѕРґ
```

Р•СЃС‚СЊ Р»Рё СЂР°Р·РЅРёС†Р° РІ РІРѕР·СЂР°СЃС‚Рµ РјРµР¶РґСѓ РґРёСЃС†РёРїР»РёРЅРёСЂРѕРІР°РЅРЅС‹РјРё РїР°С†РёРµРЅС‚Р°РјРё Рё С‚РµРјРё, РєС‚Рѕ РїСЂРѕРїСѓСЃРєР°РµС‚ РІРёР·РёС‚С‹? РџРѕРјРѕРіРёС‚Рµ СЃРѕСЃС‚Р°РІРёС‚СЊ РІРѕР·СЂР°СЃС‚РЅРѕР№ РїСЂРѕС„РёР»СЊ РіСЂСѓРїРїС‹ СЂРёСЃРєР°, С‡С‚РѕР±С‹ Р°РґРјРёРЅРёСЃС‚СЂР°С‚РѕСЂС‹ РјРѕРіР»Рё СѓРґРµР»СЏС‚СЊ РёРј Р±РѕР»СЊС€Рµ РІРЅРёРјР°РЅРёСЏ РїСЂРё РїРѕРґС‚РІРµСЂР¶РґРµРЅРёРё Р·Р°РїРёСЃРё.

```python
# H_0: СЃСЂРµРґРЅРёР№ РІРѕР·СЂР°СЃС‚ РѕРґРёРЅР°РєРѕРІ Сѓ РїСЂРёС€РµРґС€РёС… Рё Сѓ no-show  # РќСѓР»РµРІР°СЏ
# H_1: СЃСЂРµРґРЅРёР№ РІРѕР·СЂР°СЃС‚ СЂР°Р·Р»РёС‡Р°РµС‚СЃСЏ РјРµР¶РґСѓ РіСЂСѓРїРїР°РјРё No-show=No Рё No-show=Yes  # РђР»СЊС‚РµСЂРЅР°С‚РёРІР°
```

```python
from scipy.stats import ttest_ind  # t-С‚РµСЃС‚

show_age = clinic.loc[clinic['No-show'] == 'No', 'Age']  # РџСЂРёС€Р»Рё
no_show_age = clinic.loc[clinic['No-show'] == 'Yes', 'Age']  # РќРµ РїСЂРёС€Р»Рё

print('РЎСЂРµРґРЅРёР№ РІРѕР·СЂР°СЃС‚ РїСЂРёС€РµРґС€РёС…:', round(show_age.mean(), 2))  # РЎСЂРµРґРЅРµРµ
print('РЎСЂРµРґРЅРёР№ РІРѕР·СЂР°СЃС‚ РїСЂРѕРїСѓСЃС‚РёРІС€РёС…:', round(no_show_age.mean(), 2))  # РЎСЂРµРґРЅРµРµ
print('РњРµРґРёР°РЅР° РІРѕР·СЂР°СЃС‚Р° РїСЂРёС€РµРґС€РёС…:', round(show_age.median(), 2))  # РњРµРґРёР°РЅР°
print('РњРµРґРёР°РЅР° РІРѕР·СЂР°СЃС‚Р° РїСЂРѕРїСѓСЃС‚РёРІС€РёС…:', round(no_show_age.median(), 2))  # РњРµРґРёР°РЅР°

t_stat, p_value = ttest_ind(show_age, no_show_age, equal_var=False, nan_policy='omit')  # Welch
print(f't-stat={t_stat:.4f}, p-value={p_value:.6f}')  # Р РµР·СѓР»СЊС‚Р°С‚
print('Р’С‹РІРѕРґ: РІРѕР·СЂР°СЃС‚РЅС‹Рµ РіСЂСѓРїРїС‹ СЂР°Р·Р»РёС‡Р°СЋС‚СЃСЏ СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё Р·РЅР°С‡РёРјРѕ' if p_value < 0.05 else 'Р’С‹РІРѕРґ: СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё Р·РЅР°С‡РёРјРѕРіРѕ СЂР°Р·Р»РёС‡РёСЏ РїРѕ РІРѕР·СЂР°СЃС‚Сѓ РЅРµ РЅР°Р№РґРµРЅРѕ')  # Р’С‹РІРѕРґ
```

```python
print('Р РµС€РµРЅРёРµ РїРѕ H0 (2.2): H0 РѕС‚РєР»РѕРЅСЏРµРј, РІРѕР·СЂР°СЃС‚РЅС‹Рµ РіСЂСѓРїРїС‹ РѕС‚Р»РёС‡Р°СЋС‚СЃСЏ СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё Р·РЅР°С‡РёРјРѕ (p < 0.05).')  # РС‚РѕРі
```

2.3. Р¤Р°РєС‚РѕСЂ СЃРѕС†РёР°Р»СЊРЅРѕР№ РїРѕРґРґРµСЂР¶РєРё 

Р’Р»РёСЏРµС‚ Р»Рё СѓС‡Р°СЃС‚РёРµ РїР°С†РёРµРЅС‚Р° РІ РіРѕСЃСѓРґР°СЂСЃС‚РІРµРЅРЅРѕР№ РїСЂРѕРіСЂР°РјРјРµ СЃРѕС†РёР°Р»СЊРЅРѕР№ РїРѕРјРѕС‰Рё (Scholarship / Bolsa FamГ­lia) РЅР° РµРіРѕ РґРёСЃС†РёРїР»РёРЅСѓ? РЎС‚РѕРёС‚ Р»Рё РєР»РёРЅРёРєРµ РІРІРѕРґРёС‚СЊ РѕСЃРѕР±С‹Рµ СѓСЃР»РѕРІРёСЏ РёР»Рё РїСЂР°РІРёР»Р° Р±СЂРѕРЅРёСЂРѕРІР°РЅРёСЏ РґР»СЏ СѓС‡Р°СЃС‚РЅРёРєРѕРІ СЌС‚РѕР№ РїСЂРѕРіСЂР°РјРјС‹?

2.4. РћС†РµРЅРєР° РЅР°РґРµР¶РЅРѕСЃС‚Рё С‚РµРєСѓС‰РёС… РїРѕРєР°Р·Р°С‚РµР»РµР№

РќР° РѕСЃРЅРѕРІРµ РёРјРµСЋС‰РёС…СЃСЏ РёСЃС‚РѕСЂРёС‡РµСЃРєРёС… РґР°РЅРЅС‹С… РјС‹ РІРёРґРёРј РѕРїСЂРµРґРµР»РµРЅРЅС‹Р№ РїСЂРѕС†РµРЅС‚ РЅРµСЏРІРѕРє. РќР°СЃРєРѕР»СЊРєРѕ С‚РѕС‡РЅРѕ СЌС‚РѕС‚ РїРѕРєР°Р·Р°С‚РµР»СЊ РѕС‚СЂР°Р¶Р°РµС‚ СЂРµР°Р»СЊРЅСѓСЋ РґРѕР»РіРѕСЃСЂРѕС‡РЅСѓСЋ СЃРёС‚СѓР°С†РёСЋ РІ СЃРµС‚Рё РєР»РёРЅРёРє? Р’ РєР°РєРёС… РїСЂРµРґРµР»Р°С… РјРѕР¶РµС‚ РєРѕР»РµР±Р°С‚СЊСЃСЏ РёСЃС‚РёРЅРЅР°СЏ РґРѕР»СЏ РїСЂРѕРїСѓСЃРєРѕРІ РІ Р±СѓРґСѓС‰РµРј?

```python
# H_0: РёСЃС‚РёРЅРЅР°СЏ РґРѕР»СЏ no-show СЂР°РІРЅР° РЅР°Р±Р»СЋРґР°РµРјРѕР№ p_hat  # РќСѓР»РµРІР°СЏ (С‚РѕС‡РµС‡РЅР°СЏ)
# H_1: РёСЃС‚РёРЅРЅР°СЏ РґРѕР»СЏ no-show РѕС‚Р»РёС‡Р°РµС‚СЃСЏ РѕС‚ p_hat  # РђР»СЊС‚РµСЂРЅР°С‚РёРІР°
# Р”Р°Р»РµРµ РѕС†РµРЅРёРІР°РµРј РЅРµРѕРїСЂРµРґРµР»РµРЅРЅРѕСЃС‚СЊ С‡РµСЂРµР· 95% РґРѕРІРµСЂРёС‚РµР»СЊРЅС‹Р№ РёРЅС‚РµСЂРІР°Р»  # РџРѕРґС…РѕРґ
```

```python
from scipy.stats import norm  # РљРІР°РЅС‚РёР»СЊ

n = clinic['no_show'].shape[0]  # Р Р°Р·РјРµСЂ
p_hat = clinic['no_show'].mean()  # РћС†РµРЅРєР°
alpha = 0.05  # РЈСЂРѕРІРµРЅСЊ
z = norm.ppf(1 - alpha / 2)  # z-РєСЂРёС‚РµСЂРёР№
se = np.sqrt(p_hat * (1 - p_hat) / n)  # РћС€РёР±РєР°
ci_low = p_hat - z * se  # РќРёР¶РЅСЏСЏ РіСЂР°РЅРёС†Р°
ci_high = p_hat + z * se  # Р’РµСЂС…РЅСЏСЏ РіСЂР°РЅРёС†Р°

print(f'РћС†РµРЅРєР° РґРѕР»Рё no-show: {p_hat:.4f}')  # РўРѕС‡РµС‡РЅР°СЏ
print(f'95% РґРѕРІРµСЂРёС‚РµР»СЊРЅС‹Р№ РёРЅС‚РµСЂРІР°Р»: ({ci_low:.4f}, {ci_high:.4f})')  # РРЅС‚РµСЂРІР°Р»
```

```python
print('Р РµС€РµРЅРёРµ РїРѕ 2.4: 95% Р”Р РґР»СЏ РґРѕР»Рё no-show = (0.1996; 0.2043), С‚РµРєСѓС‰Р°СЏ РѕС†РµРЅРєР° СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё РЅР°РґРµР¶РЅР°.')  # РС‚РѕРі
```

2.5. РџРѕРёСЃРє В«РїСЂРѕР±Р»РµРјРЅС‹С… Р·РѕРЅВ» (С„РёР»РёР°Р»РѕРІ) 

РЎРµС‚СЊ РєР»РёРЅРёРє СЂР°Р±РѕС‚Р°РµС‚ РІ СЂР°Р·РЅС‹С… СЂР°Р№РѕРЅР°С… РіРѕСЂРѕРґР° (Neighbourhood). РљР°Рє СЂСѓРєРѕРІРѕРґСЃС‚РІСѓ РїРѕРЅСЏС‚СЊ, РІ РєР°РєРёС… СЂР°Р№РѕРЅР°С… СѓСЂРѕРІРµРЅСЊ РЅРµСЏРІРѕРє РєСЂРёС‚РёС‡РµСЃРєРё РІС‹СЃРѕРє Рё С‚СЂРµР±СѓРµС‚ РѕРїРµСЂР°С‚РёРІРЅРѕРіРѕ РІРјРµС€Р°С‚РµР»СЊСЃС‚РІР° (РЅР°РїСЂРёРјРµСЂ, РёР·РјРµРЅРµРЅРёСЏ РіСЂР°С„РёРєР° СЂР°Р±РѕС‚С‹ РёР»Рё Р»РѕРіРёСЃС‚РёРєРё), Р° РіРґРµ РѕС‚РєР»РѕРЅРµРЅРёСЏ РЅРѕСЃСЏС‚ С‡РёСЃС‚Рѕ СЃР»СѓС‡Р°Р№РЅС‹Р№ С…Р°СЂР°РєС‚РµСЂ?

```python
# H_0: РґРѕР»СЏ no-show РІ СЂР°Р№РѕРЅРµ РЅРµ РїСЂРµРІС‹С€Р°РµС‚ СЃСЂРµРґРЅСЋСЋ РїРѕ СЃРµС‚Рё  # РќСѓР»РµРІР°СЏ
# H_1: РґРѕР»СЏ no-show РІ СЂР°Р№РѕРЅРµ РІС‹С€Рµ СЃСЂРµРґРЅРµР№ РїРѕ СЃРµС‚Рё  # РђР»СЊС‚РµСЂРЅР°С‚РёРІР°
# РџСЂРѕРІРµСЂРєР° РІС‹РїРѕР»РЅСЏРµС‚СЃСЏ РґР»СЏ РєР°Р¶РґРѕРіРѕ СЂР°Р№РѕРЅР° СЃ РїРѕРїСЂР°РІРєРѕР№ Р‘РѕРЅС„РµСЂСЂРѕРЅРё  # РњРЅРѕР¶РµСЃС‚РІРµРЅРЅС‹Рµ РїСЂРѕРІРµСЂРєРё
```

```python
from scipy.stats import binomtest  # Р‘РёРЅРѕРјРёР°Р»СЊРЅС‹Р№ С‚РµСЃС‚

overall_no_show_rate = clinic['no_show'].mean()  # Р‘Р°Р·Р°

neighbourhood_stats = clinic.groupby('Neighbourhood')['no_show'].agg(['sum', 'count']).reset_index()  # РџРѕ СЂР°Р№РѕРЅР°Рј
neighbourhood_stats = neighbourhood_stats.rename(columns={'sum': 'no_show_count', 'count': 'total'})  # РџРµСЂРµРёРјРµРЅРѕРІР°С‚СЊ
neighbourhood_stats['no_show_rate'] = neighbourhood_stats['no_show_count'] / neighbourhood_stats['total']  # Р”РѕР»СЏ

neighbourhood_stats['p_value'] = neighbourhood_stats.apply(
    lambda row: binomtest(
        int(row['no_show_count']),  # РЈСЃРїРµС…Рё
        int(row['total']),  # РџРѕРїС‹С‚РєРё
        overall_no_show_rate,  # Р‘Р°Р·РѕРІР°СЏ РґРѕР»СЏ
        alternative='greater'  # РўРѕР»СЊРєРѕ РІС‹С€Рµ
    ).pvalue,
    axis=1
)

bonf_alpha = 0.05 / len(neighbourhood_stats)  # Р‘РѕРЅС„РµСЂСЂРѕРЅРё
neighbourhood_stats['critical_zone'] = neighbourhood_stats['p_value'] < bonf_alpha  # РљСЂРёС‚РёС‡РЅРѕСЃС‚СЊ

problem_areas = neighbourhood_stats[neighbourhood_stats['critical_zone']].sort_values('no_show_rate', ascending=False)  # РўРѕРї СЂР°Р№РѕРЅРѕРІ

print('РЎСЂРµРґРЅСЏСЏ РґРѕР»СЏ no-show РїРѕ СЃРµС‚Рё:', round(overall_no_show_rate, 4))  # РЎСЂРµРґРЅСЏСЏ
print('РџРѕСЂРѕРі Bonferroni alpha:', round(bonf_alpha, 6))  # РџРѕСЂРѕРі
display(problem_areas.head(10))  # РўРѕРї-10
```

```python
print('Р РµС€РµРЅРёРµ РїРѕ H0 (2.5): РІС‹СЏРІР»РµРЅС‹ СЂР°Р№РѕРЅС‹, РіРґРµ РґРѕР»СЏ no-show СЃС‚Р°С‚РёСЃС‚РёС‡РµСЃРєРё РІС‹С€Рµ СЃСЂРµРґРЅРµР№ РїРѕ СЃРµС‚Рё.')  # РС‚РѕРі
```

2.6. Р’Р»РёСЏРЅРёРµ С…СЂРѕРЅРёС‡РµСЃРєРёС… Р·Р°Р±РѕР»РµРІР°РЅРёР№ 

РЎСѓС‰РµСЃС‚РІСѓРµС‚ РіРёРїРѕС‚РµР·Р°, С‡С‚Рѕ Р»СЋРґРё СЃ С…СЂРѕРЅРёС‡РµСЃРєРёРјРё Р·Р°Р±РѕР»РµРІР°РЅРёСЏРјРё (РіРёРїРµСЂС‚РѕРЅРёСЏ, РґРёР°Р±РµС‚) Р±РѕР»РµРµ РѕС‚РІРµС‚СЃС‚РІРµРЅРЅРѕ РѕС‚РЅРѕСЃСЏС‚СЃСЏ Рє Р·РґРѕСЂРѕРІСЊСЋ Рё СЂРµР¶Рµ РїСЂРѕРїСѓСЃРєР°СЋС‚ РїСЂРёС‘РјС‹.
```python
# H_0: РЅР°Р»РёС‡РёРµ С…СЂРѕРЅРёС‡РµСЃРєРёС… Р·Р°Р±РѕР»РµРІР°РЅРёР№ РЅРµ СЃРІСЏР·Р°РЅРѕ СЃ no-show  # РќСѓР»РµРІР°СЏ
# H_1: РЅР°Р»РёС‡РёРµ С…СЂРѕРЅРёС‡РµСЃРєРёС… Р·Р°Р±РѕР»РµРІР°РЅРёР№ СЃРІСЏР·Р°РЅРѕ СЃ no-show  # РђР»СЊС‚РµСЂРЅР°С‚РёРІР°
```
