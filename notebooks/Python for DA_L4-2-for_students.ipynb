##### <font color='lightcoral'>2.1. Задача</font>

<b>Дано:</b><br>
У вас есть информация о продажах в интернет-магазине:
- Товары: ['Ноутбук', 'Смартфон', 'Планшет', 'Наушники', 'Смарт-часы']
- Цена: [75000, 30000, 20000, 5000, 12000]
- Продано единиц: [120, 340, 150, 500, 230]

Задания:
1. Создайте DataFrame удобным для Вас способом (из Series, из файла, из словаря, из списка списков). Подумайте, как лучше организовать данные: что будет индексами, что будет колонками, что будет в данных.
2. Отсортируйте данные тремя способами:
- По цене (возрастание).
- По количеству проданных единиц (убывание).
- По выручке (Цена × Продано единиц) — в порядке убывания.
3. Найдите ТОП-3 товара по выручке и выведите их в отдельный DataFrame.

```python
products = ['Ноутбук', 'Смартфон', 'Планшет', 'Наушники', 'Смарт-часы']
prices = [75000, 30000, 20000, 5000, 12000]
Quantity = [120, 340, 150, 500, 230]
data = {'Product': products, 'Price': prices, 'Quantity': Quantity}
df_products = pd.DataFrame(data)
print(df_products)
print("----------------------------")
df_products_sorted = df_products.sort_values(by = ['Price', 'Quantity'], ascending = [True, False])
print(df_products_sorted)
print("----------------------------")
df_products['Revenue'] = df_products['Price'] * df_products['Quantity']
df_products_sorted = df_products.sort_values(by = ['Revenue'], ascending = False)
print(df_products_sorted)
print("----------------------------")
df_products_top3 = df_products.sort_values(by = ['Revenue'], ascending = False).head(3)
print(df_products_top3)
```

### <font color='lightcoral'>4. Решение задач</font>

У вас есть датасет Titanic. Основные колонки: <br>
- Survived — факт выживания (0 — нет, 1 — да)
- Pclass — класс билета (1, 2, 3)
- Name — имя пассажира
- Sex — пол
- Age — возраст
- Siblings/Spouses Aboard — число родственников (братьев/сестёр/супругов) на борту
- Parents/Children Aboard — число родителей/детей на борту
- Fare — стоимость билета

Задания:
1. Выведите каждую 3-ю строку до 20-й включительно.
2. Отсортируйте пассажиров: 
- По возрасту (по возрастанию);
- По стоимости билета (Fare) в порядке убывания;
- По классу билета (Pclass) и возрасту одновременно.
3. С помощью .loc:
- Выберите имена и возраст всех женщин младше 18 лет.
4. С помощью .iloc:
- Получите данные 10-й строки (по позиции);
- Выведите первые 3 столбца для строк с 5-й по 10-ю.
5. спользуя булевы условия:
- Найдите всех пассажиров 1-го класса старше 50 лет.
- Сколько женщин выжило во 2-м классе?
- Постройте DataFrame только с пассажирами, у которых стоимость билета выше средней (Fare > mean).

```python
import pandas as pd

pd.set_option("display.width", None)
pd.set_option("display.max_columns", None)

# Загрузка датасета
df_titanic = pd.read_csv("../data/titanic.csv")
df_titanic.head()
```

```python
# ЗАДАНЕ 1: каждая 3-я строка до 20-й включительно
# iloc[:20:3] — срез первых 20 строк с шагом 3
df_titanic.iloc[:20:3]
```

```python
# ЗАДАНЕ 2.1: сортировка по возрасту (от младших к старшим)
sorted_by_age = df_titanic.sort_values(by="Age")
sorted_by_age
```

```python
# ЗАДАНЕ 2.2: сортировка по стоимости билета (от дорогих к дешёвым)
sorted_by_fare_desc = df_titanic.sort_values(by="Fare", ascending=False)
sorted_by_fare_desc
```

```python
# ЗАДАНЕ 2.3: сортировка сначала по классу, потом по возрасту
sorted_by_class_age = df_titanic.sort_values(by=["Pclass", "Age"])
sorted_by_class_age
```

```python
# ЗАДАНЕ 3: .loc — выбор по меткам
# Условие: пол == female  возраст < 18
# Выбираем только столбцы Name и Age
women_under_18 = df_titanic.loc[
    (df_titanic["Sex"] == "female") & (df_titanic["Age"] < 18),
    ["Name", "Age"]
]
women_under_18
```

```python
# ЗАДАНЕ 4.1: .iloc — выбор по позиции
# iloc[9] — 10-я строка (индексация с 0)
df_titanic.iloc[9]
```

```python
# ЗАДАНЕ 4.2: .iloc — срез строк и столбцов по позиции
# Строки 5–10 (iloc[4:10]) и первые 3 столбца (:3)
df_titanic.iloc[5:11, :3]
```

```python
# ЗАДАНЕ 5.1: булево условие — 1-й класс  возраст > 50
first_class_over_50 = df_titanic[
    (df_titanic["Pclass"] == 1) & (df_titanic["Age"] > 50)
]
first_class_over_50
```

```python
# ЗАДАНЕ 5.2: количество женщин, выживших во 2-м классе
# .shape[0] возвращает количество строк
survived_women_2nd = df_titanic[
    (df_titanic["Sex"] == "female") &
    (df_titanic["Pclass"] == 2) &
    (df_titanic["Survived"] == 1)
].shape[0]
print(survived_women_2nd)
```

```python
# ЗАДАНЕ 5.3: DataFrame пассажиров с Fare выше среднего
# Сначала считаем среднее, потом фильтруем
fare_mean = df_titanic["Fare"].mean()  # среднее значение стоимости билета
fare_above_mean_df = df_titanic[df_titanic["Fare"] > fare_mean]
fare_above_mean_df
```
