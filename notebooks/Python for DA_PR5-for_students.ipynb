(РїРѕРґСЃРєР°Р·РєР°: РёСЃРїРѕР»СЊР·СѓР№С‚Рµ pd.to_datetime Рё pd.Timestamp.today()).

10. РџРѕСЃС‡РёС‚Р°Р№С‚Рµ СЃСЂРµРґРЅСЋСЋ С†РµРЅСѓ (price) РїРѕ РєР°Р¶РґРѕРјСѓ СЂР°Р№РѕРЅСѓ (neighbourhood_group).
11. РќР°Р№РґРёС‚Рµ РєРѕР»РёС‡РµСЃС‚РІРѕ СѓРЅРёРєР°Р»СЊРЅС‹С… С…РѕР·СЏРµРІ (host_id) РІ РєР°Р¶РґРѕРј СЂР°Р№РѕРЅРµ. РџРѕРґСЃРєР°Р·РєР°: df['col'].nunique()

```python
import pandas as pd
# 1. Р—Р°РіСЂСѓР·РєР° РґР°РЅРЅС‹С…
df_AB = pd.read_csv('../data/AB_NYC_2019.csv')
```

```python
# Р’С‹РІРµСЃС‚Рё РїРµСЂРІС‹Рµ 10 СЃС‚СЂРѕРє
df_AB.head(10)
```

```python
# Р’С‹РІРµСЃС‚Рё РЅР°Р·РІР°РЅРёСЏ СЃС‚РѕР»Р±С†РѕРІ Рё РѕР±С‰СѓСЋ РёРЅС„РѕСЂРјР°С†РёСЋ
df_AB.info()
df_AB.columns
```

```python
# Р”РѕР±Р°РІРёС‚СЊ СЃС‚РѕР»Р±РµС† price_per_year = price * minimum_nights
df_AB['price_per_year'] = df_AB['price'] * df_AB['minimum_nights']
df_AB[['name', 'price', 'minimum_nights', 'price_per_year']]
```

```python
# РЈРґР°Р»РёС‚СЊ СЃС‚РѕР»Р±РµС† id
del df_AB['id']
df_AB.head()
```

```python
# Р—Р°РјРµРЅРёС‚СЊ NaN РІ name РЅР° "No name"
df_AB['name'] = df_AB['name'].fillna('No name')
df_AB['name'][df_AB['name'] == 'No name']
```

```python
# Р—РЅР°С‡РµРЅРёСЏ minimum_nights > 365 Р·Р°РјРµРЅРёС‚СЊ РЅР° 365
df_AB.loc[df_AB['minimum_nights'] > 365, 'minimum_nights'] = 365
df_AB['minimum_nights'].sort_values(ascending=False)
```

```python
# Р—Р°РїРѕР»РЅРёС‚СЊ NaN РІ reviews_per_month РЅСѓР»СЏРјРё
df_AB['reviews_per_month'] = df_AB['reviews_per_month'].fillna(0)
df_AB['reviews_per_month'][df_AB['reviews_per_month'] == 0]
```

```python
# РџСЂРµРѕР±СЂР°Р·РѕРІР°С‚СЊ last_review РІ datetime
df_AB['last_review'] = pd.to_datetime(df_AB['last_review'])
df_AB['last_review'].sort_values(ascending=False)
```

```python
# РЎРѕР·РґР°С‚СЊ days_since_last_review = СЃРµРіРѕРґРЅСЏ - last_review
df_AB['days_since_last_review'] = pd.Timestamp('today') - df_AB['last_review']
df_AB['days_since_last_review'].sort_values(ascending=False)
```

```python
# РџРѕСЃС‡РёС‚Р°С‚СЊ СЃСЂРµРґРЅСЋСЋ С†РµРЅСѓ РїРѕ РєР°Р¶РґРѕРјСѓ СЂР°Р№РѕРЅСѓ
df_AB.groupby('neighbourhood_group', as_index=False)['price'].mean()
```

```python
# РџРѕСЃС‡РёС‚Р°С‚СЊ РєРѕР»РёС‡РµСЃС‚РІРѕ СѓРЅРёРєР°Р»СЊРЅС‹С… host_id РїРѕ РєР°Р¶РґРѕРјСѓ СЂР°Р№РѕРЅСѓ
df_AB.groupby('neighbourhood_group', as_index=False)['host_id'].nunique().sort_values(by='host_id', ascending=False)
```
