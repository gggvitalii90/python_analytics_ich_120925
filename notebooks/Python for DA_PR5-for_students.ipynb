```python
import pandas as pd
# 1. Загрузка данных
df_AB = pd.read_csv('../data/AB_NYC_2019.csv')
```

```python
# Вывести первые 10 строк
df_AB.head(10)
```

```python
# Вывести названия столбцов и общую информацию
df_AB.info()
df_AB.columns
```

```python
# Добавить столбец price_per_year = price * minimum_nights
df_AB['price_per_year'] = df_AB['price'] * df_AB['minimum_nights']
df_AB[['name', 'price', 'minimum_nights', 'price_per_year']]
```

```python
# Удалить столбец id
del df_AB['id']
df_AB.head()
```

```python
# Заменить NaN в name на "No name"
df_AB['name'] = df_AB['name'].fillna('No name')
df_AB['name'][df_AB['name'] == 'No name']
```

```python
# Значения minimum_nights > 365 заменить на 365
df_AB.loc[df_AB['minimum_nights'] > 365, 'minimum_nights'] = 365
df_AB['minimum_nights'].sort_values(ascending=False)
```

```python
# Заполнить NaN в reviews_per_month нулями
df_AB['reviews_per_month'] = df_AB['reviews_per_month'].fillna(0)
df_AB['reviews_per_month'][df_AB['reviews_per_month'] == 0]
```

```python
# Преобразовать last_review в datetime
df_AB['last_review'] = pd.to_datetime(df_AB['last_review'])
df_AB['last_review'].sort_values(ascending=False)
```

```python
# Создать days_since_last_review = сегодня - last_review
df_AB['days_since_last_review'] = pd.Timestamp('today') - df_AB['last_review']
df_AB['days_since_last_review'].sort_values(ascending=False)
```

```python
# Посчитать среднюю цену по каждому району
df_AB.groupby('neighbourhood_group', as_index=False)['price'].mean()
```

```python
# Посчитать количество уникальных host_id по каждому району
df_AB.groupby('neighbourhood_group', as_index=False)['host_id'].nunique().sort_values(by='host_id', ascending=False)
```

3. Описание и агрегация данных в DataFrame:

Напишите программу, которая:

- Запрашивает у пользователя количество строк данных для ввода.

- Для каждой строки запрашивает данные по трем столбцам: "мя", "Возраст", "Регион",

разделённые пробелом.

- Создаёт DataFrame на основе введённых данных.

- Выводит общую информацию о DataFrame, включая количество записей, средний

возраст, и количество уникальных регионов.

- Выбирает и выводит на экран данные о самом молодом участнике.Пример вывода:

Введите количество строк: 3

Введите данные (мя Возраст Регион): Анна 25 Берлин

Введите данные (мя Возраст Регион): Борис 30 Гамбург

Введите данные (мя Возраст Регион): Света 22 Мюнхен

Общая информация:

Количество записей: 3

Средний возраст: 25.67

Количество уникальных регионов: 3

Данные о самом молодом участнике:

мя: Света

Возраст: 22

Регион: Сочи

```python
n = int(input('Введите количество строк для вывода:'))
header = ["мя", "Возраст", "Регион"]
data = []
for  i in range(n):
    row = input('Введите данные (мя Возраст Регион):').split()
    data.append(dict(zip(header, row)))
    # print(data[-1])
df_data = pd.DataFrame(data)
df_data["Возраст"] = df_data["Возраст"].astype(int)
```

```python
df_data.info()
df_data.describe()
```

```python
print("Общая информация: "'\n')
print("Количество записей: ", len(df_data), '\n')
print("Средний возраст: ", round(df_data['Возраст'].mean(), 2), '\n')
print("Количество уникальных регионов: ", df_data['Регион'].nunique(), '\n')

print(f"""Данные о самом молодом участнике:\n
      мя: {df_data[df_data['Возраст'] == df_data['Возраст'].min()]['мя'].values[0]}\n
      Возраст: {df_data['Возраст'].min()}\n
      Регион: {df_data[df_data['Возраст'] == df_data['Возраст'].min()]['Регион'].values[0]}""")
```
