### <font color='teal'>Задание 1</font>
Напишите программу, которая:
- Запрашивает у пользователя ввод двух списков (list1, list2) – в первом 5 элементов, во втором 4. Списки вводятся в виде чисел, разделенных пробелами.
- Объединяет введенные списки в один новый список (list_full) из 9 элементов.
- Преобразует полученный список в массив (arr) NumPy.
- Трансформирует одномерный массив в матрицу размера 3х3. Обозначим эту матрицу Х.
- Выводит полученную матрицу.

```python
import numpy as np
#Запрашиваем у пользователя ввод двух списков
list1 = map(int, input("Введите список из 5 чисел: ").split())
list2 = map(int, input("Введите список из 4 чисел: ").split())
#Объединяем введенные списки в один новый список
combined_list = list(list1) + list(list2)
 
#Преобразуем полученный список в массив NumPy.
array = np.array(combined_list)

#Трансформируем одномерный массив в матрицу размера 3х3
x = array.reshape(3, 3)

#Выводим полученную матрицу.
print(x)
```

### <font color='teal'>Задание 5</font>
Создайте матрицу A 5х5 из чисел от 1 до 25. Из этой матрицы извлеките две подматрицы: С размера 2х3 из левого верхнего угла исходной матрицы, и В размера 3х2 из правого нижнего угла исходной матрицы. Найдите матричное произведение С на B и B на С.

```python
A_1 = np.arange(1,26).reshape(5, 5)
print("Исходная матрица A:")
print(A_1)
print("--------------")
C = A_1[:2, :3]
print("Подматрица C (2x3):")
print(C)
print("--------------")
B = A_1[2:, 3:]
print("Подматрица B (3x2):")
print(B)
print("--------------")
CB = C @ B
print("C @ B = ")
print(CB)
print("--------------")
BC = B @ C
print("B @ C = ")
print(BC)
```

### <font color='teal'>Задание 6</font>
Запросите у пользователя ввод трех чисел – x, y, z. Создайте трехмерный массив (arr_3_d) размера x*y*z и заполните его числами от 10 до 20 с подходящим шагом. Выведите его. Запросите у пользователя координаты элемента (i, j, k) этого тензора и его новое значение (new_element). Замените выбранный элемент на новое значение.
Вычислите среднее, максимум, минимум, сумму, произведение по каждой оси.

```python
x, y, z = map(int, input("Введите размеры массива (x y z): ").split())

arr_3_d = np.linspace(10, 20, num=x*y*z).reshape(x, y, z)
print("Трёхмерный массив:")
print(arr_3_d)

i = int(input(f"Введите i (от 0 до {x-1}): "))
j = int(input(f"Введите j (от 0 до {y-1}): "))
k = int(input(f"Введите k (от 0 до {z-1}): "))
new_element = float(input("Введите новое значение: "))
arr_3_d[i, j, k] = new_element
print("\nМассив после замены:")
print(arr_3_d)

# Ось 0
print("\nОсь 0:")
print("Среднее:")
print(arr_3_d.mean(axis=0))
print("----------")
print("Максимум:")
print(arr_3_d.max(axis=0))
print("----------")
print("Минимум:")
print(arr_3_d.min(axis=0))
print("----------")
print("Сумма:")
print(arr_3_d.sum(axis=0))
print("----------")
print("Произведение:")
print(arr_3_d.prod(axis=0))

# Ось 1
print("\nОсь 1:")
print("Среднее:")
print(arr_3_d.mean(axis=1))
print("----------")
print("Максимум:")
print(arr_3_d.max(axis=1))
print("----------")
print("Минимум:")
print(arr_3_d.min(axis=1))
print("----------")
print("Сумма:")
print(arr_3_d.sum(axis=1))
print("----------")
print("Произведение:")
print(arr_3_d.prod(axis=1))

# Ось 2
print("\nОсь 2:")
print("Среднее:")
print(arr_3_d.mean(axis=2))
print("----------")
print("Максимум:")
print(arr_3_d.max(axis=2))
print("----------")
print("Минимум:")
print(arr_3_d.min(axis=2))
print("----------")
print("Сумма:")
print(arr_3_d.sum(axis=2))
print("----------")
print("Произведение:")
print(arr_3_d.prod(axis=2))
```
