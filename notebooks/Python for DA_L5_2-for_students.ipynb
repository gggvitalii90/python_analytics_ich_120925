### <font color='lightcoral'>1. Р¤СѓРЅРєС†РёСЏ pd.concat</font>

<b>РЎРёРЅС‚Р°РєСЃРёСЃ:</b>
pd.concat(objs, axis=0, join='outer', ignore_index=False)
- objs вЂ” РѕР±СЉРµРґРёРЅСЏРµРјС‹Рµ С‚Р°Р±Р»РёС†С‹;
- axis : {0 РёР»Рё 'index', 1 РёР»Рё 'columns'} вЂ” РѕСЃСЊ РёРЅРґРµРєСЃРѕРІ РёР»Рё РѕСЃСЊ РєРѕР»РѕРЅРѕРє, РёРЅС‹РјРё СЃР»РѕРІР°РјРё СЃРѕРµРґРёРЅРµРЅРёРµ РїРѕ РІРµСЂС‚РёРєР°Р»Рё РёР»Рё РїРѕ РіРѕСЂРёР·РѕРЅС‚Р°Р»Рё;
- join : {'inner', 'outer'} вЂ” С‚РёРї РѕР±СЉРµРґРёРЅРµРЅРёСЏ вЂ” РїРµСЂРµСЃРµС‡РµРЅРёРµ РёР»Рё РѕР±СЉРµРґРёРЅРµРЅРёРµ РёРЅРґРµРєСЃРѕРІ/РєРѕР»РѕРЅРѕРє;

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.random.randint(1, 100, size=(10, 4)),
                  columns=['A', 'B', 'C', 'D'])
df
```

```python
other = df[:4].copy()
other
```

```python
other['flag'] = other['D'] > 0
other
```

```python
# СЃРѕРµРґРёРЅРµРЅРёРµ РїРѕ РІРµСЂС‚РёРєР°Р»Рё СЃ С‚РёРїРѕРј РѕР±СЉРµРґРёРЅРµРЅРёСЏ - РїРµСЂРµСЃРµС‡РµРЅРёРµ
pd.concat([df, other], axis = 0, join = 'inner')
```

```python
# СЃРѕРµРґРёРЅРµРЅРёРµ РїРѕ РІРµСЂС‚РёРєР°Р»Рё СЃ С‚РёРїРѕРј РѕР±СЉРµРґРёРЅРµРЅРёСЏ - РѕР±СЉРµРґРёРЅРµРЅРёРµ
new_df = pd.concat([df, other], axis = 0, join = 'outer')
new_df
```

```python
# СЃРѕРµРґРёРЅРµРЅРёРµ РїРѕ РіРѕСЂРёР·РѕРЅС‚Р°Р»Рё СЃ С‚РёРїРѕРј РѕР±СЉРµРґРёРЅРµРЅРёСЏ - РѕР±СЉРµРґРёРЅРµРЅРёРµ
pd.concat([df, other], axis = 1, join = 'outer')
```

### <font color='lightcoral'>2. Р¤СѓРЅРєС†РёРё pd.merge Рё df.join</font>

<b>РЎРёРЅС‚Р°РєСЃРёСЃ merge:</b><br>
pd.merge(left, right, how='inner', on=None, left_on=None, right_on=None, left_index=False, right_index=False, suffixes=('_x', '_y'), ...)
- left Рё right вЂ” РѕР±СЉРµРґРёРЅСЏРµРјС‹Рµ С‚Р°Р±Р»РёС†С‹.
- how вЂ” С‚РёРї РѕР±СЉРµРґРёРЅРµРЅРёСЏ:
    - left вЂ” С‚РѕР»СЊРєРѕ РїРѕ РєР»СЋС‡Р°Рј РёР· Р»РµРІРѕР№ С‚Р°Р±Р»РёС†С‹ == SQL left outer join;
    - right вЂ” С‚РѕР»СЊРєРѕ РїРѕ РєР»СЋС‡Р°Рј РёР· РїСЂР°РІРѕР№ С‚Р°Р±Р»РёС†С‹ == SQL right outer join;
    - outer вЂ” РїРѕ РѕР±СЉРµРґРёРЅРµРЅРёСЋ РєР»СЋС‡РµР№ == SQL full outer join;
    - inner вЂ” РїРѕ РїРµСЂРµСЃРµС‡РµРЅРёСЋ РєР»СЋС‡РµР№ == SQL inner join. РџРѕ СѓРјРѕР»С‡Р°РЅРёСЋ.
- on вЂ” РёРјСЏ (РёР»Рё РёРјРµРЅР°) РєРѕР»РѕРЅРѕРє, РїРѕ РєРѕС‚РѕСЂС‹Рј Р±СѓРґРµС‚ РїСЂРѕРёР·РІРѕРґРёС‚СЊСЃСЏ РѕР±СЉРµРґРёРЅРµРЅРёРµ (С‚.Рµ. РєР»СЋС‡Рё). Р•СЃР»Рё РёС… РЅРµСЃРєРѕР»СЊРєРѕ, С‚Рѕ РЅСѓР¶РЅРѕ РїРµСЂРµРґР°С‚СЊ СЃРїРёСЃРѕРє РёРјРµРЅ. РРјРµРЅР° РєРѕР»РѕРЅРѕРє РІ С‚Р°Р±Р»РёС†Рµ РґРѕР»Р¶РЅС‹ СЃРѕРІРїР°РґР°С‚СЊ.
- left_on Рё right_on вЂ” Р°РЅР°Р»РѕРіРёС‡РЅРѕ on РґР»СЏ СЃР»СѓС‡Р°СЏ, РєРѕРіРґР° РІ С‚Р°Р±Р»РёС†Р°С… СЂР°Р·Р»РёС‡Р°СЋС‚СЃСЏ РёРјРµРЅР° РєРѕР»РѕРЅРѕРє, СЃРѕРѕС‚РІРµС‚СЃС‚РІСѓСЋС‰РёРµ РєР»СЋС‡Р°Рј.
- left_index Рё right_index вЂ” РёСЃРїРѕР»СЊР·РѕРІР°С‚СЊ Р»Рё РёРЅРґРµРєСЃС‹ РІ РєР°С‡РµСЃС‚РІРµ РєР»СЋС‡РµР№.
- suffixes вЂ” СЃСѓС„С„РёРєСЃС‹, РєРѕС‚РѕСЂС‹Рµ Р±СѓРґСѓС‚ РґРѕР±Р°РІР»РµРЅС‹ Рє С‚РµРј РєРѕР»РѕРЅРєР°Рј, РёРјРµРЅР° РєРѕС‚РѕСЂС‹С… РїРѕРІС‚РѕСЂСЏСЋС‚СЃСЏ.

<b>РЎРёРЅС‚Р°РєСЃРёСЃ join:</b><br>
df.join(other, on=['key'], how='left', lsuffix='', rsuffix='', sort=False)
- df вЂ” РѕСЃРЅРѕРІРЅР°СЏ С‚Р°Р±Р»РёС†Р°. Р’ РєР°С‡РµСЃС‚РІРµ РєР»СЋС‡РµР№ РёСЃРїРѕР»СЊР·СѓРµС‚СЃСЏ РёРЅРґРµРєСЃ.
- other вЂ” РґСЂСѓРіР°СЏ С‚Р°Р±Р»РёС†Р°.
- on вЂ” РєРѕР»РѕРЅРєР°(-Рё) РІ other, СЃРѕРѕС‚РІРµС‚СЃС‚РІСѓСЋС‰Р°СЏ РєР»СЋС‡Р°Рј, РїРѕ РЅРёРј РїСЂРѕРёСЃС…РѕРґРёС‚ РѕР±СЉРµРґРёРЅРµРЅРёРµ. Р•СЃР»Рё None, С‚Рѕ РёСЃРїРѕР»СЊР·СѓРµС‚СЃСЏ РёРЅРґРµРєСЃ.
- how вЂ” С‚РёРї РѕР±СЉРµРґРёРЅРµРЅРёСЏ (СЃРј. pd.merge).
- lsuffix Рё rsuffix вЂ” СЃСѓС„С„РёРєСЃС‹, РєРѕС‚РѕСЂС‹Рµ Р±СѓРґСѓС‚ РґРѕР±Р°РІР»РµРЅС‹ Рє С‚РµРј РєРѕР»РѕРЅРєР°Рј, РёРјРµРЅР° РєРѕС‚РѕСЂС‹С… РїРѕРІС‚РѕСЂСЏСЋС‚СЃСЏ.

##### <font color='lightcoral'>2.1. Р’ РѕР±РѕРёС… С‚Р°Р±Р»РёС†Р°С… РєР»СЋС‡Рё РїРѕРІС‚РѕСЂСЏСЋС‚СЃСЏ</font>

```python
left = pd.DataFrame({'key': ['A', 'A'],
                     'lval': [1, 2]})
left
```

```python
right = pd.DataFrame({'key': ['A', 'A'],
                      'rval': [4, 5]})
right
```

```python
pd.merge(left, right, on = 'key')
```

##### <font color='lightcoral'>2.2. Р’ С‚Р°Р±Р»РёС†Р°С… РєР»СЋС‡Рё РЅРµ РїРѕРІС‚РѕСЂСЏСЋС‚СЃСЏ</font>

```python
left = pd.DataFrame({'key': ['A', 'B'],
                     'lval': [1, 2]})
left
```

```python
#РћРїРёСЃР°С‚РµР»СЊРЅС‹Рµ С…Р°СЂР°РєС‚РµСЂРёСЃС‚РёРєРё
right = pd.DataFrame({'key': ['A', 'B'],
                      'rval': [4, 5]})
right
```

```python
pd.merge(left, right, on = 'key')
```

##### <font color='lightcoral'>2.3. Р’РЅРµС€РЅРµРµ СЃР»РёСЏРЅРёРµ - outer</font>

```python
left = pd.DataFrame({'lkey': ['A', 'B', 'C', 'A'],
                     'value': range(4)})
left
```

```python
right = pd.DataFrame({'rkey': ['A', 'B', 'D', 'B'],
                      'value': range(4, 8)})
right
```

```python
pd.merge(left, right,
         left_on='lkey', right_on='rkey', how='outer')
```

##### <font color='lightcoral'>2.4. Р’РЅСѓС‚СЂРµРЅРЅРµРµ СЃР»РёСЏРЅРёРµ - inner</font>

```python
pd.merge(left, right,
         left_on='lkey', right_on='rkey', how='inner')
```

##### <font color='lightcoral'>2.5. РћР±СЉРµРґРёРЅРµРЅРёРµ РїРѕ РєР»СЋС‡Р°Рј Р»РµРІРѕР№ С‚Р°Р±Р»РёС†С‹</font>

```python
pd.merge(left, right,
         left_on='lkey', right_on='rkey', how='left')
```

##### <font color='lightcoral'>2.6. РћР±СЉРµРґРёРЅРµРЅРёРµ РїРѕ РєР»СЋС‡Р°Рј РїСЂР°РІРѕР№ С‚Р°Р±Р»РёС†С‹</font>

```python
pd.merge(left, right,
         left_on='lkey', right_on='rkey', how='right')
```

```python
left.join(right, how = 'inner', lsuffix='_', rsuffix='__')
```

```python
#РЈСЃС‚Р°РЅРѕРІРєР° РєР»СЋС‡Р° РІ РєР°С‡РµСЃС‚РІРµ РёРЅРґРµРєСЃР°
pd.merge(left, right, left_on='lkey', right_on='rkey', how='inner') \
        .set_index('lkey')[['value_x', 'value_y']]
```

```python
# РўРѕ Р¶Рµ СЃР°РјРѕРµ СЃ Join
left.set_index('lkey') \
    .join(right.set_index('rkey'), rsuffix='_r', how='inner')
```

```python
# Р—Р°РіСЂСѓР·РёС‚Рµ РІС‚РѕСЂРѕР№ РґР°С‚Р°СЃРµС‚ РІ DataFrame movies_2
df_m_2 = pd.read_csv('../data/movies_2.csv')
df_m_2
```

```python
# Р—Р°РіСЂСѓР·РёС‚Рµ С‚СЂРµС‚РёР№ РґР°С‚Р°СЃРµС‚ РІ DataFrame ratings
df_r = pd.read_csv('../data/ratings_mean.csv')
df_r
```

### <font color='lightcoral'>3. Р“СЂСѓРїРїРёСЂРѕРІРєР°</font>

<b>РЎРёРЅС‚Р°РєСЃРёСЃ:</b><br>
df.groupby(by=column, axis=0, level=None, sort=True, ...)
- df вЂ” С‚Р°Р±Р»РёС†Р°, РґР°РЅРЅС‹Рµ РєРѕС‚РѕСЂРѕР№ РґРѕР»Р¶РЅС‹ Р±С‹С‚СЊ СЃРіСЂСѓРїРїРёСЂРѕРІР°РЅС‹;
- by вЂ” Р·Р°РґР°РµС‚ РїСЂРёРЅС†РёРї РіСЂСѓРїРїРёСЂРѕРІРєРё. Р§Р°С‰Рµ РІСЃРµРіРѕ СЌС‚Рѕ РёРјСЏ СЃС‚РѕР»Р±С†Р°, РїРѕ РєРѕС‚РѕСЂРѕРјСѓ РЅСѓР¶РЅРѕ СЃРіСЂСѓРїРїРёСЂРѕРІР°С‚СЊ. РњРѕР¶РµС‚ С‚Р°Рє Р¶Рµ Р±С‹С‚СЊ С„СѓРЅРєС†РёРµР№;
- axis вЂ” РѕСЃСЊ (0 = РіСЂСѓРїРїРёСЂРѕРІР°С‚СЊ СЃС‚СЂРѕРєРё, 1 = РіСЂСѓРїРїРёСЂРѕРІР°С‚СЊ СЃС‚РѕР»Р±С†С‹);
