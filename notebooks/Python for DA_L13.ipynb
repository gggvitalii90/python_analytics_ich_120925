### <font color='rebeccapurple'>2. Line Chart</font>

```python
import pandas as pd

df_finance = pd.read_csv('../data/finance_chart_apple.csv')
df_finance.head(3)
```

```python
# Первый аргумент - датафрейм
# x - название столбца для отображения данных по оси X
# y - название столбца для отображения данных по оси Y
# title - название графика
fig = px.line(data_frame = df_finance,
              x = 'Date',
              y = 'AAPL.Close',
              title = 'AAPL.Close Trend Over Days'
              )
fig.show()
```

```python
# добавим деление по признаку Direction
fig = px.line(data_frame = df_finance,
              x = 'Date', 
              y = 'AAPL.Close', 
              line_group = 'direction',
              color = 'direction',
              title = 'Total AAPL.Close Trend by Direction'
             )
fig.show()
```

```python
# Другие параметры
fig = px.line(data_frame = df_finance,
              x = 'Date', 
              y = 'AAPL.Close',
              line_group = 'direction',
              color = 'direction',
              title = 'Total AAPL.Close Trend by Direction',
              # markers = True, # Показывать/не показывать маркеры на линии
              # line_dash = 'direction', # сопоставить тип линии с полем direction
              # symbol = 'direction', # сопоставить тип символа с полем direction
              # text = 'AAPL.Close', # какое поля взять для подписи данных на графике
              labels = {'Date':'Дата', 'AAPL.Close':'Цена закрытия акций'},
              color_discrete_sequence = ['green', 'black'],
              line_shape = 'linear' # тип линии 'linear', 'spline', 'hv', 'vh', 'hvh', 'vhv'
             )
fig.show()
```

```python
# Матрица линейных графиков
fig = px.line(data_frame = df_finance,
              x = 'Date', 
              y = 'AAPL.Close',
              # line_group = 'direction',
              # color = 'direction',
              title = 'Total AAPL.Close Trend by Direction',
              # markers = True, # Показывать/не показывать маркеры на линии
              # line_dash = 'direction', # сопоставить тип линии с полем direction
              # symbol = 'direction', # сопоставить тип символа с полем direction
              # text = 'AAPL.Close', # какое поля взять для подписи данных на графике
              labels = {'Date':'Дата', 'AAPL.Close':'Цена закрытия акций'},
              facet_row = 'direction'
             )
fig.show()
```

### <font color='rebeccapurple'>4. Bar Charts</font>

```python
df_Germany = px.data.gapminder().query("country=='Germany'") 
df_Germany['popMillion'] = df_Germany['pop']/1_000_000

fig = px.bar(data_frame = df_Germany, 
             x = 'year', # значения из какого столбца таблицы на оси X
             y = 'popMillion', # значения из какого столбца таблицы на оси Y
             title = 'Total Population of Germany by Year') # заголовок диаграммы
fig.show()
```

```python
fig = px.bar(data_frame = df_Germany, 
             x = 'year',
             y = ['popMillion', 'lifeExp'], # Несколько значений по столбцу Y
             title = 'Total Population of Germany by Year',
             barmode = 'group' # способ расположения нескольких столбцов:'group', 'overlay', 'relative' 
)
fig.show()
```

```python
# другие параметры
fig = px.bar(data_frame = df_Germany, 
             x = 'year',
             y = 'popMillion',
             title = 'Total Population of Germany by Year',
             color = 'lifeExp',
             text = round(df_Germany['popMillion'], 1), # Подписи столбцов
             # text_auto = True,
             labels = {'year':'Год', 'popMillion':'Население (млн.)'}, # Подписи осей
             orientation = 'v', # Ориентация графика ('v' - вертикальная, 'h' - горизонтальная)
             width = 1000, # ширина графика в пикселях
             height = 500 # высота графика в пикселях
            )
fig.show()
```

```python
df_medals = px.data.medals_long()
display(df_medals.head(3))
fig = px.bar(data_frame = df_medals, 
             x = "nation", 
             y = "count", 
             color = "medal",
             title = "Long-Form Input",
             color_discrete_sequence = ['green', 'violet','blue'],
            )
fig.show()
```

```python
df_medals = px.data.medals_long()
fig = px.bar(data_frame = df_medals, 
             x = "medal", 
             y = "count", 
             color = "nation",
             title = "Long-Form Input",
             pattern_shape = "nation",
             pattern_shape_sequence = [".", "x", "+"]
            )
fig.show()
```

```python
# Матрица баров
df_tips = px.data.tips()
fig = px.bar(data_frame = df_tips,
             x = "sex",
             y = "total_bill",
             color = "smoker",
             barmode = "group",
             facet_row = "time",
             facet_col = "day",
             category_orders={"day": ["Thur", "Fri", "Sat", "Sun"],
                              "time": ["Lunch", "Dinner"]})
fig.show()
```

```python
# Добавим анимацию
df_gapminder = px.data.gapminder().query("country in ['Germany', 'Spain', 'France']")
df_gapminder['popMillion'] = df_gapminder['pop']/1000000
fig = px.bar(
    data_frame = df_gapminder, 
    x = 'country',                # страна вместо года (чтобы бар был один и "рос" по годам)
    y = 'popMillion',
    title = 'Total Population of Germany by Year',
    color = 'lifeExp',
    text_auto = True,
    labels = {'year':'Год', 'popMillion':'Население (млн.)'},
    orientation = 'v',
    animation_frame = 'year',
    width = 1000, # ширина графика в пикселях
    height = 500 # высота графика в пикселях
)

fig.show()
```

### <font color='rebeccapurple'>5. Круговые диаграммы</font>

```python
df_gapminder_2007 = px.data.gapminder().query("year == 2007")
df_gapminder_2007.head(20)
```

```python
fig = px.pie(data_frame = df_gapminder_2007,
             names = 'continent', # Категории для "кусочков" диаграммы
             title = 'Country Distribution by Continent in 2007')
fig.show()
```

```python
# Настройка отображения информации
fig = px.pie(data_frame = df_gapminder_2007,
             names = 'continent', # Категории для "кусочков" диаграммы
             title = 'Country Distribution by Continent in 2007',
             hover_data = ['country'], # Включает дополнительные данные (названия стран) при наведении курсора на срезы.
             # labels = {'country':'Country Name'}# Настраивает метки для наведенных данных.
            )
fig.update_traces( # обновить данные объектов
    textinfo = 'percent+label', # Как подписать данные на диаграмме
    pull = [0, 0.1, 0, 0, 0] # Выделение отдельного кусочка пирога
)
fig.show()
```

```python
# Отверстие в центре
fig = px.pie(data_frame = df_gapminder_2007,
             names = 'continent', # Категории для "кусочков" диаграммы
             title = 'Country Distribution by Continent in 2007',
             hover_data = ['country'], # Включает дополнительные данные (названия стран) при наведении курсора на срезы.
             labels = {'country':'Country Name'}, # Настраивает метки для наведенных данных.
             hole = 0.5
            )
fig.update_traces( # обновить данные объекьлв
    textinfo = 'percent+label', # Как подписать данные на диаграмме
    pull = [0, 0.1, 0, 0, 0] # Выделение отдельного кусочка пирога
)
fig.show()
```

### <font color='rebeccapurple'>7. Box plot</font>

```python
fig = px.box(data_frame = df_tips,
             y = 'total_bill',
             animation_frame= 'day',
             title = 'Distribution of Total Bills')
fig.show()
```

```python
# Добавим визуализацию доверительного интервала
# Доверительный интервал - это отрезок значений, в котором с вероятностью 95% находится значение
fig = px.box(data_frame = df_tips,
             x = 'day',
             y = 'total_bill',
             color = 'smoker',
             notched = True, # включение отображения Д
             points = 'all', # отображение всех точек
             title = 'Distribution of Total Bills')
fig.show()
```

```python
# Другие параметры
fig = px.box(data_frame = df_tips,
             x = 'day',
             y = 'total_bill',
             color = 'smoker',
             notched = True, # включение отображения Д
             # points = 'all', # отображение всех точек
             labels = ({'day':'день недели', 'total_bill':'сумма чека'}),
             boxmode = 'overlay', # Расположение боксплотов относительно друг друга: 'group','overlay'
             title = 'Distribution of Total Bills')
fig.show()
```

```python
# Матрица боксплотов
fig = px.box(data_frame = df_tips,
             x = 'day',
             y = 'total_bill',
             labels = ({'day':'день недели', 'total_bill':'сумма чека'}),
             facet_row = 'smoker',
             facet_col = 'sex',
             title = 'Distribution of Total Bills')
fig.show()
```

### <font color='rebeccapurple'>8. Heatmap</font>

```python
import seaborn as sns
df_flights = sns.load_dataset("flights")
df_flights.head(3)
```

```python
df_flights_pivot = df_flights.pivot_table(index = 'month',
                                         columns = 'year',
                                         values = 'passengers')
df_flights_pivot
```

```python
fig = px.imshow(df_flights_pivot,
                labels = ({'x':'Year', 'y':'Month','color':'Passengers'}),
                title = 'Monthly Passenger Numbers Over Years')
fig.show()
```

```python
# Настроим цветовую шкалу и добавим текст
fig = px.imshow(df_flights_pivot,
                labels = ({'x':'Year', 'y':'Month','color':'Passengers'}),
                aspect = 'auto', # Настройка соотношений сторон ('auto', 'equal', None)
                color_continuous_scale = 'Viridis', # Градиентная раскраска
                text_auto = True, # Отображение значений
                title = 'Monthly Passenger Numbers Over Years')
fig.show()
```

### <font color='rebeccapurple'>9. Sunburst charts</font>
Лучевые диаграммы эффективны для визуализации иерархических структур данных и пропорций внутри них, показывая, как одно кольцо, уровень в иерархии, разбивается на составляющие его части.

```python
fig = px.sunburst(data_frame = df_tips,
                  path = ['smoker','day', 'time', 'sex'], # Уровни иерархии для диаграммы
                  values = 'total_bill', # На основе этого поля определяется размер каждого сегмента
                  color = 'time', # Поле, на основе которого раскрашивается секторы
                  title = 'Distribution of Total Bills by Day, Time, and Sex')
fig.show()
```

```python
# зменим цветовую шкалу и настроим текст внутри графика
fig = px.sunburst(data_frame = df_tips,
                  path = ['day', 'time', 'sex'], # Уровни иерархии для диаграммы
                  values = 'total_bill', # На основе этого поля определяется размер каждого сегмента
                  color = 'total_bill', # Поле, на основе которого раскрашивается секторы
                  color_continuous_scale = 'Blues',
                  title = 'Distribution of Total Bills by Day, Time, and Sex')
fig.update_traces(textinfo = 'label+percent entry')
fig.show()
```

### <font color='rebeccapurple'>10. Настройка графиков в Plotly</font>
<b> update_layout </b> - настраивает оформление графика (рамку, оси, легенду, фон). <br>
<b> update_traces </b> - настраивает сами данные (точки, линии, бары и т.п.).

```python
import plotly.express as px

df_iris = px.data.iris()
fig = px.scatter(data_frame = df_iris,
                 x = 'sepal_length',
                 y = 'sepal_width',
                 color = 'species',
                 title = 'Iris Dataset: Sepal Width vs Sepal Length'
                )
fig.show()
```

##### <font color='rebeccapurple'>10.1. Управление заголовком и названиями осей</font>

```python
fig.update_layout(
    title = 'Iris Flower Measurements',
    xaxis_title = 'Sepal Length (cm)',
    yaxis_title = 'Sepal Width (cm)'
)
fig.show()
```

##### <font color='rebeccapurple'>10.2. Управление легендой</font>
<i> Расположение легенды: </i>
- x = 0 - левая граница области графика
- x = 1 - правая граница
- y = 0 - низ
- y = 1 - верх

```python
fig.update_layout(
    legend_title = 'Species',
    legend = {'x':1,
              'y':1,
              'bordercolor':'black',
              'borderwidth': 1}
)
fig.show()
```

##### <font color='rebeccapurple'>10.3. Аннотации</font>

```python
fig.add_annotation(
    x = 5, # Координаты, на которые должна быть направлена аннотация.
    y = 3.5,
    text = "Outlier", # Текст аннотации
    showarrow = True, # Показывать ли стрелку, указывающую на аннотируемую точку.
    arrowhead = 1, # Стиль наконечника стрелки
    ax = 20, # Положение хвоста стрелки (сдвиг по оси x)
    ay = -30  # Положение хвоста стрелки (сдвиг по оси y)
)
fig.show()
```

##### <font color='rebeccapurple'>10.4. Внешний вид графика</font>

```python
fig.update_layout(
    plot_bgcolor = 'lightgrey',  # Цвет фона области построения
    paper_bgcolor = 'white',     # Цвет фона за пределами области построения
    font = {'family':'Courier New, monospace',
            'size':12,
            'color':"RebeccaPurple"}
)
fig.show()
```

```python
from plotly.subplots import make_subplots
df_Germany = px.data.gapminder().query("country=='Germany'") 
df_Germany['popMillion'] = df_Germany['pop']

fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(
px.bar(data_frame = df_Germany,x = 'year', y = 'popMillion', color_discrete_sequence = ['green']).data[0],
secondary_y=False
)

fig.add_trace(
px.bar(data_frame = df_Germany,x = 'year', y = 'lifeExp', color_discrete_sequence = ['red']).data[0],
secondary_y=True
)
fig.show()
```
