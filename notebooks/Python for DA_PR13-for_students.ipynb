### <font color='rebeccapurple'>5. Столбчатые диаграммы</font>

```python
# зучите количество людей в каждом уровне образования + в разбивке по возрастным группам:
# - 21 - 30
# - 31 - 40
# - 41 - 50
# - 51 - 62
# 1. Переведите тип DataFrame с Pandas в Polars. DataFrame назовите df_salary_polars
import polars as pl
df_salary_polars = pl.from_pandas(df_salary)
df_salary_polars
```

```python
# 2. С помощью условных функций polars добавьте в df_salary_polars столбец age_range
df_salary_polars = df_salary_polars.with_columns(
    pl.when((pl.col('Age') >= 21) & (pl.col('Age') <= 30))
    .then(pl.lit('21 - 30'))
    .when((pl.col('Age') >= 31) & (pl.col('Age') <= 40))
    .then(pl.lit('31 - 40'))
    .when((pl.col('Age') >= 41) & (pl.col('Age') <= 50))
    .then(pl.lit('41 - 50'))
    .when((pl.col('Age') >= 51) & (pl.col('Age') <= 62))
    .then(pl.lit('51 - 62'))
    .otherwise(pl.lit('Unknown'))
    .alias('age_range')
)
df_salary_polars
```

```python
# 3. Переведите датафрейм df_salary_polars из polars в pandas (назовите df_salary)
# спользуйте функцию to_pandas()
df_salary_2 = df_salary_polars.to_pandas()
df_salary_2
```

```python
# 4. Постройте столбчатую диаграмму (количество людей в каждом уровне образования + в разбивке по возрастным группам)
df_salary_grouped = df_salary_2.groupby(['Education Level', 'age_range']).size().reset_index(name='Count')
fig = px.bar(df_salary_grouped, 
             x='Education Level', 
             y='Count',
             color='age_range', 
             color_discrete_sequence = ['green', 'violet','blue', 'orange'],
             title='Count of People by Education Level and Age Range')
fig.show()
```

```python
# справьте предыдущий график на анимированный: 
# Должен меняться Age, а каждый столбик - количество человек с определенным уровнем образования
fig = px.bar(df_salary_2, 
             x='Education Level', 
             color='age_range', 
             title='Count of People by Education Level and Age Range',
             animation_frame='Age')
fig.show()
```
