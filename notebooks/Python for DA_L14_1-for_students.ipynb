##### <font color='rebeccapurple'>1.2. 3D график поверхности</font>
Чтобы создать график поверхности, нам нужна структурированная сетка точек (x, y) и соответствующие значения z, которые определяют высоту поверхности в каждой точке.

```python
import plotly.graph_objects as go
import numpy as np

x = np.linspace(-5, 5, 50) # координаты X
y = np.linspace(-5, 5, 50) # координаты Y
x, y = np.meshgrid(x, y) # структурированная сетка точек
z = np.sin(np.sqrt(x ** 2 + y ** 2)) # значения z, которые определяют высоту поверхности в каждой точке.

fig = go.Figure(
    data = [go.Surface(
        z = z,
        x = x,
        y = y)
           ]
)
fig.update_layout(title = '3D Surface Plot',# заголовок графика
                  width = 500, # ширина графика в пикселях
                  height = 500, # высота графика в пикселях
                 )
fig.show()
```

### <font color='rebeccapurple'>2. Graphic Objects</font>
Graph Objects - это строительные блоки графиков Plotly. Каждый элемент графика (например, линии, маркеры, текст) представлен в виде объекта с собственным набором настраиваемых атрибутов.

```python
import plotly.graph_objects as go
fig = go.Figure() # инициализация нового объекта фигуры
# Добавление новой трассировки к фигуре. 
# go.Scatter используется для создания линейного графика (диаграммы рассеяния в линейном режиме).
fig.add_trace(
    go.Scatter(
        x = [1, 2, 3, 4],
        y = [10, 11, 12, 13],
        mode = 'lines'
    )
)
# Обновление настройки макета фигуры, такие как заголовки и метки осей.
fig.update_layout(
    title = 'Simple Line Chart',
    xaxis_title = 'X Axis',
    yaxis_title = 'Y Axis')
fig.show()
```

```python
fig.update_traces(line = {'color' : 'RoyalBlue', 'width' : 4})#  зменение свойств трассировок, такие как цвет и ширина линии
# Добавление аннотации к графику. Параметры x, y и text задают положение и содержание аннотации.
fig.add_annotation(x = 3, y = 12, text = "Highlight", showarrow = True, arrowhead = 1)
# Дальнейшая настройка макета, например цвет фона графика.
fig.update_layout(plot_bgcolor = 'lightgrey')
fig.show()
```

### <font color='rebeccapurple'>3. Подграфики и множественные оси в Plotly</font>

```python
# Создание фигуры с несколькими подграфиками
from plotly.subplots import make_subplots

fig = make_subplots(rows = 2, cols = 1)

fig.add_trace(go.Scatter(x = [1, 2, 3], y = [4, 5, 6], name = 'Line 1'), row = 1, col = 1)
fig.add_trace(go.Scatter(x = [1, 2, 3], y = [6, 7, 8], name = 'Line 2'), row = 2, col = 1)

fig.show()
```

```python
# Общие оси Y
fig = make_subplots(rows = 1, cols = 4, #shared_yaxes = True
                    )

fig.add_trace(go.Scatter(x = [1, 2, 3], y = [2, 3, 4]), row = 1, col = 1)
fig.add_trace(go.Scatter(x = [20, 30, 40], y = [5, 5, 5]), row = 1, col = 2)
fig.add_trace(go.Scatter(x = [2, 3, 4], y = [600, 700, 800]), row = 1, col = 3)
fig.add_trace(go.Scatter(x = [4000, 5000, 6000], y = [7000, 8000, 9000]), row = 1, col = 4)

fig.update_layout(height = 600, width = 1000, title_text = "Multiple Subplots with Shared Y-Axes")
fig.show()
```

```python
# Две оси Y
fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(go.Scatter(x = [1, 2, 3], y = [2, 3, 4]), secondary_y = False)
fig.add_trace(go.Scatter(x = [1, 2, 3], y = [5, 5, 5]), secondary_y = True)

fig.update_layout(height = 600, width = 1000, title_text = "Multiple Subplots with Shared Y-Axes")
fig.show()
```

```python
# Настройка макетов подграфиков
fig = make_subplots(
    rows = 2, cols = 2, 
    vertical_spacing = 0.1, # Расстояние между подграфиками (от 0 до 1)
    subplot_titles=("Plot 1", "Plot 2", "Plot 3", "Plot 4") # Названия графиков
)

fig.add_trace(go.Scatter(x = [1, 2, 3], y = [2, 3, 4]), row = 1, col = 1)
fig.add_trace(go.Scatter(x = [20, 30, 40], y=[5, 5, 5]), row = 1, col = 2)
fig.add_trace(go.Scatter(x = [2, 3, 4], y = [600, 700, 800]), row = 2, col = 1)
fig.add_trace(go.Scatter(x = [4000, 5000, 6000], y = [7000, 8000, 9000]), row = 2, col = 2)

fig.update_layout(height = 600, width = 1000, title_text = "Customized Subplots Layout")
fig.show()
```

### <font color='rebeccapurple'>4. Взаимодействие с графиками в Plotly</font>

##### <font color='rebeccapurple'>4.1. Масштабирование и панорамирование</font>

```python
fig = go.Figure(data = go.Scatter(x = [1, 2, 3], y = [4, 5, 6]))
fig.update_layout(
    title = 'Zoom and Pan Example',
    # Ползунок диапазона к оси x, позволяющий пользователям выбирать диапазон для увеличения.
    xaxis = {'rangeslider' : {'visible' : True}},
    # Установка этого значения в False позволяет панорамировать вдоль оси y.
    yaxis = {'fixedrange' : False}
)
fig.show()
```

##### <font color='rebeccapurple'>4.2. Всплывающие подсказки</font>

```python
fig = go.Figure(data = go.Scatter(x = [1, 2, 3], y = [4, 5, 6],
                                  mode = 'markers',
                                  # Указывает информацию, которая должна отображаться при наведении курсора на точки. 
                                  # Она может включать координаты данных (x, y), текст и многое другое.
                                  hoverinfo = 'x+y+text',
                                  # Пользовательский текст, который может отображаться во всплывающих подсказках.
                                  text = ['Point 1', 'Point 2', 'Point 3']))
fig.show()
```

##### <font color='rebeccapurple'>4.3. Добавление кнопок и выпадающих меню</font>

```python
fig = go.Figure(data = [go.Scatter(x = [1, 2, 3], y = [4, 5, 6], mode = 'markers')])

fig.update_layout(
    # Настраивает выпадающие меню или кнопки на графике
    updatemenus = [
        {
            # Определяет отдельные кнопки в выпадающем меню, 
            # указывая команду (args), метку 
            # и метод (restyle для обновления атрибутов данных, relayout для атрибутов макета).
            'buttons' : [
                {'args' : ['type', 'scatter'], 
                 'label' : 'Scatter Plot', 
                 'method' : 'restyle'},
                {'args' : ['type', 'bar'], 
                 'label' : 'Bar Chart', 
                 'method' : 'restyle'},
                {'args' : ['type', 'histogram'], 
                 'label' : 'Histogram', 
                 'method' : 'restyle'}
            ],
            'direction' : 'down',
            'showactive' : True
        }
    ]    
)
fig.show()
```

##### <font color='rebeccapurple'>4.4. Параметры конфигурации</font>

```python
fig.show(config = {
    'displayModeBar': True, # Всегда показывать панель режимов
    'scrollZoom': True # Включить масштабирование прокрутки
})
```

### <font color='rebeccapurple'>5. Задача - анализ транзакций покупок кофе</font>

##### <font color='rebeccapurple'>Описание данных</font>
- hour_of_day - час дня
- cash_type - тип оплаты (card)
- money - сумма оплаты
- coffee_name - вид кофе
- Time_of_Day - время суток
- Weekday - день недели
- Month_name - месяц
- Weekdaysort - номер дня недели
- Monthsort - номер месяца
- Date - дата
- Time - время

```python
# Загрузите данные в DataFrame df_coffee
import pandas as pd
import plotly.express as px

df_coffee = pd.read_csv('../data/Coffe_sales.csv')
df_coffee
```

```python
# В df_coffee создайте новый столбец DateTime, объединив столбцы Date и Time
# Тип данных нового столбца - datetime
df_coffee['Date+time'] = pd.to_datetime(df_coffee['Date'] +' '+ df_coffee['Time'])
df_coffee
```

```python
# Для датафрейма df_coffee нарисуйте 3D диаграмму рассеяния с параметрами:
# по оси X - месяц
# по оси Y - время суток
# по оси Z - день недели
# цвет точек зависит от вида кофе
fig = px.scatter_3d(df_coffee, 
                    x = 'Month_name', 
                    y = 'Time_of_Day', 
                    z = 'Weekday', 
                    color = 'coffee_name'
                    
                    )
fig.show()
```

```python
fig = px.histogram(
    df_coffee,
    y='Weekday',
    color='coffee_name',
    category_orders={'Weekday': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']}
)
fig.update_layout(
    title='Coffee Sales by Weekday and Coffee Type',
    xaxis_title='Number of Sales',
    yaxis_title='Weekday',
    legend_title='Type of Coffee',

)
fig.show()
```

```python
fig = px.histogram(
    df_coffee,
    y='Time_of_Day',
    color='coffee_name',
    category_orders={'Time_of_Day': ['Morning', 'Afternoon', 'Night']})
fig.show()
```
