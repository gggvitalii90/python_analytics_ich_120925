```python
# Series
ser = pd.Series([1, 3, 5, np.nan, 6, 8])
print(ser)
# Элемент с индексом 2
print(ser[2])

#Срез элементов с индексами от 2 до 4 включительно
print(ser[2:5])

# По метке (имени индекса)
series_str = pd.Series([10, 20, 30, 40, 50], index = ['Store_1', 'Store_2', 'Store_3', 'Store_4', 'Store_5'])
print(series_str)
print(series_str['Store_3'])
```

```python
# DataFrame
df_from_file = pd.read_csv('titanic.csv')
print(df_from_file.head(5))
print('-----------------------')
# Столбец с названием Age
print(df_from_file.Age)
print(df_from_file['Age'])
print('-----------------------')
#Выбор нескольких столбцов
print(df_from_file[['Name', 'Age', 'Sex']])
print('-----------------------')
# Получение элемента массива. Сначала надо указать столбец, потом строку 
print(df_from_file['Name'][5])
print('-----------------------')
#Получение среза
print(df_from_file[10:20]) #правая граница не включается
print('-----------------------')
#Вывод с шагом
print(df_from_file[1:30:2])

df_from_file[2:3]
```

```python
# loc
# loc - по меткам (названиям)
d = {'one': pd.Series(range(6), index=list('abdefg')),
     'two': pd.Series(range(7), index=list('abcdefg')),
     'three': pd.Series(np.random.normal(size=7), index=list('abcdefg'))}
df = pd.DataFrame(d)
print(df)
print('-----------------------')
# Строка с меткой b
print(df.loc['b'])
print('-----------------------')
# Ячейка на пересечении строки b и столбца one
print(df.loc['b']['one']) # = df['one'][5]
print('-----------------------')
# Ячейки на пересечении строк a, b и столбца one
print(df.loc['a':'b', 'one'])
print('-----------------------')
# Строки a, b и все столбцы
print(df.loc['a':'b', :])
print('-----------------------')
# Все строки и столбец one
print(df.loc[:, 'one'])
```

```python
# iloc
# iloc - по индексам
d = {'one': pd.Series(range(6), index=list('abdefg')),
     'two': pd.Series(range(7), index=list('abcdefg')),
     'three': pd.Series(np.random.normal(size=7), index=list('abcdefg'))}
df = pd.DataFrame(d)
print(df)
print('-----------------------')
# Строка с индексом 2
print(df.iloc[2])
print('-----------------------')
# Строки с индексами 1 и 2
print(df.iloc[1:3])
print('-----------------------')
# Ячейки на пересечении строк с 1 по 2 и столбцов с 0 по 1
print(df.iloc[1:3, 0:2])
```

```python
#Булевская индексация = фильтрация
# По столбцу Age выбираем значения больше 70
print(df_from_file[df_from_file.Age > 70])
# По столбцу Age выбираем значения больше 30 и по столбцу Sex значение male
print(df_from_file[(df_from_file.Age > 30) & (df_from_file.Sex == 'male')])
```

```python
df = pd.read_csv("country_wise_latest.csv")
df
```

```python
# 2) Первые 10 строк и размер датасета

display(df.head(10))
print('Размер датасета:', df.shape)
```

```python
# 3) Все данные по стране Italy
df[df['Country/Region'] == 'Italy']
```

```python
# 4) Первые 5 строк и первые 4 столбца через .iloc
df.iloc[:5, :4]
```

```python
# 5) Через .loc выбрать страны Germany и France и колонки
df.loc[
    df['Country/Region'].isin(['Germany', 'France']),
    ['Country/Region', 'Confirmed', 'Recovered']
]
```

```python
# 6) Все страны, где подтверждённых случаев больше 1 млн
df[df['Confirmed'] > 1_000_000][['Country/Region', 'Confirmed']]
```

```python
# 7) Активных случаев > 50000 и выздоровевших > 0.5 млн

df[(df['Active'] > 50000) & (df['Recovered'] > 500000)]
```

```python
# 8) Найти максимум, среднее и медиану
# страна с максимальным числом случаев
max_country = df.loc[df['Confirmed'].idxmax(), 'Country/Region']

# среднее число активных случаев
avg_active = round(df['Active'].mean(),2)

# медианное значение новых случаев
median_new_cases = df['New cases'].median()

print('Страна с максимальным числом случаев:', max_country)
print('Среднее число активных случаев:', avg_active)
print('Медиана новых случаев:', median_new_cases)
```

```python
# 9) Series с количеством новых выздоровлений по всем странам
new_recovered_series = df.set_index('Country/Region')['New recovered']
new_recovered_series
```

```python
# 10) Новый DataFrame только с нужными колонками

df_small = df[['Country/Region', 'Confirmed', 'Deaths', 'Recovered']]
df_small
```

```python
# 11) DataFrame только для Italy, Spain, France, Germany
df_europe = df_small[df_small['Country/Region'].isin(['Italy', 'Spain', 'France', 'Germany'])]
df_europe
```
