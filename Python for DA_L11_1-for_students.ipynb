### <font color='forestgreen'>5. Контексты и выражения в polars</font>

<b>Выражения</b> относятся к вычислениям или преобразованиям, которые выполняются над столбцами данных, и позволяют применять различные операции над данными для получения новых результатов. Выражения включают математические операции, агрегирование, сравнение, манипуляции со строками и многое другое.

<b>Контекст</b> - это фундаментальное действие, которое вы хотите выполнить над данными.
- <i>Выбор</i>: Выбор столбцов из фрейма данных
- <i>Фильтрация</i>: Уменьшение размера кадра данных путем извлечения строк, удовлетворяющих заданным условиям.
- <i>Группировка/агрегация</i>: Вычисление сводной статистики в подгруппах данных

### <font color='forestgreen'>6. Выбор и фильтрация данных в Polars</font>

##### <font color='forestgreen'>6.1. Индексация</font>

```python
# Вывод первой строки
df[0]
```

```python
# Вывод столбцов fruit и price
df[:,['fruit', "price"]]
```

##### <font color='forestgreen'>6.2. Фильтрация</font>

```python
# filter оставляет только строки, удовлетворяющие условию.
df.filter(df["price"] > 0.3) # df.filter(pl.col('price') > 0.3)
```

##### <font color='forestgreen'>6.3. Выбор</font>

```python
# Метод select
# select выбирает столбцы или вычисляет выражения, но не фильтрует строки по умолчанию.
df.select(['fruit', 'price'])
# df['fruit', 'price']
```

```python
# Метод with_columns
df_1 = df.with_columns([ #Этот метод добавляет новые столбцы в DataFrame
    (pl.col("price") * pl.col("quantity")).alias("total")
])
df_1
```

```python
df_2 = df.select([pl.col('fruit'), pl.col('price') * pl.col('quantity')])
df_2
```

### <font color='coral'>7. Задача</font>

```python
# Загрузка данных
# Используя Polars, прочитать файл sales.json в DataFrame df_sales
df_sales = pl.read_json('sales.json')
df_sales
```

```python
# Фильтрация данных
# Оставить только продажи, где price > 50.
# Отфильтрованный dataframe назовите df_sales_filtered
df_sales_filtered = df_sales.filter(pl.col('price') > 50)
df_sales_filtered
```

```python
# Выбор колонок
# Создать новый DataFrame df_sales_selected, оставив только колонки: item, price, quantity.
df_sales_selected = df_sales.select(['item', 'price', 'quantity'])
df_sales_selected
```

```python
# Добавление новой колонки
# С помощью with_columns добавить колонку total_value, которая равна price * quantity.
# Получившийся DataFrame назовите df_sales_final
df_sales_final = df_sales.with_columns(
    (pl.col('price') * pl.col('quantity')).alias('total_value')
)
df_sales_final
```
